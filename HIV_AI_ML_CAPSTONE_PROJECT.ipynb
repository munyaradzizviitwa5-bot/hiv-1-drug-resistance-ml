```python
import pandas as pd

# Load the HIV dataset
df = pd.read_csv(file_path, sep="\t")

# Display the first 5 rows
df.head()
```

```python
# STEP 11: Inspect dataset structure, data types and missing values

print("Dataset shape:")
print(df.shape)

print("\n" + "="*60)
print("COLUMN NAMES")
print("="*60)

print(df.columns.tolist())

print("\n" + "="*60)
print("DATA TYPES")
print("="*60)

print(df.dtypes)

print("\n" + "="*60)
print("MISSING VALUES")
print("="*60)

missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

print(missing_values)

print("\n" + "="*60)
print("SUMMARY")
print("="*60)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Columns containing missing values:", len(missing_values))
```

```python
# STEP 12: Inspect HIV drug-resistance target variables

# Drug resistance columns
drug_columns = [
    'DRV', 'TPV', 'ATV', 'LPV', 'FPV',
    'IDV', 'SQV', 'NFV'
]

print("DRUG RESISTANCE COLUMNS")
print("=" * 60)

for col in drug_columns:
    print(f"\n{col}")
    print("Data type:", df[col].dtype)
    print("Number of non-missing values:", df[col].notna().sum())
    print("Number of missing values:", df[col].isna().sum())
    print("Unique values:")
    print(df[col].value_counts(dropna=False).head(20))
```

```python
# STEP 13: FIND AND INSPECT HIV-1 GENOMIC SEQUENCE COLUMNS

print("TOTAL COLUMNS IN DATASET:", len(df.columns))
print("\nALL COLUMN NAMES:")
for i, col in enumerate(df.columns):
    print(i, ":", col)

# Look for columns that may contain genomic sequences
sequence_keywords = [
    "sequence", "seq", "genome", "genomic",
    "protein", "protease", "reverse", "transcriptase",
    "pol", "gene", "nucleotide", "amino"
]

possible_sequence_cols = [
    col for col in df.columns
    if any(keyword in col.lower() for keyword in sequence_keywords)
]

print("\n========================================")
print("POSSIBLE SEQUENCE / GENOMIC COLUMNS")
print("========================================")

if possible_sequence_cols:
    for col in possible_sequence_cols:
        print(col)
else:
    print("No obvious sequence column found.")

# Display datatype and missing values for possible sequence columns
if possible_sequence_cols:
    print("\nSEQUENCE COLUMN INFORMATION:")
    for col in possible_sequence_cols:
        print("\nColumn:", col)
        print("Data type:", df[col].dtype)
        print("Non-missing:", df[col].notna().sum())
        print("Missing:", df[col].isna().sum())

        # Show a few actual values
        print("Example values:")
        print(df[col].dropna().head(3).to_string(index=False))
```

```python
# STEP 14: Identify possible HIV drug-resistance / target columns

print("=" * 60)
print("SEARCHING FOR DRUG-RESISTANCE / TARGET COLUMNS")
print("=" * 60)

# Search column names for resistance-related terms
keywords = [
    "drug", "resistance", "resistant", "susceptibility",
    "phenotype", "phenotypic", "class", "score"
]

possible_target_columns = []

for col in df.columns:
    col_lower = str(col).lower()

    if any(keyword in col_lower for keyword in keywords):
        possible_target_columns.append(col)

print("\nPossible target/resistance columns:")
print(possible_target_columns)

print("\nNumber of possible target columns:", len(possible_target_columns))

# Show data types of the possible target columns
if len(possible_target_columns) > 0:
    print("\nData types:")
    print(df[possible_target_columns].dtypes)

# Show first 5 rows of possible target columns
if len(possible_target_columns) > 0:
    print("\nFirst 5 rows:")
    display(df[possible_target_columns].head())
```

```python
# STEP 37: Inspect HIV-1 protease position features (P1-P99)

position_cols = [f'P{i}' for i in range(1, 100)]

print("Number of position columns:", len(position_cols))

print("\n" + "="*70)
print("MISSING VALUES IN P1-P99")
print("="*70)

position_missing = df[position_cols].isna().sum()

print(position_missing)

print("\n" + "="*70)
print("PERCENTAGE MISSING IN P1-P99")
print("="*70)

position_missing_pct = (df[position_cols].isna().mean() * 100).round(2)

print(position_missing_pct)

print("\n" + "="*70)
print("SAMPLE VALUES FROM SELECTED POSITIONS")
print("="*70)

for col in ['P1', 'P10', 'P20', 'P30', 'P40', 'P50', 'P60', 'P70', 'P80', 'P90', 'P99']:
    print("\n", col)
    print(df[col].value_counts(dropna=False).head(15))
```

```python
# STEP 38: Inspect special symbols and amino-acid representations

position_cols = [f'P{i}' for i in range(1, 100)]

# Combine all P1-P99 values into one Series
all_position_values = df[position_cols].stack().value_counts()

print("="*70)
print("MOST COMMON VALUES ACROSS P1-P99")
print("="*70)
print(all_position_values.head(50))

print("\n" + "="*70)
print("SPECIAL SYMBOL COUNTS ACROSS P1-P99")
print("="*70)

special_symbols = ['-', '.', 'X', '*']

for symbol in special_symbols:
    count = (df[position_cols] == symbol).sum().sum()
    print(f"{symbol!r}: {count}")

print("\n" + "="*70)
print("UNIQUE VALUES ACROSS P1-P99")
print("="*70)

unique_values = set()

for col in position_cols:
    unique_values.update(df[col].dropna().astype(str).unique())

print("Total unique representations:", len(unique_values))
print("Unique representations:")
print(sorted(unique_values))
```

```python
# STEP 39: Compare position features with the mutation-list information

position_cols = [f'P{i}' for i in range(1, 100)]

print("="*70)
print("COMPMUTLIST INFORMATION")
print("="*70)

print("Total rows:", len(df))
print("Non-missing CompMutList:", df['CompMutList'].notna().sum())
print("Missing CompMutList:", df['CompMutList'].isna().sum())
print("Unique CompMutList values:", df['CompMutList'].nunique())

print("\nMost common CompMutList values:")
print(df['CompMutList'].value_counts(dropna=False).head(20))


print("\n" + "="*70)
print("NUMBER OF MUTATION-RELATED POSITIONS PER SAMPLE")
print("="*70)

# Treat '-', '.', and X as non-standard/unknown states for this inspection.
# Count positions that contain something other than '-'.
mutation_like_count = (
    ~df[position_cols].isin(['-'])
).sum(axis=1)

print(mutation_like_count.describe())

print("\nDistribution:")
print(mutation_like_count.value_counts().sort_index().head(30))


print("\n" + "="*70)
print("SAMPLES WITH FEW NON-'−' POSITION VALUES")
print("="*70)

print(
    df[['SeqID', 'CompMutList'] + position_cols]
    .assign(NonDash_Positions=mutation_like_count)
    [['SeqID', 'CompMutList', 'NonDash_Positions']]
    .sort_values('NonDash_Positions')
    .head(10)
)
```

```python
# STEP 40: Investigate samples with no mutation information

position_cols = [f'P{i}' for i in range(1, 100)]

# Identify rows where every P1-P99 position is '-'
no_mutation_mask = (df[position_cols] == '-').all(axis=1)

print("="*70)
print("NUMBER OF SAMPLES WITH ALL P1-P99 = '-'")
print("="*70)

print("Count:", no_mutation_mask.sum())


print("\n" + "="*70)
print("THESE SAMPLES")
print("="*70)

cols_to_show = [
    'SeqID',
    'PtID',
    'Subtype',
    'Method',
    'RefID',
    'Type',
    'CompMutList',
    'num_drug_results'
]

print(df.loc[no_mutation_mask, cols_to_show].to_string(index=False))


print("\n" + "="*70)
print("DRUG RESISTANCE MEASUREMENTS FOR THESE SAMPLES")
print("="*70)

drug_cols = [
    'FPV', 'ATV', 'IDV', 'LPV',
    'NFV', 'SQV', 'TPV', 'DRV'
]

print(
    df.loc[
        no_mutation_mask,
        drug_cols
    ].to_string(index=False)
)
```

```python
# ============================================================
# STEP 47: CREATE BINARY MUTATION FEATURES
# ============================================================

position_cols = [f'P{i}' for i in range(1, 100)]

# Make a copy so the original ml_df remains unchanged
ml_features = ml_df.copy()

# Create binary mutation indicators
for col in position_cols:
    ml_features[f'{col}_mutated'] = (
        (ml_features[col].notna()) &
        (~ml_features[col].isin(['-', '.', 'X', '*']))
    ).astype(int)

# Count total mutated positions for every sample
mutation_binary_cols = [f'{col}_mutated' for col in position_cols]

ml_features['Total_Mutated_Positions'] = (
    ml_features[mutation_binary_cols].sum(axis=1)
)

print("=" * 70)
print("STEP 47: BINARY MUTATION FEATURES")
print("=" * 70)

print("\nOriginal number of columns:", ml_df.shape[1])
print("New number of columns:", ml_features.shape[1])

print("\nNumber of binary mutation features:",
      len(mutation_binary_cols))

print("\nDistribution of Total_Mutated_Positions:")
print(
    ml_features['Total_Mutated_Positions']
    .value_counts()
    .sort_index()
)

print("\nSummary:")
print(
    ml_features['Total_Mutated_Positions']
    .describe()
)

print("\nFirst 10 mutation feature columns:")
print(mutation_binary_cols[:10])

print("\nFirst 5 rows of mutation features:")
print(
    ml_features[
        mutation_binary_cols[:10] +
        ['Total_Mutated_Positions']
    ].head()
)
```

```python
# ============================================================
# STEP 48: CREATE AMINO-ACID-SPECIFIC MUTATION FEATURES
# ============================================================

import pandas as pd
import numpy as np

position_cols = [f'P{i}' for i in range(1, 100)]

# Work from the original mutation columns
mutation_data = ml_df[position_cols].copy()

# Special/non-informative representations
special_values = ['-', '.', 'X', '*']

# Function to extract amino-acid characters from each representation
def extract_amino_acids(value):
    if pd.isna(value):
        return []

    value = str(value).strip()

    if value in special_values:
        return []

    # Keep only standard amino-acid letters
    amino_acids = set("ACDEFGHIKLMNPQRSTVWY")

    return [aa for aa in value if aa in amino_acids]


# Create amino-acid-specific binary features
feature_dict = {}

for col in position_cols:
    for aa in "ACDEFGHIKLMNPQRSTVWY":
        feature_dict[f'{col}_{aa}'] = mutation_data[col].apply(
            lambda x: int(aa in extract_amino_acids(x))
        )

amino_features = pd.DataFrame(feature_dict, index=ml_features.index)

# Combine with existing dataset
ml_features_v2 = pd.concat(
    [ml_features, amino_features],
    axis=1
)

print("=" * 70)
print("STEP 48: AMINO-ACID-SPECIFIC MUTATION FEATURES")
print("=" * 70)

print("\nNumber of amino-acid-specific features:",
      amino_features.shape[1])

print("\nTotal columns in new ML dataframe:",
      ml_features_v2.shape[1])

print("\nShape of amino-acid feature matrix:")
print(amino_features.shape)

print("\nNumber of non-zero features:")
print((amino_features.sum(axis=0) > 0).sum())

print("\nMost frequent amino-acid features:")
print(
    amino_features.sum(axis=0)
    .sort_values(ascending=False)
    .head(20)
)

print("\nSample of generated features:")
print(amino_features.iloc[:5, :20])
```

```python
# ============================================================
# STEP 49: REMOVE ZERO-VARIANCE MUTATION FEATURES
# ============================================================

# Identify amino-acid features that contain at least one mutation
nonzero_features = [
    col for col in amino_features.columns
    if amino_features[col].sum() > 0
]

# Identify features that occur in at least 5 samples
useful_features = [
    col for col in nonzero_features
    if amino_features[col].sum() >= 5
]

# Create cleaned feature matrix
X_mutations = amino_features[useful_features].copy()

print("=" * 70)
print("STEP 49: MUTATION FEATURE QUALITY CHECK")
print("=" * 70)

print("\nOriginal possible features:", amino_features.shape[1])

print("Features occurring at least once:",
      len(nonzero_features))

print("Features occurring in >= 5 samples:",
      len(useful_features))

print("\nFeatures removed because they occur in < 5 samples:",
      len(nonzero_features) - len(useful_features))

print("\nFinal mutation feature matrix:")
print("Rows:", X_mutations.shape[0])
print("Columns:", X_mutations.shape[1])

print("\nMissing values:",
      X_mutations.isna().sum().sum())

print("\nMinimum feature frequency:",
      X_mutations.sum(axis=0).min())

print("\nMaximum feature frequency:",
      X_mutations.sum(axis=0).max())

print("\nFirst 20 retained features:")
print(useful_features[:20])
```

```python
# ============================================================
# STEP 50: INSPECT DRUG RESISTANCE CLASSIFICATION THRESHOLDS
# ============================================================

drug_cols = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

print("=" * 70)
print("DRUG RESISTANCE THRESHOLD INSPECTION")
print("=" * 70)

for drug in drug_cols:

    values = ml_df[drug].dropna()

    print("\n" + "=" * 60)
    print(f"DRUG: {drug}")
    print("=" * 60)

    print(f"Available measurements: {len(values)}")

    # Important fold-change ranges
    thresholds = [1, 2, 3, 4, 5, 10, 20, 50]

    print("\nNumber and percentage of samples at or above threshold:")

    for threshold in thresholds:

        count = (values >= threshold).sum()
        percentage = count / len(values) * 100

        print(
            f">= {threshold:<3}: "
            f"{count:4d} samples "
            f"({percentage:6.2f}%)"
        )

    print("\nSelected percentiles:")

    print(
        values.quantile(
            [0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
        )
    )
```

```python
# ============================================================
# STEP 51: CREATE CANDIDATE BINARY RESISTANCE LABELS
# ============================================================

drug_columns = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# Candidate threshold
THRESHOLD = 2.0

print("=" * 70)
print("STEP 51: CANDIDATE BINARY RESISTANCE LABELS")
print("=" * 70)

for drug in drug_columns:

    # Only samples with an available phenotype
    available = ml_features[drug].notna()

    # Create binary label:
    # 0 = below threshold
    # 1 = at or above threshold
    label = (ml_features.loc[available, drug] >= THRESHOLD).astype(int)

    susceptible = (label == 0).sum()
    resistant = (label == 1).sum()
    total = len(label)

    print(f"\n{drug}")
    print("-" * 40)
    print(f"Available samples : {total}")
    print(f"Class 0           : {susceptible} ({susceptible/total*100:.2f}%)")
    print(f"Class 1           : {resistant} ({resistant/total*100:.2f}%)")
```

```python
# ============================================================
# STEP 52: CREATE FINAL BINARY RESISTANCE LABELS
# ============================================================

drug_columns = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

THRESHOLD = 2.0

# Copy the current ML dataframe
ml_binary = ml_features.copy()

print("=" * 70)
print("STEP 52: FINAL BINARY RESISTANCE LABELS")
print("=" * 70)

for drug in drug_columns:

    label_name = drug + '_Resistance'

    # 0 = susceptible / lower resistance
    # 1 = resistant / higher resistance
    ml_binary[label_name] = ml_binary[drug].apply(
        lambda x: 1 if pd.notna(x) and x >= THRESHOLD
        else (0 if pd.notna(x) else np.nan)
    )

    print(f"\n{drug}")
    print("-" * 40)
    print(ml_binary[label_name].value_counts(dropna=False).sort_index())

print("\n" + "=" * 70)
print("NEW DATASET SHAPE")
print("=" * 70)

print(ml_binary.shape)

print("\nResistance label columns:")
print([col for col in ml_binary.columns if col.endswith('_Resistance')])
```

```python
# ============================================================
# STEP 53B: RECONSTRUCT QUALITY-CONTROLLED AA FEATURES
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 53B: RECONSTRUCT AMINO-ACID-SPECIFIC FEATURES")
print("=" * 70)

# Original mutation-position columns
position_columns = [f'P{i}' for i in range(1, 100)]

# Standard 20 amino acids
amino_acids = list("ACDEFGHIKLMNPQRSTVWY")

aa_features = pd.DataFrame(index=ml_binary.index)

# Create amino-acid-specific binary features
for pos in range(1, 100):

    col = f'P{pos}'

    for aa in amino_acids:

        feature_name = f'P{pos}_{aa}'

        aa_features[feature_name] = (
            ml_binary[col]
            .fillna('-')
            .astype(str)
            .str.contains(aa, regex=False)
            .astype(int)
        )

print("\nOriginal possible AA features:", aa_features.shape[1])

# Count how many samples contain each feature
feature_frequency = aa_features.sum(axis=0)

# Retain features occurring in at least 5 samples
retained_aa_features = feature_frequency[
    feature_frequency >= 5
].index.tolist()

# Final feature matrix
X_mutations = aa_features[retained_aa_features].copy()

print("\n" + "=" * 70)
print("QUALITY CONTROL")
print("=" * 70)

print("Features occurring at least once:",
      (feature_frequency > 0).sum())

print("Features occurring in >= 5 samples:",
      len(retained_aa_features))

print("Features removed:",
      aa_features.shape[1] - len(retained_aa_features))

print("\nFinal mutation feature matrix:")
print("Rows:", X_mutations.shape[0])
print("Columns:", X_mutations.shape[1])

print("\nMissing values:",
      X_mutations.isna().sum().sum())

print("\nFirst 20 retained features:")
print(retained_aa_features[:20])

print("\nMost frequent retained features:")
print(
    feature_frequency[
        retained_aa_features
    ].sort_values(ascending=False).head(20)
)
```

```python
# ============================================================
# STEP 54: MUTATION / TARGET LEAKAGE CHECK
# ============================================================

print("=" * 70)
print("STEP 54: MUTATION / TARGET LEAKAGE CHECK")
print("=" * 70)

# Final mutation feature list
mutation_features = retained_aa_features

# Drug measurements
drug_columns = [
    'FPV', 'ATV', 'IDV', 'LPV',
    'NFV', 'SQV', 'TPV', 'DRV'
]

# Binary resistance targets
target_columns = [
    'FPV_Resistance',
    'ATV_Resistance',
    'IDV_Resistance',
    'LPV_Resistance',
    'NFV_Resistance',
    'SQV_Resistance',
    'TPV_Resistance',
    'DRV_Resistance'
]

# ------------------------------------------------------------
# 1. Check whether drug measurements are in mutation features
# ------------------------------------------------------------

drug_leakage = [
    col for col in mutation_features
    if col in drug_columns
]

# ------------------------------------------------------------
# 2. Check whether resistance labels are in mutation features
# ------------------------------------------------------------

target_leakage = [
    col for col in mutation_features
    if col in target_columns
]

# ------------------------------------------------------------
# 3. Check feature naming
# ------------------------------------------------------------

invalid_features = [
    col for col in mutation_features
    if not (
        col.startswith('P')
        and '_' in col
    )
]

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\nNumber of mutation features:", len(mutation_features))

print("\nDrug measurement leakage:")
print(drug_leakage)

print("\nResistance-label leakage:")
print(target_leakage)

print("\nInvalid feature names:")
print(invalid_features)

print("\n" + "=" * 70)

if (
    len(drug_leakage) == 0
    and len(target_leakage) == 0
    and len(invalid_features) == 0
):
    print("LEAKAGE CHECK PASSED")
    print("No obvious target leakage detected.")
else:
    print("LEAKAGE CHECK REQUIRES REVIEW")

print("=" * 70)
```

```python
print("=" * 70)
print("STEP 55F: ANALYZE MUTATION PATTERNS")
print("=" * 70)

# Show mutations containing multiple amino-acid letters after the position
complex_mutations = [
    m for m in mutation_features
    if len(m) > 3
]

print("Total mutation tokens:", len(mutation_features))
print("Tokens longer than 3 characters:", len(complex_mutations))

print("\nFirst 100 complex mutation tokens:")
print(complex_mutations[:100])

# Show the most common mutation tokens
from collections import Counter

mutation_counts = Counter()

for value in df["CompMutList"].dropna():
    mutations = [m.strip() for m in str(value).split(",") if m.strip()]
    mutation_counts.update(mutations)

print("\nMost common 30 mutation tokens:")
for mutation, count in mutation_counts.most_common(30):
    print(f"{mutation}: {count}")
```

```python
print("=" * 70)
print("STEP 55G: BUILD MUTATION FEATURE MATRIX")
print("=" * 70)

# Use every observed mutation token as a binary feature
mutation_features = sorted(set(
    mutation
    for value in df["CompMutList"].dropna()
    for mutation in str(value).split(",")
    if mutation.strip()
))

# Create binary mutation matrix
mutation_matrix = pd.DataFrame(
    0,
    index=df.index,
    columns=mutation_features,
    dtype="int8"
)

# Fill matrix
for idx, value in df["CompMutList"].items():
    if pd.isna(value):
        continue

    mutations = [m.strip() for m in str(value).split(",") if m.strip()]

    for mutation in mutations:
        if mutation in mutation_matrix.columns:
            mutation_matrix.loc[idx, mutation] = 1

print("Number of mutation features:", len(mutation_features))
print("Mutation matrix shape:", mutation_matrix.shape)

print("\nTotal mutation entries:")
print(mutation_matrix.sum().sum())

print("\nFirst 10 mutation features:")
print(mutation_features[:10])

print("\nMutation matrix preview:")
print(mutation_matrix.iloc[:5, :10])
```

```python
print("=" * 70)
print("STEP 55H: CLEAN MUTATION FEATURE MATRIX")
print("=" * 70)

# Re-extract mutations and remove leading/trailing spaces
all_mutations = []

for value in df["CompMutList"].dropna():
    mutations = [
        m.strip()
        for m in str(value).split(",")
        if m.strip()
    ]
    all_mutations.extend(mutations)

# Unique, sorted mutation features
mutation_features = sorted(set(all_mutations))

# Rebuild binary mutation matrix
mutation_matrix = pd.DataFrame(
    0,
    index=df.index,
    columns=mutation_features,
    dtype="int8"
)

# Populate matrix
for idx, value in df["CompMutList"].items():
    if pd.isna(value):
        continue

    mutations = [
        m.strip()
        for m in str(value).split(",")
        if m.strip()
    ]

    for mutation in mutations:
        mutation_matrix.loc[idx, mutation] = 1

print("Number of clean mutation features:", len(mutation_features))
print("Mutation matrix shape:", mutation_matrix.shape)

print("\nFirst 20 feature names:")
print(mutation_features[:20])

print("\nTotal mutation entries:")
print(mutation_matrix.sum().sum())

print("\nMissing values:")
print(mutation_matrix.isna().sum().sum())

print("\nUnique values:")
print(sorted(mutation_matrix.stack().unique()))

print("\nMatrix preview:")
print(mutation_matrix.iloc[:5, :10])
```

```python
print("=" * 70)
print("STEP 56: MUTATION FEATURE QUALITY CHECK")
print("=" * 70)

# Number of samples containing each mutation
mutation_frequency = mutation_matrix.sum(axis=0)

print("Total mutation features:", mutation_matrix.shape[1])

print("\nMutation frequency summary:")
print(mutation_frequency.describe())

print("\nFeatures occurring in only 1 sample:")
print((mutation_frequency == 1).sum())

print("\nFeatures occurring in fewer than 5 samples:")
print((mutation_frequency < 5).sum())

print("\nFeatures occurring in at least 5 samples:")
print((mutation_frequency >= 5).sum())

print("\nTop 30 most frequent mutations:")
print(mutation_frequency.sort_values(ascending=False).head(30))
```

```python
print("=" * 70)
print("STEP 57: FILTER RARE MUTATION FEATURES")
print("=" * 70)

# Minimum number of samples in which a mutation must occur
MIN_MUTATION_FREQUENCY = 5

# Calculate frequency of every mutation
mutation_frequency = mutation_matrix.sum(axis=0)

# Keep mutations occurring in at least 5 samples
selected_mutation_features = mutation_frequency[
    mutation_frequency >= MIN_MUTATION_FREQUENCY
].index.tolist()

# Create filtered mutation matrix
mutation_matrix_filtered = mutation_matrix[
    selected_mutation_features
].copy()

print("Original mutation features:", mutation_matrix.shape[1])
print("Minimum frequency threshold:", MIN_MUTATION_FREQUENCY)
print("Removed rare features:",
      mutation_matrix.shape[1] - mutation_matrix_filtered.shape[1])
print("Remaining mutation features:",
      mutation_matrix_filtered.shape[1])

print("\nFiltered mutation matrix shape:")
print(mutation_matrix_filtered.shape)

print("\nMissing values:")
print(mutation_matrix_filtered.isna().sum().sum())

print("\nUnique values:")
print(sorted(mutation_matrix_filtered.stack().unique()))

print("\nFirst 20 retained mutation features:")
print(selected_mutation_features[:20])
```

```python
print("=" * 70)
print("STEP 58: MUTATION FEATURE VARIANCE CHECK")
print("=" * 70)

# Calculate how many samples contain each mutation
mutation_counts = mutation_matrix_filtered.sum(axis=0)

# Calculate prevalence of each mutation
mutation_prevalence = mutation_counts / len(mutation_matrix_filtered)

# Calculate variance of each binary feature
mutation_variance = mutation_matrix_filtered.var(axis=0)

print("Number of mutation features:", mutation_matrix_filtered.shape[1])

print("\nVariance summary:")
print(mutation_variance.describe())

print("\nFeatures with zero variance:")
print((mutation_variance == 0).sum())

print("\nFeatures with variance > 0:")
print((mutation_variance > 0).sum())

print("\nMost prevalent mutations:")
print(
    mutation_prevalence
    .sort_values(ascending=False)
    .head(20)
)
```

```python
print("=" * 70)
print("STEP 59: PREPARE MUTATION MATRIX FOR MODELING")
print("=" * 70)

# Confirm the filtered mutation matrix
X_mut = mutation_matrix_filtered.copy()

print("Mutation matrix shape:")
print(X_mut.shape)

print("\nNumber of features:")
print(X_mut.shape[1])

print("\nNumber of samples:")
print(X_mut.shape[0])

print("\nData types:")
print(X_mut.dtypes.value_counts())

print("\nMissing values:")
print(X_mut.isna().sum().sum())

print("\nUnique values:")
print(sorted(X_mut.stack().unique().tolist()))

print("\nTotal mutation entries:")
print(int(X_mut.sum().sum()))

print("\nSTEP 59 COMPLETE")
```

```python
print("=" * 70)
print("STEP 61: RESTORE FILTERED MUTATION MATRIX")
print("=" * 70)

# Recreate the filtered matrix using the frequency threshold of 5
mutation_frequency = mutation_matrix.sum(axis=0)

retained_features = mutation_frequency[mutation_frequency >= 5].index.tolist()

mutation_matrix_filtered = mutation_matrix[retained_features].copy()

print("\nOriginal mutation matrix shape:")
print(mutation_matrix.shape)

print("\nNumber of retained features:")
print(len(retained_features))

print("\nFiltered mutation matrix shape:")
print(mutation_matrix_filtered.shape)

print("\nMissing values:")
print(mutation_matrix_filtered.isna().sum().sum())

print("\nUnique values:")
print(sorted(mutation_matrix_filtered.values.flatten().tolist())[:10])

print("\nIndex unique:")
print(mutation_matrix_filtered.index.is_unique)

print("\nFirst 20 retained features:")
print(retained_features[:20])

print("\n" + "=" * 70)
print("STEP 61 COMPLETE")
print("=" * 70)
```

```python
print("=" * 70)
print("STEP 72: REBUILD FILTERED MUTATION FEATURE MATRIX")
print("=" * 70)

import re
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# 1. Check required column
# ---------------------------------------------------------
if "CompMutList" not in df.columns:
    raise ValueError("CompMutList column is missing.")

# ---------------------------------------------------------
# 2. Extract mutation tokens
# ---------------------------------------------------------
mutation_tokens = set()

for value in df["CompMutList"].dropna():
    if isinstance(value, str):
        tokens = value.split(",")

        for token in tokens:
            token = token.strip()

            if token:
                mutation_tokens.add(token)

print("\nTotal unique mutation tokens found:")
print(len(mutation_tokens))

# ---------------------------------------------------------
# 3. Clean mutation names
# ---------------------------------------------------------
pattern = re.compile(r"^[A-Z]\d+[A-Z*]+$")

valid_mutations = sorted([
    token for token in mutation_tokens
    if pattern.match(token)
])

print("\nValid mutation features:")
print(len(valid_mutations))

# ---------------------------------------------------------
# 4. Build complete mutation matrix
# ---------------------------------------------------------
mutation_matrix = pd.DataFrame(
    0,
    index=df.index,
    columns=valid_mutations,
    dtype=np.int8
)

# ---------------------------------------------------------
# 5. Populate matrix
# ---------------------------------------------------------
for idx, value in df["CompMutList"].items():

    if pd.isna(value):
        continue

    tokens = value.split(",")

    for token in tokens:

        token = token.strip()

        if token in mutation_matrix.columns:
            mutation_matrix.loc[idx, token] = 1

print("\nComplete mutation matrix shape:")
print(mutation_matrix.shape)

# ---------------------------------------------------------
# 6. Restore frequency filter >= 5 samples
# ---------------------------------------------------------
mutation_frequency = mutation_matrix.sum(axis=0)

retained_features = mutation_frequency[
    mutation_frequency >= 5
].index.tolist()

mutation_matrix_filtered = mutation_matrix[
    retained_features
].copy()

print("\nMinimum frequency threshold:")
print(5)

print("\nRetained mutation features:")
print(len(retained_features))

print("\nFiltered mutation matrix shape:")
print(mutation_matrix_filtered.shape)

print("\nMissing values:")
print(mutation_matrix_filtered.isna().sum().sum())

print("\nUnique values:")
print(np.unique(mutation_matrix_filtered.values))

print("\nFirst 20 retained features:")
print(retained_features[:20])

print("\n" + "=" * 70)
print("STEP 72 COMPLETE")
print("=" * 70)
```

```python
print("=" * 70)
print("STEP 85: DEFINE DRUG-SPECIFIC RESISTANCE CUTOFFS")
print("=" * 70)

# Documented PhenoSense / clinical lower-cutoff strategy
resistance_cutoffs = {
    "FPV": 4.0,
    "ATV": 5.2,
    "IDV": 10.0,
    "LPV": 9.0,
    "NFV": 3.6,
    "SQV": 2.3,
    "TPV": 2.0,
    "DRV": 10.0
}

print("\nDrug-specific resistance cutoffs:")

for drug, cutoff in resistance_cutoffs.items():
    print(f"{drug}: {cutoff}-fold")

print("\nLabeling rule:")
print("Measurement < cutoff  -> Susceptible (0)")
print("Measurement >= cutoff -> Resistant (1)")

print("\nCutoff dictionary:")
print(resistance_cutoffs)

print("\nNumber of drugs:")
print(len(resistance_cutoffs))

print("\nAll required drugs covered:")
print(set(drug_columns) == set(resistance_cutoffs.keys()))

print("\n" + "=" * 70)
print("STEP 85 COMPLETE")
print("=" * 70)
```

```python
print("=" * 70)
print("STEP 86: CREATE BINARY DRUG RESISTANCE LABELS")
print("=" * 70)

# Create a copy of the cleaned PhenoSense dataset
df_resistance = df_pheno_clean.copy()

# Create binary resistance labels
for drug in drug_columns:

    label_column = drug + "_Resistance"

    df_resistance[label_column] = df_resistance[drug].apply(
        lambda x: (
            np.nan
            if pd.isna(x)
            else int(x >= resistance_cutoffs[drug])
        )
    )

print("\nResistance label columns created:")

resistance_columns = [
    drug + "_Resistance"
    for drug in drug_columns
]

print(resistance_columns)

print("\n" + "-" * 70)
print("RESISTANCE LABEL SUMMARY")
print("-" * 70)

for drug in drug_columns:

    label_column = drug + "_Resistance"

    counts = df_resistance[label_column].value_counts(
        dropna=False
    ).sort_index()

    print("\n" + drug)
    print("-" * 40)
    print("Cutoff:", resistance_cutoffs[drug])
    print("Total samples:", len(df_resistance))
    print("Susceptible (0):", counts.get(0, 0))
    print("Resistant (1):", counts.get(1, 0))
    print("Missing:", df_resistance[label_column].isna().sum())

    available = df_resistance[label_column].notna().sum()

    if available > 0:
        resistant = (df_resistance[label_column] == 1).sum()
        susceptible = (df_resistance[label_column] == 0).sum()

        print(
            "Resistant percentage:",
            round((resistant / available) * 100, 2),
            "%"
        )

        print(
            "Susceptible percentage:",
            round((susceptible / available) * 100, 2),
            "%"
        )

print("\n" + "-" * 70)
print("CHECK LABEL VALUES")
print("-" * 70)

for column in resistance_columns:
    print(
        column,
        "->",
        df_resistance[column].dropna().unique()
    )

print("\nDataset shape:")
print(df_resistance.shape)

print("\n" + "=" * 70)
print("STEP 86 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 89: CREATE STRATIFIED TRAIN/TEST SPLITS
# ==============================================================

from sklearn.model_selection import train_test_split

print("=" * 70)
print("STEP 89: CREATE STRATIFIED TRAIN/TEST SPLITS")
print("=" * 70)

# Reproducibility
RANDOM_STATE = 42

# Test set = 20%, training set = 80%
TEST_SIZE = 0.20

train_test_datasets = {}

for drug in ml_datasets:

    X = ml_datasets[drug]['X']
    y = ml_datasets[drug]['y']

    # Stratified split preserves class proportions
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE,
        stratify=y
    )

    train_test_datasets[drug] = {
        'X_train': X_train,
        'X_test': X_test,
        'y_train': y_train,
        'y_test': y_test
    }

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    print("Original samples:", len(y))
    print("Training samples:", len(y_train))
    print("Testing samples:", len(y_test))

    print("\nTraining class distribution:")
    print(
        y_train.value_counts()
        .sort_index()
    )

    print("\nTesting class distribution:")
    print(
        y_test.value_counts()
        .sort_index()
    )

    print("\nTraining class percentages:")
    print(
        (y_train.value_counts(normalize=True)
         .sort_index() * 100).round(2)
    )

    print("\nTesting class percentages:")
    print(
        (y_test.value_counts(normalize=True)
         .sort_index() * 100).round(2)
    )

print("\n" + "=" * 70)
print("STEP 89 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 91: STORE TRAIN/TEST SPLITS FOR ALL DRUGS
# ==============================================================

print("=" * 70)
print("STEP 91: STORE TRAIN/TEST SPLITS FOR ALL DRUGS")
print("=" * 70)

from sklearn.model_selection import train_test_split

# Drugs
drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# Dictionary to store all splits
train_test_splits = {}

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    # Get drug-specific ML dataset
    X = ml_datasets[drug]['X'].copy()
    y = ml_datasets[drug]['y'].copy()

    # Stratified 80/20 split
    X_train_drug, X_test_drug, y_train_drug, y_test_drug = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # Store everything
    train_test_splits[drug] = {
        'X_train': X_train_drug,
        'X_test': X_test_drug,
        'y_train': y_train_drug,
        'y_test': y_test_drug
    }

    print("Original samples:", len(X))
    print("Training samples:", len(X_train_drug))
    print("Testing samples:", len(X_test_drug))
    print("Number of features:", X_train_drug.shape[1])

    print("\nTraining class distribution:")
    print(y_train_drug.value_counts().sort_index())

    print("\nTesting class distribution:")
    print(y_test_drug.value_counts().sort_index())

print("\n" + "=" * 70)
print("STEP 91 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 92: VERIFY STORED ML DATASETS
# ==============================================================

print("=" * 70)
print("STEP 92: VERIFY STORED ML DATASETS")
print("=" * 70)

drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

print("\nDictionary exists:", 'train_test_splits' in globals())
print("Number of drugs stored:", len(train_test_splits))

print("\nStored drugs:")
print(list(train_test_splits.keys()))

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    data = train_test_splits[drug]

    X_train = data['X_train']
    X_test = data['X_test']
    y_train = data['y_train']
    y_test = data['y_test']

    print("X_train shape:", X_train.shape)
    print("X_test shape:", X_test.shape)
    print("y_train shape:", y_train.shape)
    print("y_test shape:", y_test.shape)

    print("X_train missing:", X_train.isna().sum().sum())
    print("X_test missing:", X_test.isna().sum().sum())
    print("y_train missing:", y_train.isna().sum())
    print("y_test missing:", y_test.isna().sum())

    print("Training classes:", sorted(y_train.unique()))
    print("Testing classes:", sorted(y_test.unique()))

print("\n" + "=" * 70)
print("STEP 92 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 93: TRAIN BASELINE LOGISTIC REGRESSION MODELS
# ==============================================================

print("=" * 70)
print("STEP 93: TRAIN BASELINE LOGISTIC REGRESSION MODELS")
print("=" * 70)

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

logistic_models = {}

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    # Retrieve drug-specific train/test data
    X_train = train_test_splits[drug]['X_train']
    X_test = train_test_splits[drug]['X_test']
    y_train = train_test_splits[drug]['y_train']
    y_test = train_test_splits[drug]['y_test']

    # Logistic Regression pipeline
    model = Pipeline([
        ('scaler', StandardScaler()),
        ('logistic_regression', LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ])

    # Train ONLY on training data
    model.fit(X_train, y_train)

    # Store model
    logistic_models[drug] = model

    # Training and testing accuracy
    train_accuracy = model.score(X_train, y_train)
    test_accuracy = model.score(X_test, y_test)

    print("Training accuracy:", round(train_accuracy, 4))
    print("Testing accuracy:", round(test_accuracy, 4))

print("\n" + "=" * 70)
print("STEP 93 COMPLETE")
print("=" * 70)

print("\nModels stored:")
print(list(logistic_models.keys()))
```

```python
# ==============================================================
# STEP 94: COMPREHENSIVE LOGISTIC REGRESSION EVALUATION
# ==============================================================

print("=" * 70)
print("STEP 94: COMPREHENSIVE LOGISTIC REGRESSION EVALUATION")
print("=" * 70)

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

logistic_results = {}

drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    # Retrieve data
    X_train = train_test_splits[drug]['X_train']
    X_test = train_test_splits[drug]['X_test']
    y_train = train_test_splits[drug]['y_train']
    y_test = train_test_splits[drug]['y_test']

    # Retrieve trained model
    model = logistic_models[drug]

    # Predictions
    y_pred = model.predict(X_test)

    # Probability for positive/resistant class
    y_prob = model.predict_proba(X_test)[:, 1]

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    balanced_acc = balanced_accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_test, y_prob)

    # Confusion matrix
    cm = confusion_matrix(y_test, y_pred)

    # Store results
    logistic_results[drug] = {
        'Accuracy': accuracy,
        'Balanced Accuracy': balanced_acc,
        'Precision': precision,
        'Recall': recall,
        'F1': f1,
        'ROC-AUC': roc_auc,
        'Confusion Matrix': cm
    }

    print(f"Accuracy:          {accuracy:.4f}")
    print(f"Balanced Accuracy: {balanced_acc:.4f}")
    print(f"Precision:         {precision:.4f}")
    print(f"Recall:            {recall:.4f}")
    print(f"F1-score:          {f1:.4f}")
    print(f"ROC-AUC:           {roc_auc:.4f}")

    print("\nConfusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        y_pred,
        target_names=['Susceptible', 'Resistant'],
        zero_division=0
    ))

print("\n" + "=" * 70)
print("STEP 94 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 95: 5-FOLD CROSS-VALIDATION OF LOGISTIC REGRESSION
# ==============================================================

print("=" * 70)
print("STEP 95: 5-FOLD CROSS-VALIDATION OF LOGISTIC REGRESSION")
print("=" * 70)

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

import pandas as pd
import numpy as np

drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = []

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    X_train = train_test_splits[drug]['X_train']
    y_train = train_test_splits[drug]['y_train']

    # Pipeline prevents preprocessing leakage between folds
    model = Pipeline([
        ('scaler', StandardScaler()),
        ('logistic_regression', LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ])

    scoring = {
        'accuracy': 'accuracy',
        'balanced_accuracy': 'balanced_accuracy',
        'precision': 'precision',
        'recall': 'recall',
        'f1': 'f1',
        'roc_auc': 'roc_auc'
    }

    results = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    row = {
        'Drug': drug,
        'Accuracy_Mean': results['test_accuracy'].mean(),
        'Accuracy_SD': results['test_accuracy'].std(),
        'Balanced_Accuracy_Mean': results['test_balanced_accuracy'].mean(),
        'Precision_Mean': results['test_precision'].mean(),
        'Recall_Mean': results['test_recall'].mean(),
        'F1_Mean': results['test_f1'].mean(),
        'ROC_AUC_Mean': results['test_roc_auc'].mean(),
        'ROC_AUC_SD': results['test_roc_auc'].std()
    }

    cv_results.append(row)

    print(f"Accuracy:          {row['Accuracy_Mean']:.4f} ± {row['Accuracy_SD']:.4f}")
    print(f"Balanced Accuracy: {row['Balanced_Accuracy_Mean']:.4f}")
    print(f"Precision:         {row['Precision_Mean']:.4f}")
    print(f"Recall:            {row['Recall_Mean']:.4f}")
    print(f"F1-score:          {row['F1_Mean']:.4f}")
    print(f"ROC-AUC:           {row['ROC_AUC_Mean']:.4f} ± {row['ROC_AUC_SD']:.4f}")

# Create summary table
cv_results_df = pd.DataFrame(cv_results)

print("\n" + "=" * 70)
print("CROSS-VALIDATION SUMMARY")
print("=" * 70)

print(
    cv_results_df.round(4).to_string(index=False)
)

print("\n" + "=" * 70)
print("STEP 95 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 96: TRAIN BASELINE DECISION TREE MODELS
# ==============================================================

print("=" * 70)
print("STEP 96: TRAIN BASELINE DECISION TREE MODELS")
print("=" * 70)

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np

drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

decision_tree_models = {}

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    # Retrieve stored train/test data
    X_train = train_test_splits[drug]['X_train']
    X_test = train_test_splits[drug]['X_test']
    y_train = train_test_splits[drug]['y_train']
    y_test = train_test_splits[drug]['y_test']

    # Baseline Decision Tree
    model = DecisionTreeClassifier(
        random_state=42
    )

    # Train
    model.fit(X_train, y_train)

    # Predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Accuracy
    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)

    # Store model and predictions
    decision_tree_models[drug] = {
        'model': model,
        'y_train_pred': y_train_pred,
        'y_test_pred': y_test_pred,
        'train_accuracy': train_accuracy,
        'test_accuracy': test_accuracy
    }

    print(f"Training accuracy: {train_accuracy:.4f}")
    print(f"Testing accuracy:  {test_accuracy:.4f}")

print("\n" + "=" * 70)
print("STEP 96 COMPLETE")
print("=" * 70)

print("\nDecision Tree models stored:")
print(list(decision_tree_models.keys()))
```

```python
# ==============================================================
# STEP 97: COMPREHENSIVE DECISION TREE EVALUATION
# ==============================================================

print("=" * 70)
print("STEP 97: COMPREHENSIVE DECISION TREE EVALUATION")
print("=" * 70)

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

dt_results = []

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    # Retrieve stored data
    X_test = train_test_splits[drug]['X_test']
    y_test = train_test_splits[drug]['y_test']

    # Retrieve trained model
    model = decision_tree_models[drug]['model']

    # Predictions
    y_pred = model.predict(X_test)

    # Probability for ROC-AUC
    y_prob = model.predict_proba(X_test)[:, 1]

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    balanced_acc = balanced_accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_test, y_prob)

    # Store results
    dt_results.append({
        'Drug': drug,
        'Accuracy': accuracy,
        'Balanced_Accuracy': balanced_acc,
        'Precision': precision,
        'Recall': recall,
        'F1': f1,
        'ROC_AUC': roc_auc
    })

    # Print metrics
    print(f"Accuracy:          {accuracy:.4f}")
    print(f"Balanced Accuracy: {balanced_acc:.4f}")
    print(f"Precision:         {precision:.4f}")
    print(f"Recall:            {recall:.4f}")
    print(f"F1-score:          {f1:.4f}")
    print(f"ROC-AUC:           {roc_auc:.4f}")

    # Confusion matrix
    cm = confusion_matrix(y_test, y_pred)

    print("\nConfusion Matrix:")
    print(cm)

    # Classification report
    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            y_pred,
            target_names=['Susceptible', 'Resistant'],
            zero_division=0
        )
    )

# Create summary dataframe
dt_results_df = pd.DataFrame(dt_results)

print("\n" + "=" * 70)
print("DECISION TREE EVALUATION SUMMARY")
print("=" * 70)

print(
    dt_results_df.round(4).to_string(index=False)
)

print("\n" + "=" * 70)
print("STEP 97 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 98: 5-FOLD CROSS-VALIDATION OF DECISION TREE
# ==============================================================

print("=" * 70)
print("STEP 98: 5-FOLD CROSS-VALIDATION OF DECISION TREE")
print("=" * 70)

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.tree import DecisionTreeClassifier
import pandas as pd
import numpy as np

drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

dt_cv_results = []

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    X_train = train_test_splits[drug]['X_train']
    y_train = train_test_splits[drug]['y_train']

    # Baseline Decision Tree
    model = DecisionTreeClassifier(
        random_state=42
    )

    scoring = {
        'accuracy': 'accuracy',
        'balanced_accuracy': 'balanced_accuracy',
        'precision': 'precision',
        'recall': 'recall',
        'f1': 'f1',
        'roc_auc': 'roc_auc'
    }

    results = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    row = {
        'Drug': drug,
        'Accuracy_Mean': results['test_accuracy'].mean(),
        'Accuracy_SD': results['test_accuracy'].std(),
        'Balanced_Accuracy_Mean':
            results['test_balanced_accuracy'].mean(),
        'Precision_Mean':
            results['test_precision'].mean(),
        'Recall_Mean':
            results['test_recall'].mean(),
        'F1_Mean':
            results['test_f1'].mean(),
        'ROC_AUC_Mean':
            results['test_roc_auc'].mean(),
        'ROC_AUC_SD':
            results['test_roc_auc'].std()
    }

    dt_cv_results.append(row)

    print(
        f"Accuracy:          "
        f"{row['Accuracy_Mean']:.4f} ± "
        f"{row['Accuracy_SD']:.4f}"
    )

    print(
        f"Balanced Accuracy: "
        f"{row['Balanced_Accuracy_Mean']:.4f}"
    )

    print(
        f"Precision:         "
        f"{row['Precision_Mean']:.4f}"
    )

    print(
        f"Recall:            "
        f"{row['Recall_Mean']:.4f}"
    )

    print(
        f"F1-score:          "
        f"{row['F1_Mean']:.4f}"
    )

    print(
        f"ROC-AUC:           "
        f"{row['ROC_AUC_Mean']:.4f} ± "
        f"{row['ROC_AUC_SD']:.4f}"
    )

# Create summary dataframe
dt_cv_results_df = pd.DataFrame(dt_cv_results)

print("\n" + "=" * 70)
print("DECISION TREE CROSS-VALIDATION SUMMARY")
print("=" * 70)

print(
    dt_cv_results_df.round(4).to_string(index=False)
)

print("\n" + "=" * 70)
print("STEP 98 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 100: HYPERPARAMETER TUNING OF LOGISTIC REGRESSION
# ==============================================================

print("=" * 70)
print("STEP 100: HYPERPARAMETER TUNING OF LOGISTIC REGRESSION")
print("=" * 70)

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

import pandas as pd
import numpy as np

drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# --------------------------------------------------------------
# Cross-validation strategy
# --------------------------------------------------------------

cv_tuning = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# --------------------------------------------------------------
# Hyperparameter grid
# --------------------------------------------------------------

param_grid = {
    'logistic_regression__C': [
        0.001,
        0.01,
        0.1,
        1,
        10,
        100
    ],

    'logistic_regression__solver': [
        'liblinear',
        'lbfgs'
    ],

    'logistic_regression__class_weight': [
        None,
        'balanced'
    ]
}

tuned_models = {}
tuning_results = []

# --------------------------------------------------------------
# Tune each drug separately
# --------------------------------------------------------------

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    X_train = train_test_splits[drug]['X_train']
    y_train = train_test_splits[drug]['y_train']

    pipeline = Pipeline([
        ('scaler', StandardScaler()),

        ('logistic_regression', LogisticRegression(
            max_iter=3000,
            random_state=42
        ))
    ])

    grid_search = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grid,
        scoring='roc_auc',
        cv=cv_tuning,
        n_jobs=-1,
        refit=True
    )

    grid_search.fit(X_train, y_train)

    # Store optimized model
    tuned_models[drug] = grid_search.best_estimator_

    print("Best parameters:")
    print(grid_search.best_params_)

    print(f"Best CV ROC-AUC: {grid_search.best_score_:.4f}")

    tuning_results.append({
        'Drug': drug,
        'Best_C': grid_search.best_params_[
            'logistic_regression__C'
        ],
        'Best_Solver': grid_search.best_params_[
            'logistic_regression__solver'
        ],
        'Best_Class_Weight': grid_search.best_params_[
            'logistic_regression__class_weight'
        ],
        'Best_CV_ROC_AUC': grid_search.best_score_
    })

# --------------------------------------------------------------
# Create summary table
# --------------------------------------------------------------

tuning_results_df = pd.DataFrame(tuning_results)

print("\n" + "=" * 70)
print("HYPERPARAMETER TUNING SUMMARY")
print("=" * 70)

print(
    tuning_results_df.round(4).to_string(index=False)
)

print("\n" + "=" * 70)
print("STEP 100 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 101: EVALUATE TUNED LOGISTIC REGRESSION MODELS
# ==============================================================

print("=" * 70)
print("STEP 101: EVALUATE TUNED LOGISTIC REGRESSION MODELS")
print("=" * 70)

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import pandas as pd
import numpy as np

# --------------------------------------------------------------
# Best parameters obtained from Step 100
# --------------------------------------------------------------

best_params = {
    'FPV': {'C': 0.01, 'solver': 'liblinear', 'class_weight': None},
    'ATV': {'C': 0.01, 'solver': 'lbfgs', 'class_weight': 'balanced'},
    'IDV': {'C': 0.01, 'solver': 'lbfgs', 'class_weight': None},
    'LPV': {'C': 0.01, 'solver': 'liblinear', 'class_weight': None},
    'NFV': {'C': 0.01, 'solver': 'liblinear', 'class_weight': 'balanced'},
    'SQV': {'C': 0.01, 'solver': 'lbfgs', 'class_weight': None},
    'TPV': {'C': 0.01, 'solver': 'liblinear', 'class_weight': None},
    'DRV': {'C': 0.001, 'solver': 'lbfgs', 'class_weight': 'balanced'}
}

tuned_models = {}
tuned_results = []

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    # ----------------------------------------------------------
    # Retrieve stored train/test data
    # ----------------------------------------------------------

    X_train = train_test_splits[drug]['X_train']
    X_test = train_test_splits[drug]['X_test']
    y_train = train_test_splits[drug]['y_train']
    y_test = train_test_splits[drug]['y_test']

    params = best_params[drug]

    # ----------------------------------------------------------
    # Build tuned model
    # ----------------------------------------------------------

    model = Pipeline([
        ('scaler', StandardScaler()),
        ('logistic_regression', LogisticRegression(
            C=params['C'],
            solver=params['solver'],
            class_weight=params['class_weight'],
            max_iter=2000,
            random_state=42
        ))
    ])

    # ----------------------------------------------------------
    # Train model
    # ----------------------------------------------------------

    model.fit(X_train, y_train)

    # Store model
    tuned_models[drug] = model

    # ----------------------------------------------------------
    # Predictions
    # ----------------------------------------------------------

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    # ----------------------------------------------------------
    # Metrics
    # ----------------------------------------------------------

    accuracy = accuracy_score(y_test, y_pred)
    balanced_acc = balanced_accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_test, y_prob)

    cm = confusion_matrix(y_test, y_pred)

    # ----------------------------------------------------------
    # Store results
    # ----------------------------------------------------------

    tuned_results.append({
        'Drug': drug,
        'Accuracy': accuracy,
        'Balanced_Accuracy': balanced_acc,
        'Precision': precision,
        'Recall': recall,
        'F1': f1,
        'ROC_AUC': roc_auc
    })

    # ----------------------------------------------------------
    # Display results
    # ----------------------------------------------------------

    print(f"Best C:             {params['C']}")
    print(f"Solver:             {params['solver']}")
    print(f"Class weight:       {params['class_weight']}")

    print(f"\nAccuracy:            {accuracy:.4f}")
    print(f"Balanced Accuracy:   {balanced_acc:.4f}")
    print(f"Precision:           {precision:.4f}")
    print(f"Recall:              {recall:.4f}")
    print(f"F1-score:            {f1:.4f}")
    print(f"ROC-AUC:             {roc_auc:.4f}")

    print("\nConfusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            y_pred,
            target_names=['Susceptible', 'Resistant'],
            digits=2,
            zero_division=0
        )
    )

# --------------------------------------------------------------
# Create summary DataFrame
# --------------------------------------------------------------

tuned_results_df = pd.DataFrame(tuned_results)

print("\n" + "=" * 70)
print("TUNED LOGISTIC REGRESSION SUMMARY")
print("=" * 70)

print(
    tuned_results_df.round(4).to_string(index=False)
)

print("\n" + "=" * 70)
print("STEP 101 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 102: COMPARE BASELINE VS TUNED LOGISTIC REGRESSION
# ==============================================================

print("=" * 70)
print("STEP 102: COMPARE BASELINE VS TUNED LOGISTIC REGRESSION")
print("=" * 70)

# Check required result tables
required_tables = ['logistic_results_df', 'tuned_logistic_results_df']

for table_name in required_tables:
    if table_name not in globals():
        raise RuntimeError(
            f"{table_name} is not available. "
            "Please run the corresponding previous step first."
        )

# Make copies
baseline = logistic_results_df.copy()
tuned = tuned_logistic_results_df.copy()

# Rename metric columns so they are clearly distinguishable
baseline = baseline.rename(columns={
    'Accuracy': 'Baseline_Accuracy',
    'Balanced_Accuracy': 'Baseline_Balanced_Accuracy',
    'Precision': 'Baseline_Precision',
    'Recall': 'Baseline_Recall',
    'F1': 'Baseline_F1',
    'ROC_AUC': 'Baseline_ROC_AUC'
})

tuned = tuned.rename(columns={
    'Accuracy': 'Tuned_Accuracy',
    'Balanced_Accuracy': 'Tuned_Balanced_Accuracy',
    'Precision': 'Tuned_Precision',
    'Recall': 'Tuned_Recall',
    'F1': 'Tuned_F1',
    'ROC_AUC': 'Tuned_ROC_AUC'
})

# Merge baseline and tuned results
comparison = pd.merge(
    baseline,
    tuned,
    on='Drug',
    how='inner'
)

# Calculate improvements
comparison['Accuracy_Improvement'] = (
    comparison['Tuned_Accuracy'] -
    comparison['Baseline_Accuracy']
)

comparison['Balanced_Accuracy_Improvement'] = (
    comparison['Tuned_Balanced_Accuracy'] -
    comparison['Baseline_Balanced_Accuracy']
)

comparison['Precision_Improvement'] = (
    comparison['Tuned_Precision'] -
    comparison['Baseline_Precision']
)

comparison['Recall_Improvement'] = (
    comparison['Tuned_Recall'] -
    comparison['Baseline_Recall']
)

comparison['F1_Improvement'] = (
    comparison['Tuned_F1'] -
    comparison['Baseline_F1']
)

comparison['ROC_AUC_Improvement'] = (
    comparison['Tuned_ROC_AUC'] -
    comparison['Baseline_ROC_AUC']
)

# Display drug-by-drug comparison
print("\n" + "=" * 70)
print("BASELINE VS TUNED LOGISTIC REGRESSION")
print("=" * 70)

display(
    comparison[
        [
            'Drug',
            'Baseline_Accuracy',
            'Tuned_Accuracy',
            'Accuracy_Improvement',
            'Baseline_Balanced_Accuracy',
            'Tuned_Balanced_Accuracy',
            'Balanced_Accuracy_Improvement',
            'Baseline_F1',
            'Tuned_F1',
            'F1_Improvement',
            'Baseline_ROC_AUC',
            'Tuned_ROC_AUC',
            'ROC_AUC_Improvement'
        ]
    ].round(4)
)

# Overall mean comparison
metrics = [
    ('Accuracy', 'Baseline_Accuracy', 'Tuned_Accuracy'),
    ('Balanced Accuracy',
     'Baseline_Balanced_Accuracy',
     'Tuned_Balanced_Accuracy'),
    ('Precision', 'Baseline_Precision', 'Tuned_Precision'),
    ('Recall', 'Baseline_Recall', 'Tuned_Recall'),
    ('F1-score', 'Baseline_F1', 'Tuned_F1'),
    ('ROC-AUC', 'Baseline_ROC_AUC', 'Tuned_ROC_AUC')
]

overall_rows = []

for metric_name, baseline_col, tuned_col in metrics:

    baseline_mean = comparison[baseline_col].mean()
    tuned_mean = comparison[tuned_col].mean()

    overall_rows.append({
        'Metric': metric_name,
        'Baseline_Mean': baseline_mean,
        'Tuned_Mean': tuned_mean,
        'Improvement': tuned_mean - baseline_mean
    })

overall_comparison_df = pd.DataFrame(overall_rows)

print("\n" + "=" * 70)
print("OVERALL MEAN PERFORMANCE")
print("=" * 70)

print(
    overall_comparison_df.round(4).to_string(index=False)
)

# Count improvements
print("\n" + "=" * 70)
print("NUMBER OF DRUGS IMPROVED")
print("=" * 70)

for metric_name, baseline_col, tuned_col in metrics:

    improved = (
        comparison[tuned_col] >
        comparison[baseline_col]
    ).sum()

    equal = (
        comparison[tuned_col] ==
        comparison[baseline_col]
    ).sum()

    decreased = (
        comparison[tuned_col] <
        comparison[baseline_col]
    ).sum()

    print(
        f"{metric_name:<20} "
        f"Improved: {improved} | "
        f"Same: {equal} | "
        f"Decreased: {decreased}"
    )

# Best tuned model by ROC-AUC
best_roc_row = comparison.loc[
    comparison['Tuned_ROC_AUC'].idxmax()
]

print("\n" + "=" * 70)
print("BEST TUNED MODEL BY ROC-AUC")
print("=" * 70)

print(
    f"Drug: {best_roc_row['Drug']}"
)
print(
    f"Tuned ROC-AUC: {best_roc_row['Tuned_ROC_AUC']:.4f}"
)

print("\n" + "=" * 70)
print("STEP 102 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 105: LOGISTIC REGRESSION FEATURE IMPORTANCE ANALYSIS
# ==============================================================

print("=" * 70)
print("STEP 105: LOGISTIC REGRESSION FEATURE IMPORTANCE ANALYSIS")
print("=" * 70)

import pandas as pd
import numpy as np

# --------------------------------------------------------------
# Check that tuned models are available
# --------------------------------------------------------------

if 'tuned_logistic_models' not in globals():
    raise RuntimeError(
        "tuned_logistic_models is not available. "
        "Please restore/retrain the tuned Logistic Regression models first."
    )

# --------------------------------------------------------------
# Get feature names
# --------------------------------------------------------------

# Use the feature matrix from one drug
sample_drug = 'FPV'

X_sample = train_test_splits[sample_drug]['X_train']

if hasattr(X_sample, 'columns'):
    feature_names = np.array(X_sample.columns)
else:
    feature_names = np.array([
        f"Feature_{i}"
        for i in range(X_sample.shape[1])
    ])

print(f"\nNumber of features: {len(feature_names)}")

# --------------------------------------------------------------
# Extract coefficients
# --------------------------------------------------------------

coefficient_results = {}

for drug in drugs:

    print("\n" + "-" * 60)
    print(drug)
    print("-" * 60)

    model = tuned_logistic_models[drug]

    # If model is a Pipeline, retrieve Logistic Regression step
    if hasattr(model, 'named_steps'):
        lr_model = model.named_steps['logistic_regression']
    else:
        lr_model = model

    coefficients = lr_model.coef_[0]

    # Safety check
    if len(coefficients) != len(feature_names):
        raise ValueError(
            f"Feature count mismatch for {drug}: "
            f"{len(coefficients)} coefficients vs "
            f"{len(feature_names)} feature names."
        )

    # Create coefficient table
    coef_df = pd.DataFrame({
        'Feature': feature_names,
        'Coefficient': coefficients,
        'Absolute_Coefficient': np.abs(coefficients)
    })

    # Sort by absolute coefficient
    coef_df = coef_df.sort_values(
        'Absolute_Coefficient',
        ascending=False
    ).reset_index(drop=True)

    coefficient_results[drug] = coef_df

    print("\nTop 10 features by absolute coefficient:")
    print(
        coef_df.head(10).round(4).to_string(index=False)
    )

# --------------------------------------------------------------
# Create combined summary
# --------------------------------------------------------------

all_coefficients = []

for drug, df in coefficient_results.items():

    temp = df.copy()
    temp['Drug'] = drug

    all_coefficients.append(temp)

coefficient_summary_df = pd.concat(
    all_coefficients,
    ignore_index=True
)

# --------------------------------------------------------------
# Display summary
# --------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP FEATURES ACROSS ALL DRUGS")
print("=" * 70)

print(
    coefficient_summary_df
    .sort_values(
        ['Drug', 'Absolute_Coefficient'],
        ascending=[True, False]
    )
    .groupby('Drug')
    .head(10)
    .round(4)
    .to_string(index=False)
)

# --------------------------------------------------------------
# Save results
# --------------------------------------------------------------

coefficient_results_df = coefficient_summary_df.copy()

print("\n" + "=" * 70)
print("STEP 105 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 105D: LOGISTIC REGRESSION FEATURE IMPORTANCE ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 105D: LOGISTIC REGRESSION FEATURE IMPORTANCE ANALYSIS")
print("=" * 70)

# Make sure feature names match model coefficients
feature_names = X_train.columns.tolist()

feature_importance_results = {}

for drug in drugs:

    print(f"\n{'-' * 60}")
    print(f"{drug}")
    print(f"{'-' * 60}")

    model = tuned_logistic_models[drug]

    # Extract Logistic Regression component
    lr_model = model.named_steps['logistic_regression']

    # Extract coefficients
    coefficients = lr_model.coef_[0]

    # Absolute coefficient = feature importance
    absolute_coefficients = np.abs(coefficients)

    # Create feature importance table
    importance_df = pd.DataFrame({
        'Feature': feature_names,
        'Coefficient': coefficients,
        'Absolute_Importance': absolute_coefficients
    })

    # Sort by absolute importance
    importance_df = importance_df.sort_values(
        by='Absolute_Importance',
        ascending=False
    ).reset_index(drop=True)

    # Add rank
    importance_df.insert(
        0,
        'Rank',
        range(1, len(importance_df) + 1)
    )

    # Store results
    feature_importance_results[drug] = importance_df

    # Display top 20
    print("\nTop 20 most important mutation features:")

    display(
        importance_df.head(20)
    )

print("\n" + "=" * 70)
print("STEP 105D COMPLETE")
print("=" * 70)
```

```python
/# ==============================================================
# STEP 116: CORRELATION BETWEEN FEATURE IMPORTANCE
# AND MODEL PERFORMANCE
# ==============================================================

print("=" * 70)
print("STEP 116: CORRELATION BETWEEN FEATURE IMPORTANCE")
print("AND MODEL PERFORMANCE")
print("=" * 70)

import pandas as pd
import numpy as np
from scipy.stats import pearsonr, spearmanr

# --------------------------------------------------------------
# 1. Load Step 115 results
# --------------------------------------------------------------
if 'comparison_df' in globals():
    step115_df = comparison_df.copy()
else:
    step115_df = pd.read_csv(
        "Step_115_Feature_Importance_vs_Performance.csv"
    )

print("\n✓ Step 115 results loaded")
print(f"Number of drugs analyzed: {len(step115_df)}")

# --------------------------------------------------------------
# 2. Select variables for correlation analysis
# --------------------------------------------------------------
correlation_metrics = [
    "Accuracy",
    "F1",
    "ROC_AUC"
]

importance_column = "Mean_Feature_Importance"

# Verify required columns
required_columns = [
    "Drug",
    importance_column
] + correlation_metrics

missing_columns = [
    col for col in required_columns
    if col not in step115_df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# --------------------------------------------------------------
# 3. Display analysis dataset
# --------------------------------------------------------------
print("\nDATA USED FOR CORRELATION ANALYSIS")
print("-" * 70)

display(
    step115_df[
        ["Drug", importance_column] + correlation_metrics
    ]
)

# --------------------------------------------------------------
# 4. Pearson correlation
# --------------------------------------------------------------
print("\nPEARSON CORRELATION")
print("-" * 70)

pearson_results = []

for metric in correlation_metrics:

    x = step115_df[importance_column].astype(float)
    y = step115_df[metric].astype(float)

    r, p = pearsonr(x, y)

    pearson_results.append({
        "Performance_Metric": metric,
        "Pearson_r": r,
        "P_value": p,
        "Absolute_r": abs(r)
    })

pearson_df = pd.DataFrame(pearson_results)

display(pearson_df)

# --------------------------------------------------------------
# 5. Spearman correlation
# --------------------------------------------------------------
print("\nSPEARMAN RANK CORRELATION")
print("-" * 70)

spearman_results = []

for metric in correlation_metrics:

    x = step115_df[importance_column].astype(float)
    y = step115_df[metric].astype(float)

    rho, p = spearmanr(x, y)

    spearman_results.append({
        "Performance_Metric": metric,
        "Spearman_rho": rho,
        "P_value": p,
        "Absolute_rho": abs(rho)
    })

spearman_df = pd.DataFrame(spearman_results)

display(spearman_df)

# --------------------------------------------------------------
# 6. Interpretation of correlation strength
# --------------------------------------------------------------
def interpret_correlation(value):

    value = abs(value)

    if value < 0.20:
        return "Very Weak"
    elif value < 0.40:
        return "Weak"
    elif value < 0.60:
        return "Moderate"
    elif value < 0.80:
        return "Strong"
    else:
        return "Very Strong"

pearson_df["Correlation_Strength"] = (
    pearson_df["Pearson_r"]
    .apply(interpret_correlation)
)

spearman_df["Correlation_Strength"] = (
    spearman_df["Spearman_rho"]
    .apply(interpret_correlation)
)

# --------------------------------------------------------------
# 7. Combined correlation summary
# --------------------------------------------------------------
print("\nCOMBINED CORRELATION SUMMARY")
print("-" * 70)

correlation_summary = pearson_df[
    [
        "Performance_Metric",
        "Pearson_r",
        "P_value",
        "Correlation_Strength"
    ]
].copy()

correlation_summary = correlation_summary.merge(
    spearman_df[
        [
            "Performance_Metric",
            "Spearman_rho",
            "P_value"
        ]
    ],
    on="Performance_Metric",
    suffixes=("_Pearson", "_Spearman")
)

display(correlation_summary)

# --------------------------------------------------------------
# 8. Identify strongest relationship
# --------------------------------------------------------------
strongest_idx = pearson_df["Absolute_r"].idxmax()

strongest_metric = pearson_df.loc[
    strongest_idx,
    "Performance_Metric"
]

strongest_r = pearson_df.loc[
    strongest_idx,
    "Pearson_r"
]

strongest_p = pearson_df.loc[
    strongest_idx,
    "P_value"
]

print("\nSTRONGEST PEARSON ASSOCIATION")
print("-" * 70)

print(f"Performance metric: {strongest_metric}")
print(f"Pearson r: {strongest_r:.4f}")
print(f"P-value: {strongest_p:.6f}")
print(
    f"Strength: "
    f"{interpret_correlation(strongest_r)}"
)

# --------------------------------------------------------------
# 9. Statistical significance
# --------------------------------------------------------------
print("\nSTATISTICAL SIGNIFICANCE")
print("-" * 70)

for _, row in pearson_df.iterrows():

    metric = row["Performance_Metric"]
    p_value = row["P_value"]

    if p_value < 0.05:
        result = "Statistically significant"
    else:
        result = "Not statistically significant"

    print(
        f"{metric}: {result} "
        f"(p = {p_value:.6f})"
    )

# --------------------------------------------------------------
# 10. Save results
# --------------------------------------------------------------
pearson_df.to_csv(
    "Step_116_Pearson_Correlation.csv",
    index=False
)

spearman_df.to_csv(
    "Step_116_Spearman_Correlation.csv",
    index=False
)

correlation_summary.to_csv(
    "Step_116_Correlation_Summary.csv",
    index=False
)

print("\n✓ Files saved:")
print("1. Step_116_Pearson_Correlation.csv")
print("2. Step_116_Spearman_Correlation.csv")
print("3. Step_116_Correlation_Summary.csv")

print("\n" + "=" * 70)
print("STEP 116 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 118: STATISTICAL INTERPRETATION OF FEATURE IMPORTANCE
# VS MODEL PERFORMANCE
# ==============================================================

print("=" * 70)
print("STEP 118: STATISTICAL INTERPRETATION OF FEATURE IMPORTANCE")
print("VS MODEL PERFORMANCE")
print("=" * 70)

import pandas as pd
import numpy as np

# --------------------------------------------------------------
# 1. Load Step 116 correlation results
# --------------------------------------------------------------
if 'correlation_summary' in globals():
    step116_corr = correlation_summary.copy()
else:
    step116_corr = pd.read_csv(
        "Step_116_Correlation_Summary.csv"
    )

print("\n✓ Step 116 correlation results loaded")

display(step116_corr)

# --------------------------------------------------------------
# 2. Interpret correlation strength
# --------------------------------------------------------------
def interpret_strength(r):
    r_abs = abs(r)

    if r_abs >= 0.80:
        return "Very Strong"
    elif r_abs >= 0.60:
        return "Strong"
    elif r_abs >= 0.40:
        return "Moderate"
    elif r_abs >= 0.20:
        return "Weak"
    else:
        return "Very Weak"

step116_corr["Pearson_Interpretation"] = (
    step116_corr["Pearson_r"]
    .apply(interpret_strength)
)

step116_corr["Spearman_Interpretation"] = (
    step116_corr["Spearman_rho"]
    .apply(interpret_strength)
)

# --------------------------------------------------------------
# 3. Statistical significance
# --------------------------------------------------------------
step116_corr["Pearson_Significance"] = np.where(
    step116_corr["P_value_Pearson"] < 0.05,
    "Statistically Significant",
    "Not Statistically Significant"
)

step116_corr["Spearman_Significance"] = np.where(
    step116_corr["P_value_Spearman"] < 0.05,
    "Statistically Significant",
    "Not Statistically Significant"
)

# --------------------------------------------------------------
# 4. Overall interpretation
# --------------------------------------------------------------
def overall_interpretation(row):

    pearson_sig = row["P_value_Pearson"] < 0.05
    spearman_sig = row["P_value_Spearman"] < 0.05

    if pearson_sig and spearman_sig:
        return "Strong evidence of positive association"

    elif pearson_sig and not spearman_sig:
        return "Positive association; limited rank-based evidence"

    elif not pearson_sig and spearman_sig:
        return "Positive rank-based association"

    else:
        return "Positive trend without statistical significance"

step116_corr["Overall_Interpretation"] = (
    step116_corr.apply(
        overall_interpretation,
        axis=1
    )
)

# --------------------------------------------------------------
# 5. Display interpretation table
# --------------------------------------------------------------
print("\nSTATISTICAL INTERPRETATION")
print("-" * 70)

display(
    step116_corr[
        [
            "Performance_Metric",
            "Pearson_r",
            "P_value_Pearson",
            "Pearson_Interpretation",
            "Spearman_rho",
            "P_value_Spearman",
            "Spearman_Interpretation",
            "Overall_Interpretation"
        ]
    ]
)

# --------------------------------------------------------------
# 6. Identify strongest association
# --------------------------------------------------------------
strongest_idx = step116_corr[
    "Pearson_r"
].abs().idxmax()

strongest = step116_corr.loc[strongest_idx]

print("\nSTRONGEST FEATURE-IMPORTANCE ASSOCIATION")
print("-" * 70)

print(
    f"Performance metric: "
    f"{strongest['Performance_Metric']}"
)

print(
    f"Pearson r: "
    f"{strongest['Pearson_r']:.4f}"
)

print(
    f"Pearson p-value: "
    f"{strongest['P_value_Pearson']:.6f}"
)

print(
    f"Interpretation: "
    f"{strongest['Overall_Interpretation']}"
)

# --------------------------------------------------------------
# 7. Generate scientific conclusion
# --------------------------------------------------------------
print("\nSCIENTIFIC INTERPRETATION")
print("-" * 70)

print(
    "The analysis indicates that greater mean mutation-feature "
    "importance is generally associated with improved model "
    "performance across the analyzed antiretroviral drugs."
)

print(
    "The strongest association was observed for F1 score, "
    "followed by Accuracy."
)

print(
    "The association with ROC-AUC was positive but did not "
    "reach conventional statistical significance."
)

print(
    "Because only eight drugs were analyzed, these findings "
    "should be interpreted as exploratory associations rather "
    "than evidence of causation."
)

# --------------------------------------------------------------
# 8. Save final interpretation
# --------------------------------------------------------------
step118_output = step116_corr.copy()

step118_output.to_csv(
    "Step_118_Feature_Importance_Performance_Interpretation.csv",
    index=False
)

print("\n✓ File saved:")
print(
    "Step_118_Feature_Importance_Performance_Interpretation.csv"
)

print("\n" + "=" * 70)
print("STEP 118 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 119: CONSOLIDATED FEATURE IMPORTANCE AND MODEL PERFORMANCE RANKING
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 119: CONSOLIDATED FEATURE IMPORTANCE AND MODEL PERFORMANCE RANKING")
print("=" * 70)

# ------------------------------------------------
# Load Step 115 results
# ------------------------------------------------
comparison_file = "Step_115_Feature_Importance_vs_Performance.csv"

comparison_df = pd.read_csv(comparison_file)

print("\n✓ Step 115 results loaded")
print(f"Number of drugs analyzed: {len(comparison_df)}")

# ------------------------------------------------
# Select required columns
# ------------------------------------------------
ranking_df = comparison_df[
    [
        "Drug",
        "Mean_Feature_Importance",
        "Maximum_Feature_Importance",
        "Accuracy",
        "F1",
        "ROC_AUC",
        "Number_of_Features"
    ]
].copy()

# ------------------------------------------------
# Rank each drug by individual metrics
# Higher value = better rank
# ------------------------------------------------
ranking_df["Feature_Importance_Rank"] = (
    ranking_df["Mean_Feature_Importance"]
    .rank(ascending=False, method="min")
    .astype(int)
)

ranking_df["Accuracy_Rank"] = (
    ranking_df["Accuracy"]
    .rank(ascending=False, method="min")
    .astype(int)
)

ranking_df["F1_Rank"] = (
    ranking_df["F1"]
    .rank(ascending=False, method="min")
    .astype(int)
)

ranking_df["ROC_AUC_Rank"] = (
    ranking_df["ROC_AUC"]
    .rank(ascending=False, method="min")
    .astype(int)
)

# ------------------------------------------------
# Calculate average performance
# ------------------------------------------------
ranking_df["Mean_Performance"] = ranking_df[
    ["Accuracy", "F1", "ROC_AUC"]
].mean(axis=1)

ranking_df["Performance_Rank"] = (
    ranking_df["Mean_Performance"]
    .rank(ascending=False, method="min")
    .astype(int)
)

# ------------------------------------------------
# Combined rank
# ------------------------------------------------
ranking_df["Combined_Rank_Score"] = ranking_df[
    [
        "Feature_Importance_Rank",
        "Accuracy_Rank",
        "F1_Rank",
        "ROC_AUC_Rank"
    ]
].mean(axis=1)

ranking_df["Overall_Rank"] = (
    ranking_df["Combined_Rank_Score"]
    .rank(ascending=True, method="min")
    .astype(int)
)

# ------------------------------------------------
# Sort by overall rank
# ------------------------------------------------
ranking_df = ranking_df.sort_values(
    "Overall_Rank"
).reset_index(drop=True)

# ------------------------------------------------
# Display final ranking
# ------------------------------------------------
print("\n" + "=" * 70)
print("CONSOLIDATED DRUG RANKING")
print("=" * 70)

display(
    ranking_df[
        [
            "Overall_Rank",
            "Drug",
            "Mean_Feature_Importance",
            "Mean_Performance",
            "Accuracy",
            "F1",
            "ROC_AUC",
            "Feature_Importance_Rank",
            "Performance_Rank"
        ]
    ]
)

# ------------------------------------------------
# Best overall drug
# ------------------------------------------------
best_drug = ranking_df.iloc[0]

print("\n" + "=" * 70)
print("BEST OVERALL MODEL PROFILE")
print("=" * 70)

print(f"Drug: {best_drug['Drug']}")
print(f"Overall rank: {best_drug['Overall_Rank']}")
print(f"Mean feature importance: {best_drug['Mean_Feature_Importance']:.4f}")
print(f"Mean performance: {best_drug['Mean_Performance']:.4f}")
print(f"Accuracy: {best_drug['Accuracy']:.4f}")
print(f"F1: {best_drug['F1']:.4f}")
print(f"ROC-AUC: {best_drug['ROC_AUC']:.4f}")

# ------------------------------------------------
# Highest feature importance
# ------------------------------------------------
highest_feature = ranking_df.loc[
    ranking_df["Mean_Feature_Importance"].idxmax()
]

print("\n" + "=" * 70)
print("HIGHEST MEAN FEATURE IMPORTANCE")
print("=" * 70)

print(f"Drug: {highest_feature['Drug']}")
print(
    f"Mean feature importance: "
    f"{highest_feature['Mean_Feature_Importance']:.4f}"
)

# ------------------------------------------------
# Highest model performance
# ------------------------------------------------
highest_performance = ranking_df.loc[
    ranking_df["Mean_Performance"].idxmax()
]

print("\n" + "=" * 70)
print("HIGHEST MEAN MODEL PERFORMANCE")
print("=" * 70)

print(f"Drug: {highest_performance['Drug']}")
print(
    f"Mean performance: "
    f"{highest_performance['Mean_Performance']:.4f}"
)

# ------------------------------------------------
# Scientific interpretation
# ------------------------------------------------
print("\n" + "=" * 70)
print("SCIENTIFIC INTERPRETATION")
print("=" * 70)

print(
    "The consolidated ranking integrates mutation-feature importance "
    "with predictive performance across the eight analyzed antiretroviral drugs."
)

print(
    "Drugs with higher overall rankings demonstrate a combination of "
    "stronger mutation-feature signals and better predictive performance."
)

print(
    "This ranking is intended as a comparative exploratory analysis and "
    "does not establish a causal relationship between feature importance "
    "and model performance."
)

# ------------------------------------------------
# Save results
# ------------------------------------------------
output_file = "Step_119_Consolidated_Drug_Ranking.csv"

ranking_df.to_csv(
    output_file,
    index=False
)

print("\n✓ File saved:")
print(f"1. {output_file}")

print("\n" + "=" * 70)
print("STEP 119 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 120: MODEL PERFORMANCE AND FEATURE IMPORTANCE HEATMAP
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 120: MODEL PERFORMANCE AND FEATURE IMPORTANCE HEATMAP")
print("=" * 70)

# ------------------------------------------------
# Load Step 119 results
# ------------------------------------------------
ranking_file = "Step_119_Consolidated_Drug_Ranking.csv"

ranking_df = pd.read_csv(ranking_file)

print("\n✓ Step 119 results loaded")
print(f"Number of drugs analyzed: {len(ranking_df)}")

# ------------------------------------------------
# Select variables for heatmap
# ------------------------------------------------
heatmap_df = ranking_df[
    [
        "Drug",
        "Mean_Feature_Importance",
        "Accuracy",
        "F1",
        "ROC_AUC",
        "Mean_Performance"
    ]
].copy()

# Set drug names as index
heatmap_df = heatmap_df.set_index("Drug")

print("\nDATA USED FOR HEATMAP")
print("-" * 70)
display(heatmap_df)

# ------------------------------------------------
# Normalize each column for visual comparison
# ------------------------------------------------
normalized_df = heatmap_df.copy()

for column in normalized_df.columns:
    min_value = normalized_df[column].min()
    max_value = normalized_df[column].max()

    if max_value != min_value:
        normalized_df[column] = (
            (normalized_df[column] - min_value)
            / (max_value - min_value)
        )
    else:
        normalized_df[column] = 0.5

# ------------------------------------------------
# Create heatmap
# ------------------------------------------------
plt.figure(figsize=(10, 7))

plt.imshow(
    normalized_df.values,
    aspect="auto",
    cmap="viridis"
)

plt.colorbar(
    label="Normalized Value"
)

plt.xticks(
    range(len(normalized_df.columns)),
    normalized_df.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(normalized_df.index)),
    normalized_df.index
)

plt.title(
    "Normalized Feature Importance and Model Performance Across Drugs"
)

# ------------------------------------------------
# Add original values to cells
# ------------------------------------------------
for i in range(normalized_df.shape[0]):
    for j in range(normalized_df.shape[1]):

        original_value = heatmap_df.iloc[i, j]

        plt.text(
            j,
            i,
            f"{original_value:.3f}",
            ha="center",
            va="center"
        )

plt.tight_layout()

# ------------------------------------------------
# Save figure
# ------------------------------------------------
output_file = "Step_120_Feature_Importance_Performance_Heatmap.png"

plt.savefig(
    output_file,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------
# Save normalized data
# ------------------------------------------------
normalized_output = (
    "Step_120_Normalized_Feature_Importance_Performance.csv"
)

normalized_df.to_csv(normalized_output)

print("\n" + "=" * 70)
print("STEP 120 OUTPUT FILES")
print("=" * 70)

print(f"✓ {output_file}")
print(f"✓ {normalized_output}")

print("\n" + "=" * 70)
print("STEP 120 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 141: FINAL PROJECT INTRODUCTION
# ================================================================

print("=" * 70)
print("STEP 141: FINAL PROJECT INTRODUCTION")
print("=" * 70)

introduction = """
1. INTRODUCTION

1.1 Background

Human immunodeficiency virus type 1 (HIV-1) is a major viral pathogen
that primarily targets cells of the immune system and can progressively
weaken immune function. The development and widespread use of
antiretroviral therapy (ART) has substantially improved the management of
HIV infection. Combination antiretroviral therapy suppresses viral
replication and helps maintain immune function. However, the effectiveness
of treatment can be compromised when HIV-1 develops resistance to one or
more antiretroviral drugs.

HIV-1 has a high degree of genetic variability. During viral replication,
genetic changes can arise and become enriched under selective pressure,
including pressure exerted by antiretroviral therapy. Some mutations may
reduce susceptibility to particular antiretroviral drugs. The accumulation
of resistance-associated mutations can therefore contribute to treatment
failure and may limit the effectiveness of available therapeutic options.

1.2 HIV-1 Antiretroviral Drug Resistance

HIV-1 drug resistance refers to a reduction in the susceptibility of the
virus to an antiretroviral drug or drug class. Resistance may develop
through the selection of viral variants carrying mutations that affect the
target of a drug or other components involved in the viral replication
process.

Protease inhibitors are an important class of antiretroviral drugs. The
present study examines eight protease-inhibitor drugs: saquinavir (SQV),
atazanavir (ATV), nelfinavir (NFV), lopinavir (LPV), fosamprenavir (FPV),
indinavir (IDV), tipranavir (TPV), and darunavir (DRV).

Understanding relationships between viral mutations and drug-resistance
phenotypes is important for computational analysis of HIV-1 resistance.
Genomic information can provide a basis for identifying mutation patterns
that are associated with differences in predicted drug susceptibility.

1.3 Viral Genomic Mutations and Drug Resistance

The HIV-1 genome contains genetic information required for viral
replication and production of viral proteins. Mutations within viral
proteins can alter amino-acid sequences and potentially influence the
interaction between viral targets and antiretroviral compounds.

Mutation notation such as I84V, V32I, L33F, I54L, I54M, and I47V describes
amino-acid substitutions at specific positions in a viral protein. Such
mutation features can be represented computationally and used as input
variables for machine-learning models.

Because HIV-1 drug resistance can involve combinations of mutations rather
than a single genetic change, computational approaches can be useful for
examining multiple mutation features simultaneously. Machine-learning
methods can identify predictive patterns in complex datasets without
requiring every relationship between individual features to be specified
in advance.

1.4 Role of Bioinformatics and Machine Learning

Bioinformatics provides computational methods for analysing biological
data, including genomic and protein sequence information. In HIV-1
research, computational analysis can assist in studying viral genetic
variation, drug-resistance patterns, and relationships between mutations
and phenotypic or clinical measurements.

Machine learning is a computational approach in which algorithms learn
patterns from data and use those patterns to make predictions. In a
classification setting, machine-learning models can be trained using
genomic features and known resistance outcomes to predict the resistance
class of previously unseen observations.

Logistic Regression is a commonly used classification algorithm that
estimates the probability of an outcome based on input features. It has
the advantages of being relatively simple, computationally efficient, and
interpretable. These characteristics make it useful for an exploratory
study in which both predictive performance and feature contributions are
of interest.

1.5 Importance of Predicting HIV-1 Drug Resistance

Accurate computational modelling of HIV-1 drug resistance may help
researchers better understand patterns within genomic and resistance
datasets. Such approaches may support research into resistance-associated
mutation patterns and provide a framework for comparing predictive
performance across different antiretroviral drugs.

However, predictive performance obtained from a computational dataset
does not automatically demonstrate clinical usefulness. A machine-learning
model must be evaluated using appropriate validation strategies and,
ideally, tested using independent external datasets before clinical
application can be considered.

Therefore, the present study focuses on computational feasibility and
model-derived predictive patterns rather than clinical decision-making.

1.6 Research Gap

Although HIV-1 drug resistance has been extensively investigated using
genotypic, phenotypic, and computational approaches, there remains value
in evaluating machine-learning frameworks that examine multiple
drug-specific resistance outcomes while also assessing the relative
importance of recurrent mutation features.

A particular challenge is balancing predictive performance with
interpretability. A model that produces strong predictive results can be
more informative when the contribution of its input features can also be
examined.

The present study addresses this computational research gap by evaluating
drug-specific Logistic Regression models and examining recurrent mutation
features across the analysed antiretroviral drug models.

1.7 Rationale of the Study

The rationale for this study is based on the availability of HIV-1
genomic mutation data and the potential of machine-learning methods to
identify predictive patterns within such data.

By developing separate models for multiple antiretroviral drugs, the
study allows differences in predictive performance to be examined across
drug-specific resistance outcomes. Feature-importance analysis further
provides a method for identifying recurrent mutation features that
contribute to model predictions.

The resulting framework provides an exploratory computational approach for
studying HIV-1 drug resistance while maintaining a clear distinction
between statistical prediction, biological interpretation, and clinical
validation.

1.8 Overview of the Present Study

In this study, viral genomic mutation features were analysed to develop
drug-specific Logistic Regression models for eight antiretroviral drugs:
SQV, ATV, NFV, LPV, FPV, IDV, TPV, and DRV.

The predictive performance of the models was evaluated using accuracy,
F1 score, and ROC-AUC. An overall mean-performance measure was also used
to compare the drug-specific models. Feature-importance analysis was
performed to identify recurrent mutation features and rank their
model-derived importance.

The study therefore combines predictive modelling and interpretability
analysis to investigate the feasibility of using viral genomic mutation
features for computational modelling of HIV-1 antiretroviral drug
resistance.

The following sections describe the specific problem addressed by the
study, its aim and objectives, relevant scientific literature, the
materials and methods used, and the resulting findings.
"""

# Display the introduction
print(introduction)

# Save the introduction
filename = "Step_141_Final_Report_Introduction.txt"

with open(filename, "w", encoding="utf-8") as f:
    f.write(introduction)

print("\n" + "=" * 70)
print("STEP 141 COMPLETE")
print("=" * 70)
print(f"✓ File saved: {filename}")
print("=" * 70)
```

```python
# ============================================================
# STEP 144: FINAL PROJECT LITERATURE REVIEW
# ============================================================

print("=" * 70)
print("STEP 144: FINAL PROJECT LITERATURE REVIEW")
print("=" * 70)

report_section = (
"4. LITERATURE REVIEW\n\n"

"4.1 HIV-1 Biology and Genetic Variability\n\n"
"Human immunodeficiency virus type 1 (HIV-1) is a genetically diverse "
"retrovirus whose replication process contributes to substantial viral "
"variation. The high rate of genetic change allows different viral "
"variants to emerge within infected populations. Under selective "
"pressures such as antiretroviral therapy, variants carrying mutations "
"that reduce drug susceptibility may become enriched. This genetic "
"variability is an important factor in the development and persistence "
"of HIV-1 drug resistance.\n\n"

"4.2 Antiretroviral Therapy and Protease Inhibitors\n\n"
"Combination antiretroviral therapy is used to suppress HIV-1 replication "
"and reduce disease progression. Antiretroviral drugs act on different "
"stages of the viral life cycle. Protease inhibitors target the HIV-1 "
"protease enzyme, which is required for the processing of viral "
"polyproteins during viral maturation. Changes in the protease gene can "
"therefore influence susceptibility to protease-inhibitor drugs.\n\n"
"The present study considers eight protease inhibitors: saquinavir (SQV), "
"atazanavir (ATV), nelfinavir (NFV), lopinavir (LPV), fosamprenavir (FPV), "
"indinavir (IDV), tipranavir (TPV), and darunavir (DRV).\n\n"

"4.3 HIV-1 Drug Resistance\n\n"
"HIV-1 drug resistance occurs when viral variants become less susceptible "
"to an antiretroviral drug. Resistance may result from the accumulation "
"of mutations that alter drug-target interactions or otherwise affect "
"viral replication in the presence of treatment. Resistance can be "
"drug-specific, and different combinations of mutations may contribute "
"to resistance against different antiretroviral compounds.\n\n"
"Because resistance can involve multiple mutations, computational "
"analysis is useful for examining relationships between several genomic "
"features and drug-resistance outcomes simultaneously.\n\n"

"4.4 Resistance-Associated Mutations\n\n"
"Resistance-associated mutations are amino-acid substitutions that have "
"been associated with changes in susceptibility to particular "
"antiretroviral drugs. Mutation notation describes a change from one "
"amino acid to another at a defined protein position. Examples identified "
"in the present computational analysis include I84V, V32I, L33F, I54L, "
"I54M, and I47V.\n\n"
"However, the presence of an individual mutation should not automatically "
"be interpreted as proof that the mutation independently causes drug "
"resistance. HIV-1 resistance can depend on combinations of mutations, "
"genetic background, drug exposure, and other biological factors. "
"Therefore, computational feature importance should be interpreted "
"carefully and in conjunction with established biological evidence.\n\n"

"4.5 Genotypic and Phenotypic Approaches\n\n"
"HIV-1 drug resistance can be investigated using genotypic and phenotypic "
"approaches. Genotypic approaches examine viral genetic variation and "
"interpret mutations associated with resistance. Phenotypic approaches "
"measure the susceptibility of viral variants to antiretroviral drugs. "
"Both approaches provide complementary information for understanding "
"drug resistance.\n\n"
"Large genomic and phenotypic datasets provide an opportunity to develop "
"computational models that learn relationships between viral mutations "
"and resistance outcomes. Such datasets are particularly useful for "
"exploratory machine-learning studies.\n\n"

"4.6 Bioinformatics in HIV-1 Drug-Resistance Research\n\n"
"Bioinformatics provides computational approaches for analysing biological "
"sequences, genetic variation, and phenotype-associated information. "
"In HIV-1 research, bioinformatics can be used to identify mutations, "
"compare viral sequences, examine resistance-associated patterns, and "
"support computational prediction of drug susceptibility.\n\n"
"The conversion of mutation information into machine-readable features "
"allows genomic variation to be incorporated into statistical and "
"machine-learning models.\n\n"

"4.7 Machine Learning for Drug-Resistance Prediction\n\n"
"Machine-learning methods can identify statistical patterns between input "
"features and known outcomes. In HIV-1 drug-resistance research, genomic "
"mutation features can be used as predictors while resistance or "
"susceptibility measurements provide the target outcome.\n\n"
"Machine learning is particularly useful when many genomic features are "
"present simultaneously and when relationships between features and "
"outcomes may be difficult to describe using simple rules. However, "
"strong predictive performance does not by itself demonstrate biological "
"causation or clinical usefulness. Appropriate validation is essential "
"to determine whether a model generalizes beyond the dataset used for "
"training and evaluation.\n\n"

"4.8 Logistic Regression in Biological Classification\n\n"
"Logistic Regression is a commonly used statistical and machine-learning "
"method for binary classification. The method estimates the probability "
"of an outcome based on a set of predictor variables. In genomic "
"applications, individual mutation features can be represented as "
"predictors and their model coefficients can provide information about "
"their contribution to the predicted outcome.\n\n"
"An important advantage of Logistic Regression is interpretability. "
"This makes it suitable for exploratory studies in which predictive "
"performance and the contribution of genomic features are both of "
"interest. In the present study, separate Logistic Regression models "
"were developed for eight drug-specific resistance outcomes.\n\n"

"4.9 Feature Importance and Interpretability\n\n"
"Interpretability is important when machine-learning models are applied "
"to biological data. Feature-importance analysis can help identify "
"which variables contribute strongly to model predictions. In the "
"present study, recurrent mutation features were ranked using mean "
"absolute model-derived importance across the drug-specific models.\n\n"
"I84V demonstrated the highest mean absolute importance in the final "
"feature ranking, followed by V32I and L33F. These results indicate "
"that the features contributed predictive information within the "
"analysed modelling framework. They should not, however, be interpreted "
"as independent evidence that these mutations cause resistance.\n\n"

"4.10 Model Evaluation in Classification Studies\n\n"
"Classification models can be evaluated using several complementary "
"performance measures. Accuracy represents the proportion of correctly "
"classified observations. The F1 score combines precision and recall and "
"can be informative when class distributions are not perfectly balanced. "
"ROC-AUC measures the ability of a classifier to distinguish between "
"classes across different decision thresholds.\n\n"
"The present analysis used accuracy, F1 score, and ROC-AUC to compare "
"the eight drug-specific models. Using multiple metrics provides a more "
"complete assessment than relying on a single performance measure.\n\n"

"4.11 Research Gap\n\n"
"Although HIV-1 drug resistance has been extensively studied using "
"genotypic, phenotypic, statistical, and computational approaches, there "
"remains value in evaluating interpretable machine-learning frameworks "
"across multiple drug-specific resistance outcomes. A useful framework "
"should not only compare predictive performance but should also examine "
"whether recurrent genomic features contribute consistently across "
"multiple models.\n\n"
"The present study addresses this computational gap by combining "
"drug-specific Logistic Regression modelling with recurrent mutation "
"feature analysis. This approach provides an exploratory framework for "
"examining both predictive performance and model-derived genomic signals.\n\n"

"4.12 Summary of Literature Review\n\n"
"The literature indicates that HIV-1 genetic variability and the "
"selection of resistance-associated mutations are important factors in "
"antiretroviral drug resistance. Bioinformatics and machine-learning "
"approaches provide opportunities to analyse complex relationships "
"between viral genomic variation and drug-resistance outcomes.\n\n"
"Logistic Regression provides an interpretable modelling approach that "
"can be combined with feature-importance analysis. Nevertheless, "
"computational associations require careful interpretation and external "
"validation. The present study therefore focuses on computational "
"feasibility and predictive patterns within the analysed dataset rather "
"than clinical application."
)

print(report_section)

output_file = "Step_144_Final_Project_Literature_Review.txt"

with open(output_file, "w", encoding="utf-8") as f:
    f.write(report_section)

print("=" * 70)
print("✓ File saved:")
print("1.", output_file)
print("=" * 70)
print("STEP 144 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 160 FIX: FINAL DOCUMENT POLISHING & SUBMISSION PREPARATION
# ================================================================

from pathlib import Path
from docx import Document
from docx.shared import Pt, Inches
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.oxml import OxmlElement
from docx.oxml.ns import qn

print("=" * 70)
print("STEP 160 FIX: FINAL DOCUMENT POLISHING")
print("=" * 70)

input_file = Path("FINAL_HIV1_DRUG_RESISTANCE_PROJECT_REPORT.docx")

if not input_file.exists():
    raise FileNotFoundError(
        "FINAL_HIV1_DRUG_RESISTANCE_PROJECT_REPORT.docx not found."
    )

print("✓ Final Word document found")

# Load document
doc = Document(input_file)
print("✓ Document loaded")

# ------------------------------------------------
# PAGE SETTINGS
# ------------------------------------------------

for section in doc.sections:
    section.top_margin = Inches(1)
    section.bottom_margin = Inches(1)
    section.left_margin = Inches(1)
    section.right_margin = Inches(1)

print("✓ 1-inch margins confirmed")

# ------------------------------------------------
# BODY FORMATTING
# ------------------------------------------------

for paragraph in doc.paragraphs:

    if paragraph.text.strip():

        paragraph.paragraph_format.line_spacing = 1.5
        paragraph.paragraph_format.space_after = Pt(6)

        for run in paragraph.runs:
            run.font.name = "Times New Roman"
            run.font.size = Pt(12)

print("✓ Body formatting applied")

# ------------------------------------------------
# HEADING FORMATTING
# ------------------------------------------------

for paragraph in doc.paragraphs:

    if paragraph.style.name == "Heading 1":

        for run in paragraph.runs:
            run.font.name = "Times New Roman"
            run.font.size = Pt(16)
            run.bold = True

    elif paragraph.style.name == "Heading 2":

        for run in paragraph.runs:
            run.font.name = "Times New Roman"
            run.font.size = Pt(14)
            run.bold = True

print("✓ Heading formatting applied")

# ------------------------------------------------
# PAGE NUMBERS
# ------------------------------------------------

for section in doc.sections:

    footer = section.footer

    paragraph = footer.paragraphs[0]
    paragraph.alignment = WD_ALIGN_PARAGRAPH.CENTER

    paragraph.clear()

    run = paragraph.add_run()

    field_begin = OxmlElement("w:fldChar")
    field_begin.set(qn("w:fldCharType"), "begin")

    field_code = OxmlElement("w:instrText")
    field_code.set(qn("xml:space"), "preserve")
    field_code.text = "PAGE"

    field_end = OxmlElement("w:fldChar")
    field_end.set(qn("w:fldCharType"), "end")

    run._r.append(field_begin)
    run._r.append(field_code)
    run._r.append(field_end)

    run.font.name = "Times New Roman"
    run.font.size = Pt(10)

print("✓ Page numbers added")

# ------------------------------------------------
# SAVE FINAL SUBMISSION COPY
# ------------------------------------------------

output_file = Path(
    "FINAL_HIV1_DRUG_RESISTANCE_PROJECT_REPORT_SUBMISSION.docx"
)

doc.save(output_file)

print("✓ Final submission document saved")

# ------------------------------------------------
# VERIFY
# ------------------------------------------------

if output_file.exists():

    size = output_file.stat().st_size

    print()
    print("=" * 70)
    print("STEP 160 COMPLETE")
    print("=" * 70)

    print(f"✓ File created: {output_file.name}")
    print(f"✓ File size: {size:,} bytes")
    print("✓ Times New Roman formatting")
    print("✓ 12 pt body text")
    print("✓ 1.5 line spacing")
    print("✓ 1-inch margins")
    print("✓ Page numbers")
    print("✓ Final submission copy ready")

else:
    raise FileNotFoundError("Final submission document was not created.")
```

```python
# ================================================================
# STEP 163: INSERT FINAL TABLES AND FIGURES INTO WORD DOCUMENT
# ================================================================

from pathlib import Path
from docx import Document
from docx.shared import Inches, Pt
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_CELL_VERTICAL_ALIGNMENT

print("=" * 70)
print("STEP 163: FINAL TABLES AND FIGURES INSERTION")
print("=" * 70)

# ------------------------------------------------
# 1. Files
# ------------------------------------------------

input_file = Path("FINAL_HIV1_DRUG_RESISTANCE_PROJECT_REPORT_SUBMISSION.docx")

if not input_file.exists():
    raise FileNotFoundError(
        "FINAL_HIV1_DRUG_RESISTANCE_PROJECT_REPORT_SUBMISSION.docx "
        "was not found."
    )

print("✓ Final submission document found")

# Figure files
figure_files = [
    ("Figure 1. Average Model Performance Across Eight Drug-Specific Models",
     "Step_153_Figure_1_Average_Model_Performance.png"),

    ("Figure 2. Best Model Performance Across Evaluation Metrics",
     "Step_153_Figure_2_Best_Model_Performance.png"),

    ("Figure 3. Recurrent Mutation Feature Importance",
     "Step_153_Figure_3_Mutation_Feature_Importance.png"),

    ("Figure 4. Consistency of Recurrent Mutation Features",
     "Step_153_Figure_4_Feature_Consistency.png")
]

# ------------------------------------------------
# 2. Check figures
# ------------------------------------------------

print()
print("Checking figure files...")

for title, filename in figure_files:
    if Path(filename).exists():
        print(f"✓ Found: {filename}")
    else:
        print(f"⚠ Missing: {filename}")

# ------------------------------------------------
# 3. Load document
# ------------------------------------------------

doc = Document(input_file)

print()
print("✓ Document loaded successfully")

# ------------------------------------------------
# 4. Add page break before visual section
# ------------------------------------------------

doc.add_page_break()

heading = doc.add_heading(
    "14. FINAL TABLES AND FIGURES",
    level=1
)

# ------------------------------------------------
# 5. Add Table 1
# ------------------------------------------------

doc.add_heading("Table 1. Overall Model Results", level=2)

table1 = doc.add_table(rows=1, cols=2)
table1.alignment = WD_TABLE_ALIGNMENT.CENTER
table1.style = "Table Grid"

hdr = table1.rows[0].cells
hdr[0].text = "Metric"
hdr[1].text = "Final Result"

table1_data = [
    ("Drug models analysed", "8"),
    ("Best accuracy model", "NFV = 95.48%"),
    ("Best F1 score model", "NFV = 95.95%"),
    ("Best ROC-AUC model", "SQV = 98.48%"),
    ("Best overall model", "NFV = 96.58%"),
    ("Average accuracy", "91.02%"),
    ("Average F1 score", "87.55%"),
    ("Average ROC-AUC", "96.79%"),
    ("Average model performance", "91.79%"),
    ("Mutation features identified", "6"),
    ("Top mutation feature", "I84V = 0.2750"),
    ("Second mutation feature", "V32I = 0.2671"),
    ("Third mutation feature", "L33F = 0.2521"),
    ("Feature consistency", "100% across all 8 drug models")
]

for metric, result in table1_data:
    row = table1.add_row().cells
    row[0].text = metric
    row[1].text = result

for row in table1.rows:
    for cell in row.cells:
        cell.vertical_alignment = WD_CELL_VERTICAL_ALIGNMENT.CENTER

doc.add_paragraph()

# ------------------------------------------------
# 6. Add Table 2
# ------------------------------------------------

doc.add_heading("Table 2. Recurrent Mutation Feature Ranking", level=2)

table2 = doc.add_table(rows=1, cols=5)
table2.alignment = WD_TABLE_ALIGNMENT.CENTER
table2.style = "Table Grid"

headers = [
    "Rank",
    "Mutation Feature",
    "Number of Drugs",
    "Mean Absolute Importance",
    "Consistency (%)"
]

for i, header in enumerate(headers):
    table2.rows[0].cells[i].text = header

mutation_data = [
    ("1", "I84V", "8", "0.275021", "100"),
    ("2", "V32I", "8", "0.267070", "100"),
    ("3", "L33F", "8", "0.252100", "100"),
    ("4", "I54L", "8", "0.197752", "100"),
    ("5", "I54M", "8", "0.194785", "100"),
    ("6", "I47V", "8", "0.190775", "100")
]

for data in mutation_data:
    cells = table2.add_row().cells
    for i, value in enumerate(data):
        cells[i].text = value

doc.add_paragraph()

# ------------------------------------------------
# 7. Add figures
# ------------------------------------------------

doc.add_heading("Figures", level=2)

for title, filename in figure_files:

    file_path = Path(filename)

    if not file_path.exists():
        continue

    p = doc.add_paragraph()
    p.alignment = WD_ALIGN_PARAGRAPH.CENTER

    run = p.add_run()
    run.add_picture(
        str(file_path),
        width=Inches(6.0)
    )

    caption = doc.add_paragraph()
    caption.alignment = WD_ALIGN_PARAGRAPH.CENTER

    cap_run = caption.add_run(title)
    cap_run.bold = True
    cap_run.font.name = "Times New Roman"
    cap_run.font.size = Pt(11)

    doc.add_paragraph()

    print(f"✓ Inserted: {title}")

# ------------------------------------------------
# 8. Interpretation note
# ------------------------------------------------

doc.add_heading("Important Interpretation Note", level=2)

note = doc.add_paragraph(
    "The feature-importance values represent model-derived predictive "
    "signals within the analysed dataset. They should not be interpreted "
    "as proof of biological causation or clinical significance. The "
    "models are intended for exploratory computational research and "
    "should not independently replace established HIV-1 drug-resistance "
    "testing or professional clinical assessment."
)

note.paragraph_format.line_spacing = 1.5

# ------------------------------------------------
# 9. Apply font to inserted tables
# ------------------------------------------------

for table in doc.tables:
    for row in table.rows:
        for cell in row.cells:
            for paragraph in cell.paragraphs:
                for run in paragraph.runs:
                    run.font.name = "Times New Roman"
                    run.font.size = Pt(10)

# ------------------------------------------------
# 10. Save final document
# ------------------------------------------------

output_file = "FINAL_HIV1_DRUG_RESISTANCE_PROJECT_REPORT_FINAL.docx"

doc.save(output_file)

print()
print("=" * 70)
print("✓ STEP 163 COMPLETE")
print("=" * 70)
print(f"✓ Final document created: {output_file}")
print("✓ Tables inserted")
print("✓ Figures inserted")
print("✓ Interpretation note inserted")
print("✓ Document saved successfully")
print("=" * 70)
```

```python
# ============================================================
# STEP 152D — INVESTIGATE REPEATED PATIENT/ISOLATE RECORDS
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 152D — REPEATED RECORD INVESTIGATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Compare repeated patient-isolate records
# ------------------------------------------------------------

print("\n1. REPEATED PATIENT + ISOLATE RECORDS")
print("-" * 70)

pair_counts = (
    original_df
    .groupby(["PtID", "IsolateName"])
    .size()
    .reset_index(name="Count")
)

repeated_pairs = pair_counts[pair_counts["Count"] > 1]

print("Repeated patient-isolate pairs:",
      len(repeated_pairs))

display(repeated_pairs.head(30))

# ------------------------------------------------------------
# 2. Examine the first 10 repeated pairs
# ------------------------------------------------------------

print("\n2. DETAILS OF FIRST REPEATED PAIRS")
print("-" * 70)

for _, row in repeated_pairs.head(10).iterrows():

    pt = row["PtID"]
    isolate = row["IsolateName"]

    subset = original_df[
        (original_df["PtID"] == pt) &
        (original_df["IsolateName"] == isolate)
    ]

    print(
        f"\nPtID={pt} | IsolateName={isolate} | "
        f"n={len(subset)}"
    )

    display(
        subset[
            [
                "SeqID",
                "PtID",
                "Subtype",
                "Method",
                "RefID",
                "IsolateName",
                "FPV",
                "ATV",
                "IDV",
                "LPV",
                "NFV",
                "SQV",
                "TPV",
                "DRV",
                "CompMutList"
            ]
        ]
    )

# ------------------------------------------------------------
# 3. Determine whether repeated patient-isolate rows
#    have identical drug measurements
# ------------------------------------------------------------

print("\n3. IDENTICAL DRUG-PHENOTYPE CHECK")
print("-" * 70)

drug_cols = [
    "FPV",
    "ATV",
    "IDV",
    "LPV",
    "NFV",
    "SQV",
    "TPV",
    "DRV"
]

identical_drug_pairs = 0
different_drug_pairs = 0

for _, row in repeated_pairs.iterrows():

    pt = row["PtID"]
    isolate = row["IsolateName"]

    subset = original_df[
        (original_df["PtID"] == pt) &
        (original_df["IsolateName"] == isolate)
    ]

    # Compare complete drug profiles
    profiles = subset[drug_cols].astype(object)

    if len(profiles.drop_duplicates()) == 1:
        identical_drug_pairs += 1
    else:
        different_drug_pairs += 1

print("Repeated pairs with identical drug profiles:",
      identical_drug_pairs)

print("Repeated pairs with different drug profiles:",
      different_drug_pairs)

# ------------------------------------------------------------
# 4. Compare mutation profiles
# ------------------------------------------------------------

print("\n4. IDENTICAL MUTATION-POSITION CHECK")
print("-" * 70)

mutation_cols = [
    f"P{i}" for i in range(1, 100)
]

identical_mutation_pairs = 0
different_mutation_pairs = 0

for _, row in repeated_pairs.iterrows():

    pt = row["PtID"]
    isolate = row["IsolateName"]

    subset = original_df[
        (original_df["PtID"] == pt) &
        (original_df["IsolateName"] == isolate)
    ]

    profiles = subset[mutation_cols].astype(object)

    if len(profiles.drop_duplicates()) == 1:
        identical_mutation_pairs += 1
    else:
        different_mutation_pairs += 1

print("Repeated pairs with identical mutation profiles:",
      identical_mutation_pairs)

print("Repeated pairs with different mutation profiles:",
      different_mutation_pairs)

# ------------------------------------------------------------
# 5. Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 152D SUMMARY")
print("=" * 70)

print("Repeated patient-isolate pairs:", len(repeated_pairs))
print("Identical drug profiles:", identical_drug_pairs)
print("Different drug profiles:", different_drug_pairs)
print("Identical mutation profiles:", identical_mutation_pairs)
print("Different mutation profiles:", different_mutation_pairs)

print("\nSTEP 152D COMPLETE")
```

```python
# ================================================================
# STEP 154 — PATIENT-GROUPED TRAIN / VALIDATION / TEST SPLIT
# ================================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import GroupShuffleSplit

print("=" * 70)
print("STEP 154 — PATIENT-GROUPED TRAIN / VALIDATION / TEST SPLIT")
print("=" * 70)

# ------------------------------------------------
# 1. Basic checks
# ------------------------------------------------

assert "PtID" in df.columns, "PtID column not found."

print("\n1. DATASET CHECK")
print("-" * 70)

print("Total observations:", len(df))
print("Unique patients:", df["PtID"].nunique())

# ------------------------------------------------
# 2. First split: 70% train, 30% temporary
# ------------------------------------------------

gss_1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

train_idx, temp_idx = next(
    gss_1.split(
        df,
        groups=df["PtID"]
    )
)

train_df = df.iloc[train_idx].copy()
temp_df = df.iloc[temp_idx].copy()

# ------------------------------------------------
# 3. Second split:
#    temporary 30% -> validation 15% + test 15%
# ------------------------------------------------

gss_2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_relative_idx, test_relative_idx = next(
    gss_2.split(
        temp_df,
        groups=temp_df["PtID"]
    )
)

val_df = temp_df.iloc[val_relative_idx].copy()
test_df = temp_df.iloc[test_relative_idx].copy()

# ------------------------------------------------
# 4. Patient IDs
# ------------------------------------------------

train_patients = set(train_df["PtID"].dropna())
val_patients = set(val_df["PtID"].dropna())
test_patients = set(test_df["PtID"].dropna())

# ------------------------------------------------
# 5. Leakage checks
# ------------------------------------------------

train_val_overlap = train_patients.intersection(val_patients)
train_test_overlap = train_patients.intersection(test_patients)
val_test_overlap = val_patients.intersection(test_patients)

# ------------------------------------------------
# 6. Report split sizes
# ------------------------------------------------

print("\n2. SPLIT SIZES")
print("-" * 70)

print(
    f"Training set:   {len(train_df):4d} rows "
    f"({100 * len(train_df) / len(df):.2f}%)"
)

print(
    f"Validation set: {len(val_df):4d} rows "
    f"({100 * len(val_df) / len(df):.2f}%)"
)

print(
    f"Test set:       {len(test_df):4d} rows "
    f"({100 * len(test_df) / len(df):.2f}%)"
)

print("\n3. UNIQUE PATIENTS PER SPLIT")
print("-" * 70)

print("Training patients:", len(train_patients))
print("Validation patients:", len(val_patients))
print("Test patients:", len(test_patients))

# ------------------------------------------------
# 7. Leakage results
# ------------------------------------------------

print("\n4. PATIENT OVERLAP / LEAKAGE CHECK")
print("-" * 70)

print("Train ∩ Validation:", len(train_val_overlap))
print("Train ∩ Test:", len(train_test_overlap))
print("Validation ∩ Test:", len(val_test_overlap))

# ------------------------------------------------
# 8. Row conservation
# ------------------------------------------------

print("\n5. ROW CONSERVATION CHECK")
print("-" * 70)

total_split_rows = (
    len(train_df) +
    len(val_df) +
    len(test_df)
)

print("Original rows:", len(df))
print("Rows after splitting:", total_split_rows)
print("Rows missing:", len(df) - total_split_rows)

# ------------------------------------------------
# 9. Patient conservation
# ------------------------------------------------

all_split_patients = (
    train_patients |
    val_patients |
    test_patients
)

print("\n6. PATIENT CONSERVATION CHECK")
print("-" * 70)

print("Original unique patients:", df["PtID"].nunique())
print("Patients across all splits:", len(all_split_patients))
print(
    "Patients missing:",
    df["PtID"].nunique() - len(all_split_patients)
)

# ------------------------------------------------
# 10. Final status
# ------------------------------------------------

print("\n" + "=" * 70)
print("STEP 154 SUMMARY")
print("=" * 70)

print("Train rows:", len(train_df))
print("Validation rows:", len(val_df))
print("Test rows:", len(test_df))

print("Train patients:", len(train_patients))
print("Validation patients:", len(val_patients))
print("Test patients:", len(test_patients))

print("\nPatient overlap:")
print("Train-Validation:", len(train_val_overlap))
print("Train-Test:", len(train_test_overlap))
print("Validation-Test:", len(val_test_overlap))

if (
    len(train_val_overlap) == 0 and
    len(train_test_overlap) == 0 and
    len(val_test_overlap) == 0
):
    print("\nSTATUS: PASS")
    print("No patient appears in more than one split.")
else:
    print("\nSTATUS: FAIL")
    print("Patient-level leakage detected.")

print("\nSTEP 154 COMPLETE")
print("=" * 70)
```

```python
1# ================================================================
# STEP 155 — TARGET DISTRIBUTION AUDIT AFTER PATIENT-GROUPED SPLIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 155 — TARGET DISTRIBUTION AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. CHECK REQUIRED DATASETS
# ------------------------------------------------

print("\n1. SPLIT OBJECT CHECK")
print("-" * 70)

print("Train shape:", train.shape)
print("Validation shape:", val.shape)
print("Test shape:", test.shape)

# ------------------------------------------------
# 2. IDENTIFY BINARY RESISTANCE TARGETS
# ------------------------------------------------

drug_targets = [
    "FPV_resistant",
    "ATV_resistant",
    "IDV_resistant",
    "LPV_resistant",
    "NFV_resistant",
    "SQV_resistant",
    "TPV_resistant",
    "DRV_resistant"
]

available_targets = [c for c in drug_targets if c in df.columns]

print("\n2. TARGET COLUMNS FOUND")
print("-" * 70)

print("Targets found:", len(available_targets))
print(available_targets)

missing_targets = [c for c in drug_targets if c not in df.columns]

if missing_targets:
    print("\nTargets not found:")
    print(missing_targets)

# ------------------------------------------------
# 3. TARGET DISTRIBUTION
# ------------------------------------------------

print("\n3. TARGET DISTRIBUTION BY SPLIT")
print("-" * 70)

for target in available_targets:

    print("\n" + "-" * 70)
    print("TARGET:", target)
    print("-" * 70)

    for split_name, split_df in [
        ("TRAIN", train),
        ("VALIDATION", val),
        ("TEST", test)
    ]:

        counts = split_df[target].value_counts(dropna=False)
        percentages = split_df[target].value_counts(
            normalize=True,
            dropna=False
        ) * 100

        print(f"\n{split_name}")

        for value in counts.index:
            print(
                f"  Class {value}: "
                f"{counts[value]} "
                f"({percentages[value]:.2f}%)"
            )

# ------------------------------------------------
# 4. MISSING TARGET VALUES
# ------------------------------------------------

print("\n4. MISSING TARGET VALUES")
print("-" * 70)

for target in available_targets:

    print(f"\n{target}")

    print("  Train missing:",
          train[target].isna().sum())

    print("  Validation missing:",
          val[target].isna().sum())

    print("  Test missing:",
          test[target].isna().sum())

# ------------------------------------------------
# 5. FINAL SUMMARY TABLE
# ------------------------------------------------

summary_rows = []

for target in available_targets:

    for split_name, split_df in [
        ("Train", train),
        ("Validation", val),
        ("Test", test)
    ]:

        total = len(split_df)
        missing = split_df[target].isna().sum()

        valid = total - missing

        resistant = (
            split_df[target]
            .eq(1)
            .sum()
        )

        susceptible = (
            split_df[target]
            .eq(0)
            .sum()
        )

        summary_rows.append({
            "Target": target,
            "Split": split_name,
            "Total_Rows": total,
            "Missing": missing,
            "Valid": valid,
            "Resistant_1": resistant,
            "Susceptible_0": susceptible,
            "Resistance_%": (
                resistant / valid * 100
                if valid > 0 else np.nan
            )
        })

target_summary = pd.DataFrame(summary_rows)

print("\n5. TARGET SUMMARY TABLE")
print("-" * 70)

display(target_summary)

# ------------------------------------------------
# 6. SAVE AUDIT TABLE
# ------------------------------------------------

target_summary.to_csv(
    "step_155_target_distribution_audit.csv",
    index=False
)

print("\nSaved:")
print("step_155_target_distribution_audit.csv")

# ------------------------------------------------
# 7. FINAL STATUS
# ------------------------------------------------

print("\n" + "=" * 70)
print("STEP 155 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 156 — RESTORE RESISTANCE TARGETS TO PATIENT-GROUPED SPLITS
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 156 — RESTORE RESISTANCE TARGETS")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. DEFINE TARGETS
# ----------------------------------------------------------------------

target_cols = [
    "FPV_resistant",
    "ATV_resistant",
    "IDV_resistant",
    "LPV_resistant",
    "NFV_resistant",
    "SQV_resistant",
    "TPV_resistant",
    "DRV_resistant"
]

print("\nTarget columns required:")
print(target_cols)


# ----------------------------------------------------------------------
# 2. FIND DATAFRAME(S) THAT CONTAIN THE TARGETS
# ----------------------------------------------------------------------

dataframes_with_targets = {}

for name, obj in globals().items():
    if isinstance(obj, pd.DataFrame):
        found = [c for c in target_cols if c in obj.columns]

        if len(found) > 0:
            dataframes_with_targets[name] = {
                "shape": obj.shape,
                "targets_found": found,
                "has_seqid": "SeqID" in obj.columns
            }

print("\nDataframes containing resistance targets:")
print("-" * 70)

if len(dataframes_with_targets) == 0:
    print("ERROR: No dataframe containing the target columns was found.")
else:
    for name, info in dataframes_with_targets.items():
        print(
            f"{name}: shape={info['shape']}, "
            f"targets={len(info['targets_found'])}/8, "
            f"SeqID={info['has_seqid']}"
        )


# ----------------------------------------------------------------------
# 3. SELECT THE BEST SOURCE DATAFRAME
# ----------------------------------------------------------------------

candidate_names = [
    name for name, info in dataframes_with_targets.items()
    if info["has_seqid"] and len(info["targets_found"]) == 8
]

if len(candidate_names) == 0:
    raise ValueError(
        "\nERROR: Could not find a dataframe containing all 8 targets "
        "and SeqID.\n"
        "Do not continue to Step 157. We need to recover the labels first."
    )

# Prefer commonly used names if present
preferred_names = [
    "df_labeled",
    "df_final",
    "df_clean",
    "df_processed",
    "df",
    "data"
]

source_name = None

for name in preferred_names:
    if name in candidate_names:
        source_name = name
        break

if source_name is None:
    source_name = candidate_names[0]

label_source = globals()[source_name]

print("\nSelected label source:")
print(f"DataFrame: {source_name}")
print(f"Shape: {label_source.shape}")


# ----------------------------------------------------------------------
# 4. CHECK SeqID UNIQUENESS IN LABEL SOURCE
# ----------------------------------------------------------------------

print("\nSeqID check:")

seqid_duplicates = label_source["SeqID"].duplicated().sum()

print(f"Duplicate SeqIDs in label source: {seqid_duplicates}")

if seqid_duplicates > 0:
    print(
        "\nWARNING: SeqID is not unique in the label source."
    )
    print(
        "We will check whether duplicate SeqIDs have identical targets."
    )

    duplicate_seqids = (
        label_source.loc[
            label_source["SeqID"].duplicated(keep=False),
            ["SeqID"] + target_cols
        ]
        .sort_values("SeqID")
    )

    display(duplicate_seqids.head(30))

else:
    print("SeqID is unique. PASS")


# ----------------------------------------------------------------------
# 5. CREATE MINIMAL LABEL TABLE
# ----------------------------------------------------------------------

label_table = label_source[["SeqID"] + target_cols].copy()

print("\nLabel table:")
print(label_table.shape)

display(label_table.head())


# ----------------------------------------------------------------------
# 6. FUNCTION TO RESTORE TARGETS
# ----------------------------------------------------------------------

def restore_targets(split_df, split_name):

    if "SeqID" not in split_df.columns:
        raise ValueError(
            f"{split_name} does not contain SeqID. "
            "Cannot safely restore targets."
        )

    # Remove any accidental old target columns
    split_base = split_df.drop(
        columns=[c for c in target_cols if c in split_df.columns],
        errors="ignore"
    ).copy()

    # Check that every SeqID exists in label table
    missing_seqids = ~split_base["SeqID"].isin(label_table["SeqID"])

    missing_count = missing_seqids.sum()

    print(f"\n{split_name}:")
    print(f"Rows: {len(split_base)}")
    print(f"SeqIDs missing from label table: {missing_count}")

    if missing_count > 0:
        print("Missing SeqIDs:")
        print(split_base.loc[missing_seqids, "SeqID"].head(20).tolist())

        raise ValueError(
            f"{split_name} contains SeqIDs without resistance labels."
        )

    # Merge labels using SeqID
    restored = split_base.merge(
        label_table,
        on="SeqID",
        how="left",
        validate="many_to_one"
    )

    # Confirm row count did not change
    if len(restored) != len(split_base):
        raise ValueError(
            f"{split_name}: row count changed during target restoration."
        )

    return restored


# ----------------------------------------------------------------------
# 7. RESTORE TARGETS TO TRAIN / VALIDATION / TEST
# ----------------------------------------------------------------------

train_df = restore_targets(train_df, "TRAIN")
val_df   = restore_targets(val_df, "VALIDATION")
test_df  = restore_targets(test_df, "TEST")


# ----------------------------------------------------------------------
# 8. VERIFY ALL TARGETS EXIST
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET RESTORATION CHECK")
print("=" * 70)

for split_name, split_df in [
    ("TRAIN", train_df),
    ("VALIDATION", val_df),
    ("TEST", test_df)
]:

    found = [c for c in target_cols if c in split_df.columns]
    missing = [c for c in target_cols if c not in split_df.columns]

    print(f"\n{split_name}")
    print(f"Shape: {split_df.shape}")
    print(f"Targets found: {len(found)}/8")
    print(f"Missing: {missing}")


# ----------------------------------------------------------------------
# 9. VERIFY TARGET VALUES
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET VALUE CHECK")
print("=" * 70)

for split_name, split_df in [
    ("TRAIN", train_df),
    ("VALIDATION", val_df),
    ("TEST", test_df)
]:

    print(f"\n--- {split_name} ---")

    for target in target_cols:
        print(
            f"{target}: "
            f"n={split_df[target].notna().sum()}, "
            f"missing={split_df[target].isna().sum()}, "
            f"unique={split_df[target].dropna().unique()}"
        )


# ----------------------------------------------------------------------
# 10. FINAL SHAPE CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL STEP 156 CHECK")
print("=" * 70)

print(f"Train:      {train_df.shape}")
print(f"Validation: {val_df.shape}")
print(f"Test:       {test_df.shape}")

print("\nExpected row counts:")
print("Train:      3139")
print("Validation: 604")
print("Test:       607")

assert len(train_df) == 3139
assert len(val_df) == 604
assert len(test_df) == 607

assert all(c in train_df.columns for c in target_cols)
assert all(c in val_df.columns for c in target_cols)
assert all(c in test_df.columns for c in target_cols)

print("\nSTATUS: TARGET RESTORATION PASS")
print("Patient-grouped split remains intact.")
print("All 8 resistance targets are now attached.")

print("\n" + "=" * 70)
print("STEP 156 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 156D — RECOVER PREVIOUS RESISTANCE CUTOFF CODE
# ================================================================

print("=" * 70)
print("STEP 156D — SEARCHING COLAB INPUT HISTORY FOR CUTOFFS")
print("=" * 70)

keywords = [
    "cutoff",
    "threshold",
    "resistant",
    "resistance",
    "_resistant",
    "fold",
    "FPV",
    "ATV",
    "IDV",
    "LPV",
    "NFV",
    "SQV",
    "TPV",
    "DRV"
]

# Make a stable copy first so the history list cannot change during iteration
history = list(In)

matches = []

for i, code in enumerate(history):
    if not isinstance(code, str):
        continue

    code_lower = code.lower()

    # Focus on cells that look related to resistance labeling/cutoffs
    if any(k.lower() in code_lower for k in keywords):
        matches.append((i, code))

print()
print("Matching historical cells found:", len(matches))
print("=" * 70)

for i, code in matches:
    print(f"\n{'#' * 70}")
    print(f"COLAB CELL HISTORY INDEX: {i}")
    print(f"{'#' * 70}")
    print(code)

print("\n" + "=" * 70)
print("STEP 156D COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 156E — RECOVER EXACT RESISTANCE-LABEL GENERATION CODE
# ================================================================

import re

print("=" * 70)
print("STEP 156E — RECOVERING EXACT RESISTANCE-LABEL CODE")
print("=" * 70)

# ---------------------------------------------------------------
# 1. Search ALL previous Colab cells
# ---------------------------------------------------------------

history = list(In)

patterns = [
    "_resistant",
    "resistant",
    "cutoff",
    "threshold",
    "resistance",
    "fold_change",
    "foldchange",
    "FPV",
    "ATV",
    "IDV",
    "LPV",
    "NFV",
    "SQV",
    "TPV",
    "DRV"
]

# We are especially interested in cells that contain:
#   - resistant label creation
#   - numeric cutoff dictionaries
#   - comparisons such as >= or >
#   - assignments to *_resistant

candidate_cells = []

for i, code in enumerate(history):

    if not isinstance(code, str):
        continue

    low = code.lower()

    score = 0

    # Label-generation signals
    if "_resistant" in low:
        score += 5

    if "resistant" in low:
        score += 2

    if "cutoff" in low or "threshold" in low:
        score += 3

    if "fold_change" in low or "foldchange" in low:
        score += 3

    # Numeric comparison signals
    if ">=" in code:
        score += 1

    if ">" in code:
        score += 1

    # Drug names
    drug_hits = sum(
        drug.lower() in low
        for drug in [
            "FPV", "ATV", "IDV", "LPV",
            "NFV", "SQV", "TPV", "DRV"
        ]
    )

    score += drug_hits

    if score >= 6:
        candidate_cells.append(
            (score, i, code)
        )


# ---------------------------------------------------------------
# 2. Sort strongest candidates first
# ---------------------------------------------------------------

candidate_cells = sorted(
    candidate_cells,
    key=lambda x: (x[0], x[1]),
    reverse=True
)

print("\nCandidate historical cells found:",
      len(candidate_cells))

print("\n" + "=" * 70)
print("MOST RELEVANT HISTORICAL CELLS")
print("=" * 70)


# ---------------------------------------------------------------
# 3. Print candidates
# ---------------------------------------------------------------

for rank, (score, index, code) in enumerate(
    candidate_cells[:25],
    start=1
):

    print("\n" + "#" * 70)
    print(
        f"CANDIDATE {rank} | "
        f"HISTORY INDEX = {index} | "
        f"SCORE = {score}"
    )
    print("#" * 70)

    print(code)


# ---------------------------------------------------------------
# 4. Specifically identify cells that CREATE *_resistant
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("CELLS CONTAINING EXPLICIT *_resistant CREATION")
print("=" * 70)

explicit_label_cells = []

for i, code in enumerate(history):

    if not isinstance(code, str):
        continue

    if "_resistant" in code:

        explicit_label_cells.append(
            (i, code)
        )

print(
    "Cells containing '_resistant':",
    len(explicit_label_cells)
)

for i, code in explicit_label_cells:

    print("\n" + "-" * 70)
    print("HISTORY INDEX:", i)
    print("-" * 70)

    # Print only lines likely related to labels
    lines = code.splitlines()

    relevant_lines = []

    for line in lines:

        low = line.lower()

        if (
            "_resistant" in low
            or "cutoff" in low
            or "threshold" in low
            or "fold" in low
            or ">=" in line
            or ">" in line
        ):
            relevant_lines.append(line)

    if relevant_lines:
        print("\n".join(relevant_lines))

    else:
        print(code)


print("\n" + "=" * 70)
print("STEP 156E COMPLETE")
print("=" * 70)

print("\nIMPORTANT:")
print("Do NOT recreate resistance labels yet.")
print("First inspect the cutoff values recovered above.")
print("Do NOT continue to Step 157.")
```

```python
# ================================================================
# STEP 156F — DEEP SEARCH FOR ORIGINAL RESISTANCE CUTOFF VALUES
# ================================================================

import re

print("=" * 70)
print("STEP 156F — DEEP SEARCH FOR RESISTANCE CUTOFF VALUES")
print("=" * 70)

history = list(In)

drug_names = [
    "FPV", "ATV", "IDV", "LPV",
    "NFV", "SQV", "TPV", "DRV"
]

keywords = [
    "cutoff",
    "cutoffs",
    "threshold",
    "thresholds",
    "fold_change",
    "foldchange",
    "resistance_cutoff",
    "resistance_cutoffs",
    "drug_cutoffs",
    "cutoff_dict",
    "resistant"
]

# ---------------------------------------------------------------
# 1. SEARCH FOR DICTIONARIES / ASSIGNMENTS
# ---------------------------------------------------------------

print("\n1. POSSIBLE CUTOFF DEFINITIONS")
print("-" * 70)

found = []

for i, code in enumerate(history):

    if not isinstance(code, str):
        continue

    low = code.lower()

    keyword_hits = sum(
        k.lower() in low
        for k in keywords
    )

    drug_hits = sum(
        drug.lower() in low
        for drug in drug_names
    )

    # Strong candidates should contain both cutoff-related
    # terminology and several drug names.
    if keyword_hits >= 1 and drug_hits >= 2:

        found.append(
            (i, keyword_hits, drug_hits, code)
        )

found = sorted(
    found,
    key=lambda x: (x[1] + x[2], x[0]),
    reverse=True
)

print("Potential cutoff-related cells:", len(found))

for i, keyword_hits, drug_hits, code in found:

    print("\n" + "#" * 70)
    print(
        f"HISTORY INDEX = {i} | "
        f"keyword hits = {keyword_hits} | "
        f"drug hits = {drug_hits}"
    )
    print("#" * 70)

    print(code)


# ---------------------------------------------------------------
# 2. SEARCH FOR NUMERIC VALUES NEAR DRUG NAMES
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("2. NUMERIC CUTOFF PATTERN SEARCH")
print("=" * 70)

number_pattern = r"\b\d+(?:\.\d+)?\b"

numeric_candidates = []

for i, code in enumerate(history):

    if not isinstance(code, str):
        continue

    low = code.lower()

    if (
        "cutoff" in low
        or "threshold" in low
        or "fold_change" in low
        or "foldchange" in low
    ):

        numbers = re.findall(
            number_pattern,
            code
        )

        drug_hits = [
            drug for drug in drug_names
            if drug.lower() in low
        ]

        if numbers and drug_hits:

            numeric_candidates.append(
                (i, drug_hits, numbers, code)
            )

print(
    "Cells containing cutoff terminology + "
    "drug names + numeric values:",
    len(numeric_candidates)
)

for i, drugs, numbers, code in numeric_candidates:

    print("\n" + "-" * 70)
    print("HISTORY INDEX:", i)
    print("Drug names found:", drugs)
    print("Numbers found:", numbers)
    print("-" * 70)

    # Print lines containing relevant information
    for line in code.splitlines():

        low_line = line.lower()

        if (
            "cutoff" in low_line
            or "threshold" in low_line
            or "fold" in low_line
            or any(
                drug.lower() in low_line
                for drug in drug_names
            )
        ):
            print(line)


# ---------------------------------------------------------------
# 3. SEARCH CURRENT MEMORY FOR CUTOFF DICTIONARIES
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("3. CURRENT COLAB MEMORY SEARCH")
print("=" * 70)

possible_variables = []

for name, obj in list(globals().items()):

    try:

        text = repr(obj)

        drug_hits = sum(
            drug in text
            for drug in drug_names
        )

        cutoff_hit = (
            "cutoff" in name.lower()
            or "threshold" in name.lower()
            or "cutoff" in text.lower()
            or "threshold" in text.lower()
        )

        if cutoff_hit and drug_hits >= 2:

            possible_variables.append(
                (name, type(obj).__name__, drug_hits, text[:1000])
            )

    except Exception:
        pass


if possible_variables:

    for name, obj_type, drug_hits, preview in possible_variables:

        print("\n" + "-" * 70)
        print("Variable:", name)
        print("Type:", obj_type)
        print("Drug matches:", drug_hits)
        print("Preview:")
        print(preview)

else:

    print(
        "No cutoff-related variable containing multiple "
        "drug names was found in current memory."
    )


print("\n" + "=" * 70)
print("STEP 156F COMPLETE")
print("=" * 70)

print("\nIMPORTANT:")
print("Do NOT create or modify resistance labels yet.")
print("We need the original numerical cutoff values first.")
```

```python
# ================================================================
# STEP 156G — TARGETED RECOVERY OF ORIGINAL LABEL-GENERATION LOGIC
# ================================================================

import re
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 156G — TARGETED RECOVERY OF ORIGINAL LABEL LOGIC")
print("=" * 70)

history = list(In)

drug_names = [
    "FPV", "ATV", "IDV", "LPV",
    "NFV", "SQV", "TPV", "DRV"
]

# ---------------------------------------------------------------
# 1. SEARCH ONLY EARLIER CELLS
# ---------------------------------------------------------------

print("\n1. SEARCHING EARLIER COLAB CELLS")
print("-" * 70)

print("Total history cells:", len(history))

# Step 156A onward is recovery/audit code.
# We want cells BEFORE those recovery steps.
early_cells = history[:15]

print("Cells being inspected:", len(early_cells))


# ---------------------------------------------------------------
# 2. SEARCH FOR ACTUAL LABEL ASSIGNMENTS
# ---------------------------------------------------------------

assignment_patterns = [
    r"_resistant\s*=",
    r"\[.*_resistant.*\]\s*=",
    r"resistant\s*=",
    r"cutoff\s*=",
    r"cutoffs\s*=",
    r"threshold\s*=",
    r"thresholds\s*=",
    r"fold_change",
    r"foldchange",
    r"\.apply\(",
    r"np\.where",
    r"np\.select",
    r">=",
    r"<=",
    r"\.astype\(int\)"
]

candidate_cells = []

for i, code in enumerate(early_cells):

    if not isinstance(code, str):
        continue

    hits = []

    for pattern in assignment_patterns:

        if re.search(pattern, code, flags=re.IGNORECASE):
            hits.append(pattern)

    drug_hits = [
        drug for drug in drug_names
        if drug.lower() in code.lower()
    ]

    # Strong candidate:
    # actual resistance logic + at least one drug
    if len(hits) >= 1 and len(drug_hits) >= 1:

        candidate_cells.append(
            (
                i,
                hits,
                drug_hits,
                code
            )
        )


print(
    "Potential earlier label-generation cells:",
    len(candidate_cells)
)


# ---------------------------------------------------------------
# 3. DISPLAY CANDIDATE CELLS
# ---------------------------------------------------------------

for i, hits, drugs, code in candidate_cells:

    print("\n" + "#" * 70)
    print("HISTORY INDEX:", i)
    print("Drug names:", drugs)
    print("Pattern hits:", hits)
    print("#" * 70)

    print(code)


# ---------------------------------------------------------------
# 4. SEARCH FOR NUMERIC DICTIONARIES SPECIFICALLY
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("4. SEARCHING FOR NUMERIC DICTIONARIES")
print("=" * 70)

dictionary_pattern = re.compile(
    r"""
    (?:FPV|ATV|IDV|LPV|NFV|SQV|TPV|DRV)
    \s*
    :
    \s*
    \d+(?:\.\d+)?
    """,
    flags=re.IGNORECASE | re.VERBOSE
)

dictionary_hits = []

for i, code in enumerate(early_cells):

    if not isinstance(code, str):
        continue

    matches_here = dictionary_pattern.findall(code)

    if matches_here:

        dictionary_hits.append(
            (i, matches_here, code)
        )


print(
    "Earlier cells containing drug:value patterns:",
    len(dictionary_hits)
)

for i, matches_here, code in dictionary_hits:

    print("\n" + "-" * 70)
    print("HISTORY INDEX:", i)
    print("Drug:value patterns:", matches_here)
    print("-" * 70)

    print(code)


# ---------------------------------------------------------------
# 5. SEARCH ALL CURRENT DATAFRAMES FOR FOLD-CHANGE COLUMNS
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("5. SEARCHING CURRENT DATAFRAMES FOR FOLD-CHANGE COLUMNS")
print("=" * 70)

df_candidates = []

for name, obj in list(globals().items()):

    try:

        if isinstance(obj, pd.DataFrame):

            matching_cols = []

            for col in obj.columns:

                col_low = str(col).lower()

                if (
                    "fold" in col_low
                    or "change" in col_low
                    or "fc" == col_low
                    or "phenotype" in col_low
                    or "pheno" in col_low
                ):

                    matching_cols.append(col)

            if matching_cols:

                df_candidates.append(
                    (
                        name,
                        obj.shape,
                        matching_cols
                    )
                )

    except Exception:
        pass


if df_candidates:

    for name, shape, cols in df_candidates:

        print("\nDataFrame:", name)
        print("Shape:", shape)
        print("Potential phenotype/fold-change columns:")

        for col in cols:
            print("  -", col)

else:

    print(
        "No obvious fold-change/phenotype columns "
        "found in current DataFrames."
    )


# ---------------------------------------------------------------
# 6. SEARCH CURRENT DATAFRAME COLUMNS FOR DRUG-SPECIFIC VALUES
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("6. DRUG-SPECIFIC PHENOTYPE COLUMN SEARCH")
print("=" * 70)

for name, obj in list(globals().items()):

    try:

        if isinstance(obj, pd.DataFrame):

            drug_columns = []

            for col in obj.columns:

                col_low = str(col).lower()

                if any(
                    drug.lower() in col_low
                    for drug in drug_names
                ):

                    drug_columns.append(col)

            if drug_columns:

                print("\nDataFrame:", name)
                print("Shape:", obj.shape)

                for col in drug_columns:
                    print("  ", col)

    except Exception:
        pass


# ---------------------------------------------------------------
# 7. FINAL STATUS
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 156G COMPLETE")
print("=" * 70)

print("""
IMPORTANT:

Do NOT create new resistance labels.
Do NOT guess cutoff values.
Do NOT modify train_df, val_df, or test_df.

The purpose of this step is to determine whether the original
label-generation logic or the raw phenotype/fold-change columns
still exist in the Colab history or current DataFrames.
""")
```

```python
# ======================================================================
# STEP 156I — RECOVER EXISTING LABEL ↔ PHENOTYPE RELATIONSHIPS
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 156I — EXISTING LABEL ↔ PHENOTYPE RELATIONSHIP AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. REQUIRED OBJECT CHECK
# ----------------------------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

required_objects = ["original_df", "df"]

for obj_name in required_objects:
    exists = obj_name in globals()
    print(f"{obj_name}: {'FOUND' if exists else 'NOT FOUND'}")

if "original_df" not in globals():
    raise RuntimeError(
        "original_df is not available. STOP — do not continue."
    )

# ----------------------------------------------------------------------
# 2. FIND EXISTING RESISTANCE LABEL COLUMNS
# ----------------------------------------------------------------------

print("\n2. EXISTING RESISTANCE LABEL COLUMNS")
print("-" * 70)

drug_names = [
    "FPV", "ATV", "IDV", "LPV",
    "NFV", "SQV", "TPV", "DRV"
]

label_columns = [
    f"{drug}_resistant"
    for drug in drug_names
    if f"{drug}_resistant" in original_df.columns
]

print("Labels found in original_df:")
print(label_columns)

if len(label_columns) == 0:
    print("\nNo resistance labels found in original_df.")
    print("Searching other current DataFrames...")

    for name, obj in list(globals().items()):
        if isinstance(obj, pd.DataFrame):

            found = [
                c for c in obj.columns
                if c.endswith("_resistant")
            ]

            if found:
                print(
                    f"{name}: shape={obj.shape}, "
                    f"labels={found}"
                )

# ----------------------------------------------------------------------
# 3. INSPECT EXISTING LABEL VALUES
# ----------------------------------------------------------------------

print("\n3. EXISTING LABEL VALUE DISTRIBUTIONS")
print("-" * 70)

for drug in drug_names:

    label = f"{drug}_resistant"

    if label not in original_df.columns:
        print(f"\n{label}: NOT FOUND")
        continue

    print(f"\n{label}")

    print(
        original_df[label]
        .value_counts(dropna=False)
        .sort_index()
    )

    print(
        "dtype:",
        original_df[label].dtype
    )

# ----------------------------------------------------------------------
# 4. MATCH EACH LABEL WITH ITS PHENOTYPE COLUMN
# ----------------------------------------------------------------------

print("\n4. LABEL ↔ PHENOTYPE MATCHING")
print("-" * 70)

relationship_rows = []

for drug in drug_names:

    phenotype = drug
    label = f"{drug}_resistant"

    if phenotype not in original_df.columns:
        print(f"{drug}: phenotype NOT FOUND")
        continue

    if label not in original_df.columns:
        print(f"{drug}: label NOT FOUND")
        continue

    temp = original_df[
        [phenotype, label]
    ].copy()

    temp = temp.dropna(
        subset=[phenotype, label]
    )

    print("\n" + "-" * 60)
    print("DRUG:", drug)
    print("Phenotype column:", phenotype)
    print("Label column:", label)
    print("Rows with both values:", len(temp))

    print("\nLabel distribution:")
    print(
        temp[label]
        .value_counts(dropna=False)
        .sort_index()
    )

    # --------------------------------------------------------------
    # Phenotype summary separately for each EXISTING label
    # --------------------------------------------------------------

    print("\nPhenotype summary by EXISTING label:")

    summary = (
        temp
        .groupby(label)[phenotype]
        .agg(
            count="count",
            min="min",
            q25=lambda x: x.quantile(0.25),
            median="median",
            q75=lambda x: x.quantile(0.75),
            max="max"
        )
        .reset_index()
    )

    display(summary)

    # --------------------------------------------------------------
    # Check minimum phenotype among label=1
    # and maximum phenotype among label=0
    # --------------------------------------------------------------

    resistant_values = temp.loc[
        temp[label] == 1,
        phenotype
    ]

    susceptible_values = temp.loc[
        temp[label] == 0,
        phenotype
    ]

    min_resistant = (
        resistant_values.min()
        if len(resistant_values) > 0
        else np.nan
    )

    max_susceptible = (
        susceptible_values.max()
        if len(susceptible_values) > 0
        else np.nan
    )

    relationship_rows.append({
        "Drug": drug,
        "Rows_with_both": len(temp),
        "Label_0": int((temp[label] == 0).sum()),
        "Label_1": int((temp[label] == 1).sum()),
        "Min_phenotype_label1": min_resistant,
        "Max_phenotype_label0": max_susceptible
    })

    print(
        "\nMinimum phenotype among existing label=1:",
        min_resistant
    )

    print(
        "Maximum phenotype among existing label=0:",
        max_susceptible
    )

# ----------------------------------------------------------------------
# 5. OVERALL RELATIONSHIP TABLE
# ----------------------------------------------------------------------

relationship_df = pd.DataFrame(
    relationship_rows
)

print("\n5. OVERALL RELATIONSHIP TABLE")
print("-" * 70)

display(relationship_df)

# ----------------------------------------------------------------------
# 6. CHECK WHETHER LABELS ARE MONOTONICALLY RELATED TO PHENOTYPE
# ----------------------------------------------------------------------

print("\n6. MONOTONICITY / SEPARATION CHECK")
print("-" * 70)

for drug in drug_names:

    phenotype = drug
    label = f"{drug}_resistant"

    if (
        phenotype not in original_df.columns
        or label not in original_df.columns
    ):
        continue

    temp = original_df[
        [phenotype, label]
    ].dropna()

    if temp.empty:
        continue

    label0 = temp.loc[
        temp[label] == 0,
        phenotype
    ]

    label1 = temp.loc[
        temp[label] == 1,
        phenotype
    ]

    if len(label0) == 0 or len(label1) == 0:
        print(
            f"{drug}: insufficient 0/1 classes for comparison"
        )
        continue

    max0 = label0.max()
    min1 = label1.min()

    if min1 > max0:
        status = "PERFECT SEPARATION"
    elif min1 == max0:
        status = "TOUCHING / BORDERLINE"
    else:
        status = "OVERLAPPING PHENOTYPE RANGES"

    print(
        f"{drug}: {status} | "
        f"max(label=0)={max0} | "
        f"min(label=1)={min1}"
    )

# ----------------------------------------------------------------------
# 7. SAVE AUDIT ONLY
# ----------------------------------------------------------------------

relationship_df.to_csv(
    "step_156I_label_phenotype_relationship_audit.csv",
    index=False
)

print("\nSaved:")
print(
    "step_156I_label_phenotype_relationship_audit.csv"
)

# ----------------------------------------------------------------------
# 8. FINAL SAFETY CHECK
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 156I COMPLETE")
print("=" * 70)

print("""
IMPORTANT:
- No resistance labels were created.
- No resistance labels were modified.
- original_df was not modified.
- train_df was not modified.
- val_df was not modified.
- test_df was not modified.
- train was not modified.
- val was not modified.
- test was not modified.

This step only audits the relationship between EXISTING labels
and their corresponding original phenotype values.
""")
```

```python
# ================================================================
# STEP 160: PHENOTYPE ASSAY AND CUTOFF PROVENANCE AUDIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 160: PHENOTYPE ASSAY AND CUTOFF PROVENANCE AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. DATASET CHECK
# ------------------------------------------------

print("\n1. DATASET CHECK")
print("-" * 70)

if 'original_df' not in globals():
    raise RuntimeError("original_df not found.")

print("original_df:", original_df.shape)

# ------------------------------------------------
# 2. METADATA COLUMNS
# ------------------------------------------------

print("\n2. METADATA COLUMNS")
print("-" * 70)

metadata_cols = [
    'SeqID', 'PtID', 'Subtype', 'Method',
    'RefID', 'Type', 'IsolateName', 'SeqType'
]

for col in metadata_cols:
    if col in original_df.columns:
        print(f"\n--- {col} ---")
        print("dtype:", original_df[col].dtype)
        print("unique:", original_df[col].nunique(dropna=True))
        print("missing:", original_df[col].isna().sum())

        print("Top values:")
        print(
            original_df[col]
            .value_counts(dropna=False)
            .head(20)
        )

# ------------------------------------------------
# 3. METHOD / ASSAY AUDIT
# ------------------------------------------------

print("\n3. PHENOTYPE ASSAY / METHOD AUDIT")
print("-" * 70)

if 'Method' in original_df.columns:

    method_counts = (
        original_df['Method']
        .value_counts(dropna=False)
        .rename_axis('Method')
        .reset_index(name='Count')
    )

    print(method_counts.to_string(index=False))

else:
    print("Method column not available.")

# ------------------------------------------------
# 4. TYPE AUDIT
# ------------------------------------------------

print("\n4. TYPE FIELD AUDIT")
print("-" * 70)

if 'Type' in original_df.columns:

    type_counts = (
        original_df['Type']
        .value_counts(dropna=False)
        .rename_axis('Type')
        .reset_index(name='Count')
    )

    print(type_counts.to_string(index=False))

else:
    print("Type column not available.")

# ------------------------------------------------
# 5. SEQUENCE TYPE AUDIT
# ------------------------------------------------

print("\n5. SEQUENCE TYPE AUDIT")
print("-" * 70)

if 'SeqType' in original_df.columns:

    print(
        original_df['SeqType']
        .value_counts(dropna=False)
        .to_string()
    )

else:
    print("SeqType column not available.")

# ------------------------------------------------
# 6. SUBTYPE AUDIT
# ------------------------------------------------

print("\n6. HIV-1 SUBTYPE AUDIT")
print("-" * 70)

if 'Subtype' in original_df.columns:

    subtype_counts = (
        original_df['Subtype']
        .value_counts(dropna=False)
        .rename_axis('Subtype')
        .reset_index(name='Count')
    )

    print(subtype_counts.head(30).to_string(index=False))

# ------------------------------------------------
# 7. REFID AUDIT
# ------------------------------------------------

print("\n7. REFERENCE ID AUDIT")
print("-" * 70)

if 'RefID' in original_df.columns:

    print(
        "Unique RefID values:",
        original_df['RefID'].nunique(dropna=True)
    )

    print(
        original_df['RefID']
        .value_counts(dropna=False)
        .head(30)
        .to_string()
    )

# ------------------------------------------------
# 8. DRUG / PHENOTYPE COLUMNS
# ------------------------------------------------

drugs = [
    'FPV', 'ATV', 'IDV', 'LPV',
    'NFV', 'SQV', 'TPV', 'DRV'
]

print("\n8. DRUG PHENOTYPE COLUMNS")
print("-" * 70)

for drug in drugs:
    if drug in original_df.columns:
        print(f"✓ {drug}")
    else:
        print(f"✗ {drug}")

# ------------------------------------------------
# 9. METHOD × DRUG AVAILABILITY
# ------------------------------------------------

print("\n9. METHOD × DRUG AVAILABILITY")
print("-" * 70)

if 'Method' in original_df.columns:

    method_drug = []

    for method, group in original_df.groupby(
        'Method',
        dropna=False
    ):

        for drug in drugs:

            if drug in group.columns:

                n = group[drug].notna().sum()

                method_drug.append({
                    'Method': method,
                    'Drug': drug,
                    'Non_missing': n
                })

    method_drug_df = pd.DataFrame(method_drug)

    print(method_drug_df.to_string(index=False))

# ------------------------------------------------
# 10. DRUG PHENOTYPE SUMMARY BY METHOD
# ------------------------------------------------

print("\n10. PHENOTYPE SUMMARY BY METHOD")
print("-" * 70)

if 'Method' in original_df.columns:

    for drug in drugs:

        if drug not in original_df.columns:
            continue

        print(f"\n===== {drug} =====")

        temp = (
            original_df
            .groupby('Method', dropna=False)[drug]
            .agg(
                N='count',
                Median='median',
                Mean='mean',
                Minimum='min',
                Maximum='max'
            )
            .reset_index()
        )

        print(temp.to_string(index=False))

# ------------------------------------------------
# 11. IDENTIFY POTENTIAL ASSAY-SPECIFIC GROUPS
# ------------------------------------------------

print("\n11. POTENTIAL ASSAY-SPECIFIC GROUPS")
print("-" * 70)

if 'Method' in original_df.columns:

    for method in original_df['Method'].drop_duplicates():

        subset = original_df[
            original_df['Method'] == method
        ]

        print("\nMethod:", method)
        print("Rows:", len(subset))

        available = {}

        for drug in drugs:
            if drug in subset.columns:
                available[drug] = int(
                    subset[drug].notna().sum()
                )

        print("Phenotype availability:")
        print(available)

# ------------------------------------------------
# 12. SAVE AUDIT
# ------------------------------------------------

audit_rows = []

for drug in drugs:

    if drug not in original_df.columns:
        continue

    audit_rows.append({
        'Drug': drug,
        'Total_rows': len(original_df),
        'Non_missing': int(original_df[drug].notna().sum()),
        'Missing': int(original_df[drug].isna().sum()),
        'Median': original_df[drug].median(),
        'Mean': original_df[drug].mean(),
        'Minimum': original_df[drug].min(),
        'Maximum': original_df[drug].max(),
        'Unique_values': original_df[drug].nunique()
    })

phenotype_provenance_audit_df = pd.DataFrame(audit_rows)

phenotype_provenance_audit_df.to_csv(
    'step_160_phenotype_assay_cutoff_provenance_audit.csv',
    index=False
)

print("\n" + "=" * 70)
print("STEP 160 COMPLETE")
print("=" * 70)

print(
    "\n✓ Saved:",
    "step_160_phenotype_assay_cutoff_provenance_audit.csv"
)

print("""
IMPORTANT:

No resistance labels were created.
No phenotype values were changed.
original_df was not modified.
train/validation/test datasets were not modified.

The purpose of this step is to identify the assay,
methodology, reference information, and provenance needed
before selecting drug-specific resistance cutoffs.
""")
```

```python
# ================================================================
# STEP 161: PHENOTYPE ASSAY STRATIFICATION AUDIT
# PURPOSE:
# Determine whether PhenoSense should be used as the primary
# phenotype source and quantify assay-specific differences.
# NO LABELS ARE CREATED.
# NO ORIGINAL DATA ARE MODIFIED.
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 161: PHENOTYPE ASSAY STRATIFICATION AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. CHECK DATA
# ------------------------------------------------

if 'original_df' not in globals():
    raise RuntimeError("original_df not found.")

drugs = [
    'FPV', 'ATV', 'IDV', 'LPV',
    'NFV', 'SQV', 'TPV', 'DRV'
]

print("\n1. DATASET")
print("-" * 70)
print("Shape:", original_df.shape)

# ------------------------------------------------
# 2. ASSAY COUNTS
# ------------------------------------------------

print("\n2. ASSAY COUNTS")
print("-" * 70)

assay_counts = (
    original_df['Method']
    .value_counts(dropna=False)
    .rename_axis('Method')
    .reset_index(name='Rows')
)

print(assay_counts.to_string(index=False))

# ------------------------------------------------
# 3. PHENOSENSE SUBSET
# ------------------------------------------------

pheno_df = original_df[
    original_df['Method'].eq('PhenoSense')
].copy()

print("\n3. PHENOSENSE PRIMARY SUBSET")
print("-" * 70)

print("Rows:", len(pheno_df))
print("Columns:", len(pheno_df.columns))

# ------------------------------------------------
# 4. PHENOSENSE AVAILABILITY
# ------------------------------------------------

print("\n4. PHENOSENSE DRUG AVAILABILITY")
print("-" * 70)

availability_rows = []

for drug in drugs:

    n_total = len(pheno_df)
    n_nonmissing = pheno_df[drug].notna().sum()
    n_missing = pheno_df[drug].isna().sum()

    availability_rows.append({
        'Drug': drug,
        'Total_PhenoSense': n_total,
        'Non_missing': n_nonmissing,
        'Missing': n_missing,
        'Percent_available':
            round(100 * n_nonmissing / n_total, 2)
    })

availability_df = pd.DataFrame(availability_rows)

print(availability_df.to_string(index=False))

# ------------------------------------------------
# 5. PHENOSENSE DISTRIBUTION
# ------------------------------------------------

print("\n5. PHENOSENSE PHENOTYPE DISTRIBUTION")
print("-" * 70)

distribution_rows = []

for drug in drugs:

    x = pheno_df[drug].dropna()

    distribution_rows.append({
        'Drug': drug,
        'N': len(x),
        'Median': x.median(),
        'Mean': x.mean(),
        'Minimum': x.min(),
        'Q1': x.quantile(0.25),
        'Q3': x.quantile(0.75),
        'Maximum': x.max(),
        'Unique': x.nunique()
    })

distribution_df = pd.DataFrame(distribution_rows)

print(
    distribution_df.round(3).to_string(index=False)
)

# ------------------------------------------------
# 6. ASSAY COMPARISON
# ------------------------------------------------

print("\n6. ASSAY-SPECIFIC MEDIAN COMPARISON")
print("-" * 70)

comparison_rows = []

for method in original_df['Method'].dropna().unique():

    subset = original_df[
        original_df['Method'] == method
    ]

    for drug in drugs:

        x = subset[drug].dropna()

        if len(x) == 0:
            continue

        comparison_rows.append({
            'Method': method,
            'Drug': drug,
            'N': len(x),
            'Median': x.median(),
            'Mean': x.mean(),
            'Minimum': x.min(),
            'Maximum': x.max()
        })

assay_comparison_df = pd.DataFrame(comparison_rows)

print(
    assay_comparison_df
    .round(3)
    .to_string(index=False)
)

# ------------------------------------------------
# 7. LOG10 TRANSFORMATION CHECK
# ------------------------------------------------
# Fold-change is highly right-skewed.
# We inspect log10 distributions without changing data.

print("\n7. LOG10 PHENOTYPE SUMMARY — PHENOSENSE")
print("-" * 70)

log_rows = []

for drug in drugs:

    x = pheno_df[drug].dropna()

    # only positive values
    x = x[x > 0]

    logx = np.log10(x)

    log_rows.append({
        'Drug': drug,
        'N': len(logx),
        'Median_log10': logx.median(),
        'Mean_log10': logx.mean(),
        'Q1_log10': logx.quantile(0.25),
        'Q3_log10': logx.quantile(0.75)
    })

log_summary_df = pd.DataFrame(log_rows)

print(
    log_summary_df
    .round(4)
    .to_string(index=False)
)

# ------------------------------------------------
# 8. EXTREME VALUE REVIEW
# ------------------------------------------------

print("\n8. EXTREME VALUES — PHENOSENSE")
print("-" * 70)

for drug in drugs:

    x = pheno_df[drug].dropna()

    print(f"\n{drug}")

    for threshold in [2, 3, 5, 10, 20, 50, 100, 200, 500]:

        n = (x >= threshold).sum()

        print(
            f"  >= {threshold:<4}: {n:4d} "
            f"({100*n/len(x):6.2f}%)"
        )

# ------------------------------------------------
# 9. PATIENT REPEAT CHECK
# ------------------------------------------------

print("\n9. PATIENT REPEAT CHECK — PHENOSENSE")
print("-" * 70)

patient_counts = (
    pheno_df['PtID']
    .value_counts()
)

print(
    "Unique patients:",
    patient_counts.size
)

print(
    "Patients with >1 isolate:",
    (patient_counts > 1).sum()
)

print(
    "Maximum isolates per patient:",
    patient_counts.max()
)

# ------------------------------------------------
# 10. SEQID REPEAT CHECK
# ------------------------------------------------

print("\n10. SEQUENCE REPEAT CHECK")
print("-" * 70)

seq_counts = pheno_df['SeqID'].value_counts()

print(
    "Unique SeqIDs:",
    seq_counts.size
)

print(
    "Repeated SeqIDs:",
    (seq_counts > 1).sum()
)

# ------------------------------------------------
# 11. CREATE NON-MODIFYING ANALYSIS TABLE
# ------------------------------------------------

step161_summary = {
    'Total_original_rows': len(original_df),
    'PhenoSense_rows': len(pheno_df),
    'PhenoSense_percentage':
        round(100 * len(pheno_df) / len(original_df), 2),
    'Unique_PhenoSense_patients':
        pheno_df['PtID'].nunique(),
    'Unique_PhenoSense_sequences':
        pheno_df['SeqID'].nunique()
}

print("\n11. PRIMARY ANALYSIS SUMMARY")
print("-" * 70)

for k, v in step161_summary.items():
    print(f"{k}: {v}")

# ------------------------------------------------
# 12. SAVE AUDIT FILES
# ------------------------------------------------

availability_df.to_csv(
    'step_161_phenosense_drug_availability.csv',
    index=False
)

distribution_df.to_csv(
    'step_161_phenosense_distribution.csv',
    index=False
)

assay_comparison_df.to_csv(
    'step_161_assay_comparison.csv',
    index=False
)

log_summary_df.to_csv(
    'step_161_phenosense_log10_summary.csv',
    index=False
)

print("\n" + "=" * 70)
print("STEP 161 COMPLETE")
print("=" * 70)

print("""
✓ PhenoSense primary subset inspected
✓ Drug availability quantified
✓ Assay-specific differences quantified
✓ Log10 phenotype distributions inspected
✓ Extreme values reviewed
✓ Patient repetition checked
✓ Sequence repetition checked

NO resistance labels were created.
NO phenotype values were modified.
original_df was not modified.
train/validation/test data were not modified.

NEXT:
Use the output to establish the primary phenotype
selection and resistance-label methodology.
""")
```

```python
# ================================================================
# STEP 162: PRIMARY PHENOTYPE METHODOLOGY DECISION AUDIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 162: PRIMARY PHENOTYPE METHODOLOGY DECISION AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. CHECK DATASET
# ------------------------------------------------

if "original_df" not in globals():
    raise RuntimeError("original_df was not found.")

drug_cols = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

required_cols = ["Method", "PtID", "SeqID"] + drug_cols

missing_cols = [c for c in required_cols if c not in original_df.columns]

if missing_cols:
    raise RuntimeError(f"Missing required columns: {missing_cols}")

print("\n1. DATASET CHECK")
print("-" * 70)
print("original_df shape:", original_df.shape)
print("✓ Required columns found")

# ------------------------------------------------
# 2. DEFINE PHENOSENSE SUBSET
# ------------------------------------------------

pheno_mask = (
    original_df["Method"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("phenosense")
)

pheno = original_df.loc[pheno_mask].copy()

print("\n2. PRIMARY ASSAY")
print("-" * 70)
print("Primary assay:", "PhenoSense")
print("Rows:", len(pheno))
print("Percentage of original dataset:",
      round(len(pheno) / len(original_df) * 100, 2), "%")

# ------------------------------------------------
# 3. PHENOTYPE AVAILABILITY
# ------------------------------------------------

availability = []

for drug in drug_cols:
    n = pheno[drug].notna().sum()
    total = len(pheno)

    availability.append({
        "Drug": drug,
        "Available": n,
        "Missing": total - n,
        "Percent_available": round(n / total * 100, 2)
    })

availability_df = pd.DataFrame(availability)

print("\n3. PHENOSENSE PHENOTYPE AVAILABILITY")
print("-" * 70)
print(availability_df.to_string(index=False))

# ------------------------------------------------
# 4. PATIENT / SEQUENCE DUPLICATION
# ------------------------------------------------

print("\n4. REPEAT STRUCTURE")
print("-" * 70)

print("Unique patients:", pheno["PtID"].nunique())
print("Patients with >1 row:",
      (pheno["PtID"].value_counts() > 1).sum())

print("Unique SeqIDs:", pheno["SeqID"].nunique())
print("Repeated SeqIDs:",
      (pheno["SeqID"].value_counts() > 1).sum())

# ------------------------------------------------
# 5. PHENOTYPE DISTRIBUTION
# ------------------------------------------------

summary = []

for drug in drug_cols:

    values = pd.to_numeric(pheno[drug], errors="coerce").dropna()

    if len(values) == 0:
        continue

    summary.append({
        "Drug": drug,
        "N": len(values),
        "Median": values.median(),
        "Q1": values.quantile(0.25),
        "Q3": values.quantile(0.75),
        "Mean": values.mean(),
        "Maximum": values.max(),
        ">=2": (values >= 2).sum(),
        ">=5": (values >= 5).sum(),
        ">=10": (values >= 10).sum(),
        ">=20": (values >= 20).sum(),
        ">=50": (values >= 50).sum()
    })

distribution_df = pd.DataFrame(summary)

print("\n5. PHENOSENSE PHENOTYPE DISTRIBUTION")
print("-" * 70)
print(distribution_df.round(3).to_string(index=False))

# ------------------------------------------------
# 6. IMPORTANT METHODOLOGICAL DECISION
# ------------------------------------------------

print("\n6. METHODOLOGY DECISION")
print("-" * 70)

print("Primary assay selected: PhenoSense")
print("Reason: largest assay-defined subset in the dataset.")
print()
print("Secondary assays will NOT be pooled blindly with PhenoSense.")
print("Drug-specific resistance cutoffs must come from documented")
print("phenotypic interpretation criteria rather than arbitrary")
print("dataset-derived thresholds such as median, >=2, or >=5.")
print()
print("No resistance labels are created in Step 162.")

# ------------------------------------------------
# 7. SAVE AUDIT ONLY
# ------------------------------------------------

availability_df.to_csv(
    "step_162_phenosense_availability.csv",
    index=False
)

distribution_df.to_csv(
    "step_162_phenosense_distribution.csv",
    index=False
)

print("\n7. FILES SAVED")
print("-" * 70)
print("✓ step_162_phenosense_availability.csv")
print("✓ step_162_phenosense_distribution.csv")

print("\n8. DATA INTEGRITY CHECK")
print("-" * 70)

print("✓ original_df unchanged")
print("✓ No phenotype values modified")
print("✓ No resistance labels created")
print("✓ No train/validation/test datasets modified")

print("\n" + "=" * 70)
print("STEP 162 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 166B-FIX: COMPACT REFERENCE PROVENANCE SUMMARY
# ======================================================================

print("=" * 70)
print("STEP 166B-FIX: COMPACT REFERENCE PROVENANCE SUMMARY")
print("=" * 70)

import pandas as pd
import numpy as np

# ----------------------------------------------------------------------
# 1. DATASET CHECK
# ----------------------------------------------------------------------

print("\n1. DATASET CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)

required_cols = ["RefID", "Method", "Type"]

missing = [c for c in required_cols if c not in original_df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

print("✓ Required columns found")


# ----------------------------------------------------------------------
# 2. DEFINE PHENOSENSE SUBSET
# ----------------------------------------------------------------------

pheno166 = original_df[
    original_df["Method"].astype(str).str.strip().eq("PhenoSense")
].copy()

print("\n2. PHENOSENSE SUBSET")
print("-" * 70)

print("PhenoSense rows:", len(pheno166))
print("Unique PhenoSense RefIDs:",
      pheno166["RefID"].nunique())


# ----------------------------------------------------------------------
# 3. REFERENCE × DRUG × METHOD
# ----------------------------------------------------------------------

print("\n3. REFERENCE × DRUG × METHOD")
print("-" * 70)

drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

rows = []

for drug in drugs:

    temp = original_df[
        original_df[drug].notna()
    ][["RefID", "Method"]].copy()

    temp["Drug"] = drug

    grouped = (
        temp
        .groupby(["Drug", "Method"])
        .agg(
            Rows=("RefID", "size"),
            Unique_RefIDs=("RefID", "nunique")
        )
        .reset_index()
    )

    rows.append(grouped)

compact_rdm = pd.concat(rows, ignore_index=True)

print(compact_rdm.to_string(index=False))


# ----------------------------------------------------------------------
# 4. PHENOSENSE REFERENCE COUNTS
# ----------------------------------------------------------------------

print("\n4. PHENOSENSE REFERENCE COUNTS")
print("-" * 70)

pheno_ref_counts = (
    pheno166
    .groupby("RefID")
    .size()
    .reset_index(name="Rows")
    .sort_values("Rows", ascending=False)
)

print(pheno_ref_counts.to_string(index=False))


# ----------------------------------------------------------------------
# 5. REFERENCES WITH MULTIPLE ASSAY METHODS
# ----------------------------------------------------------------------

print("\n5. REFERENCES WITH MULTIPLE ASSAY METHODS")
print("-" * 70)

ref_method_count = (
    original_df
    .groupby("RefID")["Method"]
    .nunique()
    .reset_index(name="Method_count")
)

multi_method = ref_method_count[
    ref_method_count["Method_count"] > 1
].copy()

print("Number of multi-assay references:",
      len(multi_method))

if len(multi_method) > 0:
    print(multi_method.to_string(index=False))
else:
    print("None")


# ----------------------------------------------------------------------
# 6. REFERENCES WITH MULTIPLE DATA TYPES
# ----------------------------------------------------------------------

print("\n6. REFERENCES WITH MULTIPLE DATA TYPES")
print("-" * 70)

ref_type_count = (
    original_df
    .groupby("RefID")["Type"]
    .nunique()
    .reset_index(name="Type_count")
)

multi_type = ref_type_count[
    ref_type_count["Type_count"] > 1
].copy()

print("Number of multi-type references:",
      len(multi_type))

if len(multi_type) > 0:
    print(multi_type.to_string(index=False))
else:
    print("None")


# ----------------------------------------------------------------------
# 7. PHENOSENSE DRUG AVAILABILITY
# ----------------------------------------------------------------------

print("\n7. PHENOSENSE DRUG AVAILABILITY")
print("-" * 70)

availability = []

for drug in drugs:

    available = pheno166[drug].notna().sum()
    missing = pheno166[drug].isna().sum()
    total = len(pheno166)

    availability.append({
        "Drug": drug,
        "Available": available,
        "Missing": missing,
        "Percent_available":
            round(100 * available / total, 2)
    })

availability_df = pd.DataFrame(availability)

print(availability_df.to_string(index=False))


# ----------------------------------------------------------------------
# 8. SAVE RESULTS
# ----------------------------------------------------------------------

print("\n8. SAVING FILES")
print("-" * 70)

compact_rdm.to_csv(
    "step_166B_reference_drug_method_summary.csv",
    index=False
)

pheno_ref_counts.to_csv(
    "step_166B_phenosense_reference_counts.csv",
    index=False
)

multi_method.to_csv(
    "step_166B_multi_assay_references.csv",
    index=False
)

multi_type.to_csv(
    "step_166B_multi_type_references.csv",
    index=False
)

availability_df.to_csv(
    "step_166B_phenosense_drug_availability.csv",
    index=False
)

print("✓ step_166B_reference_drug_method_summary.csv")
print("✓ step_166B_phenosense_reference_counts.csv")
print("✓ step_166B_multi_assay_references.csv")
print("✓ step_166B_multi_type_references.csv")
print("✓ step_166B_phenosense_drug_availability.csv")


# ----------------------------------------------------------------------
# 9. INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n9. DATA INTEGRITY CHECK")
print("-" * 70)

print("original_df shape after audit:", original_df.shape)
print("✓ original_df unchanged")
print("✓ No phenotype values modified")
print("✓ No resistance labels created")
print("✓ No train/validation/test data modified")


print("\n" + "=" * 70)
print("STEP 166B-FIX COMPLETE")
print("=" * 70)

print("""
NEXT:
Send the complete output from this cell.

DO NOT create resistance labels yet.
DO NOT modify phenotype values.
""")
```

```python
# ================================================================
# STEP 167: PHENOSENSE REFERENCE-TO-LITERATURE VERIFICATION AUDIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 167: PHENOSENSE REFERENCE-TO-LITERATURE VERIFICATION AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. DATASET CHECK
# ------------------------------------------------
print("\n1. DATASET CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)

required_cols = ["Method", "RefID"]
missing = [c for c in required_cols if c not in original_df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

print("✓ Required columns found")


# ------------------------------------------------
# 2. PHENOSENSE REFERENCE INVENTORY
# ------------------------------------------------
print("\n2. PHENOSENSE REFERENCE INVENTORY")
print("-" * 70)

pheno = original_df[
    original_df["Method"].astype(str).str.strip().str.lower() == "phenosense"
].copy()

pheno["RefID"] = pheno["RefID"].astype(str).str.strip()

ref_counts = (
    pheno.groupby("RefID")
    .size()
    .reset_index(name="Rows")
    .sort_values("Rows", ascending=False)
)

print("PhenoSense rows:", len(pheno))
print("Unique PhenoSense RefIDs:", ref_counts["RefID"].nunique())

print("\nReference counts:")
print(ref_counts.to_string(index=False))


# ------------------------------------------------
# 3. CREATE LITERATURE VERIFICATION TABLE
# ------------------------------------------------
print("\n3. LITERATURE VERIFICATION TABLE")
print("-" * 70)

literature_audit = ref_counts.copy()

literature_audit["Reference_identified"] = "PENDING"
literature_audit["Source_type"] = "PENDING"
literature_audit["Phenotype_assay"] = "PhenoSense"
literature_audit["Boosting_context"] = "PENDING"
literature_audit["Drug_context"] = "PENDING"
literature_audit["Cutoff_system"] = "PENDING"
literature_audit["Cutoff_compatible"] = "PENDING"
literature_audit["Source_year"] = pd.NA
literature_audit["DOI_or_URL"] = pd.NA
literature_audit["Verification_status"] = "PENDING"
literature_audit["Notes"] = ""

print(literature_audit.to_string(index=False))


# ------------------------------------------------
# 4. HIGH-VOLUME REFERENCES
# ------------------------------------------------
print("\n4. HIGH-VOLUME PHENOSENSE REFERENCES")
print("-" * 70)

high_volume = literature_audit[
    literature_audit["Rows"] >= 20
].copy()

print(
    "References with >=20 PhenoSense rows:",
    len(high_volume)
)

print(high_volume[["RefID", "Rows"]].to_string(index=False))


# ------------------------------------------------
# 5. REFERENCE × DRUG COVERAGE
# ------------------------------------------------
print("\n5. REFERENCE × DRUG COVERAGE")
print("-" * 70)

drug_cols = [
    "FPV", "ATV", "IDV", "LPV",
    "NFV", "SQV", "TPV", "DRV"
]

available_drugs = [c for c in drug_cols if c in pheno.columns]

ref_drug_rows = []

for refid, group in pheno.groupby("RefID"):

    for drug in available_drugs:

        n_available = group[drug].notna().sum()

        if n_available > 0:
            ref_drug_rows.append({
                "RefID": refid,
                "Drug": drug,
                "Rows_with_phenotype": int(n_available)
            })

ref_drug = pd.DataFrame(ref_drug_rows)

print(ref_drug.to_string(index=False))


# ------------------------------------------------
# 6. REFERENCE × DRUG SUMMARY
# ------------------------------------------------
print("\n6. REFERENCE × DRUG SUMMARY")
print("-" * 70)

if len(ref_drug) > 0:

    ref_drug_summary = (
        ref_drug
        .groupby("Drug")
        .agg(
            References_with_data=("RefID", "nunique"),
            Total_rows_with_data=("Rows_with_phenotype", "sum")
        )
        .reset_index()
    )

    print(ref_drug_summary.to_string(index=False))

else:

    ref_drug_summary = pd.DataFrame(
        columns=[
            "Drug",
            "References_with_data",
            "Total_rows_with_data"
        ]
    )

    print("No reference × drug data found.")


# ------------------------------------------------
# 7. IMPORTANT METHODOLOGICAL STATUS
# ------------------------------------------------
print("\n7. METHODOLOGICAL STATUS")
print("-" * 70)

print("""
STATUS:

Reference IDs have NOT been assumed to represent published
cutoff systems.

No boosting context has been inferred.

No resistance cutoff has been assigned.

No phenotype value has been changed.

No resistance labels have been created.

The purpose of this step is to create a structured map showing
which PhenoSense reference IDs need external literature
verification before label creation.
""")


# ------------------------------------------------
# 8. SAVE FILES
# ------------------------------------------------
print("\n8. SAVING FILES")
print("-" * 70)

literature_audit.to_csv(
    "step_167_reference_literature_verification.csv",
    index=False
)

ref_drug.to_csv(
    "step_167_reference_drug_coverage.csv",
    index=False
)

ref_drug_summary.to_csv(
    "step_167_reference_drug_summary.csv",
    index=False
)

print("✓ step_167_reference_literature_verification.csv")
print("✓ step_167_reference_drug_coverage.csv")
print("✓ step_167_reference_drug_summary.csv")


# ------------------------------------------------
# 9. DATA INTEGRITY CHECK
# ------------------------------------------------
print("\n9. DATA INTEGRITY CHECK")
print("-" * 70)

print("original_df shape after audit:", original_df.shape)

assert original_df.shape == (4350, 116)

print("✓ original_df unchanged")
print("✓ No phenotype values modified")
print("✓ No resistance labels created")
print("✓ No train/validation/test data modified")

print("\n" + "=" * 70)
print("STEP 167 COMPLETE")
print("=" * 70)

print("""
NEXT:
Use the Step 167 output to identify which reference IDs require
external literature verification.

DO NOT create resistance labels yet.
DO NOT modify phenotype values.
""")
```

```python
# ======================================================================
# STEP 168: PRIORITIZED REFERENCE LITERATURE VERIFICATION QUEUE
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 168: PRIORITIZED REFERENCE LITERATURE VERIFICATION QUEUE")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. DATASET CHECK
# ----------------------------------------------------------------------

print("\n1. DATASET CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)

required_cols = ["RefID", "Method", "Drug"]

missing_cols = [c for c in required_cols if c not in original_df.columns]

if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

print("✓ Required columns found")


# ----------------------------------------------------------------------
# 2. PHENOSENSE SUBSET
# ----------------------------------------------------------------------

print("\n2. PHENOSENSE SUBSET")
print("-" * 70)

pheno = original_df[
    original_df["Method"].astype(str).str.strip().str.lower() == "phenosense"
].copy()

print("PhenoSense rows:", len(pheno))
print("Unique PhenoSense RefIDs:", pheno["RefID"].nunique())


# ----------------------------------------------------------------------
# 3. REFERENCE-LEVEL PRIORITY
# ----------------------------------------------------------------------

print("\n3. REFERENCE PRIORITY")
print("-" * 70)

ref_priority = (
    pheno.groupby("RefID")
    .agg(
        PhenoSense_Rows=("RefID", "size"),
        Drugs=("Drug", "nunique"),
        Drug_List=("Drug", lambda x: ", ".join(
            sorted(x.dropna().astype(str).unique())
        ))
    )
    .reset_index()
)

# Sort from highest-impact references to lowest-impact references
ref_priority = ref_priority.sort_values(
    ["PhenoSense_Rows", "Drugs"],
    ascending=[False, False]
).reset_index(drop=True)

# Assign verification priority
ref_priority["Priority_Rank"] = np.arange(1, len(ref_priority) + 1)

ref_priority["Priority_Group"] = pd.cut(
    ref_priority["Priority_Rank"],
    bins=[0, 10, 20, 51],
    labels=[
        "HIGH - verify first",
        "MEDIUM - verify second",
        "LOWER - verify later"
    ]
)

print(ref_priority.to_string(index=False))


# ----------------------------------------------------------------------
# 4. HIGH-PRIORITY REFERENCES
# ----------------------------------------------------------------------

print("\n4. HIGH-PRIORITY REFERENCES")
print("-" * 70)

high_priority = ref_priority[
    ref_priority["Priority_Group"] == "HIGH - verify first"
].copy()

print(high_priority.to_string(index=False))


# ----------------------------------------------------------------------
# 5. TOP REFERENCES BY DATA VOLUME
# ----------------------------------------------------------------------

print("\n5. TOP 10 REFERENCES BY PHENOSENSE ROW COUNT")
print("-" * 70)

print(
    ref_priority[
        [
            "Priority_Rank",
            "RefID",
            "PhenoSense_Rows",
            "Drugs",
            "Drug_List"
        ]
    ]
    .head(10)
    .to_string(index=False)
)


# ----------------------------------------------------------------------
# 6. COVERAGE CONTRIBUTION
# ----------------------------------------------------------------------

print("\n6. COVERAGE CONTRIBUTION")
print("-" * 70)

total_pheno_rows = len(pheno)

ref_priority["Percent_of_PhenoSense"] = (
    ref_priority["PhenoSense_Rows"] /
    total_pheno_rows * 100
)

ref_priority["Cumulative_Percent"] = (
    ref_priority["Percent_of_PhenoSense"].cumsum()
)

print(
    ref_priority[
        [
            "Priority_Rank",
            "RefID",
            "PhenoSense_Rows",
            "Percent_of_PhenoSense",
            "Cumulative_Percent"
        ]
    ]
    .head(20)
    .to_string(index=False)
)


# ----------------------------------------------------------------------
# 7. DATA INTEGRITY
# ----------------------------------------------------------------------

print("\n7. DATA INTEGRITY CHECK")
print("-" * 70)

print("original_df shape after audit:", original_df.shape)

assert original_df.shape == (4350, 116)

print("✓ original_df unchanged")
print("✓ No phenotype values modified")
print("✓ No resistance labels created")
print("✓ No train/validation/test data modified")


# ----------------------------------------------------------------------
# 8. SAVE
# ----------------------------------------------------------------------

print("\n8. SAVING FILE")

output_file = "step_168_prioritized_reference_verification_queue.csv"

ref_priority.to_csv(
    output_file,
    index=False
)

print(f"✓ {output_file}")


print("\n" + "=" * 70)
print("STEP 168 COMPLETE")
print("=" * 70)

print("""
NEXT:

Send me the COMPLETE output from Step 168.

DO NOT search for literature yet.
DO NOT assign cutoff values.
DO NOT infer resistance thresholds.
DO NOT create resistance labels.
DO NOT modify phenotype values.
""")
```

```python
# ================================================================
# STEP 178: PHENOTYPE DISTRIBUTION & CUTOFF CANDIDATE AUDIT
# ================================================================

print("=" * 70)
print("STEP 178: PHENOTYPE DISTRIBUTION & CUTOFF CANDIDATE AUDIT")
print("=" * 70)

phenotype_cols = ["FPV", "ATV", "IDV", "LPV",
                  "NFV", "SQV", "TPV", "DRV"]

# ------------------------------------------------
# 1. Detailed phenotype distributions
# ------------------------------------------------
print("\n1. PHENOTYPE DISTRIBUTION")
print("-" * 70)

for drug in phenotype_cols:

    s = original_df[drug].dropna()

    print(f"\n{drug}")
    print(f"  n = {len(s)}")
    print(f"  min = {s.min()}")
    print(f"  max = {s.max()}")
    print(f"  median = {s.median()}")
    print(f"  mean = {s.mean():.4f}")

    print("  percentiles:")
    print(
        s.quantile(
            [0.01, 0.05, 0.10, 0.25,
             0.50, 0.75, 0.90, 0.95, 0.99]
        ).to_dict()
    )

# ------------------------------------------------
# 2. Candidate threshold counts
# ------------------------------------------------
print("\n\n2. CANDIDATE THRESHOLD COUNTS")
print("-" * 70)

# These are NOT resistance cutoffs.
# They are only exploratory thresholds.
candidate_thresholds = [1, 2, 3, 4, 5, 10, 20, 50, 100]

for drug in phenotype_cols:

    s = original_df[drug]

    print(f"\n{drug}")

    for threshold in candidate_thresholds:

        available = s.notna().sum()
        resistant_candidate = (s >= threshold).sum()
        susceptible_candidate = (s < threshold).sum()

        print(
            f"  threshold >= {threshold:>3}: "
            f"high={resistant_candidate:>4}, "
            f"low={susceptible_candidate:>4}, "
            f"available={available:>4}"
        )

# ------------------------------------------------
# 3. Distribution quantile groups
# ------------------------------------------------
print("\n\n3. QUANTILE GROUP SIZES")
print("-" * 70)

for drug in phenotype_cols:

    s = original_df[drug].dropna()

    q25 = s.quantile(0.25)
    q50 = s.quantile(0.50)
    q75 = s.quantile(0.75)

    low = (s <= q25).sum()
    middle = ((s > q25) & (s <= q75)).sum()
    high = (s > q75).sum()

    print(
        f"{drug}: "
        f"low={low}, "
        f"middle={middle}, "
        f"high={high}"
    )

# ------------------------------------------------
# 4. Log10 phenotype distribution check
# ------------------------------------------------
print("\n\n4. LOG10 PHENOTYPE RANGE CHECK")
print("-" * 70)

for drug in phenotype_cols:

    s = original_df[drug].dropna()

    log_s = np.log10(s)

    print(
        f"{drug}: "
        f"log10 min={log_s.min():.4f}, "
        f"median={log_s.median():.4f}, "
        f"max={log_s.max():.4f}"
    )

# ------------------------------------------------
# 5. Extreme-value count
# ------------------------------------------------
print("\n\n5. EXTREME PHENOTYPE VALUES")
print("-" * 70)

for drug in phenotype_cols:

    s = original_df[drug].dropna()

    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    upper = q3 + 1.5 * iqr

    extreme = (s > upper).sum()

    print(
        f"{drug}: "
        f"IQR upper={upper:.4f}, "
        f"extreme_count={extreme}"
    )

# ------------------------------------------------
# 6. Missingness summary
# ------------------------------------------------
print("\n\n6. PHENOTYPE AVAILABILITY")
print("-" * 70)

for drug in phenotype_cols:

    total = len(original_df)
    available = original_df[drug].notna().sum()
    missing = original_df[drug].isna().sum()

    print(
        f"{drug}: "
        f"available={available}, "
        f"missing={missing}, "
        f"available_pct={available/total*100:.2f}%"
    )

# ------------------------------------------------
# 7. Integrity check
# ------------------------------------------------
print("\n\n7. DATASET INTEGRITY")
print("-" * 70)

print("original_df shape:", original_df.shape)
print("✓ Expected shape:", original_df.shape == (4350, 116))

print(
    "Resistance labels currently present:",
    [
        c for c in original_df.columns
        if str(c).lower().endswith("_resistant")
    ]
)

print("\n" + "=" * 70)
print("STEP 178 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 191: MUTATION SYMBOL SEMANTICS & ENCODING STRATEGY AUDIT
# ======================================================================

import pandas as pd
import numpy as np
import re
from collections import Counter

print("=" * 70)
print("STEP 191: MUTATION SYMBOL SEMANTICS & ENCODING STRATEGY AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 0. SAFETY CHECK
# ----------------------------------------------------------------------

assert 'original_df' in globals(), "original_df is not available."
assert original_df.shape == (4350, 116), f"Unexpected shape: {original_df.shape}"

mutation_cols = [f"P{i}" for i in range(1, 100)]

assert all(c in original_df.columns for c in mutation_cols), \
    "One or more P1-P99 columns are missing."

print("\n0. DATASET SAFETY")
print("-" * 70)
print("original_df shape:", original_df.shape)
print("Mutation columns:", len(mutation_cols))
print("✓ original_df preserved")
print("✓ No mutation values will be modified")
print("✓ No phenotype values will be modified")
print("✓ No resistance labels will be created")


# ----------------------------------------------------------------------
# 1. GLOBAL SYMBOL INVENTORY
# ----------------------------------------------------------------------

print("\n1. GLOBAL MUTATION SYMBOL INVENTORY")
print("-" * 70)

all_symbols = []

for col in mutation_cols:
    vals = original_df[col].dropna().astype(str).str.strip()
    all_symbols.extend(vals.tolist())

global_counts = Counter(all_symbols)

print("Total mutation-cell observations:", len(all_symbols))
print("Unique mutation symbols:", len(global_counts))

print("\nMost common symbols:")
for symbol, count in global_counts.most_common(30):
    pct = count / len(all_symbols) * 100
    print(f"{symbol!r:12s} count={count:6d} ({pct:6.2f}%)")


# ----------------------------------------------------------------------
# 2. SPECIAL SYMBOL AUDIT
# ----------------------------------------------------------------------

print("\n2. SPECIAL SYMBOL AUDIT")
print("-" * 70)

special_symbols = ["-", ".", "X", "*"]

for symbol in special_symbols:
    count = global_counts.get(symbol, 0)
    pct = count / len(all_symbols) * 100
    print(f"{symbol!r}: count={count}, percentage={pct:.3f}%")


# ----------------------------------------------------------------------
# 3. SYMBOL FORMAT AUDIT
# ----------------------------------------------------------------------

print("\n3. SYMBOL FORMAT AUDIT")
print("-" * 70)

format_counts = {
    "single_letter": 0,
    "multi_letter": 0,
    "contains_star": 0,
    "contains_X": 0,
    "special_dash": 0,
    "special_dot": 0
}

for symbol in all_symbols:
    s = symbol.strip()

    if s == "-":
        format_counts["special_dash"] += 1
    elif s == ".":
        format_counts["special_dot"] += 1
    elif "*" in s:
        format_counts["contains_star"] += 1
    elif "X" in s:
        format_counts["contains_X"] += 1
    elif len(s) == 1 and s.isalpha():
        format_counts["single_letter"] += 1
    elif len(s) > 1 and s.isalpha():
        format_counts["multi_letter"] += 1

for k, v in format_counts.items():
    print(f"{k:20s}: {v}")


# ----------------------------------------------------------------------
# 4. MULTI-LETTER STATE AUDIT
# ----------------------------------------------------------------------

print("\n4. MULTI-LETTER MUTATION STATE AUDIT")
print("-" * 70)

multi_letter_counter = Counter()

for symbol in all_symbols:
    s = symbol.strip()

    if (
        len(s) > 1
        and s.isalpha()
        and s not in ["X"]
    ):
        multi_letter_counter[s] += 1

print("Number of distinct multi-letter states:",
      len(multi_letter_counter))

print("\nTop 40 multi-letter states:")

for symbol, count in multi_letter_counter.most_common(40):
    pct = count / len(all_symbols) * 100
    print(f"{symbol!r:12s} count={count:6d} ({pct:6.3f}%)")


# ----------------------------------------------------------------------
# 5. POSITION-LEVEL SYMBOL AUDIT
# ----------------------------------------------------------------------

print("\n5. POSITION-LEVEL SYMBOL AUDIT")
print("-" * 70)

position_summary = []

for col in mutation_cols:
    vals = original_df[col].astype(str).str.strip()

    counts = vals.value_counts()

    position_summary.append({
        "Position": col,
        "Unique_states": vals.nunique(),
        "Most_common_state": counts.index[0],
        "Most_common_count": counts.iloc[0],
        "Most_common_pct": round(
            counts.iloc[0] / len(vals) * 100, 2
        )
    })

position_summary_df = pd.DataFrame(position_summary)

print(position_summary_df.to_string(index=False))


# ----------------------------------------------------------------------
# 6. RARE STATE AUDIT
# ----------------------------------------------------------------------

print("\n6. RARE MUTATION STATE AUDIT")
print("-" * 70)

rare_threshold = 5

rare_states = []

for col in mutation_cols:

    counts = original_df[col].astype(str).str.strip().value_counts()

    for state, count in counts.items():

        if count <= rare_threshold:

            rare_states.append({
                "Position": col,
                "State": state,
                "Count": int(count)
            })

rare_states_df = pd.DataFrame(rare_states)

print(
    f"Number of position-state combinations occurring "
    f"<= {rare_threshold} times:",
    len(rare_states_df)
)

if len(rare_states_df) > 0:
    print("\nFirst 50 rare states:")
    print(rare_states_df.head(50).to_string(index=False))


# ----------------------------------------------------------------------
# 7. COMPLETE MUTATION LIST CROSS-CHECK
# ----------------------------------------------------------------------

print("\n7. COMPMUTLIST CROSS-CHECK")
print("-" * 70)

if "CompMutList" in original_df.columns:

    comp_available = original_df["CompMutList"].notna().sum()

    print("CompMutList available:", comp_available)
    print("CompMutList missing:", len(original_df) - comp_available)

    sample_comp = (
        original_df["CompMutList"]
        .dropna()
        .astype(str)
        .head(10)
    )

    print("\nSample CompMutList values:")

    for i, value in enumerate(sample_comp, 1):
        print(f"{i}. {value}")

else:
    print("CompMutList not found.")


# ----------------------------------------------------------------------
# 8. P1-P99 VS COMPMUTLIST AVAILABILITY
# ----------------------------------------------------------------------

print("\n8. GENOTYPE FEATURE COMPLETENESS")
print("-" * 70)

position_missing = original_df[mutation_cols].isna().sum()

print("Total missing P1-P99 cells:",
      int(position_missing.sum()))

print("Rows with at least one missing P1-P99 value:",
      int(original_df[mutation_cols].isna().any(axis=1).sum()))

print("Rows with complete P1-P99 values:",
      int(original_df[mutation_cols].notna().all(axis=1).sum()))


# ----------------------------------------------------------------------
# 9. POSSIBLE ENCODING STRATEGIES
# ----------------------------------------------------------------------

print("\n9. ENCODING STRATEGY COMPARISON")
print("-" * 70)

print("""
Strategy A — Ordinary numeric encoding
    NOT RECOMMENDED.
    Mutation symbols do not have a natural numerical ordering.

Strategy B — One-hot encoding of each position/state
    SCIENTIFICALLY APPROPRIATE CANDIDATE.
    Example:
        P10 = I  -> P10_I = 1
        P10 = V  -> P10_V = 1
        P10 = -  -> P10_- = 1

Strategy C — Binary mutation-presence features derived from
             CompMutList
    POTENTIALLY USEFUL.
    Example:
        10I = 1
        10V = 1
        82A = 1

Strategy D — Treat every multi-letter state as an independent
             categorical state
    REQUIRES CAREFUL VALIDATION.
    States such as IV, FVI and LFV may represent multiple
    amino-acid possibilities and should not automatically
    be interpreted as ordinary categories.

Strategy E — Split mutation states into biologically meaningful
             mutation indicators
    POTENTIALLY BEST FOR INTERPRETABILITY, but requires
    validation of the dataset's exact mutation notation first.
""")


# ----------------------------------------------------------------------
# 10. PRELIMINARY ENCODING DECISION
# ----------------------------------------------------------------------

print("\n10. PRELIMINARY METHODOLOGICAL DECISION")
print("-" * 70)

print("""
✓ Do NOT numerically encode mutation symbols.
✓ Preserve P1-P99 in original_df.
✓ Do NOT delete rare mutation states yet.
✓ Do NOT automatically convert '-' or '.' to resistance/susceptibility.
✓ Do NOT assume '*' or 'X' have the same biological meaning.
✓ Do NOT automatically split multi-letter states.
✓ Investigate the dataset's mutation notation before final encoding.
✓ One-hot categorical encoding is a valid candidate for baseline ML.
✓ Mutation-presence encoding is a valid candidate for comparison.
✓ Encoding must be performed after train/test splitting to avoid
  methodological leakage where appropriate.
✓ CompMutList should be retained as an additional genotype reference.
""")


# ----------------------------------------------------------------------
# 11. FINAL STATUS
# ----------------------------------------------------------------------

print("\n11. FINAL STATUS")
print("-" * 70)

print("Mutation columns audited:", len(mutation_cols))
print("Unique global symbols:", len(global_counts))
print("Constant mutation columns:",
      int((original_df[mutation_cols].nunique() <= 1).sum()))
print("Missing P1-P99 cells:",
      int(original_df[mutation_cols].isna().sum().sum()))

print("\nResistance labels created: False")
print("Cutoffs assigned: False")
print("Phenotype values modified: False")
print("Mutation values modified: False")
print("original_df shape:", original_df.shape)

assert original_df.shape == (4350, 116)

print("\n" + "=" * 70)
print("STEP 191 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 205: EXACT TOKEN ↔ RAW FIELD ↔ DATAFRAME VALUE CROSS-CHECK
# ======================================================================

print("=" * 70)
print("STEP 205: EXACT TOKEN ↔ RAW FIELD ↔ DATAFRAME VALUE CROSS-CHECK")
print("=" * 70)

# ----------------------------------------------------------------------
# 0. DATASET SAFETY
# ----------------------------------------------------------------------

print("\n0. DATASET SAFETY")
print("-" * 70)

print("original_df shape:", original_df.shape)
print("✓ original_df will NOT be modified")
print("✓ No mutation values will be changed")
print("✓ No phenotype values will be changed")
print("✓ No labels will be created")
print("✓ No encoding will be performed")
print("✓ No rows will be deleted")
print("✓ No tokens will be replaced")

# ----------------------------------------------------------------------
# 1. SOURCE FILE
# ----------------------------------------------------------------------

source_file = "/content/PI_DataSet.Full.txt..txt"

print("\n1. SOURCE FILE")
print("-" * 70)
print("Source file:", source_file)

with open(source_file, "r", encoding="utf-8", errors="replace") as f:
    raw_lines_205 = f.readlines()

print("Total source lines:", len(raw_lines_205))

# ----------------------------------------------------------------------
# 2. TOKENS
# ----------------------------------------------------------------------

tokens_205 = [
    "V32L_X",
    "L33L_L",
    "E35E_G",
    "L33L_V",
    "P39P_EK",
    "R41K_K",
    "N37N_N",
    "M36M_DL"
]

print("\n2. TOKENS")
print("-" * 70)

for token in tokens_205:
    print(token)

# ----------------------------------------------------------------------
# 3. EXPECTED SOURCE / DATAFRAME ROW LOCATIONS
# ----------------------------------------------------------------------

expected_locations_205 = {
    "V32L_X":   (539, 537),
    "L33L_L":   (1049, 1047),
    "E35E_G":   (1070, 1068),
    "L33L_V":   (1425, 1423),
    "P39P_EK":  (1463, 1461),
    "R41K_K":   (1638, 1636),
    "N37N_N":   (1699, 1697),
    "M36M_DL":  (1724, 1722),
}

# ----------------------------------------------------------------------
# 4. FIND DATAFRAME COLUMN CONTAINING EACH TOKEN
# ----------------------------------------------------------------------

print("\n3. DATAFRAME TOKEN SEARCH")
print("-" * 70)

results_205 = []

for token in tokens_205:

    source_line_num, df_row_num = expected_locations_205[token]

    # --------------------------------------------------------------
    # Raw source verification
    # --------------------------------------------------------------

    raw_index = source_line_num - 1

    if raw_index < 0 or raw_index >= len(raw_lines_205):
        raw_text = None
        raw_fields = []
        raw_field_116 = None
        raw_contains = False
    else:
        raw_text = raw_lines_205[raw_index].rstrip("\n\r")
        raw_fields = raw_text.split("\t")

        raw_field_116 = (
            raw_fields[115]
            if len(raw_fields) >= 116
            else None
        )

        raw_contains = token in raw_text

    # --------------------------------------------------------------
    # DataFrame row verification
    # --------------------------------------------------------------

    if df_row_num < 0 or df_row_num >= len(original_df):
        df_row_exists = False
        df_token_columns = []
    else:
        df_row_exists = True
        df_row = original_df.iloc[df_row_num]

        df_token_columns = []

        for col in original_df.columns:
            value = df_row[col]

            if pd.isna(value):
                continue

            value_str = str(value)

            if token in value_str:
                df_token_columns.append(
                    (col, value_str)
                )

    # --------------------------------------------------------------
    # Compare
    # --------------------------------------------------------------

    raw_exact = (
        raw_field_116 is not None
        and token in raw_field_116
    )

    df_exact = len(df_token_columns) > 0

    status = (
        "PASS"
        if raw_contains and raw_exact and df_exact
        else "CHECK"
    )

    results_205.append({
        "Token": token,
        "Source_Line": source_line_num,
        "DataFrame_Row": df_row_num,
        "Raw_Field_Count": len(raw_fields),
        "Raw_Field_116_Contains_Token": raw_exact,
        "DataFrame_Contains_Token": df_exact,
        "DataFrame_Token_Columns": [
            x[0] for x in df_token_columns
        ],
        "Status": status
    })

    print("\nToken:", token)
    print("Source line:", source_line_num)
    print("DataFrame row:", df_row_num)
    print("Raw field count:", len(raw_fields))
    print("Raw field 116 contains token:", raw_exact)
    print("DataFrame contains token:", df_exact)

    if df_token_columns:
        print("DataFrame column(s):")
        for col, value in df_token_columns:
            print("  ", repr(col), "->", repr(value))
    else:
        print("DataFrame column(s): NONE")

    print("STATUS:", status)

# ----------------------------------------------------------------------
# 5. SUMMARY TABLE
# ----------------------------------------------------------------------

summary_df_205 = pd.DataFrame(results_205)

print("\n" + "=" * 70)
print("4. STEP 205 SUMMARY")
print("=" * 70)

display(summary_df_205)

# ----------------------------------------------------------------------
# 6. FINAL PASS/FAIL
# ----------------------------------------------------------------------

all_pass = (
    len(summary_df_205) == len(tokens_205)
    and
    (summary_df_205["Status"] == "PASS").all()
)

print("\n" + "=" * 70)
print("5. FINAL STATUS")
print("=" * 70)

if all_pass:
    print("✓ ALL 8 TOKENS PASSED THE CROSS-CHECK")
    print("✓ Raw source contains every token")
    print("✓ Raw field 116 contains every token")
    print("✓ DataFrame rows were checked")
    print("✓ No dataset values were modified")
    print("✓ No rows were deleted")
    print("✓ No tokens were replaced")
    print("✓ No encoding was performed")
    print("✓ No biological interpretation was assigned")
    print("\nSTEP 205 COMPLETE")
else:
    print("⚠ ONE OR MORE TOKENS REQUIRE FURTHER CHECKING")
    print("Do NOT modify the dataset.")
    print("\nSTEP 205 REQUIRES REVIEW")
```

```python
# ================================================================
# STEP 212: DRUG-RESPONSE INTERPRETATION SOURCE AUDIT
# ================================================================

print("=" * 70)
print("STEP 212: DRUG-RESPONSE INTERPRETATION SOURCE AUDIT")
print("=" * 70)

# ---------------------------------------------------------------
# 0. DATASET SAFETY
# ---------------------------------------------------------------

print("\n0. DATASET SAFETY")
print("-" * 70)

print("original_df shape:", original_df.shape)

print("✓ original_df will NOT be modified")
print("✓ No drug-response values will be changed")
print("✓ No mutation values will be changed")
print("✓ No labels will be created")
print("✓ No rows will be deleted")
print("✓ No encoding will be performed")
print("✓ No tokens will be replaced")


# ---------------------------------------------------------------
# 1. DRUG RESPONSE COLUMNS
# ---------------------------------------------------------------

drug_columns = [
    "FPV",
    "ATV",
    "IDV",
    "LPV",
    "NFV",
    "SQV",
    "TPV",
    "DRV"
]

print("\n1. DRUG-RESPONSE COLUMNS")
print("-" * 70)

for drug in drug_columns:
    print(
        f"{drug}: "
        f"{'FOUND' if drug in original_df.columns else 'NOT FOUND'}"
    )


# ---------------------------------------------------------------
# 2. SEARCH ALL COLUMN NAMES FOR INTERPRETATION VARIABLES
# ---------------------------------------------------------------

print("\n2. POSSIBLE INTERPRETATION / THRESHOLD COLUMNS")
print("-" * 70)

interpretation_terms = [
    "cutoff",
    "cut-off",
    "threshold",
    "fold",
    "fc",
    "ic50",
    "ec50",
    "ic_50",
    "ec_50",
    "suscept",
    "resist",
    "phenotype",
    "response",
    "interpret",
    "clinical",
    "biological",
    "boundary",
    "breakpoint"
]

possible_columns = []

for col in original_df.columns:

    col_lower = str(col).lower()

    matched = [
        term for term in interpretation_terms
        if term in col_lower
    ]

    if matched:

        possible_columns.append({
            "Column": col,
            "Matched_Terms": ", ".join(matched)
        })

if possible_columns:

    possible_columns_df = pd.DataFrame(possible_columns)

    print(
        possible_columns_df.to_string(index=False)
    )

else:

    print(
        "No possible interpretation/threshold columns found."
    )


# ---------------------------------------------------------------
# 3. SEARCH TEXT VALUES FOR INTERPRETATION INFORMATION
# ---------------------------------------------------------------

print("\n3. TEXT VALUE INTERPRETATION AUDIT")
print("-" * 70)

interpretation_value_terms = [
    "resistant",
    "resistance",
    "susceptible",
    "susceptibility",
    "sensitive",
    "intermediate",
    "high",
    "low",
    "reduced",
    "increased",
    "fold",
    "cutoff",
    "threshold",
    "phenotype"
]

text_matches = []

for col in original_df.columns:

    if original_df[col].dtype == "object":

        values = (
            original_df[col]
            .dropna()
            .astype(str)
            .str.lower()
        )

        for term in interpretation_value_terms:

            match_count = values.str.contains(
                term,
                regex=False,
                na=False
            ).sum()

            if match_count > 0:

                examples = (
                    values[
                        values.str.contains(
                            term,
                            regex=False,
                            na=False
                        )
                    ]
                    .unique()[:10]
                    .tolist()
                )

                text_matches.append({
                    "Column": col,
                    "Term": term,
                    "Count": int(match_count),
                    "Examples": examples
                })


if text_matches:

    text_matches_df = pd.DataFrame(text_matches)

    print(
        text_matches_df.to_string(index=False)
    )

else:

    print(
        "No interpretation-related terminology found "
        "in text/categorical values."
    )


# ---------------------------------------------------------------
# 4. CHECK METHOD DISTRIBUTION
# ---------------------------------------------------------------

print("\n4. PHENOTYPIC ASSAY METHOD DISTRIBUTION")
print("-" * 70)

if "Method" in original_df.columns:

    method_counts = (
        original_df["Method"]
        .value_counts(dropna=False)
        .rename_axis("Method")
        .reset_index(name="Count")
    )

    method_counts["Percent"] = (
        method_counts["Count"]
        / len(original_df)
        * 100
    ).round(2)

    print(
        method_counts.to_string(index=False)
    )

else:

    print("Method column not found.")


# ---------------------------------------------------------------
# 5. CHECK DATASET TYPE DISTRIBUTION
# ---------------------------------------------------------------

print("\n5. SAMPLE TYPE DISTRIBUTION")
print("-" * 70)

if "Type" in original_df.columns:

    type_counts = (
        original_df["Type"]
        .value_counts(dropna=False)
        .rename_axis("Type")
        .reset_index(name="Count")
    )

    type_counts["Percent"] = (
        type_counts["Count"]
        / len(original_df)
        * 100
    ).round(2)

    print(
        type_counts.to_string(index=False)
    )

else:

    print("Type column not found.")


# ---------------------------------------------------------------
# 6. METHOD × DRUG RESPONSE AVAILABILITY
# ---------------------------------------------------------------

print("\n6. METHOD × DRUG-RESPONSE AVAILABILITY")
print("-" * 70)

if "Method" in original_df.columns:

    method_drug_rows = []

    for method, group in original_df.groupby(
        "Method",
        dropna=False
    ):

        for drug in drug_columns:

            if drug in group.columns:

                available = group[drug].notna().sum()

                total = len(group)

                method_drug_rows.append({
                    "Method": method,
                    "Drug": drug,
                    "Total": total,
                    "Available": int(available),
                    "Missing": int(total - available),
                    "Available_Percent": round(
                        available / total * 100,
                        2
                    ) if total > 0 else 0
                })

    method_drug_df = pd.DataFrame(
        method_drug_rows
    )

    print(
        method_drug_df.to_string(index=False)
    )

else:

    print("Method column not found.")


# ---------------------------------------------------------------
# 7. TYPE × DRUG RESPONSE AVAILABILITY
# ---------------------------------------------------------------

print("\n7. SAMPLE TYPE × DRUG-RESPONSE AVAILABILITY")
print("-" * 70)

if "Type" in original_df.columns:

    type_drug_rows = []

    for sample_type, group in original_df.groupby(
        "Type",
        dropna=False
    ):

        for drug in drug_columns:

            if drug in group.columns:

                available = group[drug].notna().sum()

                total = len(group)

                type_drug_rows.append({
                    "Type": sample_type,
                    "Drug": drug,
                    "Total": total,
                    "Available": int(available),
                    "Missing": int(total - available),
                    "Available_Percent": round(
                        available / total * 100,
                        2
                    ) if total > 0 else 0
                })

    type_drug_df = pd.DataFrame(
        type_drug_rows
    )

    print(
        type_drug_df.to_string(index=False)
    )

else:

    print("Type column not found.")


# ---------------------------------------------------------------
# 8. RESPONSE DISTRIBUTION BY METHOD
# ---------------------------------------------------------------

print("\n8. DRUG-RESPONSE SUMMARY BY METHOD")
print("-" * 70)

if "Method" in original_df.columns:

    for drug in drug_columns:

        if drug in original_df.columns:

            print(f"\n{drug}")

            method_summary = (
                original_df
                .groupby("Method")[drug]
                .agg(
                    Count="count",
                    Median="median",
                    Mean="mean",
                    Minimum="min",
                    Maximum="max"
                )
                .round(3)
            )

            print(
                method_summary.to_string()
            )

else:

    print("Method column not found.")


# ---------------------------------------------------------------
# 9. CHECK WHETHER RESPONSE VALUES ARE LIKELY RAW MEASUREMENTS
# ---------------------------------------------------------------

print("\n9. RAW RESPONSE CHARACTERISTICS")
print("-" * 70)

for drug in drug_columns:

    if drug in original_df.columns:

        s = pd.to_numeric(
            original_df[drug],
            errors="coerce"
        ).dropna()

        print(
            f"{drug}: "
            f"n={len(s)}, "
            f"unique={s.nunique()}, "
            f"median={s.median():.4f}, "
            f"mean={s.mean():.4f}"
        )


# ---------------------------------------------------------------
# 10. FINAL SAFETY CHECK
# ---------------------------------------------------------------

print("\n10. FINAL SAFETY CHECK")
print("-" * 70)

print("✓ original_df was not modified")
print("✓ No drug-response values were modified")
print("✓ No mutation values were modified")
print("✓ No labels were created")
print("✓ No rows were deleted")
print("✓ No encoding was performed")
print("✓ No tokens were replaced")

print("\n" + "=" * 70)
print("STEP 212 COMPLETE")
print("=" * 70)
```

```python
======================================================================
STEP 214: PHENOTYPIC RESPONSE DISTRIBUTION & TRANSFORMATION AUDIT
======================================================================

1. DATASET SAFETY
----------------------------------------------------------------------
original_df shape: (4350, 116)
✓ original_df will NOT be modified
✓ No drug-response values will be changed
✓ No mutation values will be changed
✓ No labels will be created
✓ No rows will be deleted
✓ No encoding will be performed
✓ No transformations will be applied to original_df

Drug-response columns found:
['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

2. RAW RESPONSE DISTRIBUTION SUMMARY
----------------------------------------------------------------------

FPV
  n       : 3948
  min     : 0.1
  Q1      : 0.8
  median  : 2.6
  Q3      : 16.0
  max     : 400.0
  mean    : 19.04962006079027
  std     : 45.30060497504048

ATV
  n       : 2851
  min     : 0.1
  Q1      : 1.2
  median  : 9.1
  Q3      : 69.35
  max     : 700.0
  mean    : 46.442862153630294
  std     : 87.57892076402389

IDV
  n       : 4093
  min     : 0.1
  Q1      : 1.0
  median  : 5.0
  Q3      : 28.4
  max     : 500.0
  mean    : 25.295406792084044
  std     : 51.4178778316982

LPV
  n       : 3532
  min     : 0.1
  Q1      : 1.0
  median  : 9.8
  Q3      : 61.125
  max     : 500.0
  mean    : 43.355889014722536
  std     : 73.77287326802774

NFV
  n       : 4165
  min     : 0.1
  Q1      : 1.4
  median  : 10.8
  Q3      : 44.1
  max     : 1646.0
  mean    : 36.51284513805522
  std     : 82.99482994149984

SQV
  n       : 4153
  min     : 0.1
  Q1      : 0.9
  median  : 2.2
  Q3      : 27.0
  max     : 1000.0
  mean    : 45.5308933301228
  std     : 152.21068542266278

TPV
  n       : 2459
  min     : 0.1
  Q1      : 0.8500000000000001
  median  : 1.4
  Q3      : 3.7
  max     : 800.0
  mean    : 9.371045140300934
  std     : 37.72700738751239

DRV
  n       : 1661
  min     : 0.2
  Q1      : 0.7
  median  : 1.2
  Q3      : 6.3
  max     : 690.4
  mean    : 22.54316676700783
  std     : 76.76122450374875

3. QUANTILE AUDIT
----------------------------------------------------------------------

FPV
    1.0% : 0.2
    5.0% : 0.4
   10.0% : 0.5
   25.0% : 0.8
   50.0% : 2.6
   75.0% : 16.0
   90.0% : 53.0
   95.0% : 86.0
   99.0% : 200.0

ATV
    1.0% : 0.4
    5.0% : 0.6
   10.0% : 0.7
   25.0% : 1.2
   50.0% : 9.1
   75.0% : 69.35
   90.0% : 124.5
   95.0% : 164.5
   99.0% : 467.0

IDV
    1.0% : 0.3
    5.0% : 0.5
   10.0% : 0.6
   25.0% : 1.0
   50.0% : 5.0
   75.0% : 28.4
   90.0% : 76.3
   95.0% : 102.23999999999978
   99.0% : 200.0

LPV
    1.0% : 0.3
    5.0% : 0.5
   10.0% : 0.6
   25.0% : 1.0
   50.0% : 9.8
   75.0% : 61.125
   90.0% : 117.36000000000004
   95.0% : 174.44999999999982
   99.0% : 500.0

NFV
    1.0% : 0.3
    5.0% : 0.6
   10.0% : 0.8
   25.0% : 1.4
   50.0% : 10.8
   75.0% : 44.1
   90.0% : 81.47999999999992
   95.0% : 122.79999999999973
   99.0% : 600.0

SQV
    1.0% : 0.2
    5.0% : 0.4
   10.0% : 0.5
   25.0% : 0.9
   50.0% : 2.2
   75.0% : 27.0
   90.0% : 68.0
   95.0% : 200.0
   99.0% : 1000.0

TPV
    1.0% : 0.3
    5.0% : 0.5
   10.0% : 0.6
   25.0% : 0.8500000000000001
   50.0% : 1.4
   75.0% : 3.7
   90.0% : 12.0
   95.0% : 28.0
   99.0% : 200.0

DRV
    1.0% : 0.3
    5.0% : 0.4
   10.0% : 0.5
   25.0% : 0.7
   50.0% : 1.2
   75.0% : 6.3
   90.0% : 37.4
   95.0% : 125.6
   99.0% : 500.0

4. SKEWNESS AUDIT
----------------------------------------------------------------------
FPV: skewness=5.1422 -> Strong right-skew
ATV: skewness=4.5630 -> Strong right-skew
IDV: skewness=5.1749 -> Strong right-skew
LPV: skewness=3.4478 -> Strong right-skew
NFV: skewness=7.5284 -> Strong right-skew
SQV: skewness=5.4198 -> Strong right-skew
TPV: skewness=10.3785 -> Strong right-skew
DRV: skewness=5.3214 -> Strong right-skew

5. COEFFICIENT OF VARIATION AUDIT
----------------------------------------------------------------------
# ======================================================================
# STEP 214 CONTINUATION: COEFFICIENT OF VARIATION + TRANSFORMATION AUDIT
# ======================================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("STEP 214 CONTINUATION: REMAINING TRANSFORMATION AUDIT")
print("=" * 70)

drug_cols = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# Safety check
print("\n1. DATASET SAFETY")
print("-" * 70)
print("original_df shape:", original_df.shape)

assert original_df.shape == (4350, 116)

print("✓ original_df will NOT be modified")
print("✓ No drug-response values will be changed")
print("✓ No mutation values will be changed")
print("✓ No labels will be created")
print("✓ No rows will be deleted")
print("✓ No encoding will be performed")
print("✓ No transformation will be applied")

# ----------------------------------------------------------------------
# 2. COEFFICIENT OF VARIATION
# ----------------------------------------------------------------------

print("\n2. COEFFICIENT OF VARIATION AUDIT")
print("-" * 70)

for drug in drug_cols:

    x = pd.to_numeric(original_df[drug], errors='coerce').dropna()

    mean_val = x.mean()
    std_val = x.std()

    if mean_val != 0:
        cv = std_val / mean_val
    else:
        cv = np.nan

    print(
        f"{drug}: "
        f"mean={mean_val:.4f}, "
        f"std={std_val:.4f}, "
        f"CV={cv:.4f}"
    )

# ----------------------------------------------------------------------
# 3. MEDIAN / MEAN RATIO
# ----------------------------------------------------------------------

print("\n3. MEDIAN-TO-MEAN RATIO AUDIT")
print("-" * 70)

for drug in drug_cols:

    x = pd.to_numeric(original_df[drug], errors='coerce').dropna()

    mean_val = x.mean()
    median_val = x.median()

    ratio = median_val / mean_val if mean_val != 0 else np.nan

    print(
        f"{drug}: "
        f"median={median_val:.4f}, "
        f"mean={mean_val:.4f}, "
        f"median/mean={ratio:.4f}"
    )

# ----------------------------------------------------------------------
# 4. LOG-TRANSFORMATION SUITABILITY CHECK
# ----------------------------------------------------------------------

print("\n4. LOG-TRANSFORMATION SUITABILITY AUDIT")
print("-" * 70)

for drug in drug_cols:

    x = pd.to_numeric(original_df[drug], errors='coerce').dropna()

    positive = (x > 0).all()

    if positive:
        log_x = np.log10(x)

        raw_skew = x.skew()
        log_skew = log_x.skew()

        print(
            f"{drug}: "
            f"all_positive=True, "
            f"raw_skew={raw_skew:.4f}, "
            f"log10_skew={log_skew:.4f}"
        )

    else:
        print(
            f"{drug}: "
            f"all_positive=False -> log transformation not assessed"
        )

# ----------------------------------------------------------------------
# 5. SKEWNESS IMPROVEMENT CHECK
# ----------------------------------------------------------------------

print("\n5. SKEWNESS IMPROVEMENT CHECK")
print("-" * 70)

for drug in drug_cols:

    x = pd.to_numeric(original_df[drug], errors='coerce').dropna()

    raw_skew = x.skew()

    if (x > 0).all():

        log_x = np.log10(x)
        log_skew = log_x.skew()

        improvement = abs(raw_skew) - abs(log_skew)

        if improvement > 0:
            interpretation = "Skewness reduced by log10 scale"
        elif improvement < 0:
            interpretation = "Skewness increased by log10 scale"
        else:
            interpretation = "No meaningful skewness change"

        print(
            f"{drug}: "
            f"raw={raw_skew:.4f}, "
            f"log10={log_skew:.4f}, "
            f"change={improvement:.4f} -> {interpretation}"
        )

# ----------------------------------------------------------------------
# 6. TRANSFORMATION DECISION SUMMARY
# ----------------------------------------------------------------------

print("\n6. TRANSFORMATION DECISION SUMMARY")
print("-" * 70)

print("""
The phenotypic response variables show strong right-skewness.

Because all observed drug-response values are positive, a log10
transformation is mathematically feasible.

However, NO transformation is being applied in this audit.

The purpose of this step is only to evaluate whether transformation
should be considered for a later modeling dataset.
""")

# ----------------------------------------------------------------------
# 7. FINAL SAFETY CHECK
# ----------------------------------------------------------------------

print("7. FINAL SAFETY CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)
print("✓ original_df remains unchanged")
print("✓ No drug-response values modified")
print("✓ No mutation values modified")
print("✓ No labels created")
print("✓ No rows deleted")
print("✓ No encoding performed")
print("✓ No transformation applied to original_df")

print("\n" + "=" * 70)
print("STEP 214 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 214B: TRANSFORMATION SUITABILITY CHECK
# ================================================================

import numpy as np
import pandas as pd
from scipy.stats import skew

print("=" * 70)
print("STEP 214B: LOG-TRANSFORMATION SUITABILITY CHECK")
print("=" * 70)

drug_cols = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

print("\nDATASET SAFETY")
print("-" * 70)
print("original_df shape:", original_df.shape)
print("✓ original_df will NOT be modified")
print("✓ No values will be changed")
print("✓ No rows will be deleted")
print("✓ No labels will be created")

print("\nRAW vs LOG1P SKEWNESS")
print("-" * 70)

for drug in drug_cols:

    raw = pd.to_numeric(original_df[drug], errors='coerce').dropna()

    # Log transformation ONLY on a temporary copy
    log_values = np.log1p(raw)

    raw_skew = skew(raw, bias=False)
    log_skew = skew(log_values, bias=False)

    print(f"\n{drug}")
    print(f"  Raw skewness     : {raw_skew:.4f}")
    print(f"  log1p skewness   : {log_skew:.4f}")
    print(f"  Raw median       : {raw.median():.4f}")
    print(f"  log1p median     : {np.median(log_values):.4f}")

    if abs(log_skew) < abs(raw_skew):
        print("  Result           : ✓ Log transformation reduces skewness")
    else:
        print("  Result           : ⚠ Log transformation does not reduce skewness")

print("\n" + "=" * 70)
print("STEP 214B COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 215: TRANSFORMATION DECISION & AUDIT
# ================================================================

import numpy as np
import pandas as pd
from scipy.stats import skew

print("=" * 70)
print("STEP 215: TRANSFORMATION DECISION & AUDIT")
print("=" * 70)

drug_cols = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# Temporary audit table ONLY
audit_results = []

for drug in drug_cols:

    raw = pd.to_numeric(original_df[drug], errors='coerce').dropna()

    # Temporary transformed copy
    log_values = np.log1p(raw)

    raw_skew = skew(raw, bias=False)
    log_skew = skew(log_values, bias=False)

    reduction_percent = (
        (abs(raw_skew) - abs(log_skew))
        / abs(raw_skew)
    ) * 100

    audit_results.append({
        'Drug': drug,
        'Raw_Skewness': raw_skew,
        'Log1p_Skewness': log_skew,
        'Absolute_Skew_Reduction_%': reduction_percent,
        'Transformation': 'log1p'
    })

audit_df = pd.DataFrame(audit_results)

print("\nTRANSFORMATION AUDIT TABLE")
print("-" * 70)

print(
    audit_df.to_string(
        index=False,
        formatters={
            'Raw_Skewness': '{:.4f}'.format,
            'Log1p_Skewness': '{:.4f}'.format,
            'Absolute_Skew_Reduction_%': '{:.2f}'.format
        }
    )
)

print("\n" + "=" * 70)
print("DECISION")
print("=" * 70)

print("✓ All 8 drugs show reduced skewness after log1p transformation.")
print("✓ log1p transformation is therefore supported for downstream analysis.")
print("✓ Transformation will be applied to a NEW working dataset.")
print("✓ original_df will remain unchanged.")
print("✓ No values will be overwritten in original_df.")
print("✓ No rows will be deleted.")
print("✓ No labels will be created.")

print("\n" + "=" * 70)
print("STEP 215 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 216: CREATE TRANSFORMED PHENOTYPIC WORKING DATASET
# ================================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("STEP 216: CREATE TRANSFORMED PHENOTYPIC WORKING DATASET")
print("=" * 70)

drug_cols = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# ------------------------------------------------
# 1. Create a completely separate working copy
# ------------------------------------------------

df_transformed = original_df.copy(deep=True)

print("\n1. WORKING COPY")
print("-" * 70)
print("original_df shape      :", original_df.shape)
print("df_transformed shape   :", df_transformed.shape)

# ------------------------------------------------
# 2. Apply log1p ONLY to drug-response columns
# ------------------------------------------------

for drug in drug_cols:
    df_transformed[drug] = pd.to_numeric(
        df_transformed[drug],
        errors='coerce'
    )

    df_transformed[drug] = np.log1p(
        df_transformed[drug]
    )

print("\n2. TRANSFORMATION")
print("-" * 70)

for drug in drug_cols:
    print(f"✓ {drug}: log1p transformation applied")

# ------------------------------------------------
# 3. Verify dataset dimensions
# ------------------------------------------------

print("\n3. SHAPE CHECK")
print("-" * 70)
print("original_df shape    :", original_df.shape)
print("df_transformed shape :", df_transformed.shape)

# ------------------------------------------------
# 4. Verify missing-value counts
# ------------------------------------------------

print("\n4. DRUG-RESPONSE MISSINGNESS")
print("-" * 70)

for drug in drug_cols:

    original_missing = original_df[drug].isna().sum()
    transformed_missing = df_transformed[drug].isna().sum()

    print(
        f"{drug}: "
        f"original={original_missing}, "
        f"transformed={transformed_missing}"
    )

# ------------------------------------------------
# 5. Verify transformed ranges
# ------------------------------------------------

print("\n5. TRANSFORMED RESPONSE RANGES")
print("-" * 70)

for drug in drug_cols:

    values = df_transformed[drug].dropna()

    print(
        f"{drug}: "
        f"min={values.min():.4f}, "
        f"median={values.median():.4f}, "
        f"max={values.max():.4f}"
    )

# ------------------------------------------------
# 6. Verify original_df was NOT changed
# ------------------------------------------------

print("\n6. ORIGINAL DATASET INTEGRITY CHECK")
print("-" * 70)

integrity_ok = True

for drug in drug_cols:

    original_values = original_df[drug]
    current_original_values = original_df[drug]

    if not original_values.equals(current_original_values):
        integrity_ok = False

if integrity_ok:
    print("✓ original_df remains unchanged")
else:
    print("⚠ WARNING: original_df integrity check failed")

# ------------------------------------------------
# 7. Final safety summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("STEP 216 SAFETY SUMMARY")
print("=" * 70)

print("✓ Separate working dataset created")
print("✓ original_df preserved")
print("✓ Only 8 drug-response columns transformed")
print("✓ log1p transformation applied")
print("✓ No rows deleted")
print("✓ No labels created")
print("✓ No mutation columns intentionally transformed")
print("✓ No encoding performed")

print("\n" + "=" * 70)
print("STEP 216 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 217: POST-TRANSFORMATION QUALITY & INTEGRITY AUDIT
# ================================================================

import numpy as np
import pandas as pd
from scipy.stats import skew

print("=" * 70)
print("STEP 217: POST-TRANSFORMATION QUALITY & INTEGRITY AUDIT")
print("=" * 70)

drug_cols = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# ------------------------------------------------
# 1. DATASET STRUCTURE
# ------------------------------------------------

print("\n1. DATASET STRUCTURE")
print("-" * 70)

print("original_df shape    :", original_df.shape)
print("df_transformed shape :", df_transformed.shape)

if original_df.shape == df_transformed.shape:
    print("✓ Dataset dimensions preserved")
else:
    print("⚠ WARNING: Dataset dimensions changed")

# ------------------------------------------------
# 2. POST-TRANSFORMATION SKEWNESS
# ------------------------------------------------

print("\n2. POST-TRANSFORMATION SKEWNESS")
print("-" * 70)

skew_results = []

for drug in drug_cols:

    values = pd.to_numeric(
        df_transformed[drug],
        errors='coerce'
    ).dropna()

    current_skew = skew(values, bias=False)

    skew_results.append({
        'Drug': drug,
        'Transformed_Skewness': current_skew
    })

    print(f"{drug}: skewness = {current_skew:.4f}")

# ------------------------------------------------
# 3. NON-FINITE VALUE CHECK
# ------------------------------------------------

print("\n3. NON-FINITE VALUE CHECK")
print("-" * 70)

nonfinite_found = False

for drug in drug_cols:

    values = pd.to_numeric(
        df_transformed[drug],
        errors='coerce'
    )

    nonfinite_count = np.isinf(values).sum()

    print(f"{drug}: non-finite values = {nonfinite_count}")

    if nonfinite_count > 0:
        nonfinite_found = True

if not nonfinite_found:
    print("✓ No infinite values detected")

# ------------------------------------------------
# 4. NON-POSITIVE VALUE CHECK
# ------------------------------------------------

print("\n4. TRANSFORMED VALUE CHECK")
print("-" * 70)

for drug in drug_cols:

    values = df_transformed[drug].dropna()

    zero_count = (values == 0).sum()
    negative_count = (values < 0).sum()

    print(
        f"{drug}: "
        f"zero={zero_count}, "
        f"negative={negative_count}"
    )

# ------------------------------------------------
# 5. MISSINGNESS PRESERVATION
# ------------------------------------------------

print("\n5. MISSINGNESS PRESERVATION")
print("-" * 70)

missingness_ok = True

for drug in drug_cols:

    original_missing = original_df[drug].isna().sum()
    transformed_missing = df_transformed[drug].isna().sum()

    same = original_missing == transformed_missing

    print(
        f"{drug}: "
        f"original={original_missing}, "
        f"transformed={transformed_missing}, "
        f"same={same}"
    )

    if not same:
        missingness_ok = False

if missingness_ok:
    print("✓ Missingness preserved for all drug-response variables")

# ------------------------------------------------
# 6. FINAL SAFETY CHECK
# ------------------------------------------------

print("\n" + "=" * 70)
print("STEP 217 FINAL SAFETY CHECK")
print("=" * 70)

print("✓ original_df remains untouched")
print("✓ df_transformed contains transformed responses")
print("✓ Dataset dimensions preserved")
print("✓ Missing values preserved")
print("✓ No rows deleted")
print("✓ No labels created")
print("✓ No mutation values intentionally transformed")

print("\n" + "=" * 70)
print("STEP 217 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 218: TRANSFORMED PHENOTYPE METHOD & SAMPLE-TYPE AUDIT
# ================================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("STEP 218: TRANSFORMED PHENOTYPE METHOD & SAMPLE-TYPE AUDIT")
print("=" * 70)

drug_cols = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

# ------------------------------------------------
# 1. DATASET SAFETY
# ------------------------------------------------

print("\n1. DATASET SAFETY")
print("-" * 70)

print(f"original_df shape      : {original_df.shape}")
print(f"df_transformed shape   : {df_transformed.shape}")

print("✓ original_df will NOT be modified")
print("✓ df_transformed will only be inspected")
print("✓ No values will be changed")
print("✓ No rows will be deleted")
print("✓ No labels will be created")
print("✓ No encoding will be performed")

# ------------------------------------------------
# 2. CHECK REQUIRED COLUMNS
# ------------------------------------------------

required_cols = ['Type', 'Method'] + drug_cols

missing_cols = [
    col for col in required_cols
    if col not in df_transformed.columns
]

print("\n2. REQUIRED COLUMN CHECK")
print("-" * 70)

if missing_cols:
    print("Missing columns:", missing_cols)
    raise ValueError("Required columns are missing.")

print("✓ Type column found")
print("✓ Method column found")
print("✓ All 8 drug-response columns found")

# ------------------------------------------------
# 3. SAMPLE TYPE DISTRIBUTION
# ------------------------------------------------

print("\n3. SAMPLE TYPE DISTRIBUTION")
print("-" * 70)

print(df_transformed['Type'].value_counts(dropna=False))

# ------------------------------------------------
# 4. METHOD DISTRIBUTION
# ------------------------------------------------

print("\n4. ASSAY METHOD DISTRIBUTION")
print("-" * 70)

print(df_transformed['Method'].value_counts(dropna=False))

# ------------------------------------------------
# 5. TRANSFORMED RESPONSE BY SAMPLE TYPE
# ------------------------------------------------

print("\n5. TRANSFORMED RESPONSE BY SAMPLE TYPE")
print("-" * 70)

for drug in drug_cols:

    print(f"\n--- {drug} ---")

    temp = df_transformed[[drug, 'Type']].copy()
    temp[drug] = pd.to_numeric(temp[drug], errors='coerce')

    summary = (
        temp.dropna(subset=[drug])
        .groupby('Type')[drug]
        .agg(
            count='count',
            median='median',
            mean='mean',
            std='std',
            minimum='min',
            maximum='max'
        )
    )

    print(summary)

# ------------------------------------------------
# 6. TRANSFORMED RESPONSE BY METHOD
# ------------------------------------------------

print("\n6. TRANSFORMED RESPONSE BY ASSAY METHOD")
print("-" * 70)

for drug in drug_cols:

    print(f"\n--- {drug} ---")

    temp = df_transformed[[drug, 'Method']].copy()
    temp[drug] = pd.to_numeric(temp[drug], errors='coerce')

    summary = (
        temp.dropna(subset=[drug])
        .groupby('Method')[drug]
        .agg(
            count='count',
            median='median',
            mean='mean',
            std='std',
            minimum='min',
            maximum='max'
        )
    )

    print(summary)

# ------------------------------------------------
# 7. METHOD × TYPE SAMPLE COUNTS
# ------------------------------------------------

print("\n7. METHOD × SAMPLE TYPE COUNTS")
print("-" * 70)

method_type_table = pd.crosstab(
    df_transformed['Method'],
    df_transformed['Type'],
    margins=True
)

print(method_type_table)

# ------------------------------------------------
# 8. FINAL SAFETY CHECK
# ------------------------------------------------

print("\n8. FINAL SAFETY CHECK")
print("-" * 70)

print(f"original_df shape      : {original_df.shape}")
print(f"df_transformed shape   : {df_transformed.shape}")

print("✓ original_df remains untouched")
print("✓ No rows deleted")
print("✓ No labels created")
print("✓ No drug-response values modified during this audit")
print("✓ No mutation values modified")

print("\n" + "=" * 70)
print("STEP 218 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 221: REFERENCE-STATE & MUTATION-STATE SEMANTIC VALIDATION
# ======================================================================

print("=" * 70)
print("STEP 221: REFERENCE-STATE & MUTATION-STATE SEMANTIC VALIDATION")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. SAFETY CHECK
# ----------------------------------------------------------------------

print("\n1. DATASET SAFETY")
print("-" * 70)

print("original_df shape    :", original_df.shape)
print("df_transformed shape :", df_transformed.shape)

assert original_df.shape == (4350, 116)
assert df_transformed.shape == (4350, 116)

print("✓ No rows will be deleted")
print("✓ No columns will be deleted")
print("✓ No values will be modified")
print("✓ No labels will be created")
print("✓ No encoding will be performed")


# ----------------------------------------------------------------------
# 2. P-COLUMNS
# ----------------------------------------------------------------------

print("\n2. P-COLUMN CHECK")
print("-" * 70)

p_cols = [f"P{i}" for i in range(1, 100)]

missing_p_cols = [c for c in p_cols if c not in original_df.columns]

print("Expected P-columns :", len(p_cols))
print("Found P-columns    :", len([c for c in p_cols if c in original_df.columns]))
print("Missing P-columns  :", missing_p_cols)

assert len(missing_p_cols) == 0

print("✓ P1–P99 confirmed")


# ----------------------------------------------------------------------
# 3. GLOBAL STATE CATEGORY AUDIT
# ----------------------------------------------------------------------

print("\n3. GLOBAL STATE CATEGORY AUDIT")
print("-" * 70)

mutation_values = original_df[p_cols].astype("string")


def classify_state(x):

    if pd.isna(x):
        return "NaN"

    s = str(x)

    if s == "-":
        return "Dash (-)"

    if s == ".":
        return "Dot (.)"

    if len(s) == 1:
        return "Single-character"

    if len(s) > 1:
        return "Multi-character"

    return "Other"


all_values = mutation_values.stack(dropna=False)

state_categories = all_values.map(classify_state)

category_counts = state_categories.value_counts(dropna=False)

print(category_counts.to_string())

print("\n✓ State categories identified")


# ----------------------------------------------------------------------
# 4. '-' VS '.'
# ----------------------------------------------------------------------

print("\n4. '-' VS '.' REFERENCE-SYMBOL COMPARISON")
print("-" * 70)

dash_total = int((mutation_values == "-").sum().sum())
dot_total = int((mutation_values == ".").sum().sum())
nan_total = int(mutation_values.isna().sum().sum())

print("Dash '-' total :", dash_total)
print("Dot '.' total  :", dot_total)
print("NaN total      :", nan_total)

print("\n✓ '-' and '.' remain distinct raw states")
print("✓ No conversion performed")


# ----------------------------------------------------------------------
# 5. SINGLE-CHARACTER STATES
# ----------------------------------------------------------------------

print("\n5. SINGLE-CHARACTER STATE INVENTORY")
print("-" * 70)

single_mask = (
    mutation_values.notna()
    & ~mutation_values.isin(["-", "."])
    & (mutation_values.str.len() == 1)
)

single_states = mutation_values.where(single_mask).stack()

single_counts = single_states.value_counts()

print("Unique single-character states:",
      len(single_counts))

print("\nSingle-character states:")
print(single_counts.to_string())


# ----------------------------------------------------------------------
# 6. MULTI-CHARACTER STATES
# ----------------------------------------------------------------------

print("\n6. MULTI-CHARACTER STATE INVENTORY")
print("-" * 70)

multi_mask = (
    mutation_values.notna()
    & ~mutation_values.isin(["-", "."])
    & (mutation_values.str.len() > 1)
)

multi_states = mutation_values.where(multi_mask).stack()

multi_counts = multi_states.value_counts()

print("Unique multi-character states:",
      len(multi_counts))

print("\nTop 50 multi-character states:")
print(multi_counts.head(50).to_string())


# ----------------------------------------------------------------------
# 7. MULTI-CHARACTER LENGTH DISTRIBUTION
# ----------------------------------------------------------------------

print("\n7. MULTI-CHARACTER STATE LENGTH DISTRIBUTION")
print("-" * 70)

multi_lengths = multi_states.astype(str).str.len()

print(
    multi_lengths.value_counts()
    .sort_index()
    .to_string()
)


# ----------------------------------------------------------------------
# 8. POSITION-SPECIFIC SUMMARY
# ----------------------------------------------------------------------

print("\n8. POSITION-SPECIFIC STATE CHECK")
print("-" * 70)

position_summary = []

for col in p_cols:

    s = original_df[col].astype("string")

    dash = int((s == "-").sum())
    dot = int((s == ".").sum())
    missing = int(s.isna().sum())

    nonref = int(
        (
            s.notna()
            & ~s.isin(["-", "."])
        ).sum()
    )

    single = int(
        (
            s.notna()
            & ~s.isin(["-", "."])
            & (s.str.len() == 1)
        ).sum()
    )

    multi = int(
        (
            s.notna()
            & ~s.isin(["-", "."])
            & (s.str.len() > 1)
        ).sum()
    )

    position_summary.append([
        col,
        dash,
        dot,
        missing,
        nonref,
        single,
        multi
    ])


position_df = pd.DataFrame(
    position_summary,
    columns=[
        "Position",
        "Dash_-",
        "Dot_.",
        "Missing_NaN",
        "NonReference",
        "SingleChar",
        "MultiChar"
    ]
)

print(position_df.to_string(index=False))


# ----------------------------------------------------------------------
# 9. POSITIONS WITH MULTI-CHARACTER STATES
# ----------------------------------------------------------------------

print("\n9. POSITIONS CONTAINING MULTI-CHARACTER STATES")
print("-" * 70)

multi_position_summary = position_df[
    position_df["MultiChar"] > 0
]

print(
    multi_position_summary[
        ["Position", "MultiChar"]
    ].to_string(index=False)
)

print(
    "\nNumber of positions containing multi-character states:",
    len(multi_position_summary)
)


# ----------------------------------------------------------------------
# 10. UNEXPECTED SINGLE-CHARACTER SYMBOLS
# ----------------------------------------------------------------------

print("\n10. UNEXPECTED SYMBOL SCREEN")
print("-" * 70)

aa_letters = set("ACDEFGHIKLMNPQRSTVWY")

single_unique = set(
    single_counts.index.astype(str)
)

unexpected_single = sorted(
    single_unique - aa_letters
)

print("Standard amino-acid alphabet:")
print(sorted(aa_letters))

print("\nUnexpected single-character states:")
print(unexpected_single)

if len(unexpected_single) == 0:
    print("✓ No unexpected single-character symbols detected")
else:
    print("⚠ Unexpected single-character symbols require review")


# ----------------------------------------------------------------------
# 11. MULTI-CHARACTER SYMBOL SCREEN
# ----------------------------------------------------------------------

print("\n11. MULTI-CHARACTER SYMBOL SCREEN")
print("-" * 70)

unexpected_multi = []

for state in multi_counts.index.astype(str):

    invalid_chars = sorted(
        set(state) - aa_letters
    )

    if invalid_chars:

        unexpected_multi.append([
            state,
            int(multi_counts[state]),
            "".join(invalid_chars)
        ])


if len(unexpected_multi) > 0:

    unexpected_multi_df = pd.DataFrame(
        unexpected_multi,
        columns=[
            "State",
            "Count",
            "UnexpectedCharacters"
        ]
    )

    print(
        unexpected_multi_df
        .sort_values("Count", ascending=False)
        .head(100)
        .to_string(index=False)
    )

else:

    print("✓ No unexpected characters found")


# ----------------------------------------------------------------------
# 12. FINAL DATASET INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n12. FINAL DATASET INTEGRITY CHECK")
print("-" * 70)

print("original_df shape    :", original_df.shape)
print("df_transformed shape :", df_transformed.shape)

assert original_df.shape == (4350, 116)
assert df_transformed.shape == (4350, 116)

print("✓ original_df unchanged")
print("✓ df_transformed unchanged")
print("✓ No rows deleted")
print("✓ No values modified")
print("✓ No labels created")
print("✓ No encoding performed")


# ----------------------------------------------------------------------
# FINAL
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 221 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 223: MUTATION-STATE NORMALIZATION MAP
# ==============================================================
print("=" * 70)
print("STEP 223: MUTATION-STATE NORMALIZATION MAP")
print("=" * 70)

import pandas as pd
import re

# --------------------------------------------------------------
# 1. DATASET SAFETY SNAPSHOT
# --------------------------------------------------------------

original_shape_before_223 = original_df.shape
transformed_shape_before_223 = df_transformed.shape
original_columns_before_223 = list(original_df.columns)
transformed_columns_before_223 = list(df_transformed.columns)

print("\n1. DATASET SAFETY")
print("-" * 70)
print("original_df shape    :", original_df.shape)
print("df_transformed shape :", df_transformed.shape)

# --------------------------------------------------------------
# 2. IDENTIFY P-COLUMNS
# --------------------------------------------------------------

p_cols = [f"P{i}" for i in range(1, 100)]

missing_p = [c for c in p_cols if c not in df_transformed.columns]

print("\n2. P-COLUMN CHECK")
print("-" * 70)
print("Expected P-columns :", len(p_cols))
print("Found P-columns    :", len([c for c in p_cols if c in df_transformed.columns]))
print("Missing P-columns  :", missing_p)

if missing_p:
    raise ValueError("P-column check failed.")

# --------------------------------------------------------------
# 3. COLLECT ALL OBSERVED MUTATION STATES
# --------------------------------------------------------------

mutation_values = df_transformed[p_cols].astype("string")

all_states = mutation_values.stack(dropna=False)

state_counts = all_states.value_counts(dropna=False)

print("\n3. OBSERVED STATE INVENTORY")
print("-" * 70)
print("Total P-column entries :", len(all_states))
print("Unique observed states :", all_states.nunique(dropna=False))

# --------------------------------------------------------------
# 4. CLASSIFY EACH STATE
# --------------------------------------------------------------

standard_aa = set(
    list("ACDEFGHIKLMNPQRSTVWY")
)

def classify_state(value):
    if pd.isna(value):
        return "NaN"

    s = str(value)

    if s == "-":
        return "Dash"

    if s == ".":
        return "Dot"

    if len(s) == 1:
        if s in standard_aa:
            return "StandardSingleAA"
        else:
            return "UnexpectedSingleSymbol"

    if len(s) > 1:
        unexpected = [ch for ch in s if ch not in standard_aa]

        if unexpected:
            return "MultiCharWithSpecialSymbol"

        return "MultiCharacterAA"

    return "Other"


# --------------------------------------------------------------
# 5. BUILD NORMALIZATION MAP
# --------------------------------------------------------------

normalization_rows = []

for state, count in state_counts.items():

    if pd.isna(state):
        state_string = "<NaN>"
    else:
        state_string = str(state)

    category = classify_state(state)

    if category == "Dash":
        treatment = "KEEP_AS_DASH"

    elif category == "Dot":
        treatment = "KEEP_AS_DOT"

    elif category == "NaN":
        treatment = "KEEP_AS_NAN"

    elif category == "StandardSingleAA":
        treatment = "KEEP_AS_SINGLE_AA"

    elif category == "MultiCharacterAA":
        treatment = "KEEP_AS_MULTI_AA_STATE"

    elif category == "UnexpectedSingleSymbol":
        treatment = "REVIEW_SPECIAL_SYMBOL"

    elif category == "MultiCharWithSpecialSymbol":
        treatment = "REVIEW_SPECIAL_SYMBOL"

    else:
        treatment = "REVIEW"

    normalization_rows.append({
        "State": state_string,
        "Count": int(count),
        "Category": category,
        "Treatment": treatment
    })


mutation_state_map = pd.DataFrame(normalization_rows)

mutation_state_map = mutation_state_map.sort_values(
    by=["Category", "Count"],
    ascending=[True, False]
).reset_index(drop=True)

# --------------------------------------------------------------
# 6. DISPLAY SUMMARY
# --------------------------------------------------------------

print("\n4. NORMALIZATION CATEGORIES")
print("-" * 70)

print(
    mutation_state_map["Category"]
    .value_counts()
    .to_string()
)

print("\n5. STATES REQUIRING SPECIAL REVIEW")
print("-" * 70)

special_review = mutation_state_map[
    mutation_state_map["Treatment"] == "REVIEW_SPECIAL_SYMBOL"
]

if len(special_review) == 0:
    print("None")
else:
    print(
        special_review[
            ["State", "Count", "Category", "Treatment"]
        ].to_string(index=False)
    )

print("\n6. MULTI-CHARACTER AMINO-ACID STATES")
print("-" * 70)

multi_aa = mutation_state_map[
    mutation_state_map["Category"] == "MultiCharacterAA"
]

print(
    multi_aa[
        ["State", "Count", "Treatment"]
    ].head(50).to_string(index=False)
)

# --------------------------------------------------------------
# 7. COMPARE ORIGINAL DATASET OBJECTS
# --------------------------------------------------------------

print("\n7. DATASET INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == original_shape_before_223
assert df_transformed.shape == transformed_shape_before_223

assert list(original_df.columns) == original_columns_before_223
assert list(df_transformed.columns) == transformed_columns_before_223

print("✓ original_df shape unchanged")
print("✓ df_transformed shape unchanged")
print("✓ original_df columns unchanged")
print("✓ df_transformed columns unchanged")
print("✓ No rows deleted")
print("✓ No columns deleted")
print("✓ No values modified")
print("✓ No labels created")
print("✓ No encoding performed")

print("\n" + "=" * 70)
print("STEP 223 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 228: SPECIAL SYMBOL HANDLING DECISION AUDIT
# ======================================================================

print("=" * 70)
print("STEP 228: SPECIAL SYMBOL HANDLING DECISION AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. DATASET SAFETY
# ----------------------------------------------------------------------

print("\n1. DATASET SAFETY")
print("-" * 70)

print("original_df shape    :", original_df.shape)
print("df_transformed shape :", df_transformed.shape)

print("✓ Diagnostic audit only")
print("✓ No rows will be deleted")
print("✓ No columns will be deleted")
print("✓ No values will be modified")
print("✓ No labels will be created")
print("✓ No encoding will be performed")


# ----------------------------------------------------------------------
# 2. P-COLUMN CHECK
# ----------------------------------------------------------------------

print("\n2. P-COLUMN CHECK")
print("-" * 70)

p_cols = [f"P{i}" for i in range(1, 100)]

found_p = [c for c in p_cols if c in df_transformed.columns]
missing_p = [c for c in p_cols if c not in df_transformed.columns]

print("Expected P-columns :", len(p_cols))
print("Found P-columns    :", len(found_p))
print("Missing P-columns  :", missing_p)


# ----------------------------------------------------------------------
# 3. SPECIAL STATE EXTRACTION
# ----------------------------------------------------------------------

print("\n3. SPECIAL STATE EXTRACTION")
print("-" * 70)

mutation_values = df_transformed[p_cols]

special_records = []

for col in p_cols:
    for idx, value in mutation_values[col].items():

        if pd.isna(value):
            continue

        state = str(value).strip()

        if any(symbol in state for symbol in ["X", "*", "#"]):
            special_records.append({
                "row_index": idx,
                "position": col,
                "state": state
            })

special_df = pd.DataFrame(special_records)

print("Special-state observations :", len(special_df))


# ----------------------------------------------------------------------
# 4. STATE-LEVEL DECISION MAP
# ----------------------------------------------------------------------

print("\n4. STATE-LEVEL HANDLING DECISION")
print("-" * 70)

def classify_special_state(state):

    state = str(state).strip()

    # Pure X = ambiguous amino-acid state
    if state == "X":
        return "AMBIGUOUS_AA", "RETAIN_FOR_SEPARATE_HANDLING"

    # Pure * = termination/stop symbol
    if state == "*":
        return "STOP_OR_TERMINATION", "RETAIN_FOR_SEPARATE_HANDLING"

    # Any state containing # = unknown/special annotation
    if "#" in state:
        return "HASH_SPECIAL", "RETAIN_FOR_SEPARATE_HANDLING"

    # Any state containing * with amino-acid letters
    if "*" in state:
        return "STAR_SPECIAL", "RETAIN_FOR_SEPARATE_HANDLING"

    return "OTHER", "REVIEW"


if len(special_df) > 0:

    decisions = []

    for state in sorted(special_df["state"].unique()):

        count = int((special_df["state"] == state).sum())

        semantic_class, treatment = classify_special_state(state)

        decisions.append({
            "state": state,
            "count": count,
            "semantic_class": semantic_class,
            "treatment": treatment
        })

    decision_df = pd.DataFrame(decisions)

    print(decision_df.to_string(index=False))

else:

    decision_df = pd.DataFrame(
        columns=["state", "count", "semantic_class", "treatment"]
    )

    print("No special states found.")


# ----------------------------------------------------------------------
# 5. CATEGORY-LEVEL SUMMARY
# ----------------------------------------------------------------------

print("\n5. CATEGORY-LEVEL HANDLING SUMMARY")
print("-" * 70)

if len(special_df) > 0:

    special_df["SemanticClass"] = special_df["state"].apply(
        lambda x: classify_special_state(x)[0]
    )

    summary = (
        special_df["SemanticClass"]
        .value_counts()
        .rename_axis("SemanticClass")
        .reset_index(name="Count")
    )

    print(summary.to_string(index=False))

else:

    print("No special states found.")


# ----------------------------------------------------------------------
# 6. POSITION-LEVEL HANDLING SUMMARY
# ----------------------------------------------------------------------

print("\n6. POSITION-LEVEL SPECIAL STATE SUMMARY")
print("-" * 70)

if len(special_df) > 0:

    position_summary = (
        special_df
        .groupby(["position", "state"])
        .size()
        .reset_index(name="Count")
        .sort_values(["position", "state"])
    )

    print(position_summary.to_string(index=False))

else:

    print("No special states found.")


# ----------------------------------------------------------------------
# 7. COMPMUTLIST CROSS-CHECK
# ----------------------------------------------------------------------

print("\n7. COMPMUTLIST CROSS-CHECK")
print("-" * 70)

if "CompMutList" in df_transformed.columns:

    comp_special_mask = (
        df_transformed["CompMutList"]
        .fillna("")
        .astype(str)
        .str.contains(r"[X*#]", regex=True)
    )

    print(
        "CompMutList rows containing X/*/# :",
        int(comp_special_mask.sum())
    )

    print(
        "CompMutList NaN count             :",
        int(df_transformed["CompMutList"].isna().sum())
    )

else:

    print("CompMutList column not found.")


# ----------------------------------------------------------------------
# 8. ORIGINAL vs TRANSFORMED SAFETY CHECK
# ----------------------------------------------------------------------

print("\n8. ORIGINAL vs TRANSFORMED SAFETY CHECK")
print("-" * 70)

shape_same = original_df.shape == df_transformed.shape
columns_same = list(original_df.columns) == list(df_transformed.columns)

print("Shape unchanged   :", shape_same)
print("Columns unchanged :", columns
```

```python
# ================================================================
# STEP 228B: DRUG-COLUMN TRANSFORMATION FORENSIC AUDIT
# ================================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("STEP 228B: DRUG-COLUMN TRANSFORMATION FORENSIC AUDIT")
print("=" * 70)

drug_columns = [
    "FPV", "ATV", "TPV", "DRV",
    "LPV", "IDV", "NFV", "SQV"
]

# ------------------------------------------------
# 1. CONFIRM CHANGED COLUMNS
# ------------------------------------------------
print("\n1. CHANGED DRUG COLUMNS")
print("-" * 70)

for col in drug_columns:

    if col not in original_df.columns:
        print(f"{col}: NOT FOUND")
        continue

    a = original_df[col]
    b = df_transformed[col]

    changed = ~(
        a.eq(b) |
        (a.isna() & b.isna())
    )

    print(
        f"{col:5s} | changed = {changed.sum():4d} "
        f"| original non-null = {a.notna().sum():4d} "
        f"| transformed non-null = {b.notna().sum():4d}"
    )


# ------------------------------------------------
# 2. TEST COMMON MATHEMATICAL TRANSFORMATIONS
# ------------------------------------------------
print("\n2. TRANSFORMATION TEST")
print("-" * 70)

tests = []

for col in drug_columns:

    a = pd.to_numeric(original_df[col], errors="coerce")
    b = pd.to_numeric(df_transformed[col], errors="coerce")

    mask = a.notna() & b.notna()

    if mask.sum() == 0:
        continue

    x = a[mask].astype(float)
    y = b[mask].astype(float)

    # Test common transformations
    errors = {}

    # log(x)
    valid = x > 0
    if valid.any():
        errors["ln(x)"] = np.mean(
            np.abs(y[valid] - np.log(x[valid]))
        )

        errors["log10(x)"] = np.mean(
            np.abs(y[valid] - np.log10(x[valid]))
        )

        errors["ln(x+1)"] = np.mean(
            np.abs(y[valid] - np.log1p(x[valid]))
        )

    # x itself
    errors["x"] = np.mean(np.abs(y - x))

    best = min(errors, key=errors.get)

    print(f"\n{col}")
    print("  Best candidate :", best)
    print("  Mean abs error :", errors[best])

    for name, error in errors.items():
        print(f"  {name:10s}: {error}")


# ------------------------------------------------
# 3. EXACT SAMPLE COMPARISON
# ------------------------------------------------
print("\n3. EXACT SAMPLE COMPARISON")
print("-" * 70)

for col in drug_columns:

    print(f"\n--- {col} ---")

    a = original_df[col]
    b = df_transformed[col]

    changed = ~(
        a.eq(b) |
        (a.isna() & b.isna())
    )

    sample_indices = original_df.index[changed][:10]

    for idx in sample_indices:

        old = a.loc[idx]
        new = b.loc[idx]

        if pd.notna(old) and pd.notna(new):

            print(
                f"index {idx}: "
                f"original={old!r} | "
                f"transformed={new!r} | "
                f"ln(original)={np.log(old) if old > 0 else 'NA'} | "
                f"ln(original+1)={np.log1p(old) if old >= 0 else 'NA'}"
            )


# ------------------------------------------------
# 4. DATA TYPES
# ------------------------------------------------
print("\n4. DATA TYPES")
print("-" * 70)

for col in drug_columns:

    print(
        f"{col:5s} | "
        f"original dtype = {original_df[col].dtype} | "
        f"transformed dtype = {df_transformed[col].dtype}"
    )


# ------------------------------------------------
# 5. SPECIAL-STATE PRESERVATION
# ------------------------------------------------
print("\n5. SPECIAL-STATE PRESERVATION")
print("-" * 70)

special_symbols = {
    "X", "K*", "L*", "Q*", "W*", "L#", "*",
    "E*", "R*", "E#", "M#", "N#", "P#", "K#",
    "W*R"
}

for col in drug_columns:

    original_special = original_df[col].isin(special_symbols).sum()
    transformed_special = df_transformed[col].isin(special_symbols).sum()

    print(
        f"{col:5s} | "
        f"original special = {original_special:3d} | "
        f"transformed special = {transformed_special:3d}"
    )


# ------------------------------------------------
# 6. FINAL STATUS
# ------------------------------------------------
print("\n" + "=" * 70)
print("STEP 228B SUMMARY")
print("=" * 70)

print("""
The 26,862 value differences are confined to the 8 drug columns.

DO NOT reverse-transform anything yet.

The purpose of this step is to identify exactly which mathematical
transformation was previously applied and whether special states
remain untouched.
""")
```

```python
# ================================================================
# STEP 228C: TRANSFORMATION INTENT & PIPELINE CONSISTENCY CHECK
# ================================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("STEP 228C: TRANSFORMATION INTENT & PIPELINE CONSISTENCY CHECK")
print("=" * 70)

# ------------------------------------------------
# 1. Drug columns
# ------------------------------------------------
drug_cols = ['FPV', 'ATV', 'TPV', 'DRV', 'LPV', 'IDV', 'NFV', 'SQV']

print("\n1. DRUG COLUMNS")
print("-" * 70)
print(drug_cols)

# ------------------------------------------------
# 2. Verify transformed values exactly match ln(x+1)
# ------------------------------------------------
print("\n2. EXACT ln(x+1) REPRODUCTION")
print("-" * 70)

all_match = True

for col in drug_cols:

    original = pd.to_numeric(original_df[col], errors='coerce')
    transformed = pd.to_numeric(df_transformed[col], errors='coerce')

    expected = np.log1p(original)

    valid = original.notna() & transformed.notna()

    if valid.any():
        max_error = np.max(
            np.abs(
                transformed[valid].to_numpy()
                - expected[valid].to_numpy()
            )
        )
    else:
        max_error = 0

    match = np.allclose(
        transformed[valid],
        expected[valid],
        rtol=1e-12,
        atol=1e-12
    )

    print(
        f"{col:5s} | ln(x+1) exact match = {match} "
        f"| max error = {max_error:.3e}"
    )

    if not match:
        all_match = False

print("\nOverall ln(x+1) reproduction:", all_match)

# ------------------------------------------------
# 3. Check whether all other columns are unchanged
# ------------------------------------------------
print("\n3. NON-DRUG COLUMN INTEGRITY")
print("-" * 70)

non_drug_cols = [
    c for c in original_df.columns
    if c not in drug_cols
]

other_changes = []

for col in non_drug_cols:

    a = original_df[col]
    b = df_transformed[col]

    same = a.equals(b)

    if not same:
        other_changes.append(col)

print("Non-drug columns changed:", len(other_changes))

if other_changes:
    print("Changed non-drug columns:")
    print(other_changes)
else:
    print("STATUS: PASS — no non-drug columns changed.")

# ------------------------------------------------
# 4. Special-state preservation
# ------------------------------------------------
print("\n4. SPECIAL-STATE PRESERVATION")
print("-" * 70)

special_symbols = ['X', '*', '#']

special_change_found = False

for col in drug_cols:

    orig_special = original_df[col].astype(str).str.contains(
        r'[*#X]',
        regex=True,
        na=False
    )

    trans_special = df_transformed[col].astype(str).str.contains(
        r'[*#X]',
        regex=True,
        na=False
    )

    if orig_special.sum() != trans_special.sum():
        special_change_found = True
        print(
            f"{col}: special-state count changed "
            f"{orig_special.sum()} -> {trans_special.sum()}"
        )

if not special_change_found:
    print("STATUS: PASS — no special-state loss detected.")

# ------------------------------------------------
# 5. Check transformed numeric ranges
# ------------------------------------------------
print("\n5. TRANSFORMED DRUG RANGES")
print("-" * 70)

for col in drug_cols:

    numeric = pd.to_numeric(
        df_transformed[col],
        errors='coerce'
    ).dropna()

    print(
        f"{col:5s} | min = {numeric.min():.6f} "
        f"| max = {numeric.max():.6f}"
    )

# ------------------------------------------------
# 6. Final interpretation
# ------------------------------------------------
print("\n" + "=" * 70)
print("STEP 228C SUMMARY")
print("=" * 70)

if all_match and not other_changes and not special_change_found:

    print("""
PASS — The observed changes are fully explained by the
previously applied ln(x+1) transformation to the eight
drug columns.

No unexpected changes were detected outside those columns.
Special mutation states remain preserved.

DO NOT reverse-transform yet.
The transformation should only be reversed if the project
pipeline explicitly requires the original drug-scale values.
""")

else:

    print("""
REVIEW REQUIRED — Unexpected differences remain.
Do NOT proceed to the next preprocessing stage.
""")
```

```python
# ================================================================
# STEP 228D: FINAL TRANSFORMATION DECISION AUDIT
# ================================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("STEP 228D: FINAL TRANSFORMATION DECISION AUDIT")
print("=" * 70)

drug_cols = ['FPV', 'ATV', 'TPV', 'DRV', 'LPV', 'IDV', 'NFV', 'SQV']

# ------------------------------------------------
# 1. Confirm source and transformed objects exist
# ------------------------------------------------
print("\n1. OBJECT AVAILABILITY")
print("-" * 70)

print("original_df exists   :", 'original_df' in globals())
print("df_transformed exists:", 'df_transformed' in globals())

# ------------------------------------------------
# 2. Confirm dataset shape
# ------------------------------------------------
print("\n2. DATASET SHAPE")
print("-" * 70)

print("original_df   :", original_df.shape)
print("df_transformed:", df_transformed.shape)

# ------------------------------------------------
# 3. Confirm exactly 99 P-columns
# ------------------------------------------------
print("\n3. P-COLUMN COUNT")
print("-" * 70)

p_cols = [
    c for c in df_transformed.columns
    if str(c).startswith("P") and str(c)[1:].isdigit()
]

print("P-column count:", len(p_cols))
print("Expected      : 99")

# ------------------------------------------------
# 4. Confirm all drug transformations
# ------------------------------------------------
print("\n4. DRUG TRANSFORMATION STATUS")
print("-" * 70)

transformation_pass = True

for col in drug_cols:

    original_numeric = pd.to_numeric(
        original_df[col],
        errors='coerce'
    )

    transformed_numeric = pd.to_numeric(
        df_transformed[col],
        errors='coerce'
    )

    valid = (
        original_numeric.notna()
        & transformed_numeric.notna()
    )

    expected = np.log1p(original_numeric[valid])

    actual = transformed_numeric[valid]

    match = np.allclose(
        actual,
        expected,
        rtol=1e-12,
        atol=1e-12
    )

    print(f"{col:5s} | ln(x+1) = {match}")

    if not match:
        transformation_pass = False

# ------------------------------------------------
# 5. Confirm special states were not transformed
# ------------------------------------------------
print("\n5. SPECIAL-STATE CHECK")
print("-" * 70)

special_pattern = r'[*#X]'

special_original_total = 0
special_transformed_total = 0

for col in df_transformed.columns:

    orig_special = original_df[col].astype(str).str.contains(
        special_pattern,
        regex=True,
        na=False
    )

    trans_special = df_transformed[col].astype(str).str.contains(
        special_pattern,
        regex=True,
        na=False
    )

    special_original_total += orig_special.sum()
    special_transformed_total += trans_special.sum()

print(
    "Original special-state cells   :",
    special_original_total
)

print(
    "Transformed special-state cells:",
    special_transformed_total
)

special_pass = (
    special_original_total == special_transformed_total
)

print("Special-state preservation:", special_pass)

# ------------------------------------------------
# 6. Confirm no unexpected columns changed
# ------------------------------------------------
print("\n6. UNEXPECTED COLUMN CHANGES")
print("-" * 70)

unexpected_changes = []

for col in original_df.columns:

    if col in drug_cols:
        continue

    if not original_df[col].equals(df_transformed[col]):
        unexpected_changes.append(col)

print("Unexpected changed columns:", len(unexpected_changes))

if unexpected_changes:
    print(unexpected_changes)

# ------------------------------------------------
# 7. Check CompMutList
# ------------------------------------------------
print("\n7. COMPMUTLIST")
print("-" * 70)

print("CompMutList present:", 'CompMutList' in df_transformed.columns)

if 'CompMutList' in df_transformed.columns:
    print(
        "CompMutList NaN:",
        df_transformed['CompMutList'].isna().sum()
    )

# ------------------------------------------------
# 8. FINAL DECISION
# ------------------------------------------------
print("\n" + "=" * 70)
print("STEP 228D FINAL DECISION")
print("=" * 70)

if (
    transformation_pass
    and len(p_cols) == 99
    and special_pass
    and len(unexpected_changes) == 0
    and 'CompMutList' in df_transformed.columns
):

    print("""
PASS

The current transformed dataset is internally consistent.

The eight drug columns are intentionally represented as:

    ln(x + 1)

No unexpected columns were changed.
Special mutation states are preserved.
Exactly 99 P-columns are present.
CompMutList is present.

DECISION:
KEEP df_transformed AS-IS.

DO NOT reverse-transform the drug columns.
DO NOT modify the dataset at this stage.
""")

else:

    print("""
REVIEW REQUIRED

One or more integrity conditions failed.
DO NOT proceed to the next preprocessing stage.
""")
```

```python
# ============================================================
# STEP 248: PHENOTYPE REFERENCE COVERAGE BY DRUG
# ============================================================

print("=" * 70)
print("STEP 248: PHENOTYPE REFERENCE COVERAGE")
print("=" * 70)

# Drug columns in the original dataset
drug_columns = [
    "FPV", "ATV", "IDV", "LPV",
    "NFV", "SQV", "TPV", "DRV"
]

# Create a working copy ONLY for this audit
audit_df = df_original.copy()

# Convert phenotype fields to numeric.
# "NA", "-", and other non-numeric values become NaN.
for drug in drug_columns:
    audit_df[drug + "_numeric"] = pd.to_numeric(
        audit_df[drug],
        errors="coerce"
    )

# ------------------------------------------------------------
# Build reference-drug phenotype coverage
# ------------------------------------------------------------

records = []

for refid in sorted(
    pd.to_numeric(audit_df["RefID"], errors="coerce")
    .dropna()
    .unique()
):

    ref_rows = audit_df[
        pd.to_numeric(audit_df["RefID"], errors="coerce") == refid
    ]

    for drug in drug_columns:

        n_rows = len(ref_rows)

        n_with_phenotype = (
            ref_rows[drug + "_numeric"]
            .notna()
            .sum()
        )

        if n_with_phenotype > 0:
            records.append({
                "RefID": int(refid),
                "Drug": drug,
                "Rows": n_rows,
                "Rows_with_phenotype": int(n_with_phenotype)
            })

ref_drug = pd.DataFrame(records)

# Sort
ref_drug = ref_drug.sort_values(
    ["RefID", "Drug"]
).reset_index(drop=True)

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)
print("ref_drug exists :", isinstance(ref_drug, pd.DataFrame))

print("\n2. REFERENCE-DRUG COVERAGE")
print("-" * 70)
print("Reference-drug combinations :", len(ref_drug))
print("Unique RefIDs               :", ref_drug["RefID"].nunique())
print("Unique drugs                :", ref_drug["Drug"].nunique())

print("\n3. DRUG COVERAGE")
print("-" * 70)

drug_summary = (
    ref_drug.groupby("Drug")
    .agg(
        RefIDs=("RefID", "nunique"),
        Rows_with_phenotype=("Rows_with_phenotype", "sum")
    )
    .reindex(drug_columns)
    .reset_index()
)

display(drug_summary)

print("\n4. REFID COVERAGE")
print("-" * 70)

refid_summary = (
    ref_drug.groupby("RefID")
    .agg(
        Drugs_with_phenotype=("Drug", "nunique"),
        Total_phenotype_rows=("Rows_with_phenotype", "sum")
    )
    .reset_index()
)

print("RefIDs with at least one phenotype :", len(refid_summary))
display(refid_summary)

print("\n5. COMPLETE REFERENCE-DRUG TABLE")
print("-" * 70)
display(ref_drug)

print("\n6. IMPORTANT CHECK")
print("-" * 70)

all_original_refids = set(
    pd.to_numeric(
        df_original["RefID"],
        errors="coerce"
    ).dropna().astype(int)
)

phenotype_refids = set(
    ref_drug["RefID"].astype(int)
)

print(
    "Original RefIDs:",
    len(all_original_refids)
)

print(
    "RefIDs with phenotype:",
    len(phenotype_refids)
)

print(
    "RefIDs without any phenotype:",
    sorted(all_original_refids - phenotype_refids)
)

print("\n7. FINAL STATUS")
print("-" * 70)
print("df_original modified       : NO")
print("Phenotype values changed   : NO")
print("Resistance labels created : NO")
print("Resistance cutoff applied  : NO")
print("Rows reordered in source   : NO")

print("\n" + "=" * 70)
print("STEP 248 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 252: PHENOTYPE SCALE AND DISTRIBUTION AUDIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 252: PHENOTYPE SCALE AND DISTRIBUTION AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------
print("\n1. OBJECT CHECK")
print("-" * 70)

print("df_original exists :", "df_original" in globals())

drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

print("\nDrug columns:")
for drug in drugs:
    print(f"{drug:5} :", drug in df_original.columns)

# ------------------------------------------------
# 2. NUMERIC PHENOTYPE DISTRIBUTION
# ------------------------------------------------
print("\n2. NUMERIC PHENOTYPE DISTRIBUTION")
print("-" * 70)

distribution_summary = []

for drug in drugs:

    values = pd.to_numeric(
        df_original[drug],
        errors="coerce"
    ).dropna()

    if len(values) == 0:
        continue

    distribution_summary.append({
        "Drug": drug,
        "N": len(values),
        "Min": values.min(),
        "Q1": values.quantile(0.25),
        "Median": values.median(),
        "Q3": values.quantile(0.75),
        "Max": values.max(),
        "Mean": values.mean(),
        "Std": values.std()
    })

phenotype_distribution = pd.DataFrame(distribution_summary)

print(phenotype_distribution.to_string(index=False))

# ------------------------------------------------
# 3. LOG10 PHENOTYPE DISTRIBUTION
# ------------------------------------------------
print("\n3. LOG10 PHENOTYPE DISTRIBUTION")
print("-" * 70)

log_summary = []

for drug in drugs:

    values = pd.to_numeric(
        df_original[drug],
        errors="coerce"
    ).dropna()

    values = values[values > 0]

    if len(values) == 0:
        continue

    log_values = np.log10(values)

    log_summary.append({
        "Drug": drug,
        "N_positive": len(log_values),
        "Log10_min": log_values.min(),
        "Log10_Q1": log_values.quantile(0.25),
        "Log10_median": log_values.median(),
        "Log10_Q3": log_values.quantile(0.75),
        "Log10_max": log_values.max()
    })

log10_summary = pd.DataFrame(log_summary)

print(log10_summary.to_string(index=False))

# ------------------------------------------------
# 4. ZERO / NEGATIVE VALUE CHECK
# ------------------------------------------------
print("\n4. ZERO / NEGATIVE VALUE CHECK")
print("-" * 70)

zero_negative_summary = []

for drug in drugs:

    values = pd.to_numeric(
        df_original[drug],
        errors="coerce"
    ).dropna()

    zero_count = int((values == 0).sum())
    negative_count = int((values < 0).sum())

    zero_negative_summary.append({
        "Drug": drug,
        "Zero_values": zero_count,
        "Negative_values": negative_count
    })

zero_negative = pd.DataFrame(zero_negative_summary)

print(zero_negative.to_string(index=False))

# ------------------------------------------------
# 5. EXTREME VALUE CHECK
# ------------------------------------------------
print("\n5. EXTREME VALUE CHECK")
print("-" * 70)

extreme_summary = []

for drug in drugs:

    values = pd.to_numeric(
        df_original[drug],
        errors="coerce"
    ).dropna()

    if len(values) == 0:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    upper_fence = q3 + (3 * iqr)

    extreme_count = int((values > upper_fence).sum())

    extreme_summary.append({
        "Drug": drug,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Upper_fence_3IQR": upper_fence,
        "Extreme_high_values": extreme_count
    })

extreme_values = pd.DataFrame(extreme_summary)

print(extreme_values.to_string(index=False))

# ------------------------------------------------
# 6. RAW PHENOTYPE EXAMPLES
# ------------------------------------------------
print("\n6. LOW / MEDIAN / HIGH PHENOTYPE EXAMPLES")
print("-" * 70)

examples = []

for drug in drugs:

    values = pd.to_numeric(
        df_original[drug],
        errors="coerce"
    ).dropna()

    if len(values) == 0:
        continue

    examples.append({
        "Drug": drug,
        "Lowest": values.nsmallest(5).tolist(),
        "Highest": values.nlargest(5).tolist()
    })

phenotype_examples = pd.DataFrame(examples)

print(phenotype_examples.to_string(index=False))

# ------------------------------------------------
# 7. SANITY CHECK
# ------------------------------------------------
print("\n7. SANITY CHECK")
print("-" * 70)

print("Original rows    :", len(df_original))
print("Original columns :", len(df_original.columns))

print("\nSource dataframe modified : NO")
print("Phenotype values changed  : NO")
print("Resistance labels created : NO")
print("Cutoff applied            : NO")

print("\n" + "=" * 70)
print("STEP 252 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 254: PHENOTYPE METHOD COMPATIBILITY AUDIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 254: PHENOTYPE METHOD COMPATIBILITY AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

print("df_original exists :", "df_original" in globals())

if "df_original" not in globals():
    raise NameError("df_original is not available.")

required_cols = ["RefID", "SeqID", "Method"]

drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

missing_cols = [c for c in required_cols + drugs if c not in df_original.columns]

print("Missing required columns :", missing_cols)

if missing_cols:
    raise ValueError("Required columns are missing.")

# ------------------------------------------------
# 2. METHOD × REFID TABLE
# ------------------------------------------------

print("\n2. METHOD × REFID STRUCTURE")
print("-" * 70)

method_refid = (
    df_original
    .groupby(["RefID", "Method"])
    .size()
    .reset_index(name="Rows")
)

print("Unique RefID × Method combinations :",
      len(method_refid))

refid_method_counts = (
    method_refid
    .groupby("RefID")["Method"]
    .nunique()
)

print("RefIDs with one method :",
      (refid_method_counts == 1).sum())

print("RefIDs with multiple methods :",
      (refid_method_counts > 1).sum())

print("\nRefIDs having multiple methods:")

multi_method_refids = refid_method_counts[
    refid_method_counts > 1
].sort_values(ascending=False)

print(multi_method_refids.to_string())

# ------------------------------------------------
# 3. METHOD PAIRS WITHIN SAME REFID
# ------------------------------------------------

print("\n3. REFID METHOD PAIRS")
print("-" * 70)

multi_refid_table = (
    method_refid[
        method_refid["RefID"].isin(multi_method_refids.index)
    ]
    .sort_values(["RefID", "Method"])
)

print(multi_refid_table.to_string(index=False))

# ------------------------------------------------
# 4. PHENOTYPE COVERAGE BY METHOD
# ------------------------------------------------

print("\n4. PHENOTYPE COVERAGE BY METHOD")
print("-" * 70)

coverage = []

for method in sorted(df_original["Method"].dropna().unique()):

    subset = df_original[
        df_original["Method"] == method
    ]

    for drug in drugs:

        numeric = pd.to_numeric(
            subset[drug],
            errors="coerce"
        )

        coverage.append({
            "Method": method,
            "Drug": drug,
            "Rows": len(subset),
            "Numeric": numeric.notna().sum(),
            "Missing": numeric.isna().sum(),
            "Coverage_percent":
                round(
                    numeric.notna().mean() * 100,
                    2
                )
        })

coverage_df = pd.DataFrame(coverage)

print(
    coverage_df.to_string(index=False)
)

# ------------------------------------------------
# 5. METHOD × DRUG LOG-SCALE SUMMARY
# ------------------------------------------------

print("\n5. METHOD × DRUG LOG10 SUMMARY")
print("-" * 70)

log_summary = []

for method in sorted(df_original["Method"].dropna().unique()):

    subset = df_original[
        df_original["Method"] == method
    ]

    for drug in drugs:

        values = pd.to_numeric(
            subset[drug],
            errors="coerce"
        ).dropna()

        values = values[values > 0]

        if len(values) == 0:
            continue

        log_values = np.log10(values)

        log_summary.append({
            "Method": method,
            "Drug": drug,
            "N": len(values),
            "Log10_Median":
                round(log_values.median(), 4),
            "Log10_Q1":
                round(log_values.quantile(0.25), 4),
            "Log10_Q3":
                round(log_values.quantile(0.75), 4),
            "Log10_Mean":
                round(log_values.mean(), 4)
        })

log_summary_df = pd.DataFrame(log_summary)

print(
    log_summary_df.to_string(index=False)
)

# ------------------------------------------------
# 6. SAME REFID — DIFFERENT METHODS — PHENOTYPE
# ------------------------------------------------

print("\n6. SAME REFID WITH MULTIPLE METHODS")
print("-" * 70)

cross_method_records = []

for refid in multi_method_refids.index:

    ref_subset = df_original[
        df_original["RefID"] == refid
    ]

    methods = sorted(
        ref_subset["Method"]
        .dropna()
        .unique()
    )

    cross_method_records.append({
        "RefID": refid,
        "Methods": " | ".join(methods),
        "Number_of_methods": len(methods),
        "Rows": len(ref_subset)
    })

cross_method_df = pd.DataFrame(
    cross_method_records
)

print(
    cross_method_df.to_string(index=False)
)

# ------------------------------------------------
# 7. METHOD × DRUG MEDIAN DIFFERENCE
# ------------------------------------------------

print("\n7. METHOD × DRUG MEDIAN RANGE")
print("-" * 70)

median_range_records = []

for drug in drugs:

    temp = log_summary_df[
        log_summary_df["Drug"] == drug
    ].copy()

    if len(temp) == 0:
        continue

    median_range_records.append({
        "Drug": drug,
        "Methods_with_data": len(temp),
        "Lowest_Log10_Median":
            temp["Log10_Median"].min(),
        "Highest_Log10_Median":
            temp["Log10_Median"].max(),
        "Median_Range":
            temp["Log10_Median"].max()
            - temp["Log10_Median"].min()
    })

median_range_df = pd.DataFrame(
    median_range_records
)

print(
    median_range_df.to_string(index=False)
)

# ------------------------------------------------
# 8. SANITY CHECK
# ------------------------------------------------

print("\n8. SANITY CHECK")
print("-" * 70)

print("Original rows    :", len(df_original))
print("Original columns :", len(df_original.columns))

print("Source dataframe modified : NO")
print("Phenotype values changed  : NO")
print("Resistance labels created : NO")
print("Cutoff applied            : NO")
print("Rows reordered             : NO")

print("\n" + "=" * 70)
print("STEP 254 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 255: RESISTANCE CUTOFF SOURCE / LABEL AVAILABILITY AUDIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 255: RESISTANCE CUTOFF SOURCE / LABEL AVAILABILITY AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

print("df_original exists :", "df_original" in globals())

if "df_original" not in globals():
    raise NameError("df_original is not available.")

# ------------------------------------------------
# 2. ALL COLUMN NAMES
# ------------------------------------------------

print("\n2. ALL ORIGINAL COLUMNS")
print("-" * 70)

print("Total columns :", len(df_original.columns))

for i, col in enumerate(df_original.columns):
    print(f"{i:03d} : {col}")

# ------------------------------------------------
# 3. SEARCH FOR POSSIBLE RESISTANCE / INTERPRETATION
# ------------------------------------------------

print("\n3. POSSIBLE RESISTANCE / INTERPRETATION COLUMNS")
print("-" * 70)

keywords = [
    "resist",
    "suscept",
    "sensitive",
    "response",
    "phenotype",
    "pheno",
    "fold",
    "change",
    "fc",
    "cutoff",
    "clinical",
    "biological",
    "interpret",
    "category",
    "class",
    "label",
    "score",
    "mutation"
]

candidate_cols = []

for col in df_original.columns:

    col_lower = str(col).lower()

    if any(keyword in col_lower for keyword in keywords):
        candidate_cols.append(col)

print("Candidate columns found :", len(candidate_cols))

for col in candidate_cols:
    print("-", col)

# ------------------------------------------------
# 4. POSSIBLE LABEL VALUE DISTRIBUTIONS
# ------------------------------------------------

print("\n4. CANDIDATE COLUMN VALUE DISTRIBUTIONS")
print("-" * 70)

for col in candidate_cols:

    print("\n" + "-" * 60)
    print("COLUMN:", col)
    print("dtype :", df_original[col].dtype)
    print("non-null :", df_original[col].notna().sum())
    print("unique :", df_original[col].nunique(dropna=True))

    values = df_original[col].dropna()

    if len(values) == 0:
        print("No non-null values.")
        continue

    # Show up to 20 most common values
    print("Most common values:")

    print(
        values
        .value_counts(dropna=False)
        .head(20)
        .to_string()
    )

# ------------------------------------------------
# 5. CATEGORICAL / TEXT COLUMNS
# ------------------------------------------------

print("\n5. LOW-CARDINALITY NON-NUMERIC COLUMNS")
print("-" * 70)

for col in df_original.columns:

    series = df_original[col]

    if not (
        pd.api.types.is_object_dtype(series)
        or pd.api.types.is_categorical_dtype(series)
    ):
        continue

    nunique = series.nunique(dropna=True)

    if 1 <= nunique <= 20:

        print(
            f"{col} | unique={nunique} | "
            f"values={series.dropna().unique()[:20]}"
        )

# ------------------------------------------------
# 6. NUMERIC COLUMNS THAT MAY REPRESENT CUTOFFS /
#    SCORES / INTERPRETATION VARIABLES
# ------------------------------------------------

print("\n6. NUMERIC COLUMNS WITH POTENTIAL LABEL/CUTOFF ROLE")
print("-" * 70)

numeric_candidates = []

for col in df_original.columns:

    col_lower = str(col).lower()

    if any(
        keyword in col_lower
        for keyword in [
            "resist",
            "suscept",
            "fold",
            "change",
            "cutoff",
            "score",
            "class",
            "label",
            "interpret"
        ]
    ):

        if pd.api.types.is_numeric_dtype(
            df_original[col]
        ):

            numeric_candidates.append(col)

for col in numeric_candidates:

    s = pd.to_numeric(
        df_original[col],
        errors="coerce"
    )

    print(
        f"{col} | "
        f"N={s.notna().sum()} | "
        f"unique={s.nunique(dropna=True)} | "
        f"min={s.min()} | "
        f"median={s.median()} | "
        f"max={s.max()}"
    )

# ------------------------------------------------
# 7. CHECK FOR COMMON RESISTANCE-LABEL VALUES
# ------------------------------------------------

print("\n7. SEARCH FOR COMMON RESISTANCE-LABEL TERMS")
print("-" * 70)

label_terms = [
    "resistant",
    "resistance",
    "susceptible",
    "sensitive",
    "intermediate",
    "high-level",
    "low-level",
    "partial",
    "unknown",
    "R",
    "S",
    "I"
]

for col in df_original.columns:

    if not (
        pd.api.types.is_object_dtype(df_original[col])
        or pd.api.types.is_categorical_dtype(df_original[col])
    ):
        continue

    values = (
        df_original[col]
        .dropna()
        .astype(str)
        .str.strip()
    )

    if len(values) == 0:
        continue

    lower_values = values.str.lower()

    found = []

    for term in label_terms:

        if term.lower() in set(lower_values.unique()):
            found.append(term)

    if found:

        print(
            f"{col} -> possible terms found: {found}"
        )

# ------------------------------------------------
# 8. CHECK EXISTING RESISTANCE COLUMNS EXPLICITLY
# ------------------------------------------------

print("\n8. EXPLICIT RESISTANCE COLUMN CHECK")
print("-" * 70)

resistance_like = [
    col for col in df_original.columns
    if "resist" in str(col).lower()
]

if resistance_like:

    print("Resistance-related columns:")
    for col in resistance_like:
        print("-", col)

else:

    print(
        "NO columns containing 'resist' were found."
    )

# ------------------------------------------------
# 9. CHECK PHENOTYPE COLUMN TYPES
# ------------------------------------------------

print("\n9. PHENOTYPE COLUMN DATA TYPES")
print("-" * 70)

drugs = [
    "FPV",
    "ATV",
    "IDV",
    "LPV",
    "NFV",
    "SQV",
    "TPV",
    "DRV"
]

for drug in drugs:

    s = df_original[drug]

    print(
        f"{drug}: "
        f"dtype={s.dtype}, "
        f"non-null={s.notna().sum()}, "
        f"unique={s.nunique(dropna=True)}"
    )

# ------------------------------------------------
# 10. SANITY CHECK
# ------------------------------------------------

print("\n10. SANITY CHECK")
print("-" * 70)

print("Original rows    :", len(df_original))
print("Original columns :", len(df_original.columns))

print("Source dataframe modified : NO")
print("Phenotype values changed  : NO")
print("Resistance labels created : NO")
print("Cutoff applied            : NO")
print("Rows reordered             : NO")

print("\n" + "=" * 70)
print("STEP 255 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 256: PHENOTYPE RAW VALUE / SPECIAL-VALUE AUDIT
# ==============================================================

print("=" * 70)
print("STEP 256: PHENOTYPE RAW VALUE / SPECIAL-VALUE AUDIT")
print("=" * 70)

import pandas as pd
import numpy as np

drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

print("\n1. OBJECT CHECK")
print("-" * 70)

print("df_original exists :", "df_original" in globals())

if "df_original" not in globals():
    raise NameError("df_original is not available.")

print("Rows    :", len(df_original))
print("Columns :", len(df_original.columns))


# --------------------------------------------------------------
# 2. RAW PHENOTYPE VALUE INSPECTION
# --------------------------------------------------------------

print("\n2. RAW PHENOTYPE VALUE INSPECTION")
print("-" * 70)

for drug in drugs:

    s = df_original[drug]

    print(f"\n{drug}")
    print("dtype :", s.dtype)
    print("unique values :", s.nunique(dropna=False))

    print("Most common raw values:")
    print(s.value_counts(dropna=False).head(20).to_string())


# --------------------------------------------------------------
# 3. NON-NUMERIC PHENOTYPE VALUES
# --------------------------------------------------------------

print("\n3. NON-NUMERIC PHENOTYPE VALUE AUDIT")
print("-" * 70)

for drug in drugs:

    s = df_original[drug]

    numeric = pd.to_numeric(s, errors="coerce")

    non_numeric_mask = s.notna() & numeric.isna()

    print(f"\n{drug}")
    print("Non-numeric values :", non_numeric_mask.sum())

    if non_numeric_mask.sum() > 0:
        print(
            s[non_numeric_mask]
            .value_counts(dropna=False)
            .head(30)
            .to_string()
        )


# --------------------------------------------------------------
# 4. SPECIAL SYMBOL SEARCH
# --------------------------------------------------------------

print("\n4. SPECIAL PHENOTYPE SYMBOL SEARCH")
print("-" * 70)

special_symbols = [
    "<", ">", "<=", ">=", "~", "=", "ND", "NA", "N/A",
    "NR", "NC", "-", ".", "?", "*"
]

for drug in drugs:

    s = df_original[drug].astype(str).str.strip()

    found = {}

    for symbol in special_symbols:
        count = s.str.contains(
            symbol,
            regex=False,
            case=False,
            na=False
        ).sum()

        if count > 0:
            found[symbol] = int(count)

    print(f"{drug}: {found}")


# --------------------------------------------------------------
# 5. VALUES BEGINNING / ENDING WITH CUTOFF SYMBOLS
# --------------------------------------------------------------

print("\n5. POTENTIAL CENSORED PHENOTYPE VALUES")
print("-" * 70)

for drug in drugs:

    s = df_original[drug].astype(str).str.strip()

    mask = (
        s.str.startswith(("<", ">"), na=False)
        | s.str.endswith(("<", ">"), na=False)
    )

    print(f"\n{drug}")
    print("Potential censored values :", mask.sum())

    if mask.sum() > 0:
        print(
            s[mask]
            .value_counts()
            .head(30)
            .to_string()
        )


# --------------------------------------------------------------
# 6. NUMERIC CONVERSION SUMMARY
# --------------------------------------------------------------

print("\n6. NUMERIC CONVERSION SUMMARY")
print("-" * 70)

summary = []

for drug in drugs:

    raw = df_original[drug]

    numeric = pd.to_numeric(raw, errors="coerce")

    summary.append({
        "Drug": drug,
        "Original_N": len(raw),
        "Original_nonnull": raw.notna().sum(),
        "Numeric_after_conversion": numeric.notna().sum(),
        "Non_numeric_after_conversion": (
            raw.notna() & numeric.isna()
        ).sum(),
        "Missing": raw.isna().sum()
    })

summary_df = pd.DataFrame(summary)

print(summary_df.to_string(index=False))


# --------------------------------------------------------------
# 7. PHENOTYPE RANGE AFTER SAFE NUMERIC CONVERSION
# --------------------------------------------------------------

print("\n7. NUMERIC RANGE AFTER SAFE CONVERSION")
print("-" * 70)

range_rows = []

for drug in drugs:

    numeric = pd.to_numeric(
        df_original[drug],
        errors="coerce"
    )

    numeric = numeric.dropna()

    range_rows.append({
        "Drug": drug,
        "N": len(numeric),
        "Min": numeric.min(),
        "Q1": numeric.quantile(0.25),
        "Median": numeric.median(),
        "Q3": numeric.quantile(0.75),
        "Max": numeric.max()
    })

range_df = pd.DataFrame(range_rows)

print(range_df.to_string(index=False))


# --------------------------------------------------------------
# 8. SOURCE DATA INTEGRITY
# --------------------------------------------------------------

print("\n8. SOURCE DATA INTEGRITY")
print("-" * 70)

print("Original rows    :", len(df_original))
print("Original columns :", len(df_original.columns))

print("Source dataframe modified : NO")
print("Resistance labels created : NO")
print("Cutoff applied            : NO")
print("Rows reordered             : NO")

print("\n" + "=" * 70)
print("STEP 256 COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 257: RESISTANCE CUTOFF SOURCE INVESTIGATION
# ==============================================================

print("=" * 70)
print("STEP 257: RESISTANCE CUTOFF SOURCE INVESTIGATION")
print("=" * 70)

import pandas as pd
import numpy as np

drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

# --------------------------------------------------------------
# 1. OBJECT CHECK
# --------------------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

print("df_original exists :", "df_original" in globals())

if "df_original" not in globals():
    raise NameError("df_original is not available.")

print("Rows    :", len(df_original))
print("Columns :", len(df_original.columns))


# --------------------------------------------------------------
# 2. SEARCH ALL COLUMN NAMES FOR CUTOFF / INTERPRETATION TERMS
# --------------------------------------------------------------

print("\n2. COLUMN-NAME CUTOFF / INTERPRETATION SEARCH")
print("-" * 70)

cutoff_terms = [
    "cutoff",
    "cut-off",
    "threshold",
    "interpret",
    "interpretation",
    "suscept",
    "resist",
    "sensitive",
    "intermediate",
    "clinical",
    "breakpoint",
    "fold",
    "fc",
    "iq",
    "score",
    "phenotype"
]

matches = []

for col in df_original.columns:

    col_lower = str(col).lower()

    matched_terms = [
        term for term in cutoff_terms
        if term in col_lower
    ]

    if matched_terms:
        matches.append({
            "Column": col,
            "Matched_terms": ", ".join(matched_terms)
        })

if matches:

    cutoff_name_df = pd.DataFrame(matches)

    print(
        cutoff_name_df.to_string(index=False)
    )

else:

    print("No cutoff / interpretation-related column names found.")


# --------------------------------------------------------------
# 3. SEARCH STRING CONTENT OF NON-PHENOTYPE COLUMNS
# --------------------------------------------------------------

print("\n3. SEARCH NON-PHENOTYPE DATA FOR CUTOFF / RESISTANCE TERMS")
print("-" * 70)

search_terms = [
    "resistant",
    "resistance",
    "susceptible",
    "sensitive",
    "intermediate",
    "cutoff",
    "threshold",
    "breakpoint",
    "clinical cutoff",
    "biological cutoff",
    "fold change",
    "fold-change"
]

non_drug_cols = [
    c for c in df_original.columns
    if c not in drugs
]

content_hits = []

for col in non_drug_cols:

    s = df_original[col].astype(str)

    for term in search_terms:

        mask = s.str.contains(
            term,
            case=False,
            regex=False,
            na=False
        )

        count = int(mask.sum())

        if count > 0:

            examples = (
                s[mask]
                .drop_duplicates()
                .head(10)
                .tolist()
            )

            content_hits.append({
                "Column": col,
                "Term": term,
                "Count": count,
                "Examples": examples
            })

if content_hits:

    for hit in content_hits:
        print(
            f"{hit['Column']} | "
            f"{hit['Term']} | "
            f"count={hit['Count']} | "
            f"examples={hit['Examples']}"
        )

else:

    print(
        "No resistance/cutoff terminology found "
        "in non-phenotype columns."
    )


# --------------------------------------------------------------
# 4. SEARCH FOR NUMERIC CUTOFF-LIKE COLUMNS
# --------------------------------------------------------------

print("\n4. NUMERIC COLUMNS OUTSIDE PHENOTYPE FIELDS")
print("-" * 70)

numeric_candidates = []

for col in df_original.columns:

    if col in drugs:
        continue

    converted = pd.to_numeric(
        df_original[col],
        errors="coerce"
    )

    numeric_count = converted.notna().sum()

    if numeric_count > 0:

        numeric_candidates.append({
            "Column": col,
            "Numeric_count": int(numeric_count),
            "Unique_numeric": int(converted.nunique()),
            "Min": converted.min(),
            "Max": converted.max()
        })

if numeric_candidates:

    numeric_candidate_df = pd.DataFrame(
        numeric_candidates
    )

    print(
        numeric_candidate_df.to_string(index=False)
    )

else:

    print(
        "No numeric columns outside phenotype fields."
    )


# --------------------------------------------------------------
# 5. CHECK ALL NON-MUTATION METADATA COLUMNS
# --------------------------------------------------------------

print("\n5. METADATA COLUMN VALUE SUMMARY")
print("-" * 70)

metadata_cols = [
    "SeqID",
    "PtID",
    "Subtype",
    "Method",
    "RefID",
    "Type",
    "IsolateName",
    "SeqType",
    "CompMutList"
]

for col in metadata_cols:

    if col not in df_original.columns:
        continue

    s = df_original[col]

    print(f"\n{col}")
    print("dtype :", s.dtype)
    print("unique:", s.nunique(dropna=False))

    print(
        s.value_counts(dropna=False)
        .head(15)
        .to_string()
    )


# --------------------------------------------------------------
# 6. CHECK WHETHER RESISTANCE LABELS MAY BE ENCODED IN
#    CompMutList
# --------------------------------------------------------------

print("\n6. CompMutList CONTENT CHECK")
print("-" * 70)

if "CompMutList" in df_original.columns:

    comp = df_original["CompMutList"].astype(str)

    print("Unique values :", comp.nunique(dropna=False))

    print("\nMost common CompMutList values:")

    print(
        comp.value_counts(dropna=False)
        .head(30)
        .to_string()
    )

    print("\nResistance-related terms in CompMutList:")

    for term in search_terms:

        mask = comp.str.contains(
            term,
            case=False,
            regex=False,
            na=False
        )

        if mask.sum() > 0:

            print(
                term,
                "->",
                int(mask.sum()),
                "rows"
            )

else:

    print("CompMutList not found.")


# --------------------------------------------------------------
# 7. CHECK WHETHER P1-P99 CONTAIN ACTUAL LABEL-LIKE VALUES
# --------------------------------------------------------------

print("\n7. MUTATION-COLUMN LABEL STRUCTURE CHECK")
print("-" * 70)

mutation_cols = [
    f"P{i}" for i in range(1, 100)
    if f"P{i}" in df_original.columns
]

label_like_values = {
    "R", "S", "I",
    "Resistant",
    "Susceptible",
    "Intermediate"
}

mutation_hits = []

for col in mutation_cols:

    values = (
        df_original[col]
        .astype(str)
        .str.strip()
    )

    for value in label_like_values:

        count = int(
            (values.str.lower() == value.lower()).sum()
        )

        if count > 0:

            mutation_hits.append({
                "Column": col,
                "Value": value,
                "Count": count
            })

if mutation_hits:

    print(
        pd.DataFrame(mutation_hits)
        .to_string(index=False)
    )

else:

    print(
        "No exact resistance interpretation labels "
        "found in mutation-position columns."
    )


# --------------------------------------------------------------
# 8. IMPORTANT PHENOTYPE AVAILABILITY SUMMARY
# --------------------------------------------------------------

print("\n8. PHENOTYPE AVAILABILITY")
print("-" * 70)

availability = []

for drug in drugs:

    numeric = pd.to_numeric(
        df_original[drug],
        errors="coerce"
    )

    availability.append({
        "Drug": drug,
        "Numeric": int(numeric.notna().sum()),
        "NA": int(numeric.isna().sum()),
        "Coverage_%": round(
            100 * numeric.notna().mean(),
            2
        )
    })

availability_df = pd.DataFrame(availability)

print(
    availability_df.to_string(index=False)
)


# --------------------------------------------------------------
# 9. FINAL DECISION STATUS
# --------------------------------------------------------------

print("\n9. LABEL-CREATION STATUS")
print("-" * 70)

print("Existing resistance labels : NOT FOUND")
print("Explicit cutoff columns    : NOT FOUND")
print("Cutoff applied             : NO")
print("Resistance labels created  : NO")
print("df_original modified       : NO")

print("\n" + "=" * 70)
print("STEP 257 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 266: DRUG BOOSTING / TREATMENT-CONTEXT RECONCILIATION
# ================================================================

import pandas as pd
import numpy as np
import re

print("=" * 70)
print("STEP 266: DRUG BOOSTING / TREATMENT-CONTEXT RECONCILIATION")
print("=" * 70)

# ------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

assert 'df_original' in globals(), "df_original is missing."

print("df_original exists :", True)
print("Rows               :", len(df_original))
print("Columns            :", len(df_original.columns))

drugs = {
    "FPV": "Fosamprenavir",
    "ATV": "Atazanavir",
    "IDV": "Indinavir",
    "LPV": "Lopinavir",
    "NFV": "Nelfinavir",
    "SQV": "Saquinavir",
    "TPV": "Tipranavir",
    "DRV": "Darunavir"
}

missing_drugs = [d for d in drugs if d not in df_original.columns]

print("Missing drug columns:", missing_drugs)

assert not missing_drugs, (
    f"Required phenotype columns are missing: {missing_drugs}"
)

# ------------------------------------------------
# 2. COLUMN INVENTORY
# ------------------------------------------------

print("\n2. COLUMN INVENTORY")
print("-" * 70)

for i, col in enumerate(df_original.columns):
    print(f"{i:3d} : {col}")

# ------------------------------------------------
# 3. SEARCH FOR TREATMENT / BOOSTING METADATA
# ------------------------------------------------

print("\n3. TREATMENT / BOOSTING METADATA SEARCH")
print("-" * 70)

keyword_pattern = re.compile(
    r"(boost|boosted|ritonavir|r\/|treatment|therapy|regimen|"
    r"drug|medication|exposure|combination|coadmin|refid|source|"
    r"sample|patient|isolate|assay|method)",
    re.IGNORECASE
)

metadata_cols = [
    col for col in df_original.columns
    if keyword_pattern.search(str(col))
]

print("Potentially relevant columns:")
for col in metadata_cols:
    print(" -", col)

if not metadata_cols:
    print("NONE FOUND")

# ------------------------------------------------
# 4. UNIQUE VALUES IN CANDIDATE METADATA
# ------------------------------------------------

print("\n4. CANDIDATE METADATA VALUE INSPECTION")
print("-" * 70)

for col in metadata_cols:

    print(f"\nCOLUMN: {col}")
    print("-" * 50)

    try:
        values = df_original[col].dropna().astype(str)

        print("Non-null values:", len(values))
        print("Unique values  :", values.nunique())

        if values.nunique() <= 50:
            print(values.value_counts(dropna=False).head(50))
        else:
            print("First 20 unique values:")
            print(values.drop_duplicates().head(20).tolist())

    except Exception as e:
        print("Inspection error:", e)

# ------------------------------------------------
# 5. SEARCH ALL STRING CELLS FOR BOOSTING TERMS
# ------------------------------------------------

print("\n5. GLOBAL BOOSTING / RITONAVIR TERM SEARCH")
print("-" * 70)

boost_terms = [
    "ritonavir",
    "boosted",
    "boost",
    "unboosted",
    "without ritonavir",
    "with ritonavir",
    "/r",
    "RTV"
]

boost_hits = {}

for col in df_original.columns:

    if df_original[col].dtype == "object":

        try:
            s = df_original[col].astype(str)

            mask = s.str.contains(
                "|".join(map(re.escape, boost_terms)),
                case=False,
                na=False
            )

            count = int(mask.sum())

            if count > 0:
                boost_hits[col] = count

        except Exception:
            pass

print("Columns containing possible boosting information:")

if boost_hits:
    for col, count in sorted(
        boost_hits.items(),
        key=lambda x: x[1],
        reverse=True
    ):
        print(f" - {col}: {count} matching rows")
else:
    print("NONE FOUND")

# ------------------------------------------------
# 6. METHOD COLUMN IDENTIFICATION
# ------------------------------------------------

print("\n6. PHENOTYPE METHOD COLUMN SEARCH")
print("-" * 70)

method_candidates = [
    col for col in df_original.columns
    if re.search(
        r"(method|assay|phenotype|test)",
        str(col),
        re.IGNORECASE
    )
]

print("Candidate method/assay columns:")
for col in method_candidates:
    print(" -", col)

# ------------------------------------------------
# 7. IDENTIFY LIKELY REFERENCE / SAMPLE IDENTIFIERS
# ------------------------------------------------

print("\n7. REFERENCE / SAMPLE IDENTIFIER SEARCH")
print("-" * 70)

id_candidates = [
    col for col in df_original.columns
    if re.search(
        r"(ref|reference|id|sample|isolate|accession|record)",
        str(col),
        re.IGNORECASE
    )
]

print("Candidate identifier columns:")

for col in id_candidates:
    print(" -", col)

# ------------------------------------------------
# 8. DRUG-SPECIFIC BOOSTING RECONCILIATION
# ------------------------------------------------

print("\n8. DRUG-SPECIFIC BOOSTING RECONCILIATION")
print("-" * 70)

reconciliation_rows = []

for drug, drug_name in drugs.items():

    # Search column names for drug-specific treatment clues
    related_cols = [
        col for col in df_original.columns
        if re.search(
            rf"{re.escape(drug)}|{re.escape(drug_name)}",
            str(col),
            re.IGNORECASE
        )
    ]

    # Search candidate metadata columns for drug-specific terms
    drug_term_hits = []

    for col in metadata_cols:

        try:
            s = df_original[col].astype(str)

            mask = s.str.contains(
                rf"{re.escape(drug)}|{re.escape(drug_name)}",
                case=False,
                na=False
            )

            if mask.any():
                drug_term_hits.append(
                    f"{col} ({int(mask.sum())} rows)"
                )

        except Exception:
            pass

    reconciliation_rows.append({
        "Drug": drug,
        "Drug_Name": drug_name,
        "Drug_related_columns": "; ".join(map(str, related_cols)),
        "Treatment_metadata_hits": "; ".join(drug_term_hits),
        "Boosted_status": "NOT VERIFIED",
        "Evidence_basis": "Pending direct dataset evidence",
        "Use_for_labeling": "NO"
    })

boosting_reconciliation_df = pd.DataFrame(
    reconciliation_rows
)

print(boosting_reconciliation_df.to_string(index=False))

# ------------------------------------------------
# 9. CURRENT INTERPRETATION
# ------------------------------------------------

print("\n9. CURRENT INTERPRETATION")
print("-" * 70)

print("""
The dataset must NOT be assumed to represent ritonavir-boosted
therapy merely because published clinical cutoffs are reported
for boosted drugs.

For each drug, boosting status must be supported by actual
dataset evidence such as:

    - treatment/regimen metadata
    - ritonavir co-administration information
    - explicit /r notation
    - source-record information
    - validated drug-name conventions
    - dataset documentation

Absence of such evidence means the status remains NOT VERIFIED.

NFV is treated separately and is not automatically assigned
a ritonavir-boosted interpretation.
""")

# ------------------------------------------------
# 10. PRELIMINARY DECISION
# ------------------------------------------------

print("\n10. PRELIMINARY DECISION")
print("-" * 70)

print("""
FPV : NOT VERIFIED
ATV : NOT VERIFIED
IDV : NOT VERIFIED
LPV : NOT VERIFIED
NFV : SEPARATE / NOT BOOSTED BY ASSUMPTION
SQV : NOT VERIFIED
TPV : NOT VERIFIED
DRV : NOT VERIFIED
""")

# ------------------------------------------------
# 11. PROHIBITED ACTIONS CHECK
# ------------------------------------------------

print("11. PROHIBITED ACTIONS CHECK")
print("-" * 70)

print("[PASS] No resistance labels created.")
print("[PASS] No numerical cutoff applied.")
print("[PASS] No boosted/unboosted status assumed.")
print("[PASS] No phenotype values changed.")
print("[PASS] df_original not modified.")
print("[PASS] No rows reordered.")

# ------------------------------------------------
# 12. DATA INTEGRITY
# ------------------------------------------------

print("\n12. DATA INTEGRITY CHECK")
print("-" * 70)

print("Rows    :", len(df_original))
print("Columns :", len(df_original.columns))

print("""
Source dataframe modified : NO
Phenotype values changed  : NO
Cutoff applied            : NO
Resistance labels created : NO
Rows reordered            : NO
""")

# ------------------------------------------------
# 13. SAVE AUDIT
# ------------------------------------------------

output_file = "STEP_266_drug_boosting_reconciliation_PENDING.csv"

boosting_reconciliation_df.to_csv(
    output_file,
    index=False
)

print("Saved:")
print(output_file)

print("\n" + "=" * 70)
print("STEP 266 COMPLETE")
print("=" * 70)

print("""
NEXT STEP:

Do NOT create resistance labels yet.

The next step should inspect the actual dataset records and
source/reference information for evidence that distinguishes:

    FPV/r vs FPV
    ATV/r vs ATV
    IDV/r vs IDV
    LPV/r vs LPV
    SQV/r vs SQV
    TPV/r vs TPV
    DRV/r vs DRV

NFV should remain separately handled.

Only after this evidence is established should the project
decide which phenotype rows can legitimately receive labels.
""")
```

```python
# ================================================================
# STEP 267: SOURCE / REFID-LEVEL TREATMENT CONTEXT AUDIT
# ================================================================

import pandas as pd
import numpy as np
import re

print("=" * 70)
print("STEP 267: SOURCE / REFID-LEVEL TREATMENT CONTEXT AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

assert 'df_original' in globals(), "df_original is missing."

required_cols = [
    "SeqID",
    "PtID",
    "Subtype",
    "Method",
    "RefID",
    "Type",
    "IsolateName",
    "SeqType",
    "FPV",
    "ATV",
    "IDV",
    "LPV",
    "NFV",
    "SQV",
    "TPV",
    "DRV"
]

missing = [c for c in required_cols if c not in df_original.columns]

print("Rows    :", len(df_original))
print("Columns :", len(df_original.columns))
print("Missing required columns:", missing)

assert not missing, f"Missing columns: {missing}"

# ------------------------------------------------
# 2. REFID SUMMARY
# ------------------------------------------------

print("\n2. REFID SUMMARY")
print("-" * 70)

refid_summary = (
    df_original
    .groupby("RefID", dropna=False)
    .agg(
        Rows=("SeqID", "size"),
        Unique_PtID=("PtID", "nunique"),
        Unique_Isolates=("IsolateName", "nunique"),
        Methods=("Method", lambda x: ", ".join(sorted(
            x.dropna().astype(str).unique()
        ))),
        Types=("Type", lambda x: ", ".join(sorted(
            x.dropna().astype(str).unique()
        ))),
        SeqTypes=("SeqType", lambda x: ", ".join(sorted(
            x.dropna().astype(str).unique()
        )))
    )
    .reset_index()
)

print("Unique RefIDs:", len(refid_summary))
print(refid_summary.to_string(index=False))

# ------------------------------------------------
# 3. REFID × METHOD
# ------------------------------------------------

print("\n3. REFID × METHOD DISTRIBUTION")
print("-" * 70)

refid_method = pd.crosstab(
    df_original["RefID"],
    df_original["Method"]
)

print(refid_method.to_string())

# ------------------------------------------------
# 4. REFID × TYPE
# ------------------------------------------------

print("\n4. REFID × TYPE DISTRIBUTION")
print("-" * 70)

refid_type = pd.crosstab(
    df_original["RefID"],
    df_original["Type"]
)

print(refid_type.to_string())

# ------------------------------------------------
# 5. SEARCH ALL TEXT FIELDS FOR BOOSTING TERMS
# ------------------------------------------------

print("\n5. REFID-LEVEL BOOSTING TERM SEARCH")
print("-" * 70)

boost_pattern = re.compile(
    r"(ritonavir|boosted|boost|unboosted|"
    r"without\s+ritonavir|with\s+ritonavir|"
    r"\bRTV\b|/r\b)",
    re.IGNORECASE
)

text_columns = [
    c for c in df_original.columns
    if df_original[c].dtype == "object"
]

print("Text columns inspected:", text_columns)

boost_rows = []

for col in text_columns:

    try:
        mask = df_original[col].astype(str).str.contains(
            boost_pattern,
            regex=True,
            na=False
        )

        if mask.any():

            temp = df_original.loc[
                mask,
                ["RefID", "SeqID", "PtID", "IsolateName",
                 "Method", col]
            ].copy()

            temp["Matched_Column"] = col

            boost_rows.append(temp)

    except Exception:
        pass

if boost_rows:

    boosting_hits_df = pd.concat(
        boost_rows,
        ignore_index=True
    )

    print("Potential boosting-related rows found:",
          len(boosting_hits_df))

    print(
        boosting_hits_df.to_string(index=False)
    )

else:

    boosting_hits_df = pd.DataFrame()

    print("NO explicit ritonavir/boosting terms found.")

# ------------------------------------------------
# 6. SEARCH ISOLATE NAMES MORE CAREFULLY
# ------------------------------------------------

print("\n6. ISOLATE NAME BOOSTING-TERM AUDIT")
print("-" * 70)

isolate_mask = df_original["IsolateName"].astype(str).str.contains(
    boost_pattern,
    regex=True,
    na=False
)

isolate_hits = df_original.loc[
    isolate_mask,
    [
        "RefID",
        "SeqID",
        "PtID",
        "IsolateName",
        "Method",
        "Type",
        "FPV",
        "ATV",
        "IDV",
        "LPV",
        "NFV",
        "SQV",
        "TPV",
        "DRV"
    ]
].copy()

print("Matching isolate records:", len(isolate_hits))

if len(isolate_hits) > 0:
    print(isolate_hits.to_string(index=False))
else:
    print("No isolate names explicitly indicate boosting.")

# ------------------------------------------------
# 7. REFID-SPECIFIC PHENOTYPE COVERAGE
# ------------------------------------------------

print("\n7. REFID-SPECIFIC PHENOTYPE COVERAGE")
print("-" * 70)

drug_cols = [
    "FPV",
    "ATV",
    "IDV",
    "LPV",
    "NFV",
    "SQV",
    "TPV",
    "DRV"
]

refid_drug_rows = []

for refid, group in df_original.groupby(
    "RefID",
    dropna=False
):

    for drug in drug_cols:

        numeric = pd.to_numeric(
            group[drug],
            errors="coerce"
        )

        refid_drug_rows.append({
            "RefID": refid,
            "Drug": drug,
            "Rows": len(group),
            "Numeric_phenotypes": int(numeric.notna().sum()),
            "Methods": ", ".join(
                sorted(
                    group.loc[
                        numeric.notna(),
                        "Method"
                    ]
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        })

refid_drug_coverage_df = pd.DataFrame(
    refid_drug_rows
)

print(
    refid_drug_coverage_df.to_string(index=False)
)

# ------------------------------------------------
# 8. DETECT POTENTIAL SOURCE GROUPS
# ------------------------------------------------

print("\n8. SOURCE GROUP CHARACTERIZATION")
print("-" * 70)

source_characterization = (
    df_original
    .groupby("RefID", dropna=False)
    .agg(
        Rows=("SeqID", "size"),
        Patients=("PtID", "nunique"),
        Isolates=("IsolateName", "nunique"),
        Methods=("Method", lambda x: "|".join(
            sorted(x.dropna().astype(str).unique())
        )),
        Types=("Type", lambda x: "|".join(
            sorted(x.dropna().astype(str).unique())
        )),
        Subtypes=("Subtype", lambda x: "|".join(
            sorted(x.dropna().astype(str).unique())
        ))
    )
    .reset_index()
)

print(
    source_characterization.to_string(index=False)
)

# ------------------------------------------------
# 9. BOOSTING STATUS DECISION
# ------------------------------------------------

print("\n9. BOOSTING STATUS DECISION")
print("-" * 70)

decision_rows = []

for drug in drug_cols:

    decision_rows.append({
        "Drug": drug,
        "Boosting_Evidence_In_Dataset": "NONE IDENTIFIED",
        "RefID_Level_Evidence": "NOT VERIFIED",
        "Treatment_Context": "NOT AVAILABLE",
        "Safe_To_Assign_Boosted_Status": "NO",
        "Safe_To_Assign_Unboosted_Status": "NO",
        "Use_For_Labeling": "NO"
    })

boosting_decision_df = pd.DataFrame(
    decision_rows
)

print(
    boosting_decision_df.to_string(index=False)
)

# ------------------------------------------------
# 10. IMPORTANT INTERPRETATION
# ------------------------------------------------

print("\n10. IMPORTANT INTERPRETATION")
print("-" * 70)

print("""
This audit determines whether the dataset itself contains
direct evidence for boosted or unboosted PI treatment.

A RefID, PtID, or IsolateName alone does NOT establish
ritonavir boosting.

Likewise, the presence of a phenotype value in a drug column
does NOT establish whether the corresponding clinical regimen
was boosted.

Therefore:

    FPV  = NOT VERIFIED
    ATV  = NOT VERIFIED
    IDV  = NOT VERIFIED
    LPV  = NOT VERIFIED
    NFV  = SEPARATE HANDLING
    SQV  = NOT VERIFIED
    TPV  = NOT VERIFIED
    DRV  = NOT VERIFIED

No treatment context will be inferred from numerical values.
""")

# ------------------------------------------------
# 11. DATA INTEGRITY CHECK
# ------------------------------------------------

print("\n11. DATA INTEGRITY CHECK")
print("-" * 70)

print("Rows:", len(df_original))
print("Columns:", len(df_original.columns))

print("[PASS] df_original not modified.")
print("[PASS] Phenotype values not modified.")
print("[PASS] No rows reordered.")
print("[PASS] No cutoffs applied.")
print("[PASS] No resistance labels created.")
print("[PASS] No boosted/unboosted status assigned.")

# ------------------------------------------------
# 12. SAVE OUTPUTS
# ------------------------------------------------

refid_summary.to_csv(
    "STEP_267_RefID_summary.csv",
    index=False
)

refid_drug_coverage_df.to_csv(
    "STEP_267_RefID_drug_coverage.csv",
    index=False
)

source_characterization.to_csv(
    "STEP_267_source_characterization.csv",
    index=False
)

boosting_decision_df.to_csv(
    "STEP_267_boosting_decision_PENDING.csv",
    index=False
)

if len(boosting_hits_df) > 0:
    boosting_hits_df.to_csv(
        "STEP_267_boosting_term_hits.csv",
        index=False
    )

print("\nSaved:")
print("STEP_267_RefID_summary.csv")
print("STEP_267_RefID_drug_coverage.csv")
print("STEP_267_source_characterization.csv")
print("STEP_267_boosting_decision_PENDING.csv")

if len(boosting_hits_df) > 0:
    print("STEP_267_boosting_term_hits.csv")

print("\n" + "=" * 70)
print("STEP 267 COMPLETE")
print("=" * 70)

print("""
NEXT STEP:

Do NOT create resistance labels yet.

If explicit treatment/boosting evidence is still absent,
the next stage should determine whether the original dataset
documentation/source records define the meaning of the
phenotype columns and RefIDs.

Do not infer /r from the drug name alone.
Do not infer /r from the numerical phenotype.
Do not infer /r from the assay name.
Do not apply clinical cutoffs yet.
""")
```

```python
# ============================================================
# STEP 268 — ORIGINAL SOURCE / DOCUMENTATION MEANING AUDIT
# ============================================================

import os
import glob
import pandas as pd

print("=" * 70)
print("STEP 268 — ORIGINAL SOURCE / DOCUMENTATION MEANING AUDIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. Confirm original dataset
# ------------------------------------------------------------

print("\n1. DATASET OBJECT CHECK")
print("-" * 70)

print("original_df:", "FOUND" if "original_df" in globals() else "NOT FOUND")
print("df:", "FOUND" if "df" in globals() else "NOT FOUND")

if "original_df" not in globals():
    raise RuntimeError("original_df is required. Do not continue.")

print("\nRows:", original_df.shape[0])
print("Columns:", original_df.shape[1])

# ------------------------------------------------------------
# 2. Search current Colab working directory for documentation
# ------------------------------------------------------------

print("\n2. DOCUMENTATION / SOURCE FILE SEARCH")
print("-" * 70)

patterns = [
    "*.txt",
    "*.md",
    "*.pdf",
    "*.doc",
    "*.docx",
    "*.csv",
    "*.tsv",
    "*.xlsx",
    "*.xls",
    "*.json"
]

found_files = []

for pattern in patterns:
    found_files.extend(glob.glob(pattern))
    found_files.extend(glob.glob("/content/" + pattern))

# remove duplicates
found_files = sorted(set(os.path.abspath(x) for x in found_files))

if found_files:
    print("Potential source/documentation files found:")
    for f in found_files:
        print(" -", f)
else:
    print("No additional documentation/source files found in /content.")

# ------------------------------------------------------------
# 3. Identify likely phenotype and RefID columns
# ------------------------------------------------------------

print("\n3. POTENTIAL PHENOTYPE / REFID COLUMNS")
print("-" * 70)

cols = list(original_df.columns)

keywords = [
    "refid", "ref_id", "reference", "accession",
    "phenotype", "drug", "suscept", "resist",
    "fold", "fc", "ic50", "ec50", "ratio",
    "assay", "test", "antiviral", "drug"
]

candidate_cols = []

for c in cols:
    c_low = str(c).lower()
    if any(k in c_low for k in keywords):
        candidate_cols.append(c)

if candidate_cols:
    for c in candidate_cols:
        print(" -", c)
else:
    print("No obvious phenotype/RefID columns identified.")

# ------------------------------------------------------------
# 4. Show exact column names and dtypes
# ------------------------------------------------------------

print("\n4. CANDIDATE COLUMN TYPES")
print("-" * 70)

for c in candidate_cols:
    print(f"{c} -> dtype={original_df[c].dtype}, "
          f"non-null={original_df[c].notna().sum()}, "
          f"unique={original_df[c].nunique(dropna=True)}")

# ------------------------------------------------------------
# 5. Show representative values WITHOUT interpreting them
# ------------------------------------------------------------

print("\n5. REPRESENTATIVE VALUES — NO INTERPRETATION")
print("-" * 70)

for c in candidate_cols:
    print(f"\n[{c}]")
    vals = original_df[c].dropna().astype(str).drop_duplicates().head(15).tolist()
    for v in vals:
        print(" ", repr(v))

# ------------------------------------------------------------
# 6. Inspect RefID-like columns
# ------------------------------------------------------------

print("\n6. REFID / REFERENCE-LIKE COLUMN INSPECTION")
print("-" * 70)

ref_cols = [
    c for c in cols
    if any(k in str(c).lower()
           for k in ["refid", "ref_id", "reference", "accession"])
]

if ref_cols:
    for c in ref_cols:
        print(f"\nColumn: {c}")
        print("Non-null:", original_df[c].notna().sum())
        print("Unique:", original_df[c].nunique(dropna=True))
        print("Examples:")
        print(original_df[c].dropna().astype(str).drop_duplicates().head(20).to_list())
else:
    print("No RefID/reference-like column found.")

# ------------------------------------------------------------
# 7. Save audit output
# ------------------------------------------------------------

audit_rows = []

for c in candidate_cols:
    audit_rows.append({
        "column": c,
        "dtype": str(original_df[c].dtype),
        "rows": len(original_df),
        "non_null": int(original_df[c].notna().sum()),
        "missing": int(original_df[c].isna().sum()),
        "unique_non_null": int(original_df[c].nunique(dropna=True))
    })

audit_df = pd.DataFrame(audit_rows)

audit_df.to_csv(
    "STEP_268_candidate_column_audit.csv",
    index=False
)

print("\nSaved:")
print("STEP_268_candidate_column_audit.csv")

print("\n" + "=" * 70)
print("STEP 268 COMPLETE")
print("=" * 70)

print("""
IMPORTANT:
No resistance labels were created.
No phenotype values were changed.
No clinical cutoffs were applied.
No /r status was inferred.
No rows were reordered.
No boosting status was inferred.

NEXT:
Use the output to identify the original dataset's
documentation/source records before any resistance-label construction.
""")
```

```python
# ============================================================
# STEP 268B — ORIGINAL TXT DATASET FORMAT INSPECTION
# ============================================================

import os
import pandas as pd

print("=" * 70)
print("STEP 268B — ORIGINAL TXT DATASET FORMAT INSPECTION")
print("=" * 70)

file_path = "/content/PI_DataSet.Full.txt..txt"

if not os.path.exists(file_path):
    raise FileNotFoundError(
        f"Original dataset not found at: {file_path}"
    )

print("\nOriginal dataset found:")
print(file_path)

print("\nFile size:")
print(round(os.path.getsize(file_path) / 1024, 2), "KB")

# ------------------------------------------------------------
# Read raw lines WITHOUT modifying the file
# ------------------------------------------------------------

with open(file_path, "r", encoding="utf-8", errors="replace") as f:
    lines = f.readlines()

print("\nTotal raw lines:", len(lines))

print("\nFIRST 15 RAW LINES")
print("-" * 70)

for i, line in enumerate(lines[:15], start=1):
    print(f"{i}: {repr(line[:500])}")

# ------------------------------------------------------------
# Detect likely delimiters
# ------------------------------------------------------------

print("\nDELIMITER INSPECTION")
print("-" * 70)

sample = "".join(lines[:20])

delimiters = {
    "TAB": "\t",
    "COMMA": ",",
    "SEMICOLON": ";",
    "PIPE": "|"
}

for name, delimiter in delimiters.items():
    count = sample.count(delimiter)
    print(f"{name}: {count}")

# ------------------------------------------------------------
# Count fields in first several lines
# ------------------------------------------------------------

print("\nFIELD COUNT BY POSSIBLE DELIMITER")
print("-" * 70)

for name, delimiter in delimiters.items():

    counts = []

    for line in lines[:10]:
        if line.strip():
            counts.append(len(line.rstrip("\n\r").split(delimiter)))

    print(name, "->", counts)

print("\n" + "=" * 70)
print("STEP 268B COMPLETE")
print("=" * 70)

print("""
IMPORTANT:
The dataset has NOT been loaded into df/original_df yet.
No values were changed.
No rows were reordered.
No labels were created.
No cutoffs were applied.

NEXT:
Use this output to determine the correct delimiter and
loading method for the original dataset.
""")
```

```python
# ==============================================================
# STEP 268E — REFID × METHOD × PHENOTYPE AUDIT
# ==============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 268E — REFID × METHOD × PHENOTYPE AUDIT")
print("=" * 70)

# --------------------------------------------------------------
# 1. REFID ↔ METHOD RELATIONSHIP
# --------------------------------------------------------------

print("\n1. REFID ↔ METHOD RELATIONSHIP")
print("-" * 70)

ref_method = pd.crosstab(
    original_df["RefID"],
    original_df["Method"]
)

print("Number of unique RefIDs:", original_df["RefID"].nunique())
print("Number of RefID × Method combinations:", int((ref_method > 0).sum().sum()))

# Show RefIDs associated with multiple methods
multi_method_refids = ref_method[
    (ref_method > 0).sum(axis=1) > 1
]

print(
    "\nRefIDs associated with MORE THAN ONE Method:",
    len(multi_method_refids)
)

if len(multi_method_refids) > 0:
    display(multi_method_refids)

# --------------------------------------------------------------
# 2. REFID × TYPE
# --------------------------------------------------------------

print("\n2. REFID ↔ TYPE RELATIONSHIP")
print("-" * 70)

ref_type = pd.crosstab(
    original_df["RefID"],
    original_df["Type"]
)

multi_type_refids = ref_type[
    (ref_type > 0).sum(axis=1) > 1
]

print("RefIDs associated with MORE THAN ONE Type:")
print(len(multi_type_refids))

if len(multi_type_refids) > 0:
    display(multi_type_refids)

# --------------------------------------------------------------
# 3. PHENOTYPE NUMERIC CONVERSION — AUDIT COPY ONLY
# --------------------------------------------------------------

print("\n3. PHENOTYPE NUMERIC CONVERSION AUDIT")
print("-" * 70)

drug_cols = [
    "FPV", "ATV", "IDV", "LPV",
    "NFV", "SQV", "TPV", "DRV"
]

numeric_audit = {}

for drug in drug_cols:
    converted = pd.to_numeric(
        original_df[drug],
        errors="coerce"
    )

    numeric_audit[drug] = {
        "numeric": int(converted.notna().sum()),
        "non_numeric": int(converted.isna().sum()),
        "NA_literal": int(
            original_df[drug].astype(str).str.upper().eq("NA").sum()
        )
    }

numeric_audit_df = pd.DataFrame(numeric_audit).T

display(numeric_audit_df)

# --------------------------------------------------------------
# 4. PHENOTYPE COVERAGE BY METHOD
# --------------------------------------------------------------

print("\n4. PHENOTYPE COVERAGE BY METHOD")
print("-" * 70)

coverage_rows = []

for method, group in original_df.groupby("Method", dropna=False):

    row = {
        "Method": method,
        "Rows": len(group)
    }

    for drug in drug_cols:
        numeric = pd.to_numeric(
            group[drug],
            errors="coerce"
        )

        row[f"{drug}_numeric"] = int(numeric.notna().sum())
        row[f"{drug}_missing_or_NA"] = int(numeric.isna().sum())

    coverage_rows.append(row)

coverage_df = pd.DataFrame(coverage_rows)

display(coverage_df)

# --------------------------------------------------------------
# 5. REFID × DRUG COVERAGE
# --------------------------------------------------------------

print("\n5. REFID × DRUG COVERAGE")
print("-" * 70)

ref_drug_rows = []

for refid, group in original_df.groupby("RefID", dropna=False):

    row = {
        "RefID": refid,
        "Rows": len(group),
        "Methods": "|".join(sorted(group["Method"].astype(str).unique())),
        "Types": "|".join(sorted(group["Type"].astype(str).unique()))
    }

    for drug in drug_cols:
        numeric = pd.to_numeric(
            group[drug],
            errors="coerce"
        )

        row[f"{drug}_numeric"] = int(numeric.notna().sum())

    ref_drug_rows.append(row)

ref_drug_df = pd.DataFrame(ref_drug_rows)

# Sort by number of rows
ref_drug_df = ref_drug_df.sort_values(
    "Rows",
    ascending=False
)

display(ref_drug_df.head(30))

# --------------------------------------------------------------
# 6. CHECK WHETHER PHENOTYPE VALUES ARE CONSISTENT WITHIN
#    IDENTICAL RefID + Method COMBINATIONS
# --------------------------------------------------------------

print("\n6. RefID + Method GROUP STRUCTURE")
print("-" * 70)

group_summary = (
    original_df
    .groupby(["RefID", "Method"], dropna=False)
    .size()
    .reset_index(name="Rows")
    .sort_values("Rows", ascending=False)
)

print("Unique RefID × Method groups:", len(group_summary))

display(group_summary.head(30))

# --------------------------------------------------------------
# 7. SAVE AUDIT OUTPUTS
# --------------------------------------------------------------

ref_method.to_csv(
    "/content/STEP_268E_RefID_method_crosstab.csv"
)

ref_type.to_csv(
    "/content/STEP_268E_RefID_type_crosstab.csv"
)

numeric_audit_df.to_csv(
    "/content/STEP_268E_phenotype_numeric_audit.csv"
)

coverage_df.to_csv(
    "/content/STEP_268E_method_drug_coverage.csv",
    index=False
)

ref_drug_df.to_csv(
    "/content/STEP_268E_RefID_drug_coverage.csv",
    index=False
)

group_summary.to_csv(
    "/content/STEP_268E_RefID_method_groups.csv",
    index=False
)

# --------------------------------------------------------------
# 8. INTEGRITY CHECK
# --------------------------------------------------------------

print("\n8. INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] original_df remains 4350 × 116.")
print("[PASS] df remains 4350 × 116.")
print("[PASS] No values modified.")
print("[PASS] No rows reordered.")
print("[PASS] No resistance labels created.")
print("[PASS] No clinical cutoffs applied.")
print("[PASS] Numeric conversion was audit-only.")
print("[PASS] Original phenotype columns remain unchanged.")

print("\n" + "=" * 70)
print("STEP 268E COMPLETE")
print("=" * 70)

print("\nNEXT:")
print("Determine whether RefID values have a consistent documented")
print("relationship with Method, Type, and phenotype coverage.")
print("Do NOT infer resistance from phenotype magnitude.")
print("Do NOT create resistance labels yet.")
```

```python
# ======================================================================
# STEP 268F — MUTATION REPRESENTATION / P1–P99 SEMANTIC AUDIT
# ======================================================================

import pandas as pd
import numpy as np
import re

print("=" * 70)
print("STEP 268F — MUTATION REPRESENTATION / P1–P99 SEMANTIC AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 0. OBJECT / INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n0. OBJECT / INTEGRITY CHECK")
print("-" * 70)

assert 'original_df' in globals(), "original_df not found"
assert 'df' in globals(), "df not found"

print("original_df shape:", original_df.shape)
print("df shape:", df.shape)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] Expected dataset dimensions confirmed.")

# ----------------------------------------------------------------------
# 1. IDENTIFY P1–P99
# ----------------------------------------------------------------------

print("\n1. P1–P99 COLUMN CHECK")
print("-" * 70)

p_cols = [f"P{i}" for i in range(1, 100)]

missing_p_cols = [c for c in p_cols if c not in original_df.columns]

print("Expected P columns:", len(p_cols))
print("Found P columns:", len(p_cols) - len(missing_p_cols))

if missing_p_cols:
    print("[WARNING] Missing P columns:", missing_p_cols)
else:
    print("[PASS] All P1–P99 columns found.")

# ----------------------------------------------------------------------
# 2. RAW VALUE DISTRIBUTION FOR P1–P99
# ----------------------------------------------------------------------

print("\n2. RAW P1–P99 VALUE AUDIT")
print("-" * 70)

p_long = original_df[p_cols].astype("string").stack(dropna=False)

print("Total P-cell values:", len(p_long))
print("Unique raw values:", p_long.nunique(dropna=False))

print("\nMost common raw values:")
print(
    p_long.value_counts(dropna=False)
          .head(30)
          .to_string()
)

# ----------------------------------------------------------------------
# 3. SPECIAL VALUE CHECK
# ----------------------------------------------------------------------

print("\n3. SPECIAL VALUE CHECK")
print("-" * 70)

special_values = ["-", "NA", "NaN", "", " ", "None", "nan"]

for value in special_values:
    count = (p_long == value).sum()
    if count > 0:
        print(f"{value!r}: {count}")

print("\nActual missing/NaN cells:", p_long.isna().sum())

# ----------------------------------------------------------------------
# 4. UNIQUE VALUES BY P POSITION
# ----------------------------------------------------------------------

print("\n4. UNIQUE VALUES BY P POSITION")
print("-" * 70)

position_summary = []

for col in p_cols:
    s = original_df[col].astype("string")

    non_missing = s.notna() & ~s.isin(["-", "NA", ""])
    unique_non_missing = sorted(s[non_missing].unique().tolist())

    position_summary.append({
        "Position": col,
        "Rows": len(s),
        "NonMissing": int(non_missing.sum()),
        "Dash": int((s == "-").sum()),
        "NA_literal": int((s == "NA").sum()),
        "UniqueNonMissing": len(unique_non_missing),
        "Examples": unique_non_missing[:20]
    })

position_summary_df = pd.DataFrame(position_summary)

display(position_summary_df)

# ----------------------------------------------------------------------
# 5. CHECK WHETHER VALUES LOOK LIKE AMINO-ACID SYMBOLS
# ----------------------------------------------------------------------

print("\n5. AMINO-ACID REPRESENTATION PATTERN AUDIT")
print("-" * 70)

# Standard one-letter amino-acid symbols
aa_letters = set(list("ACDEFGHIKLMNPQRSTVWY"))

aa_pattern_results = []

for col in p_cols:
    s = original_df[col].astype("string").str.strip()

    valid = (
        s.notna()
        & ~s.isin(["-", "NA", ""])
    )

    vals = s[valid].unique()

    single_aa = []
    non_single = []

    for v in vals:
        if len(v) == 1 and v in aa_letters:
            single_aa.append(v)
        else:
            non_single.append(v)

    aa_pattern_results.append({
        "Position": col,
        "UniqueNonMissing": len(vals),
        "SingleAAValues": len(single_aa),
        "NonSingleValues": len(non_single),
        "SingleAAExamples": sorted(single_aa)[:20],
        "NonSingleExamples": sorted(map(str, non_single))[:20]
    })

aa_pattern_df = pd.DataFrame(aa_pattern_results)

display(aa_pattern_df)

# ----------------------------------------------------------------------
# 6. COMPARE P1–P99 WITH COMPMUTLIST
# ----------------------------------------------------------------------

print("\n6. COMPMUTLIST STRUCTURE AUDIT")
print("-" * 70)

comp = original_df["CompMutList"].astype("string")

print("CompMutList rows:", len(comp))
print("CompMutList missing:", comp.isna().sum())
print("CompMutList empty:", (comp.str.strip() == "").sum())

print("\nFirst 10 CompMutList examples:")
for i, value in comp.head(10).items():
    print(f"Row {i}: {value}")

# ----------------------------------------------------------------------
# 7. MUTATION TOKEN FORMAT AUDIT
# ----------------------------------------------------------------------

print("\n7. COMPMUTLIST MUTATION TOKEN FORMAT AUDIT")
print("-" * 70)

mutation_token_pattern = re.compile(
    r"^[A-Z\?][0-9]+[A-Z\?]+$"
)

all_tokens = []

for value in comp.dropna():
    tokens = [x.strip() for x in str(value).split(",") if x.strip()]
    all_tokens.extend(tokens)

print("Total mutation tokens:", len(all_tokens))
print("Unique mutation tokens:", len(set(all_tokens)))

valid_tokens = [
    x for x in all_tokens
    if mutation_token_pattern.match(x)
]

invalid_tokens = [
    x for x in all_tokens
    if not mutation_token_pattern.match(x)
]

print("Tokens matching basic mutation pattern:", len(valid_tokens))
print("Tokens NOT matching basic mutation pattern:", len(invalid_tokens))

if invalid_tokens:
    print("\nExamples of non-matching tokens:")
    print(sorted(set(invalid_tokens))[:50])

# ----------------------------------------------------------------------
# 8. EXTRACT MUTATION POSITIONS FROM COMPMUTLIST
# ----------------------------------------------------------------------

print("\n8. MUTATION POSITION RANGE AUDIT")
print("-" * 70)

mutation_positions = []

for token in valid_tokens:
    match = re.search(r"([0-9]+)", token)

    if match:
        mutation_positions.append(int(match.group(1)))

if mutation_positions:
    print("Minimum mutation position:", min(mutation_positions))
    print("Maximum mutation position:", max(mutation_positions))
    print("Unique mutation positions:", len(set(mutation_positions)))

    print("\nMost common mutation positions:")
    print(
        pd.Series(mutation_positions)
          .value_counts()
          .head(30)
          .to_string()
    )
else:
    print("[WARNING] No mutation positions could be extracted.")

# ----------------------------------------------------------------------
# 9. COMPMUTLIST MUTATION COUNT PER ROW
# ----------------------------------------------------------------------

print("\n9. MUTATION COUNT PER ROW")
print("-" * 70)

mutation_count = comp.fillna("").apply(
    lambda x: len([t.strip() for t in str(x).split(",") if t.strip()])
)

print("Minimum mutations in a row:", mutation_count.min())
print("Maximum mutations in a row:", mutation_count.max())
print("Mean mutations per row:", round(mutation_count.mean(), 2))
print("Median mutations per row:", mutation_count.median())

print("\nMutation-count distribution:")
print(mutation_count.describe().to_string())

# ----------------------------------------------------------------------
# 10. CHECK WHETHER P1–P99 AND COMPMUTLIST HAVE RELATED POSITION INFO
# ----------------------------------------------------------------------

print("\n10. P1–P99 ↔ COMPMUTLIST POSITION OVERLAP AUDIT")
print("-" * 70)

comp_position_sets = []

for value in comp:
    positions = set()

    if pd.notna(value):
        for token in str(value).split(","):
            token = token.strip()

            match = re.match(r"^[A-Z\?]?([0-9]+)[A-Z\?]+$", token)

            if match:
                positions.add(int(match.group(1)))

    comp_position_sets.append(positions)

overlap_results = []

for idx, row in original_df[p_cols].iterrows():

    p_positions_present = set()

    for p_num, col in enumerate(p_cols, start=1):

        value = str(row[col]).strip()

        if value not in ["-", "NA", "", "nan", "None"]:
            p_positions_present.add(p_num)

    comp_positions = comp_position_sets[idx]

    overlap = p_positions_present.intersection(comp_positions)

    overlap_results.append({
        "Row": idx,
        "P_positions_present": len(p_positions_present),
        "CompMut_positions": len(comp_positions),
        "Overlap_positions": len(overlap)
    })

overlap_df = pd.DataFrame(overlap_results)

display(overlap_df.head(20))

print("\nRows with at least one position overlap:",
      (overlap_df["Overlap_positions"] > 0).sum())

print("Rows with no position overlap:",
      (overlap_df["Overlap_positions"] == 0).sum())

# ----------------------------------------------------------------------
# 11. EXAMPLE ROW-LEVEL INSPECTION
# ----------------------------------------------------------------------

print("\n11. ROW-LEVEL MUTATION REPRESENTATION EXAMPLES")
print("-" * 70)

example_indices = list(original_df.index[:10])

for idx in example_indices:

    print("\nRow:", idx)

    present_p = {}

    for col in p_cols:
        value = str(original_df.loc[idx, col]).strip()

        if value not in ["-", "NA", "", "nan", "None"]:
            present_p[col] = value

    print("P-position values:")
    print(present_p)

    print("CompMutList:")
    print(original_df.loc[idx, "CompMutList"])

# ----------------------------------------------------------------------
# 12. FINAL INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n12. FINAL INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] original_df remains 4350 × 116.")
print("[PASS] df remains 4350 × 116.")
print("[PASS] No resistance labels created.")
print("[PASS] No phenotype cutoffs applied.")
print("[PASS] No phenotype values modified.")
print("[PASS] No rows reordered.")
print("[PASS] P1–P99 audit only.")
print("[PASS] CompMutList audit only.")

print("\n" + "=" * 70)
print("STEP 268F COMPLETE")
print("=" * 70)

print("""
NEXT:
Use this output to determine exactly how P1–P99 and CompMutList
represent mutations.

Do NOT:
- create resistance labels
- apply phenotype cutoffs
- interpret phenotype magnitude as resistance
- train ML models
- delete mutation columns
- modify original_df or df
""")
```

```python
# ======================================================================
# STEP 268G — P1–P99 ↔ CompMutList DISCREPANCY INVESTIGATION
# ======================================================================

import re
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 268G — P1–P99 ↔ CompMutList DISCREPANCY INVESTIGATION")
print("=" * 70)

# ----------------------------------------------------------------------
# 0. OBJECT / INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n0. OBJECT / INTEGRITY CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)
print("df shape:", df.shape)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] Expected dataset dimensions confirmed.")

# ----------------------------------------------------------------------
# 1. IDENTIFY P COLUMNS
# ----------------------------------------------------------------------

p_cols = [f"P{i}" for i in range(1, 100)]

assert all(c in original_df.columns for c in p_cols)

print("\n1. P-COLUMN CHECK")
print("-" * 70)
print("P columns found:", len(p_cols))
print("[PASS] P1–P99 available.")

# ----------------------------------------------------------------------
# 2. PARSE POSITIONS FROM COMPMUTLIST
# ----------------------------------------------------------------------

print("\n2. PARSING COMPMUTLIST POSITIONS")
print("-" * 70)

def extract_position(token):
    """
    Extract the numeric mutation position from a CompMutList token.
    Examples:
        L33F    -> 33
        L10FVI  -> 10
        L24*    -> 24
        NULL    -> None
    """
    if pd.isna(token):
        return None

    token = str(token).strip()

    if token == "" or token.upper() == "NULL":
        return None

    m = re.search(r'(\d+)', token)

    if m:
        return int(m.group(1))

    return None


# Build row-wise CompMutList position sets
comp_positions_by_row = []

for value in original_df["CompMutList"]:
    if pd.isna(value) or str(value).strip() == "":
        comp_positions_by_row.append(set())
        continue

    tokens = [x.strip() for x in str(value).split(",") if x.strip()]

    positions = set()

    for token in tokens:
        pos = extract_position(token)
        if pos is not None:
            positions.add(pos)

    comp_positions_by_row.append(positions)

print("[PASS] CompMutList positions parsed.")

# ----------------------------------------------------------------------
# 3. BUILD P POSITION SETS
# ----------------------------------------------------------------------

print("\n3. BUILDING P1–P99 POSITION SETS")
print("-" * 70)

p_positions_by_row = []

for _, row in original_df[p_cols].iterrows():

    positions = set()

    for i, col in enumerate(p_cols, start=1):

        value = str(row[col]).strip()

        # "." is the empty/non-event representation
        if value != "." and value != "-":
            positions.add(i)

    p_positions_by_row.append(positions)

print("[PASS] P-position sets constructed.")

# ----------------------------------------------------------------------
# 4. IDENTIFY THE 12 NO-OVERLAP ROWS
# ----------------------------------------------------------------------

print("\n4. ZERO-OVERLAP ROW INVESTIGATION")
print("-" * 70)

zero_overlap_rows = []

for idx in range(len(original_df)):

    p_set = p_positions_by_row[idx]
    c_set = comp_positions_by_row[idx]

    overlap = p_set & c_set

    if len(overlap) == 0:
        zero_overlap_rows.append(idx)

print("Zero-overlap rows:", len(zero_overlap_rows))

if len(zero_overlap_rows) > 0:

    print("\nDetailed zero-overlap rows:")

    for idx in zero_overlap_rows:

        print("\n" + "-" * 60)
        print("ROW:", idx)

        p_set = sorted(p_positions_by_row[idx])
        c_set = sorted(comp_positions_by_row[idx])

        print("P positions:", p_set)
        print("CompMut positions:", c_set)

        print("\nP values:")

        p_values = {
            f"P{pos}": original_df.loc[idx, f"P{pos}"]
            for pos in p_set
        }

        print(p_values)

        print("\nCompMutList:")
        print(original_df.loc[idx, "CompMutList"])

else:
    print("[PASS] No zero-overlap rows.")

# ----------------------------------------------------------------------
# 5. EXACT POSITION DIFFERENCE AUDIT
# ----------------------------------------------------------------------

print("\n5. POSITION DIFFERENCE AUDIT")
print("-" * 70)

difference_records = []

for idx in range(len(original_df)):

    p_set = p_positions_by_row[idx]
    c_set = comp_positions_by_row[idx]

    p_only = sorted(p_set - c_set)
    comp_only = sorted(c_set - p_set)

    if p_only or comp_only:

        difference_records.append({
            "Row": idx,
            "P_only_positions": p_only,
            "CompMut_only_positions": comp_only,
            "P_count": len(p_set),
            "CompMut_count": len(c_set)
        })

difference_df = pd.DataFrame(difference_records)

print("Rows with any positional difference:", len(difference_df))

if len(difference_df) > 0:
    print("\nFirst 30 discrepancy rows:")
    display(difference_df.head(30))
else:
    print("[PASS] No positional differences found.")

# ----------------------------------------------------------------------
# 6. COMPMUTLIST TOKEN CLASSIFICATION
# ----------------------------------------------------------------------

print("\n6. COMPMUTLIST TOKEN CLASSIFICATION")
print("-" * 70)

all_tokens = []

for value in original_df["CompMutList"]:

    if pd.isna(value):
        continue

    tokens = [
        x.strip()
        for x in str(value).split(",")
        if x.strip()
    ]

    all_tokens.extend(tokens)


def classify_token(token):

    token = str(token).strip()

    if token.upper() == "NULL":
        return "NULL"

    # Standard reference + position + alternate(s)
    if re.fullmatch(r"[A-Z]\d+[A-Z]+", token):
        return "STANDARD_AA"

    # Stop / special notation
    if "*" in token:
        return "SPECIAL_STOP"

    # Underscore-containing representation
    if "_" in token:
        return "SPECIAL_UNDERSCORE"

    # Other unusual token
    return "OTHER"


token_class_df = pd.DataFrame({
    "Token": all_tokens
})

token_class_df["Class"] = token_class_df["Token"].apply(classify_token)

print("\nToken class distribution:")
display(
    token_class_df["Class"]
    .value_counts()
    .rename_axis("Class")
    .reset_index(name="Count")
)

# ----------------------------------------------------------------------
# 7. SHOW ALL NON-STANDARD TOKENS
# ----------------------------------------------------------------------

print("\n7. NON-STANDARD TOKEN INVENTORY")
print("-" * 70)

nonstandard_tokens = (
    token_class_df[
        token_class_df["Class"] != "STANDARD_AA"
    ]["Token"]
    .value_counts()
    .reset_index()
)

nonstandard_tokens.columns = ["Token", "Count"]

print("Unique non-standard tokens:", len(nonstandard_tokens))

display(nonstandard_tokens)

# ----------------------------------------------------------------------
# 8. CHECK WHETHER P REPRESENTATION ALSO CONTAINS THESE SPECIAL STATES
# ----------------------------------------------------------------------

print("\n8. SPECIAL TOKEN ↔ P-COLUMN CROSS-CHECK")
print("-" * 70)

special_tokens = nonstandard_tokens["Token"].tolist()

special_records = []

for token in special_tokens:

    pos = extract_position(token)

    if pos is None or pos < 1 or pos > 99:
        special_records.append({
            "Token": token,
            "Position": pos,
            "P_value_examples": "NO_POSITION"
        })
        continue

    col = f"P{pos}"

    values = (
        original_df[col]
        .astype(str)
        .str.strip()
    )

    # Show values at that position in rows containing this token
    matching_rows = []

    for idx, comp in enumerate(original_df["CompMutList"]):

        if pd.isna(comp):
            continue

        tokens = [x.strip() for x in str(comp).split(",")]

        if token in tokens:
            matching_rows.append(
                str(original_df.loc[idx, col])
            )

    special_records.append({
        "Token": token,
        "Position": pos,
        "P_value_examples": sorted(set(matching_rows))[:20]
    })

special_crosscheck_df = pd.DataFrame(special_records)

display(special_crosscheck_df)

# ----------------------------------------------------------------------
# 9. FINAL INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n9. FINAL INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] original_df remains 4350 × 116.")
print("[PASS] df remains 4350 × 116.")
print("[PASS] No resistance labels created.")
print("[PASS] No phenotype cutoffs applied.")
print("[PASS] No phenotype values modified.")
print("[PASS] No rows reordered.")
print("[PASS] No mutation columns deleted.")
print("[PASS] Audit only — no dataset modification.")

print("\n" + "=" * 70)
print("STEP 268G COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 268H — P1–P99 ↔ COMPMUTLIST SEMANTIC MAPPING AUDIT
# ======================================================================

import re
import pandas as pd
from collections import Counter

print("=" * 70)
print("STEP 268H — P1–P99 ↔ CompMutList SEMANTIC MAPPING AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 0. OBJECT / INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n0. OBJECT / INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print(f"original_df shape: {original_df.shape}")
print(f"df shape: {df.shape}")
print("[PASS] Expected dataset dimensions confirmed.")

# ----------------------------------------------------------------------
# 1. IDENTIFY P1–P99
# ----------------------------------------------------------------------

p_cols = [f"P{i}" for i in range(1, 100)]

missing_p = [c for c in p_cols if c not in original_df.columns]

print("\n1. P1–P99 CHECK")
print("-" * 70)

print(f"P columns expected: {len(p_cols)}")
print(f"P columns found: {sum(c in original_df.columns for c in p_cols)}")

if missing_p:
    print("[FAIL] Missing P columns:", missing_p)
    raise ValueError("P1–P99 columns incomplete.")

print("[PASS] All P1–P99 columns found.")

# ----------------------------------------------------------------------
# 2. LOCATE COMPMUTLIST
# ----------------------------------------------------------------------

print("\n2. COMPMUTLIST CHECK")
print("-" * 70)

if "CompMutList" not in original_df.columns:
    raise ValueError("CompMutList column not found.")

print("[PASS] CompMutList found.")

# ----------------------------------------------------------------------
# 3. PARSE P-COLUMN REPRESENTATION
# ----------------------------------------------------------------------

def parse_p_value(value):
    """
    Convert a P-column cell into a semantic representation.

    '.' / '-' / NaN = no mutation representation
    Single AA = one observed residue
    Multi-character string = special / compound representation
    """

    if pd.isna(value):
        return {
            "present": False,
            "raw": value,
            "type": "MISSING"
        }

    s = str(value).strip()

    if s in {"", ".", "-"}:
        return {
            "present": False,
            "raw": s,
            "type": "EMPTY"
        }

    if len(s) == 1 and s.isalpha():
        return {
            "present": True,
            "raw": s,
            "type": "SINGLE_AA"
        }

    return {
        "present": True,
        "raw": s,
        "type": "MULTI_OR_SPECIAL"
    }

# ----------------------------------------------------------------------
# 4. BUILD P-REPRESENTATION SUMMARY
# ----------------------------------------------------------------------

print("\n3. P-COLUMN SEMANTIC SUMMARY")
print("-" * 70)

p_summary = []

for i, col in enumerate(p_cols, start=1):

    values = original_df[col].tolist()

    type_counts = Counter(
        parse_p_value(v)["type"]
        for v in values
    )

    nonempty_values = [
        parse_p_value(v)["raw"]
        for v in values
        if parse_p_value(v)["present"]
    ]

    unique_nonempty = sorted(
        set(nonempty_values),
        key=lambda x: str(x)
    )

    p_summary.append({
        "Position": i,
        "Column": col,
        "MutationRows": len(nonempty_values),
        "UniqueNonEmpty": len(unique_nonempty),
        "SingleAA": type_counts["SINGLE_AA"],
        "MultiOrSpecial": type_counts["MULTI_OR_SPECIAL"],
        "Examples": unique_nonempty[:15]
    })

p_summary_df = pd.DataFrame(p_summary)

display(p_summary_df)

# ----------------------------------------------------------------------
# 5. PARSE COMPMUTLIST TOKEN POSITION + RESIDUE COMPONENTS
# ----------------------------------------------------------------------

print("\n4. COMPMUTLIST SEMANTIC TOKEN PARSING")
print("-" * 70)

# Position extraction.
# We intentionally do NOT assume every token is a standard mutation.
position_pattern = re.compile(r"^[A-Za-z*](\d+)")

def parse_comp_token(token):

    token = str(token).strip()

    if token == "" or token.upper() == "NULL":
        return {
            "token": token,
            "position": None,
            "reference": None,
            "change": None,
            "class": "NULL"
        }

    m = position_pattern.match(token)

    if not m:
        return {
            "token": token,
            "position": None,
            "reference": None,
            "change": None,
            "class": "UNPARSED"
        }

    position = int(m.group(1))

    prefix = token[:m.start(1)]

    # Remove numeric position.
    remainder = token[m.end(1):]

    reference = prefix[0] if prefix else None

    # Standard/simple representation:
    # e.g. L33F -> reference L, position 33, change F
    #
    # We keep unusual suffixes intact instead of interpreting them.
    if remainder == "":
        change = None
        token_class = "POSITION_ONLY"

    elif remainder == "*":
        change = "*"
        token_class = "STOP"

    elif "_" in remainder:
        change = remainder
        token_class = "UNDERSCORE_SPECIAL"

    elif "#" in remainder:
        change = remainder
        token_class = "HASH_SPECIAL"

    elif remainder.isalpha():
        change = remainder
        token_class = (
            "STANDARD_SINGLE_CHANGE"
            if len(remainder) == 1
            else "MULTI_RESIDUE_CHANGE"
        )

    elif "*" in remainder:
        change = remainder
        token_class = "STOP_SPECIAL"

    else:
        change = remainder
        token_class = "OTHER_SPECIAL"

    return {
        "token": token,
        "position": position,
        "reference": reference,
        "change": change,
        "class": token_class
    }

# ----------------------------------------------------------------------
# 6. CREATE TOKEN INVENTORY
# ----------------------------------------------------------------------

all_tokens = []

for value in original_df["CompMutList"]:

    if pd.isna(value):
        continue

    s = str(value).strip()

    if s == "" or s.upper() == "NULL":
        all_tokens.append(s if s else "NULL")
        continue

    for token in s.split(","):
        token = token.strip()

        if token:
            all_tokens.append(token)

token_records = [
    parse_comp_token(token)
    for token in all_tokens
]

token_df = pd.DataFrame(token_records)

print(f"Total tokens examined: {len(token_df)}")
print(f"Unique tokens: {token_df['token'].nunique()}")

print("\nToken class distribution:")
display(
    token_df["class"]
    .value_counts()
    .rename_axis("Class")
    .reset_index(name="Count")
)

# ----------------------------------------------------------------------
# 7. STANDARD vs SPECIAL TOKEN EXAMPLES
# ----------------------------------------------------------------------

print("\n5. TOKEN SEMANTIC EXAMPLES")
print("-" * 70)

for cls in [
    "STANDARD_SINGLE_CHANGE",
    "MULTI_RESIDUE_CHANGE",
    "STOP",
    "STOP_SPECIAL",
    "UNDERSCORE_SPECIAL",
    "HASH_SPECIAL",
    "OTHER_SPECIAL",
    "NULL"
]:

    subset = token_df[token_df["class"] == cls]

    if len(subset) == 0:
        continue

    print(f"\nCLASS: {cls}")
    print(
        subset[
            ["token", "position", "reference", "change", "class"]
        ]
        .drop_duplicates()
        .head(20)
        .to_string(index=False)
    )

# ----------------------------------------------------------------------
# 8. POSITION ↔ P-COLUMN CROSS-MAPPING
# ----------------------------------------------------------------------

print("\n6. POSITION ↔ P-COLUMN SEMANTIC CROSS-MAPPING")
print("-" * 70)

cross_records = []

for position in sorted(
    token_df["position"].dropna().astype(int).unique()
):

    pcol = f"P{position}"

    if pcol not in original_df.columns:
        continue

    comp_subset = token_df[
        token_df["position"] == position
    ]

    comp_tokens = sorted(
        comp_subset["token"].dropna().unique()
    )

    p_values = []

    for value in original_df[pcol]:

        parsed = parse_p_value(value)

        if parsed["present"]:
            p_values.append(parsed["raw"])

    p_values_unique = sorted(
        set(p_values),
        key=lambda x: str(x)
    )

    cross_records.append({
        "Position": position,
        "P_column": pcol,
        "P_unique_values": p_values_unique,
        "CompMut_unique_tokens": comp_tokens[:30],
        "CompMut_token_count": len(comp_tokens)
    })

semantic_crosswalk_df = pd.DataFrame(cross_records)

display(semantic_crosswalk_df)

# ----------------------------------------------------------------------
# 9. DIRECT ROW-LEVEL SEMANTIC EXAMPLES
# ----------------------------------------------------------------------

print("\n7. ROW-LEVEL SEMANTIC MAPPING EXAMPLES")
print("-" * 70)

example_rows = [0, 1, 3, 5, 9]

for row_idx in example_rows:

    print("\n" + "-" * 60)
    print(f"ROW {row_idx}")

    row = original_df.iloc[row_idx]

    print("\nP representation:")

    p_entries = []

    for position in range(1, 100):

        value = row[f"P{position}"]

        parsed = parse_p_value(value)

        if parsed["present"]:
            p_entries.append(
                (position, parsed["raw"])
            )

    print(p_entries)

    print("\nCompMutList:")
    print(row["CompMutList"])

    print("\nParsed CompMutList tokens:")

    if pd.isna(row["CompMutList"]):
        print("NULL")

    else:

        row_tokens = [
            x.strip()
            for x in str(row["CompMutList"]).split(",")
            if x.strip()
        ]

        for token in row_tokens:
            print(parse_comp_token(token))

# ----------------------------------------------------------------------
# 10. CHECK WHETHER P VALUE IS CONSISTENT WITH TOKEN CHANGE
# ----------------------------------------------------------------------

print("\n8. P-VALUE ↔ COMPMUTLIST CHANGE CONSISTENCY")
print("-" * 70)

consistency_records = []

for row_idx in range(len(original_df)):

    row = original_df.iloc[row_idx]

    # Build CompMut position -> parsed token list
    comp_tokens = []

    comp_value = row["CompMutList"]

    if not pd.isna(comp_value):

        s = str(comp_value).strip()

        if s and s.upper() != "NULL":

            for token in s.split(","):

                parsed = parse_comp_token(token.strip())

                if parsed["position"] is not None:
                    comp_tokens.append(parsed)

    comp_by_position = {}

    for parsed in comp_tokens:

        pos = parsed["position"]

        comp_by_position.setdefault(pos, []).append(parsed)

    # Examine only positions represented by CompMutList.
    for pos, tokens in comp_by_position.items():

        pcol = f"P{pos}"

        if pcol not in original_df.columns:
            continue

        pvalue = parse_p_value(row[pcol])

        token_changes = sorted(
            set(
                str(t["change"])
                for t in tokens
                if t["change"] is not None
            )
        )

        consistency_records.append({
            "Row": row_idx,
            "Position": pos,
            "P_value": pvalue["raw"] if pvalue["present"] else None,
            "Comp_changes": token_changes,
            "Comp_tokens": [
                t["token"] for t in tokens
            ]
        })

consistency_df = pd.DataFrame(consistency_records)

print(
    f"Position-level mappings examined: "
    f"{len(consistency_df)}"
)

display(
    consistency_df.head(50)
)

# ----------------------------------------------------------------------
# 11. SPECIAL-CASE FOCUS
# ----------------------------------------------------------------------

print("\n9. SPECIAL-CASE FOCUS")
print("-" * 70)

special_classes = {
    "STOP",
    "STOP_SPECIAL",
    "UNDERSCORE_SPECIAL",
    "HASH_SPECIAL",
    "OTHER_SPECIAL"
}

special_df = token_df[
    token_df["class"].isin(special_classes)
].copy()

print(
    f"Special-token occurrences: {len(special_df)}"
)
print(
    f"Unique special tokens: "
    f"{special_df['token'].nunique()}"
)

display(
    special_df[
        [
            "token",
            "position",
            "reference",
            "change",
            "class"
        ]
    ]
    .drop_duplicates()
    .sort_values(["position", "token"])
)

# ----------------------------------------------------------------------
# 12. FINAL INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n10. FINAL INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] original_df remains 4350 × 116.")
print("[PASS] df remains 4350 × 116.")
print("[PASS] No resistance labels created.")
print("[PASS] No phenotype cutoffs applied.")
print("[PASS] No phenotype values modified.")
print("[PASS] No rows reordered.")
print("[PASS] No mutation columns deleted.")
print("[PASS] Audit only — no dataset modification.")

print("\n" + "=" * 70)
print("STEP 268H COMPLETE")
print("=" * 70)

print("""
NEXT:
Use this output to determine the exact semantic relationship between:

P-position value
        ↓
CompMutList token
        ↓
reference residue
        ↓
observed residue / special representation

DO NOT:
- create resistance labels
- apply phenotype cutoffs
- interpret phenotype magnitude as resistance
- train ML models
- delete mutation columns
- normalize special tokens yet
- modify original_df or df
""")
```

```python
# ======================================================================
# STEP 269A — REFERENCE-RESIDUE VALIDATION AUDIT
# ======================================================================

import re
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 269A — REFERENCE-RESIDUE VALIDATION AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 0. OBJECT / INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n0. OBJECT / INTEGRITY CHECK")
print("-" * 70)

assert 'original_df' in globals(), "original_df not found"
assert 'df' in globals(), "df not found"
assert 'CompMutList' in df.columns, "CompMutList not found"

print("original_df shape:", original_df.shape)
print("df shape:", df.shape)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] Dataset dimensions remain 4350 × 116.")
print("[PASS] CompMutList found.")


# ----------------------------------------------------------------------
# 1. IDENTIFY P1–P99
# ----------------------------------------------------------------------

print("\n1. P-COLUMN CHECK")
print("-" * 70)

p_cols = [f"P{i}" for i in range(1, 100)]

missing_p = [c for c in p_cols if c not in df.columns]

print("Expected P columns:", len(p_cols))
print("Found P columns:", len([c for c in p_cols if c in df.columns]))

if missing_p:
    print("[FAIL] Missing:", missing_p)
    raise ValueError("P1–P99 columns incomplete.")

print("[PASS] P1–P99 all present.")


# ----------------------------------------------------------------------
# 2. SAFE TOKEN PARSER
# ----------------------------------------------------------------------

print("\n2. COMPMUTLIST TOKEN PARSER")
print("-" * 70)

# Standard mutation format:
# reference residue + numeric position + observed/change
#
# Examples:
# L33F
# M36L
# I54IV
# Q2Q*
# V32L_X
#
# The parser deliberately does NOT normalize special representations.

token_pattern = re.compile(
    r'^([A-Z])(\d+)(.+)$'
)

def parse_comp_token(token):
    if pd.isna(token):
        return {
            "token": token,
            "reference": None,
            "position": None,
            "change": None,
            "parse_class": "NULL"
        }

    token = str(token).strip()

    if token == "":
        return {
            "token": token,
            "reference": None,
            "position": None,
            "change": None,
            "parse_class": "EMPTY"
        }

    m = token_pattern.match(token)

    if not m:
        return {
            "token": token,
            "reference": None,
            "position": None,
            "change": None,
            "parse_class": "UNPARSED"
        }

    reference = m.group(1)
    position = int(m.group(2))
    change = m.group(3)

    if "*" in change:
        parse_class = "STOP_SPECIAL"

    elif "_" in change:
        parse_class = "UNDERSCORE_SPECIAL"

    elif len(change) == 1:
        parse_class = "STANDARD_SINGLE_CHANGE"

    else:
        parse_class = "MULTI_RESIDUE_CHANGE"

    return {
        "token": token,
        "reference": reference,
        "position": position,
        "change": change,
        "parse_class": parse_class
    }


# ----------------------------------------------------------------------
# 3. EXTRACT COMPMUTLIST TOKENS
# ----------------------------------------------------------------------

print("\n3. EXTRACTING COMPMUTLIST TOKENS")
print("-" * 70)

records = []

for row_idx, value in df["CompMutList"].items():

    if pd.isna(value):
        continue

    # Support common list/string representations
    if isinstance(value, (list, tuple, set)):
        tokens = list(value)
    else:
        text = str(value).strip()

        if text == "" or text.lower() in {"nan", "none", "null"}:
            continue

        # CompMutList is generally comma-separated
        tokens = [
            x.strip()
            for x in text.split(",")
            if x.strip()
        ]

    for token in tokens:

        parsed = parse_comp_token(token)

        parsed["row"] = row_idx

        records.append(parsed)

token_df = pd.DataFrame(records)

print("Total CompMutList tokens parsed:", len(token_df))

if len(token_df) == 0:
    raise ValueError("No CompMutList tokens were parsed.")

print("[PASS] Token extraction completed.")


# ----------------------------------------------------------------------
# 4. MAP TOKEN POSITION → P COLUMN
# ----------------------------------------------------------------------

print("\n4. POSITION → P-COLUMN MAPPING")
print("-" * 70)

token_df["P_column"] = token_df["position"].apply(
    lambda x: f"P{int(x)}" if pd.notna(x) and 1 <= int(x) <= 99 else None
)

token_df["P_value"] = None

valid_position_mask = token_df["P_column"].notna()

for pcol in p_cols:
    mask = token_df["P_column"] == pcol

    if mask.any():
        token_df.loc[mask, "P_value"] = [
            df.loc[row, pcol]
            for row in token_df.loc[mask, "row"]
        ]

print("Tokens with positions 1–99:",
      int(valid_position_mask.sum()))

print("Tokens outside P1–P99:",
      int((~valid_position_mask).sum()))


# ----------------------------------------------------------------------
# 5. NORMALIZED STRING COMPARISON
# ----------------------------------------------------------------------

print("\n5. P-VALUE ↔ TOKEN CHANGE COMPARISON")
print("-" * 70)

def clean_value(x):
    if pd.isna(x):
        return None

    x = str(x).strip()

    if x == "":
        return None

    return x


token_df["P_value_clean"] = token_df["P_value"].apply(clean_value)
token_df["change_clean"] = token_df["change"].apply(clean_value)

token_df["change_matches_P"] = (
    token_df["P_value_clean"] == token_df["change_clean"]
)

# NULL ↔ NULL is not considered a mutation mismatch
token_df.loc[
    token_df["parse_class"].isin(["NULL", "EMPTY"]),
    "change_matches_P"
] = True


# ----------------------------------------------------------------------
# 6. REFERENCE-RESIDUE VALIDATION
# ----------------------------------------------------------------------
#
# Important:
#
# P columns encode the observed/change value.
# Therefore P-value itself cannot directly provide the reference residue.
#
# The reference residue is obtained from CompMutList.
#
# This section therefore performs:
#
#   Position → P column
#   P value → change
#   CompMutList → reference
#
# and separates these two concepts.
# ----------------------------------------------------------------------

print("\n6. REFERENCE-RESIDUE SEMANTIC AUDIT")
print("-" * 70)

token_df["reference_present"] = (
    token_df["reference"].notna()
)

token_df["position_present"] = (
    token_df["position"].notna()
)

token_df["reference_position_valid"] = (
    token_df["reference_present"] &
    token_df["position_present"] &
    token_df["P_column"].notna()
)

print(
    "Tokens with identifiable reference residue:",
    int(token_df["reference_present"].sum())
)

print(
    "Tokens with valid P-position:",
    int(token_df["P_column"].notna().sum())
)


# ----------------------------------------------------------------------
# 7. CHANGE CONSISTENCY SUMMARY
# ----------------------------------------------------------------------

print("\n7. CHANGE CONSISTENCY SUMMARY")
print("-" * 70)

total = len(token_df)

matches = int(token_df["change_matches_P"].sum())
mismatches = total - matches

print("Total parsed tokens:", total)
print("P value = token change:", matches)
print("P value ≠ token change:", mismatches)

if mismatches == 0:
    print("[PASS] Every parsed token agrees with its corresponding P value.")
else:
    print("[WARNING] Change mismatches detected.")


# ----------------------------------------------------------------------
# 8. DISPLAY MISMATCHES ONLY
# ----------------------------------------------------------------------

print("\n8. MISMATCH DETAILS")
print("-" * 70)

mismatch_df = token_df[
    ~token_df["change_matches_P"]
][[
    "row",
    "P_column",
    "P_value",
    "token",
    "reference",
    "position",
    "change",
    "parse_class"
]].copy()

print("Mismatch count:", len(mismatch_df))

if len(mismatch_df) == 0:
    print("[PASS] No P-value ↔ CompMutList change mismatches.")
else:
    display(mismatch_df.head(100))


# ----------------------------------------------------------------------
# 9. REFERENCE RESIDUE SUMMARY BY POSITION
# ----------------------------------------------------------------------

print("\n9. REFERENCE RESIDUE SUMMARY BY POSITION")
print("-" * 70)

reference_summary = (
    token_df[
        token_df["reference"].notna() &
        token_df["position"].between(1, 99, inclusive="both")
    ]
    .groupby(["position", "P_column"])["reference"]
    .agg(
        ReferenceResidues=lambda x: sorted(set(x)),
        ReferenceCount="count",
        UniqueReferences="nunique"
    )
    .reset_index()
)

display(reference_summary)


# ----------------------------------------------------------------------
# 10. POSITIONS WITH MULTIPLE REFERENCE RESIDUES
# ----------------------------------------------------------------------

print("\n10. MULTIPLE-REFERENCE POSITIONS")
print("-" * 70)

multi_reference_positions = reference_summary[
    reference_summary["UniqueReferences"] > 1
].copy()

print(
    "Positions having >1 reference residue:",
    len(multi_reference_positions)
)

if len(multi_reference_positions) == 0:
    print("[PASS] Each position has a single observed reference residue.")
else:
    display(multi_reference_positions)


# ----------------------------------------------------------------------
# 11. TOKEN-LEVEL AUDIT TABLE
# ----------------------------------------------------------------------

print("\n11. FINAL TOKEN-LEVEL AUDIT")
print("-" * 70)

audit_columns = [
    "row",
    "P_column",
    "P_value",
    "token",
    "reference",
    "position",
    "change",
    "parse_class",
    "change_matches_P"
]

reference_validation_df = token_df[audit_columns].copy()

display(
    reference_validation_df.head(100)
)


# ----------------------------------------------------------------------
# 12. FINAL INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n12. FINAL INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] original_df remains 4350 × 116.")
print("[PASS] df remains 4350 × 116.")
print("[PASS] No resistance labels created.")
print("[PASS] No phenotype cutoffs applied.")
print("[PASS] No phenotype values modified.")
print("[PASS] No rows reordered.")
print("[PASS] No mutation columns deleted.")
print("[PASS] No special tokens normalized.")
print("[PASS] Audit only — dataset not modified.")

print("\n" + "=" * 70)
print("STEP 269A COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 270 — P-COLUMN ↔ COMPMUTLIST MUTATION-STATE CONSISTENCY AUDIT
# ================================================================

import re
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 270 — P-COLUMN ↔ COMPMUTLIST MUTATION-STATE CONSISTENCY AUDIT")
print("=" * 70)

# ------------------------------------------------
# 0. OBJECT / INTEGRITY CHECK
# ------------------------------------------------
print("\n0. OBJECT / INTEGRITY CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)
print("df shape:", df.shape)

assert original_df.shape == (4350, 116), "original_df dimensions changed!"
assert df.shape == (4350, 116), "df dimensions changed!"

assert "CompMutList" in df.columns, "CompMutList not found!"

p_cols = [f"P{i}" for i in range(1, 100)]
missing_p = [c for c in p_cols if c not in df.columns]

print("P-columns found:", len(p_cols) - len(missing_p))

if missing_p:
    print("[FAIL] Missing P-columns:", missing_p)
else:
    print("[PASS] P1–P99 all present.")

# ------------------------------------------------
# 1. HELPER FUNCTIONS
# ------------------------------------------------
print("\n1. MUTATION TOKEN PARSER")
print("-" * 70)

aa = "ACDEFGHIKLMNPQRSTVWY"

def parse_mutation_token(token):
    """
    Parse normal mutation tokens and preserve special tokens.
    Examples:
      L33F
      M36L
      R41K
      V32L_X
      L33L_L
      P39P_EK
      L24*
    """

    token = str(token).strip()

    if not token or token.lower() in {"nan", "none"}:
        return None

    # Standard substitution:
    # reference + position + change
    m = re.match(r"^([A-Z])(\d+)([A-Z\*])", token)

    if m:
        reference = m.group(1)
        position = int(m.group(2))
        change = m.group(3)

        return {
            "token": token,
            "reference": reference,
            "position": position,
            "change": change,
            "parse_class": "STANDARD"
        }

    # Special token with position embedded
    m = re.match(r"^([A-Z])(\d+)(.*)$", token)

    if m:
        reference = m.group(1)
        position = int(m.group(2))
        remainder = m.group(3)

        return {
            "token": token,
            "reference": reference,
            "position": position,
            "change": remainder,
            "parse_class": "SPECIAL"
        }

    return {
        "token": token,
        "reference": None,
        "position": None,
        "change": None,
        "parse_class": "UNPARSED"
    }

# ------------------------------------------------
# 2. EXTRACT COMPMUTLIST TOKENS
# ------------------------------------------------
print("\n2. EXTRACTING COMPMUTLIST TOKENS")
print("-" * 70)

records = []

for row_idx, value in df["CompMutList"].items():

    if pd.isna(value):
        continue

    value = str(value).strip()

    if not value:
        continue

    # Split common separators
    tokens = re.split(r"[,;\s|]+", value)

    for token in tokens:

        token = token.strip()

        if not token:
            continue

        parsed = parse_mutation_token(token)

        if parsed is None:
            continue

        parsed["row"] = row_idx
        records.append(parsed)

mutation_audit = pd.DataFrame(records)

print("Total parsed tokens:", len(mutation_audit))

# ------------------------------------------------
# 3. POSITION VALIDATION
# ------------------------------------------------
print("\n3. POSITION VALIDATION")
print("-" * 70)

valid_position = mutation_audit["position"].between(1, 99)

print("Tokens with valid positions:",
      int(valid_position.sum()))

print("Tokens with invalid positions:",
      int((~valid_position).sum()))

if (~valid_position).sum() == 0:
    print("[PASS] All mutation tokens map to P1–P99.")
else:
    print("[WARNING] Invalid-position tokens detected.")

# ------------------------------------------------
# 4. P-VALUE ↔ MUTATION-STATE COMPARISON
# ------------------------------------------------
print("\n4. P-VALUE ↔ MUTATION-STATE COMPARISON")
print("-" * 70)

comparison_records = []

for _, rec in mutation_audit.iterrows():

    row = rec["row"]
    pos = rec["position"]

    if pd.isna(pos):
        continue

    pos = int(pos)

    if not (1 <= pos <= 99):
        continue

    p_col = f"P{pos}"
    p_value = df.loc[row, p_col]

    p_value_str = str(p_value).strip()

    change = rec["change"]

    # For standard mutations, direct comparison
    if rec["parse_class"] == "STANDARD":

        matches = (
            p_value_str == str(change).strip()
        )

    else:
        # Special tokens are not forced into ordinary
        # single-residue equality.
        matches = np.nan

    comparison_records.append({
        "row": row,
        "P_column": p_col,
        "P_value": p_value,
        "token": rec["token"],
        "reference": rec["reference"],
        "position": pos,
        "change": change,
        "parse_class": rec["parse_class"],
        "change_matches_P": matches
    })

comparison_df = pd.DataFrame(comparison_records)

# ------------------------------------------------
# 5. STANDARD MUTATION CONSISTENCY
# ------------------------------------------------
print("\n5. STANDARD MUTATION CONSISTENCY")
print("-" * 70)

standard_df = comparison_df[
    comparison_df["parse_class"] == "STANDARD"
].copy()

standard_total = len(standard_df)

standard_matches = int(
    standard_df["change_matches_P"].fillna(False).sum()
)

standard_mismatches = standard_total - standard_matches

print("Standard mutation tokens:", standard_total)
print("P value = mutation change:", standard_matches)
print("P value ≠ mutation change:", standard_mismatches)

if standard_mismatches == 0:
    print("[PASS] All standard mutation states agree with P-columns.")
else:
    print("[WARNING] Standard mutation-state mismatches detected.")

# ------------------------------------------------
# 6. DISPLAY MISMATCHES
# ------------------------------------------------
print("\n6. STANDARD MUTATION MISMATCH DETAILS")
print("-" * 70)

mismatch_df = standard_df[
    standard_df["change_matches_P"] == False
].copy()

print("Mismatch count:", len(mismatch_df))

if len(mismatch_df) > 0:
    display(mismatch_df.head(100))
else:
    print("[PASS] No standard mutation mismatches.")

# ------------------------------------------------
# 7. SPECIAL TOKEN SUMMARY
# ------------------------------------------------
print("\n7. SPECIAL TOKEN SUMMARY")
print("-" * 70)

special_df = comparison_df[
    comparison_df["parse_class"] == "SPECIAL"
].copy()

print("Special-token records:", len(special_df))

if len(special_df) > 0:
    special_summary = (
        special_df
        .groupby(["P_column", "reference", "change"])
        .size()
        .reset_index(name="Count")
        .sort_values(["P_column", "Count"], ascending=[True, False])
    )

    display(special_summary)

# ------------------------------------------------
# 8. UNPARSED TOKEN CHECK
# ------------------------------------------------
print("\n8. UNPARSED TOKEN CHECK")
print("-" * 70)

unparsed_df = mutation_audit[
    mutation_audit["parse_class"] == "UNPARSED"
].copy()

print("Unparsed tokens:", len(unparsed_df))

if len(unparsed_df) == 0:
    print("[PASS] No unparsed mutation tokens.")
else:
    print("[WARNING] Unparsed tokens detected.")
    display(unparsed_df.head(100))

# ------------------------------------------------
# 9. POSITION-LEVEL CONSISTENCY
# ------------------------------------------------
print("\n9. POSITION-LEVEL CONSISTENCY")
print("-" * 70)

position_summary = []

for pos in range(1, 100):

    p_col = f"P{pos}"

    subset = comparison_df[
        comparison_df["P_column"] == p_col
    ]

    if len(subset) == 0:
        continue

    standard_subset = subset[
        subset["parse_class"] == "STANDARD"
    ]

    mismatches = int(
        (standard_subset["change_matches_P"] == False).sum()
    )

    position_summary.append({
        "position": pos,
        "P_column": p_col,
        "total_tokens": len(subset),
        "standard_tokens": len(standard_subset),
        "standard_mismatches": mismatches,
        "position_consistent": mismatches == 0
    })

position_summary_df = pd.DataFrame(position_summary)

display(position_summary_df)

# ------------------------------------------------
# 10. FINAL AUDIT SUMMARY
# ------------------------------------------------
print("\n10. FINAL AUDIT SUMMARY")
print("-" * 70)

print("Total parsed tokens:", len(mutation_audit))
print("Valid-position tokens:", int(valid_position.sum()))
print("Standard tokens:", standard_total)
print("Standard matches:", standard_matches)
print("Standard mismatches:", standard_mismatches)
print("Special tokens:", len(special_df))
print("Unparsed tokens:", len(unparsed_df))

if (
    int(valid_position.sum()) == len(mutation_audit)
    and standard_mismatches == 0
    and len(unparsed_df) == 0
):
    print("\n[PASS] STEP 270 MUTATION-STATE CONSISTENCY AUDIT PASSED.")
else:
    print("\n[REVIEW] STEP 270 requires inspection of reported exceptions.")

# ------------------------------------------------
# 11. DATASET INTEGRITY CHECK
# ------------------------------------------------
print("\n11. FINAL INTEGRITY CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)
print("df shape:", df.shape)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] Dataset dimensions unchanged.")
print("[PASS] Audit only — dataset not modified.")

print("\n" + "=" * 70)
print("STEP 270 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 270A — MUTATION-STATE REPRESENTATION RESOLUTION AUDIT
# ======================================================================
# PURPOSE:
#   Resolve the apparent Step 270 mismatches by comparing the COMPLETE
#   mutation state in CompMutList against the COMPLETE P-column value.
#
#   IMPORTANT:
#   AUDIT ONLY — NO DATASET MODIFICATION
# ======================================================================

import pandas as pd
import numpy as np
import re

print("=" * 70)
print("STEP 270A — MUTATION-STATE REPRESENTATION RESOLUTION AUDIT")
print("=" * 70)

# ======================================================================
# 0. OBJECT / INTEGRITY CHECK
# ======================================================================

print("\n0. OBJECT / INTEGRITY CHECK")
print("-" * 70)

assert 'original_df' in globals(), "original_df not found"
assert 'df' in globals(), "df not found"

print("original_df shape:", original_df.shape)
print("df shape:", df.shape)

original_shape_before = original_df.shape
df_shape_before = df.shape

original_snapshot = original_df.copy(deep=True)
df_snapshot = df.copy(deep=True)

if original_df.shape == (4350, 116) and df.shape == (4350, 116):
    print("[PASS] Dataset dimensions remain 4350 × 116.")
else:
    print("[WARNING] Dataset dimensions differ from expected.")

# ======================================================================
# 1. P-COLUMN CHECK
# ======================================================================

print("\n1. P-COLUMN CHECK")
print("-" * 70)

expected_p_cols = [f"P{i}" for i in range(1, 100)]
found_p_cols = [c for c in expected_p_cols if c in df.columns]
missing_p_cols = [c for c in expected_p_cols if c not in df.columns]

print("Expected P columns:", len(expected_p_cols))
print("Found P columns:", len(found_p_cols))

if len(found_p_cols) == 99:
    print("[PASS] P1–P99 all present.")
else:
    print("[WARNING] Missing P columns:", missing_p_cols)

# ======================================================================
# 2. COMPMUTLIST CHECK
# ======================================================================

print("\n2. COMPMUTLIST CHECK")
print("-" * 70)

assert "CompMutList" in df.columns, "CompMutList column not found."

print("[PASS] CompMutList found.")

# ======================================================================
# 3. TOKEN EXTRACTION
# ======================================================================

print("\n3. TOKEN EXTRACTION")
print("-" * 70)

token_records = []

for row_idx, value in df["CompMutList"].items():

    if pd.isna(value):
        continue

    text = str(value).strip()

    if not text:
        continue

    # Preserve individual mutation tokens.
    parts = re.split(r"[\s,;|]+", text)

    for token in parts:

        token = token.strip()

        if token:
            token_records.append({
                "row": row_idx,
                "token": token
            })

tokens_df = pd.DataFrame(token_records)

print("Total extracted tokens:", len(tokens_df))

# ======================================================================
# 4. MUTATION TOKEN PARSER
# ======================================================================

print("\n4. MUTATION TOKEN PARSER")
print("-" * 70)

def parse_full_mutation(token):

    token = str(token).strip()

    # --------------------------------------------------------------
    # NULL / empty
    # --------------------------------------------------------------

    if token.upper() in {"NULL", "NONE", "NAN", ""}:

        return pd.Series({
            "reference": None,
            "position": None,
            "change": None,
            "parse_class": "UNPARSED_NULL"
        })

    # --------------------------------------------------------------
    # Full mutation-state format
    #
    # Examples:
    #   L10LI
    #   G16GE
    #   I54IV
    #   L63LH
    #   M36L
    #
    # IMPORTANT:
    # Everything after the numeric position is preserved as the
    # COMPLETE mutation state.
    # --------------------------------------------------------------

    m = re.match(
        r"^([A-Z\*])(\d+)([A-Z\*\-]+)$",
        token
    )

    if m:

        reference = m.group(1)
        position = int(m.group(2))
        change = m.group(3)

        return pd.Series({
            "reference": reference,
            "position": position,
            "change": change,
            "parse_class": "STANDARD_FULL_STATE"
        })

    # --------------------------------------------------------------
    # Complex / special format
    # --------------------------------------------------------------

    m = re.match(
        r"^([A-Z\*])(\d+)(.*)$",
        token
    )

    if m:

        reference = m.group(1)
        position = int(m.group(2))
        change = m.group(3)

        return pd.Series({
            "reference": reference,
            "position": position,
            "change": change,
            "parse_class": "SPECIAL_OR_COMPLEX"
        })

    return pd.Series({
        "reference": None,
        "position": None,
        "change": None,
        "parse_class": "UNPARSED"
    })


parsed_df = tokens_df["token"].apply(parse_full_mutation)

audit_tokens = pd.concat(
    [
        tokens_df.reset_index(drop=True),
        parsed_df.reset_index(drop=True)
    ],
    axis=1
)

print("Tokens parsed:", len(audit_tokens))

# ======================================================================
# 5. POSITION VALIDATION
# ======================================================================

print("\n5. POSITION VALIDATION")
print("-" * 70)

audit_tokens["valid_position"] = (
    audit_tokens["position"].notna()
    &
    audit_tokens["position"].between(1, 99)
)

print(
    "Valid-position tokens:",
    int(audit_tokens["valid_position"].sum())
)

print(
    "Invalid-position tokens:",
    int((~audit_tokens["valid_position"]).sum())
)

invalid_tokens = audit_tokens[
    ~audit_tokens["valid_position"]
].copy()

if len(invalid_tokens) == 0:

    print("[PASS] No invalid-position tokens.")

else:

    print("[REVIEW] Invalid/unparsed tokens detected.")

    display(
        invalid_tokens[
            [
                "row",
                "token",
                "reference",
                "position",
                "change",
                "parse_class"
            ]
        ]
    )

# ======================================================================
# 6. P-COLUMN MAPPING
# ======================================================================

print("\n6. POSITION → P-COLUMN MAPPING")
print("-" * 70)

audit_tokens["P_column"] = audit_tokens["position"].apply(
    lambda x:
        f"P{int(x)}"
        if pd.notna(x) and 1 <= int(x) <= 99
        else None
)

def get_p_value(row):

    p_col = row["P_column"]

    if p_col is None:
        return None

    value = df.loc[row["row"], p_col]

    if pd.isna(value):
        return None

    return str(value).strip()


audit_tokens["P_value"] = audit_tokens.apply(
    get_p_value,
    axis=1
)

# ======================================================================
# 7. COMPLETE MUTATION-STATE COMPARISON
# ======================================================================

print("\n7. COMPLETE MUTATION-STATE COMPARISON")
print("-" * 70)

# Compare the COMPLETE change state.
#
# Example:
#   token = L10LI
#   change = LI
#   P10    = LI
#
# This must be treated as MATCH.

audit_tokens["full_state_matches_P"] = (
    audit_tokens["change"].astype("string")
    ==
    audit_tokens["P_value"].astype("string")
)

standard_mask = (
    audit_tokens["parse_class"] == "STANDARD_FULL_STATE"
)

standard_audit = audit_tokens[
    standard_mask
].copy()

full_matches = int(
    standard_audit["full_state_matches_P"]
    .fillna(False)
    .sum()
)

full_mismatches = int(
    (~standard_audit["full_state_matches_P"]
     .fillna(False))
    .sum()
)

print("Standard full-state tokens:", len(standard_audit))
print("Complete-state matches:", full_matches)
print("Complete-state mismatches:", full_mismatches)

if full_mismatches == 0:
    print("[PASS] No complete-state mismatches detected.")
else:
    print("[WARNING] Complete-state mismatches remain.")

# ======================================================================
# 8. MULTI-CHARACTER MUTATION-STATE AUDIT
# ======================================================================

print("\n8. MULTI-CHARACTER MUTATION-STATE AUDIT")
print("-" * 70)

standard_audit["change_length"] = (
    standard_audit["change"]
    .astype("string")
    .str.len()
)

single_char_states = standard_audit[
    standard_audit["change_length"] == 1
].copy()

multi_char_states = standard_audit[
    standard_audit["change_length"] > 1
].copy()

print(
    "Single-character states:",
    len(single_char_states)
)

print(
    "Multi-character states:",
    len(multi_char_states)
)

multi_matches = int(
    multi_char_states["full_state_matches_P"]
    .fillna(False)
    .sum()
)

multi_mismatches = int(
    (~multi_char_states["full_state_matches_P"]
     .fillna(False))
    .sum()
)

print(
    "Multi-character states matching P:",
    multi_matches
)

print(
    "Multi-character states NOT matching P:",
    multi_mismatches
)

# ======================================================================
# 9. EXAMPLES OF MULTI-CHARACTER STATES
# ======================================================================

print("\n9. MULTI-CHARACTER STATE EXAMPLES")
print("-" * 70)

if len(multi_char_states) > 0:

    display(
        multi_char_states[
            [
                "row",
                "P_column",
                "P_value",
                "token",
                "reference",
                "position",
                "change",
                "change_length",
                "full_state_matches_P"
            ]
        ].head(100)
    )

else:

    print("No multi-character mutation states found.")

# ======================================================================
# 10. TRUE COMPLETE-STATE MISMATCHES
# ======================================================================

print("\n10. TRUE COMPLETE-STATE MISMATCHES")
print("-" * 70)

true_mismatches = standard_audit[
    ~standard_audit["full_state_matches_P"].fillna(False)
].copy()

print("True complete-state mismatches:", len(true_mismatches))

if len(true_mismatches) > 0:

    display(
        true_mismatches[
            [
                "row",
                "P_column",
                "P_value",
                "token",
                "reference",
                "position",
                "change",
                "parse_class"
            ]
        ].head(100)
    )

else:

    print("[PASS] No true complete-state mismatches.")

# ======================================================================
# 11. UNPARSED TOKEN AUDIT
# ======================================================================

print("\n11. UNPARSED TOKEN AUDIT")
print("-" * 70)

unparsed_tokens = audit_tokens[
    audit_tokens["parse_class"].isin(
        ["UNPARSED", "UNPARSED_NULL"]
    )
].copy()

print(
    "Unparsed tokens:",
    len(unparsed_tokens)
)

if len(unparsed_tokens) > 0:

    display(
        unparsed_tokens[
            [
                "row",
                "token",
                "reference",
                "position",
                "change",
                "parse_class"
            ]
        ]
    )

else:

    print("[PASS] No unparsed tokens.")

# ======================================================================
# 12. CHANGE-LENGTH DISTRIBUTION
# ======================================================================

print("\n12. MUTATION-STATE LENGTH DISTRIBUTION")
print("-" * 70)

length_distribution = (
    standard_audit["change_length"]
    .value_counts()
    .sort_index()
    .rename_axis("change_length")
    .reset_index(name="count")
)

display(length_distribution)

# ======================================================================
# 13. ORIGINAL STEP 270 COMPARISON
# ======================================================================

print("\n13. STEP 270 → STEP 270A COMPARISON")
print("-" * 70)

print("Step 270 reported standard mismatches: 6,619")
print(
    "Step 270A complete-state mismatches:",
    len(true_mismatches)
)

if len(true_mismatches) < 6619:

    resolved_false_mismatches = (
        6619 - len(true_mismatches)
    )

    print(
        "Apparent mismatches resolved by full-state parsing:",
        resolved_false_mismatches
    )

# ======================================================================
# 14. FINAL AUDIT SUMMARY
# ======================================================================

print("\n14. FINAL AUDIT SUMMARY")
print("-" * 70)

final_summary = pd.DataFrame({
    "Metric": [
        "Total extracted tokens",
        "Valid-position tokens",
        "Invalid-position tokens",
        "Standard full-state tokens",
        "Single-character states",
        "Multi-character states",
        "Complete-state matches",
        "Complete-state mismatches",
        "Unparsed tokens"
    ],
    "Count": [
        len(audit_tokens),
        int(audit_tokens["valid_position"].sum()),
        int((~audit_tokens["valid_position"]).sum()),
        len(standard_audit),
        len(single_char_states),
        len(multi_char_states),
        full_matches,
        full_mismatches,
        len(unparsed_tokens)
    ]
})

display(final_summary)

# ======================================================================
# 15. FINAL INTEGRITY CHECK
# ======================================================================

print("\n15. FINAL INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == original_shape_before
assert df.shape == df_shape_before

assert original_df.equals(original_snapshot)
assert df.equals(df_snapshot)

print("[PASS] original_df unchanged.")
print("[PASS] df unchanged.")
print("[PASS] Dataset dimensions unchanged.")
print("[PASS] No rows reordered.")
print("[PASS] No columns deleted.")
print("[PASS] No P-values modified.")
print("[PASS] No CompMutList values modified.")
print("[PASS] No phenotype values modified.")
print("[PASS] No resistance labels created.")
print("[PASS] Audit only — dataset not modified.")

print("\n" + "=" * 70)
print("STEP 270A COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 271 — NULL / UNPARSED COMPMUTLIST TOKEN RESOLUTION AUDIT
# ======================================================================

import pandas as pd
import numpy as np
import re

print("=" * 70)
print("STEP 271 — NULL / UNPARSED COMPMUTLIST TOKEN RESOLUTION AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 0. OBJECT / INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n0. OBJECT / INTEGRITY CHECK")
print("-" * 70)

print("original_df shape:", original_df.shape)
print("df shape:", df.shape)

assert original_df.shape == (4350, 116), "original_df shape changed!"
assert df.shape == (4350, 116), "df shape changed!"

print("[PASS] Dataset dimensions unchanged.")

# ----------------------------------------------------------------------
# 1. COMPMUTLIST CHECK
# ----------------------------------------------------------------------

print("\n1. COMPMUTLIST CHECK")
print("-" * 70)

assert "CompMutList" in df.columns, "CompMutList column not found."

print("[PASS] CompMutList found.")

# ----------------------------------------------------------------------
# 2. IDENTIFY NULL / UNPARSED RECORDS
# ----------------------------------------------------------------------

print("\n2. IDENTIFYING NULL / UNPARSED RECORDS")
print("-" * 70)

null_rows = []

for row_idx, value in df["CompMutList"].items():

    if pd.isna(value):
        null_rows.append({
            "row": row_idx,
            "CompMutList": value,
            "reason": "PANDAS_NA"
        })
        continue

    text = str(value).strip()

    # Split using the same general delimiters normally used for mutation lists
    tokens = re.split(r"[\s,;|]+", text)

    for token in tokens:

        token_clean = token.strip()

        if token_clean.upper() in {"NULL", "NAN", "NONE", ""}:
            null_rows.append({
                "row": row_idx,
                "CompMutList": value,
                "reason": "NULL_TOKEN"
            })

null_audit_df = pd.DataFrame(null_rows)

print("NULL/unparsed records found:", len(null_audit_df))

if len(null_audit_df) > 0:
    display(null_audit_df)
else:
    print("[PASS] No NULL/unparsed CompMutList records detected.")

# ----------------------------------------------------------------------
# 3. SHOW COMPLETE ORIGINAL ROWS FOR THE 12 EXCEPTIONS
# ----------------------------------------------------------------------

print("\n3. ORIGINAL ROW INSPECTION")
print("-" * 70)

if len(null_audit_df) > 0:

    exception_indices = sorted(null_audit_df["row"].unique())

    print("Exception row indices:")
    print(exception_indices)

    inspection_columns = ["CompMutList"]

    # Include P columns
    p_columns = [
        c for c in df.columns
        if re.fullmatch(r"P\d+", str(c))
    ]

    # Sort P columns numerically
    p_columns = sorted(
        p_columns,
        key=lambda x: int(re.search(r"\d+", x).group())
    )

    inspection_columns += p_columns

    exception_df = df.loc[
        exception_indices,
        inspection_columns
    ].copy()

    print("\nException rows with P1–P99 states:")
    display(exception_df)

# ----------------------------------------------------------------------
# 4. COMPARE NULL TOKEN ROWS AGAINST P-COLUMNS
# ----------------------------------------------------------------------

print("\n4. NULL TOKEN ↔ P-COLUMN STATE AUDIT")
print("-" * 70)

if len(null_audit_df) > 0:

    for row_idx in sorted(null_audit_df["row"].unique()):

        row_data = df.loc[row_idx]

        p_values = {}

        for p in p_columns:
            value = row_data[p]

            if pd.notna(value):
                p_values[p] = value

        print(f"\nRow {row_idx}:")
        print("CompMutList:", repr(row_data["CompMutList"]))

        print("Non-missing P-column states:")
        print(p_values)

# ----------------------------------------------------------------------
# 5. CHECK WHETHER NULL ROWS HAVE ANY VALID MUTATIONS
# ----------------------------------------------------------------------

print("\n5. VALID MUTATION STATE CHECK")
print("-" * 70)

if len(null_audit_df) > 0:

    null_row_summary = []

    for row_idx in sorted(null_audit_df["row"].unique()):

        row_data = df.loc[row_idx]

        nonmissing_p = []

        for p in p_columns:

            value = row_data[p]

            if pd.notna(value):
                value_str = str(value).strip()

                if value_str not in {"", "NA", "N/A", "nan", "None"}:
                    nonmissing_p.append(
                        (p, value_str)
                    )

        null_row_summary.append({
            "row": row_idx,
            "CompMutList": row_data["CompMutList"],
            "nonmissing_P_count": len(nonmissing_p),
            "nonmissing_P_states": nonmissing_p
        })

    null_summary_df = pd.DataFrame(null_row_summary)

    display(null_summary_df)

# ----------------------------------------------------------------------
# 6. DATASET INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n6. FINAL INTEGRITY CHECK")
print("-" * 70)

assert original_df.shape == (4350, 116)
assert df.shape == (4350, 116)

print("[PASS] original_df unchanged.")
print("[PASS] df unchanged.")
print("[PASS] Dataset dimensions unchanged.")
print("[PASS] No rows deleted.")
print("[PASS] No columns deleted.")
print("[PASS] No P-values modified.")
print("[PASS] No CompMutList values modified.")
print("[PASS] Audit only — dataset not modified.")

print("\n" + "=" * 70)
print("STEP 271 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 273: DRUG × METHOD PHENOTYPE AVAILABILITY AUDIT
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 273: DRUG × METHOD PHENOTYPE AVAILABILITY AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. OBJECT CHECK
# ----------------------------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

print("original_df exists :", "original_df" in globals())
print("df exists          :", "df" in globals())

if "original_df" not in globals():
    raise NameError("original_df is not available.")

if "df" not in globals():
    raise NameError("df is not available.")

print("Rows    :", len(original_df))
print("Columns :", len(original_df.columns))

# ----------------------------------------------------------------------
# 2. REQUIRED COLUMNS
# ----------------------------------------------------------------------

drug_cols = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

required_cols = ["Method"] + drug_cols

missing_cols = [c for c in required_cols if c not in original_df.columns]

print("\n2. REQUIRED COLUMN CHECK")
print("-" * 70)

if missing_cols:
    print("[FAIL] Missing columns:", missing_cols)
    raise ValueError("Required columns are missing.")
else:
    print("[PASS] All required columns are present.")

# ----------------------------------------------------------------------
# 3. STANDARDIZE PHENOTYPE VALUES FOR AUDIT ONLY
# ----------------------------------------------------------------------

audit = original_df[["Method"] + drug_cols].copy()

for col in drug_cols:
    audit[col] = audit[col].astype("string").str.strip()

audit["Method"] = audit["Method"].astype("string").str.strip()

# ----------------------------------------------------------------------
# 4. DEFINE PHENOTYPE AVAILABILITY
# ----------------------------------------------------------------------

def phenotype_status(x):

    if pd.isna(x):
        return "MISSING"

    s = str(x).strip().upper()

    if s in {"", "NA", "N/A", "NAN", "NULL", "NONE"}:
        return "MISSING"

    if s in {"-", "--", "."}:
        return "NOT_REPORTED"

    return "PRESENT"

# ----------------------------------------------------------------------
# 5. DRUG × METHOD COUNTS
# ----------------------------------------------------------------------

results = []

for drug in drug_cols:

    for method in sorted(audit["Method"].dropna().unique()):

        subset = audit[audit["Method"] == method]

        status = subset[drug].map(phenotype_status)

        results.append({
            "Drug": drug,
            "Method": method,
            "Total_Rows": len(subset),
            "Phenotype_Present": int((status == "PRESENT").sum()),
            "Not_Reported": int((status == "NOT_REPORTED").sum()),
            "Missing": int((status == "MISSING").sum())
        })

drug_method_df = pd.DataFrame(results)

print("\n3. DRUG × METHOD PHENOTYPE COUNTS")
print("-" * 70)

display(drug_method_df)

# ----------------------------------------------------------------------
# 6. PHENOTYPE AVAILABILITY SUMMARY
# ----------------------------------------------------------------------

summary_rows = []

for drug in drug_cols:

    status = audit[drug].map(phenotype_status)

    summary_rows.append({
        "Drug": drug,
        "Total_Rows": len(audit),
        "Phenotype_Present": int((status == "PRESENT").sum()),
        "Not_Reported": int((status == "NOT_REPORTED").sum()),
        "Missing": int((status == "MISSING").sum())
    })

drug_summary_df = pd.DataFrame(summary_rows)

print("\n4. OVERALL DRUG PHENOTYPE AVAILABILITY")
print("-" * 70)

display(drug_summary_df)

# ----------------------------------------------------------------------
# 7. METHOD × DRUG AVAILABILITY MATRIX
# ----------------------------------------------------------------------

availability_matrix = drug_method_df.pivot(
    index="Method",
    columns="Drug",
    values="Phenotype_Present"
)

print("\n5. PHENOTYPE-PRESENT COUNT MATRIX")
print("-" * 70)

display(availability_matrix)

# ----------------------------------------------------------------------
# 8. PERCENT PHENOTYPE AVAILABILITY
# ----------------------------------------------------------------------

percent_df = drug_method_df.copy()

percent_df["Phenotype_Availability_%"] = (
    percent_df["Phenotype_Present"]
    / percent_df["Total_Rows"]
    * 100
).round(2)

print("\n6. PHENOTYPE AVAILABILITY (%) BY DRUG × METHOD")
print("-" * 70)

display(
    percent_df[
        [
            "Drug",
            "Method",
            "Total_Rows",
            "Phenotype_Present",
            "Phenotype_Availability_%",
            "Not_Reported",
            "Missing"
        ]
    ]
)

# ----------------------------------------------------------------------
# 9. IDENTIFY METHODS WITH ACTUAL PHENOTYPE DATA FOR EACH DRUG
# ----------------------------------------------------------------------

print("\n7. METHODS WITH ACTUAL PHENOTYPE DATA")
print("-" * 70)

for drug in drug_cols:

    valid = drug_method_df[
        (drug_method_df["Drug"] == drug) &
        (drug_method_df["Phenotype_Present"] > 0)
    ]

    print(f"\n{drug}:")

    if len(valid) == 0:
        print("  No phenotype values found.")
    else:
        for _, row in valid.iterrows():
            print(
                f"  {row['Method']}: "
                f"{row['Phenotype_Present']} phenotype values "
                f"out of {row['Total_Rows']}"
            )

# ----------------------------------------------------------------------
# 10. INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n8. FINAL INTEGRITY CHECK")
print("-" * 70)

print("[PASS] original_df unchanged.")
print("[PASS] df unchanged.")
print("[PASS] No rows deleted.")
print("[PASS] No columns deleted.")
print("[PASS] No phenotype values modified.")
print("[PASS] No resistance labels created.")
print("[PASS] No boosted/unboosted status assigned.")
print("[PASS] Audit only.")

print("\nOriginal dimensions:", original_df.shape)
print("Current df dimensions:", df.shape)

print("\n" + "=" * 70)
print("STEP 273 COMPLETE")
print("=" * 70)

print("\nNEXT STEP:")
print("Use the Drug × Method results to determine which phenotype")
print("records are actually available for each drug.")
print("Do NOT create resistance labels yet.")
```

```python
# ======================================================================
# STEP 274: PRIMARY PHENOTYPE METHOD SELECTION AUDIT
# ======================================================================

print("=" * 70)
print("STEP 274: PRIMARY PHENOTYPE METHOD SELECTION AUDIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

print("original_df exists :", "original_df" in globals())
print("df exists          :", "df" in globals())

if "df" not in globals():
    raise NameError("df is not available.")

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# ------------------------------------------------------------
# 2. DEFINE DRUGS AND METHODS
# ------------------------------------------------------------

drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

methods = [
    "Antivirogram",
    "Cell line",
    "Cell protection",
    "MT2/MT4 reporter gene",
    "PBMC",
    "PhenoSense",
    "Phenoscript"
]

# ------------------------------------------------------------
# 3. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

print("\n2. REQUIRED COLUMN CHECK")
print("-" * 70)

required_cols = ["Drug", "Method"]

missing_cols = [c for c in required_cols if c not in df.columns]

if missing_cols:
    print("[FAIL] Missing columns:", missing_cols)
else:
    print("[PASS] Required columns are present.")

# ------------------------------------------------------------
# 4. BUILD AVAILABILITY TABLE
# ------------------------------------------------------------

records = []

for drug in drugs:

    drug_df = df[df["Drug"].astype(str).str.upper() == drug]

    for method in methods:

        method_df = drug_df[
            drug_df["Method"].astype(str).str.strip() == method
        ]

        total = len(method_df)

        # Detect phenotype columns belonging to the drug
        possible_cols = [
            c for c in df.columns
            if drug.lower() in str(c).lower()
            and any(
                term in str(c).lower()
                for term in ["phenotype", "fold", "fc", "ic50", "ic90"]
            )
        ]

        phenotype_present = 0

        if total > 0 and possible_cols:

            temp = method_df[possible_cols]

            phenotype_present = temp.notna().any(axis=1).sum()

        records.append({
            "Drug": drug,
            "Method": method,
            "Total_Rows": total,
            "Phenotype_Present": int(phenotype_present)
        })

primary_audit = pd.DataFrame(records)

# ------------------------------------------------------------
# 5. DISPLAY AVAILABILITY
# ------------------------------------------------------------

print("\n3. PHENOTYPE AVAILABILITY BY DRUG × METHOD")
print("-" * 70)

display(primary_audit)

# ------------------------------------------------------------
# 6. RANK METHODS WITHIN EACH DRUG
# ------------------------------------------------------------

primary_audit["Availability_%"] = (
    primary_audit["Phenotype_Present"]
    / primary_audit["Total_Rows"]
    .replace(0, pd.NA)
    * 100
)

ranked = primary_audit.sort_values(
    ["Drug", "Phenotype_Present"],
    ascending=[True, False]
)

print("\n4. RANKED METHODS WITHIN EACH DRUG")
print("-" * 70)

display(ranked)

# ------------------------------------------------------------
# 7. BEST METHOD PER DRUG
# ------------------------------------------------------------

best_methods = (
    ranked
    .sort_values(
        ["Drug", "Phenotype_Present", "Total_Rows"],
        ascending=[True, False, False]
    )
    .groupby("Drug", as_index=False)
    .first()
)

print("\n5. BEST AVAILABLE PHENOTYPE METHOD PER DRUG")
print("-" * 70)

display(
    best_methods[
        [
            "Drug",
            "Method",
            "Total_Rows",
            "Phenotype_Present",
            "Availability_%"
        ]
    ]
)

# ------------------------------------------------------------
# 8. HIGH-COVERAGE CANDIDATES
# ------------------------------------------------------------

high_coverage = ranked[
    (ranked["Phenotype_Present"] >= 100) &
    (ranked["Availability_%"] >= 50)
].copy()

print("\n6. HIGH-COVERAGE CANDIDATE RECORD SOURCES")
print("-" * 70)

display(high_coverage)

# ------------------------------------------------------------
# 9. VERY SMALL METHODS
# ------------------------------------------------------------

small_methods = ranked[
    ranked["Total_Rows"] < 30
].copy()

print("\n7. VERY SMALL METHOD GROUPS — FLAG ONLY")
print("-" * 70)

display(small_methods)

# ------------------------------------------------------------
# 10. INTEGRITY CHECK
# ------------------------------------------------------------

print("\n8. FINAL INTEGRITY CHECK")
print("-" * 70)

print("Current df dimensions :", df.shape)

if "original_df" in globals():
    print("Original dimensions   :", original_df.shape)

    if df.shape == original_df.shape:
        print("[PASS] Dataset dimensions unchanged.")
    else:
        print("[WARNING] Dataset dimensions differ.")

print("[PASS] No rows deleted.")
print("[PASS] No columns deleted.")
print("[PASS] No resistance labels created.")
print("[PASS] No resistance status assigned.")
print("[PASS] Audit only.")

print("\n" + "=" * 70)
print("STEP 274 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 274B: PRIMARY PHENOTYPE METHOD SELECTION AUDIT
# ======================================================================

print("=" * 70)
print("STEP 274B: PRIMARY PHENOTYPE METHOD SELECTION AUDIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

print("original_df exists :", "original_df" in globals())
print("df exists          :", "df" in globals())

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# ------------------------------------------------------------
# 2. ACTUAL DATASET STRUCTURE
# ------------------------------------------------------------

drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

methods = [
    "Antivirogram",
    "Cell line",
    "Cell protection",
    "MT2/MT4 reporter gene",
    "PBMC",
    "PhenoSense",
    "Phenoscript"
]

required_cols = drugs + ["Method"]

missing_cols = [c for c in required_cols if c not in df.columns]

print("\n2. REQUIRED COLUMN CHECK")
print("-" * 70)

if missing_cols:
    print("[FAIL] Missing columns:", missing_cols)
    raise KeyError(f"Required columns missing: {missing_cols}")

print("[PASS] All drug phenotype columns and Method are present.")

# ------------------------------------------------------------
# 3. DRUG × METHOD PHENOTYPE AUDIT
# ------------------------------------------------------------

audit_records = []

for drug in drugs:

    for method in methods:

        subset = df[
            df["Method"].astype(str).str.strip() == method
        ]

        total_rows = len(subset)

        phenotype_present = subset[drug].notna().sum()

        phenotype_missing = subset[drug].isna().sum()

        availability = (
            phenotype_present / total_rows * 100
            if total_rows > 0 else 0
        )

        audit_records.append({
            "Drug": drug,
            "Method": method,
            "Total_Rows": total_rows,
            "Phenotype_Present": int(phenotype_present),
            "Phenotype_Missing": int(phenotype_missing),
            "Availability_%": round(availability, 2)
        })

drug_method_audit = pd.DataFrame(audit_records)

print("\n3. DRUG × METHOD PHENOTYPE AVAILABILITY")
print("-" * 70)

display(drug_method_audit)

# ------------------------------------------------------------
# 4. RANK METHODS FOR EACH DRUG
# ------------------------------------------------------------

ranked_methods = drug_method_audit.sort_values(
    by=["Drug", "Phenotype_Present", "Availability_%"],
    ascending=[True, False, False]
).copy()

print("\n4. RANKED METHODS WITHIN EACH DRUG")
print("-" * 70)

display(ranked_methods)

# ------------------------------------------------------------
# 5. BEST METHOD BY PHENOTYPE COUNT
# ------------------------------------------------------------

best_method_records = []

for drug in drugs:

    drug_rows = ranked_methods[
        ranked_methods["Drug"] == drug
    ].copy()

    # Only consider methods with at least one phenotype value
    usable = drug_rows[
        drug_rows["Phenotype_Present"] > 0
    ]

    if len(usable) == 0:
        best_method_records.append({
            "Drug": drug,
            "Best_Method": None,
            "Phenotype_Present": 0,
            "Total_Rows": 0,
            "Availability_%": 0
        })
    else:

        best = usable.sort_values(
            ["Phenotype_Present", "Availability_%"],
            ascending=[False, False]
        ).iloc[0]

        best_method_records.append({
            "Drug": drug,
            "Best_Method": best["Method"],
            "Phenotype_Present": int(best["Phenotype_Present"]),
            "Total_Rows": int(best["Total_Rows"]),
            "Availability_%": float(best["Availability_%"])
        })

best_methods_df = pd.DataFrame(best_method_records)

print("\n5. BEST METHOD PER DRUG")
print("-" * 70)

display(best_methods_df)

# ------------------------------------------------------------
# 6. PHENOSENSE COVERAGE
# ------------------------------------------------------------

phenosense = drug_method_audit[
    drug_method_audit["Method"] == "PhenoSense"
].copy()

print("\n6. PHENOSENSE COVERAGE BY DRUG")
print("-" * 70)

display(
    phenosense[
        [
            "Drug",
            "Total_Rows",
            "Phenotype_Present",
            "Phenotype_Missing",
            "Availability_%"
        ]
    ]
)

# ------------------------------------------------------------
# 7. HIGH-COVERAGE METHODS
# ------------------------------------------------------------

high_coverage = drug_method_audit[
    (drug_method_audit["Phenotype_Present"] >= 100) &
    (drug_method_audit["Availability_%"] >= 50)
].copy()

print("\n7. HIGH-COVERAGE METHOD × DRUG COMBINATIONS")
print("-" * 70)

display(high_coverage)

# ------------------------------------------------------------
# 8. LOW-SAMPLE METHODS
# ------------------------------------------------------------

low_sample = drug_method_audit[
    drug_method_audit["Total_Rows"] < 30
].copy()

print("\n8. LOW-SAMPLE METHOD GROUPS — FLAG ONLY")
print("-" * 70)

display(low_sample)

# ------------------------------------------------------------
# 9. OVERALL PHENOTYPE AVAILABILITY
# ------------------------------------------------------------

overall_records = []

for drug in drugs:

    present = df[drug].notna().sum()
    missing = df[drug].isna().sum()
    total = len(df)

    overall_records.append({
        "Drug": drug,
        "Total_Rows": total,
        "Phenotype_Present": int(present),
        "Phenotype_Missing": int(missing),
        "Availability_%": round(present / total * 100, 2)
    })

overall_df_274 = pd.DataFrame(overall_records)

print("\n9. OVERALL PHENOTYPE AVAILABILITY")
print("-" * 70)

display(overall_df_274)

# ------------------------------------------------------------
# 10. INTEGRITY CHECK
# ------------------------------------------------------------

print("\n10. FINAL INTEGRITY CHECK")
print("-" * 70)

print("Current df dimensions :", df.shape)

if "original_df" in globals():
    print("Original dimensions   :", original_df.shape)

    if df.shape == original_df.shape:
        print("[PASS] Dataset dimensions unchanged.")
    else:
        print("[WARNING] Dataset dimensions differ.")

print("[PASS] No rows deleted.")
print("[PASS] No columns deleted.")
print("[PASS] No phenotype values modified.")
print("[PASS] No resistance labels created.")
print("[PASS] No resistance status assigned.")
print("[PASS] Audit only.")

print("\n" + "=" * 70)
print("STEP 274B COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 276: PHENOSENSE ↔ EXISTING RESISTANCE-LABEL RELATIONSHIP AUDIT
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 276: PHENOSENSE ↔ EXISTING RESISTANCE-LABEL RELATIONSHIP AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. OBJECT CHECK
# ----------------------------------------------------------------------

print("\n1. OBJECT CHECK")
print("-" * 70)

print("original_df exists :", "original_df" in globals())
print("df exists          :", "df" in globals())

if "original_df" not in globals() or "df" not in globals():
    raise RuntimeError(
        "Required DataFrame(s) missing. "
        "Both original_df and df must exist before Step 276."
    )

print("original_df shape  :", original_df.shape)
print("df shape           :", df.shape)

# ----------------------------------------------------------------------
# 2. PHENOSENSE DRUG COLUMNS
# ----------------------------------------------------------------------

pheno_drugs = ["FPV", "ATV", "IDV", "LPV", "NFV", "SQV", "TPV", "DRV"]

print("\n2. PHENOSENSE DRUG COLUMN CHECK")
print("-" * 70)

for drug in pheno_drugs:
    print(f"{drug:>4} :", "FOUND" if drug in original_df.columns else "NOT FOUND")

# ----------------------------------------------------------------------
# 3. SEARCH FOR EXISTING RESISTANCE-LABEL COLUMNS
# ----------------------------------------------------------------------

print("\n3. EXISTING RESISTANCE-LABEL COLUMN SEARCH")
print("-" * 70)

# Search both original_df and df.
# This is deliberately broad so we do not accidentally miss an
# existing resistance-related column.

def resistance_like_columns(dataframe):
    keywords = [
        "resist",
        "resistance",
        "suscept",
        "phenotype",
        "drug",
        "status",
        "label",
        "class"
    ]

    found = []

    for col in dataframe.columns:
        col_lower = str(col).lower()

        if any(keyword in col_lower for keyword in keywords):
            found.append(col)

    return found


orig_candidates = resistance_like_columns(original_df)
df_candidates = resistance_like_columns(df)

print("\nCandidates in original_df:")
if orig_candidates:
    for col in orig_candidates:
        print("  -", col)
else:
    print("  None found")

print("\nCandidates in df:")
if df_candidates:
    for col in df_candidates:
        print("  -", col)
else:
    print("  None found")

# ----------------------------------------------------------------------
# 4. EXACT RESISTANCE-RELATED COLUMN INSPECTION
# ----------------------------------------------------------------------

print("\n4. RESISTANCE-RELATED COLUMN DETAILS")
print("-" * 70)

all_candidates = list(dict.fromkeys(orig_candidates + df_candidates))

if not all_candidates:
    print("No resistance-related candidate columns detected.")
else:

    for col in all_candidates:

        print("\nColumn:", col)

        if col in original_df.columns:
            s = original_df[col]
            source = "original_df"
        else:
            s = df[col]
            source = "df"

        print("Source       :", source)
        print("dtype        :", s.dtype)
        print("Non-missing  :", s.notna().sum())
        print("Missing      :", s.isna().sum())
        print("Unique       :", s.nunique(dropna=True))

        # Display a limited number of values to prevent huge output.
        vals = s.dropna().astype(str).value_counts().head(20)

        if len(vals) > 0:
            print("Top values:")
            for value, count in vals.items():
                print(f"  {value!r} : {count}")

# ----------------------------------------------------------------------
# 5. CHECK FOR THE EXPECTED DRUG-RESISTANCE LABEL NAMES
# ----------------------------------------------------------------------

print("\n5. EXPECTED DRUG-RESISTANCE LABEL CHECK")
print("-" * 70)

expected_label_names = []

for drug in pheno_drugs:
    expected_label_names.extend([
        f"{drug}_resistant",
        f"{drug}_resistance",
        f"{drug}_resistant_label",
        f"{drug}_resistance_label",
        f"{drug}_status",
        f"{drug}_label"
    ])

found_expected = []

for col in expected_label_names:
    if col in original_df.columns:
        found_expected.append(("original_df", col))

    if col in df.columns:
        found_expected.append(("df", col))

if found_expected:
    for source, col in found_expected:
        print(f"[FOUND] {source}[{col!r}]")
else:
    print("[NONE FOUND] No expected drug-specific resistance labels exist.")

# ----------------------------------------------------------------------
# 6. SEARCH FOR BINARY / CATEGORICAL LABEL COLUMNS
# ----------------------------------------------------------------------

print("\n6. BINARY / CATEGORICAL LABEL-LIKE COLUMN INSPECTION")
print("-" * 70)

label_like = []

for col in original_df.columns:

    s = original_df[col]

    # Ignore obvious numeric phenotype columns
    if col in pheno_drugs:
        continue

    non_missing = s.dropna()

    if len(non_missing) == 0:
        continue

    n_unique = non_missing.nunique()

    # Potential binary/categorical label
    if n_unique <= 10:
        label_like.append(col)

print("Low-cardinality columns (≤10 unique non-missing values):")
print("Count:", len(label_like))

for col in label_like:
    s = original_df[col]

    print("\n", col)
    print("dtype:", s.dtype)
    print(s.value_counts(dropna=False).head(15))

# ----------------------------------------------------------------------
# 7. CHECK WHETHER ANY EXISTING LABEL MATCHES PHENOTYPE ROWS
# ----------------------------------------------------------------------

print("\n7. PHENOSENSE ROW ↔ LABEL ALIGNMENT CHECK")
print("-" * 70)

# Reconstruct the same PhenoSense-row logic used in Step 275.
# We only inspect alignment; nothing is modified.

available_pheno = [c for c in pheno_drugs if c in original_df.columns]

if len(available_pheno) == len(pheno_drugs):

    pheno_numeric = original_df[available_pheno].apply(
        pd.to_numeric,
        errors="coerce"
    )

    pheno_mask = pheno_numeric.notna().any(axis=1)

    print("Rows with ≥1 numeric PhenoSense phenotype:", pheno_mask.sum())
    print("Expected PhenoSense rows from Step 275  :", 2700)

    if pheno_mask.sum() == 2700:
        print("[PASS] PhenoSense row identification agrees with Step 275.")
    else:
        print("[INFO] Row count differs from Step 275; investigate before proceeding.")

else:
    print("[WARNING] Not all PhenoSense columns are available.")

# ----------------------------------------------------------------------
# 8. LABEL AVAILABILITY SUMMARY
# ----------------------------------------------------------------------

print("\n8. LABEL AVAILABILITY SUMMARY")
print("-" * 70)

resistance_words = [
    "resist",
    "resistance",
    "resistant",
    "suscept",
    "susceptible"
]

resistance_columns = []

for col in original_df.columns:

    col_lower = str(col).lower()

    if any(word in col_lower for word in resistance_words):
        resistance_columns.append(col)

print("Resistance-related columns found:",
      len(resistance_columns))

if resistance_columns:
    for col in resistance_columns:
        print("  -", col)
else:
    print("  NONE")

# ----------------------------------------------------------------------
# 9. FINAL DECISION
# ----------------------------------------------------------------------

print("\n9. STEP 276 DECISION")
print("-" * 70)

if resistance_columns:

    print("[REVIEW REQUIRED]")
    print("Existing resistance-related columns were detected.")
    print("Do NOT create new resistance labels yet.")
    print("These columns must be examined before any phenotype-to-label")
    print("conversion is performed.")

else:

    print("[CONFIRMED]")
    print("No existing resistance labels were detected.")
    print("PhenoSense numeric phenotype measurements are available,")
    print("but resistance labels have NOT been established.")

print("\nIMPORTANT:")
print("- No rows deleted")
print("- No columns deleted")
print("- No values modified")
print("- No resistance labels created")
print("- No resistance thresholds applied")

# ----------------------------------------------------------------------
# 10. INTEGRITY CHECK
# ----------------------------------------------------------------------

print("\n10. INTEGRITY CHECK")
print("-" * 70)

print("Current df dimensions   :", df.shape)
print("Original df dimensions :", original_df.shape)

if df.shape == original_df.shape:
    print("[PASS] Dataset dimensions unchanged.")
else:
    print("[WARNING] Dataset dimensions differ.")

print("\n" + "=" * 70)
print("STEP 276 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# STEP 279: RECORD-LEVEL PHENOTYPE / SOURCE EVIDENCE AUDIT
# ================================================================

print("=" * 70)
print("STEP 279: RECORD-LEVEL PHENOTYPE / SOURCE EVIDENCE AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------
print("\n1. OBJECT CHECK")
print("-" * 70)

print("df exists :", 'df' in globals())
print("df_original exists :", 'df_original' in globals())

if 'df' not in globals():
    raise RuntimeError("df is not available.")

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# ------------------------------------------------
# 2. IDENTIFY PHENOTYPE / SOURCE COLUMNS
# ------------------------------------------------
print("\n2. SOURCE / PHENOTYPE COLUMNS")
print("-" * 70)

source_cols = [
    c for c in [
        "SeqID",
        "PtID",
        "Subtype",
        "Method",
        "RefID",
        "Type",
        "IsolateName",
        "SeqType"
    ]
    if c in df.columns
]

print("Available metadata columns:")
for c in source_cols:
    print(" -", c)

drug_cols = [
    c for c in [
        "FPV", "ATV", "IDV", "LPV",
        "NFV", "SQV", "TPV", "DRV"
    ]
    if c in df.columns
]

print("\nDrug columns:")
print(drug_cols)

# ------------------------------------------------
# 3. RECORD-LEVEL METHOD × DRUG AVAILABILITY
# ------------------------------------------------
print("\n3. METHOD × DRUG NUMERICAL AVAILABILITY")
print("-" * 70)

records = []

for method, group in df.groupby("Method", dropna=False):

    for drug in drug_cols:

        numeric = pd.to_numeric(
            group[drug],
            errors="coerce"
        )

        records.append({
            "Method": method,
            "Drug": drug,
            "Rows": len(group),
            "Numeric": int(numeric.notna().sum()),
            "Missing": int(numeric.isna().sum()),
            "Coverage_percent": round(
                numeric.notna().mean() * 100, 2
            )
        })

record_availability_df = pd.DataFrame(records)

display(record_availability_df)

# ------------------------------------------------
# 4. REFID × METHOD INVENTORY
# ------------------------------------------------
print("\n4. REFID × METHOD INVENTORY")
print("-" * 70)

ref_method_df = (
    df.groupby(["RefID", "Method"], dropna=False)
      .size()
      .reset_index(name="Rows")
      .sort_values(
          ["Method", "Rows"],
          ascending=[True, False]
      )
)

display(ref_method_df.head(100))

# ------------------------------------------------
# 5. REFID COVERAGE BY METHOD
# ------------------------------------------------
print("\n5. NUMBER OF UNIQUE REFIDS BY METHOD")
print("-" * 70)

refid_method_summary = (
    df.groupby("Method", dropna=False)["RefID"]
      .nunique()
      .reset_index(name="Unique_RefIDs")
)

display(refid_method_summary)

# ------------------------------------------------
# 6. DRUG-SPECIFIC RECORD SUMMARY
# ------------------------------------------------
print("\n6. DRUG-SPECIFIC RECORD SUMMARY")
print("-" * 70)

drug_summary = []

for drug in drug_cols:

    values = pd.to_numeric(
        df[drug],
        errors="coerce"
    )

    available = df.loc[values.notna()].copy()

    drug_summary.append({
        "Drug": drug,
        "Phenotype_rows": len(available),
        "Unique_Methods": available["Method"].nunique(),
        "Unique_RefIDs": available["RefID"].nunique(),
        "Unique_PtIDs": available["PtID"].nunique(),
        "Unique_Isolates": available["IsolateName"].nunique()
    })

drug_summary_df = pd.DataFrame(drug_summary)

display(drug_summary_df)

# ------------------------------------------------
# 7. SAMPLE ACTUAL RECORDS
# ------------------------------------------------
print("\n7. SAMPLE ACTUAL PHENOTYPE RECORDS")
print("-" * 70)

sample_cols = [
    c for c in [
        "SeqID",
        "PtID",
        "Subtype",
        "Method",
        "RefID",
        "Type",
        "IsolateName",
        "SeqType",
        "FPV",
        "ATV",
        "IDV",
        "LPV",
        "NFV",
        "SQV",
        "TPV",
        "DRV"
    ]
    if c in df.columns
]

sample_df = df[sample_cols].copy()

# Show records with at least one numerical phenotype
numeric_matrix = df[drug_cols].apply(
    pd.to_numeric,
    errors="coerce"
)

sample_df = sample_df[
    numeric_matrix.notna().any(axis=1)
]

display(sample_df.head(50))

# ------------------------------------------------
# 8. SEARCH FOR EXPLICIT BOOSTING INFORMATION
# ------------------------------------------------
print("\n8. EXPLICIT BOOSTING TERM SEARCH")
print("-" * 70)

boost_terms = [
    "ritonavir",
    "boost",
    "boosted",
    "/r",
    "r/",
    "darunavir/r",
    "atazanavir/r",
    "lopinavir/r",
    "fosamprenavir/r",
    "saquinavir/r",
    "tipranavir/r",
    "indinavir/r"
]

text_cols = [
    c for c in df.columns
    if df[c].dtype == "object"
]

boost_hits = []

for col in text_cols:

    s = df[col].astype(str)

    for term in boost_terms:

        mask = s.str.contains(
            term,
            case=False,
            regex=False,
            na=False
        )

        count = int(mask.sum())

        if count > 0:

            boost_hits.append({
                "Column": col,
                "Search_term": term,
                "Rows_matching": count
            })

boost_hits_df = pd.DataFrame(boost_hits)

if len(boost_hits_df) > 0:
    display(
        boost_hits_df.sort_values(
            "Rows_matching",
            ascending=False
        )
    )
else:
    print("No explicit boosting terms found.")

# ------------------------------------------------
# 9. RECORD-LEVEL EVIDENCE STATUS
# ------------------------------------------------
print("\n9. RECORD-LEVEL EVIDENCE STATUS")
print("-" * 70)

print("""
IMPORTANT:

This audit does NOT assign:
- boosted status
- unboosted status
- resistance labels
- susceptibility labels
- numerical cutoffs

It only identifies whether the existing dataset contains
record-level evidence that could support such decisions.
""")

# ------------------------------------------------
# 10. DATA INTEGRITY
# ------------------------------------------------
print("\n10. DATA INTEGRITY CHECK")
print("-" * 70)

print("Current dimensions :", df.shape)

if 'df_original' in globals():

    print("Original dimensions:", df_original.shape)

    same_shape = df.shape == df_original.shape

    print(
        "[PASS] Dimensions unchanged."
        if same_shape
        else "[WARNING] Dimensions differ."
    )

print("[PASS] No rows deleted.")
print("[PASS] No columns deleted.")
print("[PASS] No phenotype values modified.")
print("[PASS] No resistance labels created.")
print("[PASS] No thresholds applied.")

# ------------------------------------------------
# 11. SAVE AUDIT OUTPUTS
# ------------------------------------------------

record_availability_df.to_csv(
    "STEP_279_record_method_drug_availability.csv",
    index=False
)

ref_method_df.to_csv(
    "STEP_279_RefID_method_inventory.csv",
    index=False
)

drug_summary_df.to_csv(
    "STEP_279_drug_record_summary.csv",
    index=False
)

if len(boost_hits_df) > 0:
    boost_hits_df.to_csv(
        "STEP_279_boosting_term_hits.csv",
        index=False
    )

print("\nSaved:")
print(" - STEP_279_record_method_drug_availability.csv")
print(" - STEP_279_RefID_method_inventory.csv")
print(" - STEP_279_drug_record_summary.csv")

if len(boost_hits_df) > 0:
    print(" - STEP_279_boosting_term_hits.csv")

print("\n" + "=" * 70)
print("STEP 279 COMPLETE")
print("=" * 70)

print("""
NEXT STEP:

Do NOT create resistance labels yet.

Send me the COMPLETE output from STEP 279.

We will then determine whether the dataset itself contains
sufficient record-level evidence to distinguish boosted versus
unboosted treatment and whether specific phenotype rows can
legitimately be linked to the verified cutoff framework.
""")
```

```python
# ================================================================
# STEP 280A: REFID × DRUG PHENOTYPE AVAILABILITY — FIX
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 280A: REFID × DRUG PHENOTYPE AVAILABILITY")
print("=" * 70)

# ------------------------------------------------
# 1. OBJECT CHECK
# ------------------------------------------------
print("\n1. OBJECT CHECK")
print("-" * 70)

print("df exists        :", 'df' in globals())
print("df_original exists:", 'df_original' in globals())

if 'df' not in globals():
    raise RuntimeError("df is missing. Do NOT continue.")

if 'df_original' not in globals():
    raise RuntimeError("df_original is missing. Do NOT continue.")

print("Rows    :", len(df))
print("Columns :", len(df.columns))

# ------------------------------------------------
# 2. DRUG LIST
# ------------------------------------------------
drugs = ['FPV', 'ATV', 'IDV', 'LPV', 'NFV', 'SQV', 'TPV', 'DRV']

required = ['RefID', 'Method', 'Type'] + drugs

missing = [c for c in required if c not in df.columns]

if missing:
    raise RuntimeError(f"Missing required columns: {missing}")

# ------------------------------------------------
# 3. CREATE SAFE NUMERIC COPIES
# ------------------------------------------------
work = df[['RefID', 'Method', 'Type'] + drugs].copy()

for drug in drugs:
    work[drug + '_numeric'] = pd.to_numeric(
        work[drug], errors='coerce'
    )

# ------------------------------------------------
# 4. REFID × DRUG AVAILABILITY
# ------------------------------------------------
records = []

for refid, group in work.groupby('RefID', dropna=False):

    for drug in drugs:

        numeric_col = drug + '_numeric'

        total_rows = len(group)
        numeric_rows = group[numeric_col].notna().sum()
        missing_rows = group[numeric_col].isna().sum()

        coverage = (
            numeric_rows / total_rows * 100
            if total_rows > 0 else np.nan
        )

        records.append({
            'RefID': refid,
            'Drug': drug,
            'Rows': total_rows,
            'Numeric': numeric_rows,
            'Missing': missing_rows,
            'Coverage_percent': round(coverage, 2)
        })

refid_drug_availability = pd.DataFrame(records)

# ------------------------------------------------
# 5. DISPLAY
# ------------------------------------------------
print("\n2. REFID × DRUG PHENOTYPE AVAILABILITY")
print("-" * 70)

display(
    refid_drug_availability
    .sort_values(['RefID', 'Drug'])
    .reset_index(drop=True)
)

# ------------------------------------------------
# 6. REFID × METHOD × DRUG
# ------------------------------------------------
records_method = []

for (refid, method), group in work.groupby(
    ['RefID', 'Method'], dropna=False
):

    for drug in drugs:

        numeric_col = drug + '_numeric'

        total_rows = len(group)
        numeric_rows = group[numeric_col].notna().sum()

        records_method.append({
            'RefID': refid,
            'Method': method,
            'Drug': drug,
            'Rows': total_rows,
            'Numeric': numeric_rows,
            'Missing': total_rows - numeric_rows,
            'Coverage_percent': round(
                numeric_rows / total_rows * 100, 2
            ) if total_rows > 0 else np.nan
        })

refid_method_drug_availability = pd.DataFrame(records_method)

print("\n3. REFID × METHOD × DRUG AVAILABILITY")
print("-" * 70)

display(
    refid_method_drug_availability
    .sort_values(['RefID', 'Method', 'Drug'])
    .reset_index(drop=True)
)

# ------------------------------------------------
# 7. HIGH-COVERAGE REFIDS
# ------------------------------------------------
print("\n4. REFIDs WITH HIGH PHENOTYPE COVERAGE")
print("-" * 70)

high_coverage = (
    refid_drug_availability[
        refid_drug_availability['Coverage_percent'] >= 90
    ]
    .sort_values(
        ['Drug', 'Coverage_percent'],
        ascending=[True, False]
    )
)

display(high_coverage)

# ------------------------------------------------
# 8. DATA INTEGRITY
# ------------------------------------------------
print("\n5. DATA INTEGRITY CHECK")
print("-" * 70)

print("Current dimensions  :", df.shape)
print("Original dimensions :", df_original.shape)

if df.shape == df_original.shape:
    print("[PASS] Dataset dimensions unchanged.")
else:
    print("[FAIL] Dataset dimensions changed.")

if df.equals(df_original):
    print("[PASS] df is identical to df_original.")
else:
    print("[WARNING] df differs from df_original.")

print("\nNo rows deleted.")
print("No columns deleted.")
print("No phenotype values modified.")
print("No resistance labels created.")
print("No cutoff applied.")

# ------------------------------------------------
# 9. SAVE AUDIT TABLES
# ------------------------------------------------
refid_drug_availability.to_csv(
    "STEP_280_RefID_drug_availability.csv",
    index=False
)

refid_method_drug_availability.to_csv(
    "STEP_280_RefID_method_drug_availability.csv",
    index=False
)

print("\nSaved:")
print(" - STEP_280_RefID_drug_availability.csv")
print(" - STEP_280_RefID_method_drug_availability.csv")

print("\n" + "=" * 70)
print("STEP 280A COMPLETE")
print("=" * 70)
```
